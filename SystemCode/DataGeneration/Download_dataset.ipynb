```
import os

url = "https://zenodo.org/records/15719919/files/2025-06-23%20ThePublicJiraDataset.zip?download=1"

output_file = os.path.join(
    SAVE_DIR,
    "2025-06-23_ThePublicJiraDataset.zip"
)

print("Download URL:")
print(url)

print("\nFile will be saved to:")
print(output_file)
```

```
import zipfile
import os

print("File size:",
      round(os.path.getsize(output_file) / (1024**3), 2),
      "GB")

try:
    with zipfile.ZipFile(output_file, "r") as z:
        bad_file = z.testzip()

    if bad_file is None:
        print("SUCCESS: ZIP integrity check passed.")
    else:
        print("ERROR: Corrupted file:", bad_file)

except zipfile.BadZipFile:
    print("ERROR: Not a valid ZIP file.")
```

```
import zipfile

with zipfile.ZipFile(output_file, "r") as z:
    files = z.namelist()

print("Total files:", len(files))

print("\nFirst 50 files:")
for f in files[:50]:
    print(f)
```

```
import zipfile

with zipfile.ZipFile(zip_path, "r") as z:
    files = z.infolist()

print("Total items:", len(files))
print()

for f in files:
    size_mb = f.file_size / (1024**2)
    print(f"{size_mb:10.2f} MB   {f.filename}")
```
