```python
from pathlib import Path

# ============================================================
# GC1 PATH CONFIGURATION
# ============================================================

GC1_ROOT = Path("/content/drive/MyDrive/GC1")

JIRA_ROOT = GC1_ROOT / "GC1_Jira_Dataset"
PRD_MEETING_ROOT = GC1_ROOT / "GC1_Generated_PRD_Meetings"

# New folder for YOUR extraction output
EXTRACTION_ROOT = GC1_ROOT / "GC1_Entity_Extraction"

EXTRACTION_ROOT.mkdir(parents=True, exist_ok=True)

print("GC1 root           :", GC1_ROOT)
print("Jira root          :", JIRA_ROOT)
print("PRD/Meeting root   :", PRD_MEETING_ROOT)
print("Extraction output  :", EXTRACTION_ROOT)

print("\nFolder checks:")
print("Jira exists        :", JIRA_ROOT.exists())
print("PRD/Meeting exists :", PRD_MEETING_ROOT.exists())
print("Output exists      :", EXTRACTION_ROOT.exists())
```

```python
# ============================================================
# STEP 3: VERIFY PROJECT FOLDER ALIGNMENT
# ============================================================
# Purpose:
# Check that the Jira dataset and PRD/Meeting dataset contain
# the same project folders.
#
# Expected:
#   Jira projects        = 30
#   PRD/Meeting projects = 30
#
# Also checks whether any project exists in one dataset but
# not the other.
# ============================================================


# ------------------------------------------------------------
# Get all project folder names from the Jira dataset.
# Only directories are included.
# ------------------------------------------------------------

jira_projects = sorted([
    folder.name
    for folder in JIRA_ROOT.iterdir()
    if folder.is_dir()
])


# ------------------------------------------------------------
# Get all project folder names from the PRD/Meeting dataset.
# ------------------------------------------------------------

text_projects = sorted([
    folder.name
    for folder in PRD_MEETING_ROOT.iterdir()
    if folder.is_dir()
])


# ------------------------------------------------------------
# Display project counts.
# ------------------------------------------------------------

print("Number of Jira project folders       :", len(jira_projects))
print("Number of PRD/Meeting project folders:", len(text_projects))


# ------------------------------------------------------------
# Display Jira project names.
# ------------------------------------------------------------

print("\n================ JIRA PROJECTS ================")

for project in jira_projects:
    print(project)


# ------------------------------------------------------------
# Display PRD/Meeting project names.
# ------------------------------------------------------------

print("\n============= PRD/MEETING PROJECTS =============")

for project in text_projects:
    print(project)


# ------------------------------------------------------------
# Check whether Jira contains projects that do not exist
# in the PRD/Meeting dataset.
# ------------------------------------------------------------

only_in_jira = sorted(
    set(jira_projects) - set(text_projects)
)


# ------------------------------------------------------------
# Check whether PRD/Meeting contains projects that do not exist
# in the Jira dataset.
# ------------------------------------------------------------

only_in_text = sorted(
    set(text_projects) - set(jira_projects)
)


print("\nProjects only in Jira:")
print(only_in_jira)

print("\nProjects only in PRD/Meeting:")
print(only_in_text)


# ------------------------------------------------------------
# Final simple alignment result.
# ------------------------------------------------------------

if len(only_in_jira) == 0 and len(only_in_text) == 0:
    print("\nSUCCESS: Jira and PRD/Meeting project folders match.")
else:
    print("\nWARNING: Project folders do not perfectly match.")
```

```python
# ============================================================
# STEP 4: SELECT ONE PILOT PROJECT
# ============================================================
# Purpose:
# Test the complete Entity Extraction pipeline on ONE project
# before processing all 30 projects.
#
# We use 24_DOCS because we already inspected examples of its
# PRD and meeting transcript.
#
# Once extraction works correctly for this project, the same
# pipeline will be scaled to all 30 projects.
# ============================================================


# ------------------------------------------------------------
# Project selected for development/testing.
# ------------------------------------------------------------

PILOT_PROJECT = "24_DOCS"


# ------------------------------------------------------------
# Construct the Jira folder path for this project.
# ------------------------------------------------------------

jira_folder = JIRA_ROOT / PILOT_PROJECT


# ------------------------------------------------------------
# Construct the PRD/Meeting folder path for this project.
# ------------------------------------------------------------

text_folder = PRD_MEETING_ROOT / PILOT_PROJECT


# ------------------------------------------------------------
# Verify both folders exist.
# ------------------------------------------------------------

print("Pilot project:", PILOT_PROJECT)

print("\nJira folder:")
print(jira_folder)
print("Exists:", jira_folder.exists())

print("\nPRD/Meeting folder:")
print(text_folder)
print("Exists:", text_folder.exists())


# ------------------------------------------------------------
# List ALL files available inside the Jira project folder.
# rglob("*") searches subfolders as well.
# ------------------------------------------------------------

print("\n================ JIRA FILES ================\n")

for file_path in sorted(jira_folder.rglob("*")):

    if file_path.is_file():

        # Convert file size from bytes to KB
        size_kb = file_path.stat().st_size / 1024

        print(
            file_path.relative_to(jira_folder),
            "|",
            round(size_kb, 2),
            "KB"
        )


# ------------------------------------------------------------
# List ALL files available inside the PRD/Meeting project folder.
# ------------------------------------------------------------

print("\n============= PRD / MEETING FILES =============\n")

for file_path in sorted(text_folder.rglob("*")):

    if file_path.is_file():

        size_kb = file_path.stat().st_size / 1024

        print(
            file_path.relative_to(text_folder),
            "|",
            round(size_kb, 2),
            "KB"
        )
```

```python
# ============================================================
# STEP 5: INSPECT ONE RAW JIRA TICKET
# ============================================================
# Purpose:
# Inspect the actual structure of ONE ticket from the curated
# Jira dataset that will be used for Entity Extraction.
#
# IMPORTANT:
# The Jira folder contains multiple representations:
#
#   jira_tickets.jsonl
#       -> Curated 300-ticket dataset.
#       -> This will be our PRIMARY Jira extraction input.
#
#   raw_issues.jsonl
#       -> Raw Jira representation.
#       -> Keep as reference/fallback if a useful field is
#          missing from jira_tickets.jsonl.
#
#   all_300_tickets_full.json
#       -> Full JSON representation of the selected tickets.
#       -> Also useful as a reference/fallback.
#
# For now we inspect jira_tickets.jsonl so that the extraction
# pipeline is designed around its REAL field structure.
# ============================================================

import json


# ------------------------------------------------------------
# Explicitly select the curated Jira file.
# ------------------------------------------------------------

jira_file = jira_folder / "jira_tickets.jsonl"


# ------------------------------------------------------------
# Verify that the expected file exists.
# ------------------------------------------------------------

print("Jira extraction input:")
print(jira_file)

print("\nFile exists:")
print(jira_file.exists())


# ------------------------------------------------------------
# Read the first NON-EMPTY line.
#
# JSONL means:
#   one JSON object per line
#
# Therefore each line should correspond to one Jira ticket.
# ------------------------------------------------------------

first_line = None

with open(jira_file, "r", encoding="utf-8") as file:

    for line in file:

        if line.strip():

            first_line = line
            break


# ------------------------------------------------------------
# Convert the JSON text into a Python dictionary.
# ------------------------------------------------------------

ticket = json.loads(first_line)


# ------------------------------------------------------------
# Show all top-level fields.
#
# This tells us which information is directly available
# before we design the extraction schema.
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("TOP-LEVEL JIRA FIELDS")
print("=" * 70)

for key in ticket.keys():

    value = ticket[key]

    print(
        f"{key:30s}",
        "| Type:",
        type(value).__name__
    )


# ------------------------------------------------------------
# Print the COMPLETE first Jira ticket.
#
# Do not worry if this output is long.
#
# We specifically need to inspect:
#
#   issue ID/key
#   issue type
#   summary
#   description
#   status
#   status history
#   priority
#   assignee
#   reporter
#   created date
#   updated date
#   resolution date
#   due date
#   components
#   labels
#   sprint
#   story points
#   comments
#   issue links
#   blockers
#   dependencies
#   change history
#
# These fields may later contribute to:
#
#   Temporal KG
#   blocker_age
#   dependency_complexity
#   ownership_gaps
#   status_change_frequency
#   issue_age
#   cross-team dependencies
#   scope/change features
#   schedule-related TabPFN features
# ============================================================

print("\n" + "=" * 70)
print("FULL FIRST JIRA RECORD")
print("=" * 70)

print(
    json.dumps(
        ticket,
        indent=2,
        ensure_ascii=False
    )
)
```

```python
# ============================================================
# STEP 6: DISCOVER GENERATED PROJECT INSTANCES - CORRECTED
# ============================================================
# Purpose:
# Identify ONLY actual synthetic project-instance folders.
#
# Example valid instance folders:
#     DOCS-I01
#     DOCS-I02
#
# The repository folder may also contain helper/output folders:
#     meetings/
#     prds/
#
# These are NOT project instances and must not be counted.
#
# We therefore identify instances using the naming convention:
#
#     <PROJECT_KEY>-I<number>
#
# Example:
#     DOCS-I01
#     DOCS-I02
# ============================================================

import re


# ------------------------------------------------------------
# Pattern for a valid generated instance ID.
#
# Examples matched:
#     DOCS-I01
#     DOCS-I02
#     JBEAP-I01
#
# Examples NOT matched:
#     meetings
#     prds
# ------------------------------------------------------------

instance_pattern = re.compile(
    r"^[A-Za-z0-9_]+-I\d+$"
)


# ------------------------------------------------------------
# Find only directories matching the instance naming pattern.
# ------------------------------------------------------------

instance_folders = sorted([
    folder
    for folder in text_folder.iterdir()
    if (
        folder.is_dir()
        and instance_pattern.match(folder.name)
    )
])


print("=" * 70)
print("GENERATED PROJECT INSTANCES")
print("=" * 70)

print("\nRepository project:", PILOT_PROJECT)

print(
    "Number of valid instance folders found:",
    len(instance_folders)
)


# ------------------------------------------------------------
# Display valid instances.
# ------------------------------------------------------------

for folder in instance_folders:

    print("\nInstance:")
    print("  ID   :", folder.name)
    print("  Path :", folder)


# ------------------------------------------------------------
# Also show ignored directories.
#
# This is useful for debugging and documenting why folders
# such as meetings/ and prds/ were not treated as instances.
# ------------------------------------------------------------

ignored_folders = sorted([
    folder.name
    for folder in text_folder.iterdir()
    if (
        folder.is_dir()
        and not instance_pattern.match(folder.name)
    )
])


print("\nIgnored non-instance folders:")
for folder_name in ignored_folders:
    print("  -", folder_name)


# ------------------------------------------------------------
# Select DOCS-I01 as our pilot synthetic instance.
#
# IMPORTANT:
# We develop and validate the complete extraction pipeline
# using ONE instance before scaling to the full dataset.
# ------------------------------------------------------------

PILOT_INSTANCE = "DOCS-I01"

instance_folder = (
    text_folder / PILOT_INSTANCE
)


print("\n" + "=" * 70)
print("SELECTED PILOT INSTANCE")
print("=" * 70)

print("Instance ID :", PILOT_INSTANCE)
print("Folder      :", instance_folder)
print("Exists      :", instance_folder.exists())


# ------------------------------------------------------------
# Safety check:
# Ensure the selected pilot instance was actually discovered
# as a valid instance.
# ------------------------------------------------------------

valid_instance_names = [
    folder.name
    for folder in instance_folders
]


if PILOT_INSTANCE in valid_instance_names:

    print(
        "\nSUCCESS: Pilot instance is valid."
    )

else:

    print(
        "\nWARNING: Pilot instance was not found "
        "among valid instance folders."
    )
```

```python
# ============================================================
# STEP 7: IDENTIFY ACTUAL PRD AND MEETING DOCUMENTS
# ============================================================
# Purpose:
# Locate ONLY the final generated enterprise artefacts.
#
# IMPORTANT:
#
# Files ending in:
#
#     .prompt.txt
#
# are generation prompts.
#
# They are NOT project evidence and must NEVER be included
# in Entity Extraction, the Temporal KG, or TabPFN features.
#
# We only use the final:
#
#     PRD-V1.txt
#     M001.txt
#     M002.txt
#
# Metadata JSON files may be used separately later.
# ============================================================


# ------------------------------------------------------------
# PRD directory
# ------------------------------------------------------------

prd_directory = instance_folder / "prds"


# ------------------------------------------------------------
# Meeting directory
# ------------------------------------------------------------

meeting_directory = instance_folder / "meetings"


# ------------------------------------------------------------
# Find final PRD TXT files.
#
# Explicitly exclude .prompt.txt.
# ------------------------------------------------------------

prd_files = sorted([
    file
    for file in prd_directory.glob("*.txt")
    if not file.name.endswith(".prompt.txt")
])


# ------------------------------------------------------------
# Find final meeting transcript TXT files.
#
# Again explicitly exclude .prompt.txt.
# ------------------------------------------------------------

meeting_files = sorted([
    file
    for file in meeting_directory.glob("*.txt")
    if not file.name.endswith(".prompt.txt")
])


print("=" * 70)
print("ACTUAL EXTRACTION INPUT DOCUMENTS")
print("=" * 70)


print("\nPRD files found:", len(prd_files))

for file in prd_files:
    print("  -", file.name)


print("\nMeeting files found:", len(meeting_files))

for file in meeting_files:
    print("  -", file.name)


# ------------------------------------------------------------
# Also identify metadata files.
#
# We are NOT extracting entities from these.
# However, they may contain useful generation metadata that
# we may preserve as source metadata/provenance.
# ------------------------------------------------------------

prd_meta_files = sorted(
    prd_directory.glob("*.meta.json")
)

meeting_meta_files = sorted(
    meeting_directory.glob("*.meta.json")
)


print("\nPRD metadata files:", len(prd_meta_files))

for file in prd_meta_files:
    print("  -", file.name)


print("\nMeeting metadata files:", len(meeting_meta_files))

for file in meeting_meta_files:
    print("  -", file.name)
```

```python
# ============================================================
# STEP 8: PREVIEW ACTUAL PRD AND MEETING INPUTS
# ============================================================
# Purpose:
# Verify that the selected files contain the expected project
# information before building extraction code.
#
# This is our FINAL raw-data inspection step.
# ============================================================


# ------------------------------------------------------------
# PRD PREVIEW
# ------------------------------------------------------------

print("=" * 70)
print("PRD PREVIEW")
print("=" * 70)


if prd_files:

    with open(
        prd_files[0],
        "r",
        encoding="utf-8"
    ) as file:

        prd_text = file.read()


    print("\nFile:")
    print(prd_files[0].name)

    print("\nLength:")
    print(len(prd_text), "characters")

    print("\nFirst 3000 characters:\n")

    print(prd_text[:3000])


else:

    print("No PRD found.")


# ------------------------------------------------------------
# MEETING PREVIEWS
# ------------------------------------------------------------

print("\n\n" + "=" * 70)
print("MEETING PREVIEWS")
print("=" * 70)


for meeting_file in meeting_files:

    with open(
        meeting_file,
        "r",
        encoding="utf-8"
    ) as file:

        meeting_text = file.read()


    print("\n" + "-" * 70)

    print("File:", meeting_file.name)

    print("Length:", len(meeting_text), "characters")

    print("-" * 70)

    print("\nFirst 2000 characters:\n")

    print(meeting_text[:2000])
```

```python
# ============================================================
# STEP 9: FINAL PILOT INPUT VALIDATION
# ============================================================
# Purpose:
# Confirm that all data needed for the pilot Entity Extraction
# pipeline is available.
#
# Required:
#
#   300 Jira tickets
#   1 PRD
#   2 Meeting transcripts
#
# We also verify that prompt files were excluded.
# ============================================================


# ------------------------------------------------------------
# Count Jira tickets from jira_tickets.jsonl.
# ------------------------------------------------------------

with open(
    jira_file,
    "r",
    encoding="utf-8"
) as file:

    jira_ticket_count = sum(
        1
        for line in file
        if line.strip()
    )


# ------------------------------------------------------------
# Check whether any prompt file accidentally entered our
# selected extraction inputs.
# ------------------------------------------------------------

selected_input_files = (
    prd_files +
    meeting_files
)


prompt_files_included = [
    file
    for file in selected_input_files
    if ".prompt." in file.name
]


# ------------------------------------------------------------
# Display validation report.
# ------------------------------------------------------------

print("=" * 70)
print("GC1 ENTITY EXTRACTION - FINAL PILOT INPUT CHECK")
print("=" * 70)


print("\nRepository project :", PILOT_PROJECT)
print("Synthetic instance :", PILOT_INSTANCE)


print("\n---------------- JIRA ----------------")

print("Input file     :", jira_file.name)
print("Ticket records :", jira_ticket_count)


print("\n---------------- PRD -----------------")

print("PRD files      :", len(prd_files))


print("\n-------------- MEETINGS --------------")

print("Meeting files  :", len(meeting_files))


print("\n--------------- SAFETY ---------------")

print(
    "Prompt files accidentally included:",
    len(prompt_files_included)
)


# ------------------------------------------------------------
# Individual checks.
# ------------------------------------------------------------

jira_ok = jira_ticket_count == 300

prd_ok = len(prd_files) == 1

meetings_ok = len(meeting_files) == 2

prompt_ok = len(prompt_files_included) == 0


print("\n------------- VALIDATION -------------")

print("300 Jira tickets :", jira_ok)
print("1 PRD            :", prd_ok)
print("2 Meetings       :", meetings_ok)
print("No prompt files  :", prompt_ok)


# ------------------------------------------------------------
# Final result.
# ------------------------------------------------------------

if (
    jira_ok
    and prd_ok
    and meetings_ok
    and prompt_ok
):

    print("\nSUCCESS")
    print("Pilot extraction inputs are valid.")

    print(
        "\nNext stage:"
        "\nMASTER EXTRACTION JSON SCHEMA"
    )

else:

    print("\nCHECK REQUIRED")

    print(
        "One or more input validation checks failed."
    )
```

```python
# ============================================================
# STEP 10: DEFINE THE MASTER EXTRACTION JSON SCHEMA
# ============================================================
# Purpose:
# Define ONE common output structure for information extracted
# from Jira, PRDs and Meeting Transcripts.
#
# This schema is the interface between:
#
#     YOUR WORK
#     Data Preparation + Information Extraction
#
#                         ↓
#
#     TEAMMATE'S WORK
#     Entity Resolution + Temporal Knowledge Graph
#
#                         ↓
#
#     LATER WORK
#     Temporal Snapshot Feature Engineering
#                         ↓
#                       TabPFN
#                         ↓
#     Schedule / Scope / Dependency / Decision Risk
#
#
# DESIGN PRINCIPLES
# ------------------------------------------------------------
#
# 1. Preserve facts rather than predicting risk.
#
# 2. Preserve original values whenever possible.
#
# 3. Preserve timestamps so historical project states can
#    later be reconstructed.
#
# 4. Preserve provenance/evidence so GraphRAG can trace
#    knowledge back to Jira, PRDs or Meetings.
#
# 5. Use LOCAL entity IDs only.
#    Entity Resolution will later merge equivalent entities.
#
# 6. Do NOT discard fields simply because they are not
#    immediately needed. They may become useful ML features.
#
# ============================================================


# ------------------------------------------------------------
# ENTITY TYPES
# ------------------------------------------------------------
# These represent the main nodes/concepts that may eventually
# appear in the Temporal Knowledge Graph.
#
# GLiNER will NOT necessarily extract all of them.
#
# Example:
#     ISSUE       -> mostly obtained directly from Jira
#     REQUIREMENT -> mostly parsed from PRD
#     PERSON      -> structured + GLiNER
#     COMPONENT   -> Jira + GLiNER
#
# ------------------------------------------------------------

ENTITY_TYPES = [
    "PROJECT",
    "REQUIREMENT",
    "FEATURE",
    "ISSUE",
    "PERSON",
    "TEAM",
    "MEETING",
    "MILESTONE",
    "COMPONENT",
    "SYSTEM",
    "DECISION",
    "BLOCKER",
    "DEPENDENCY",
    "ACTION_ITEM"
]


# ------------------------------------------------------------
# RELATIONSHIP TYPES
# ------------------------------------------------------------
# These relationships are useful for the Temporal KG.
#
# We start with a controlled vocabulary so that the same
# relationship is not represented using many different names.
#
# More can be added later if the actual data requires them.
# ------------------------------------------------------------

RELATIONSHIP_TYPES = [
    "HAS_REQUIREMENT",
    "HAS_FEATURE",
    "HAS_ISSUE",
    "HAS_COMPONENT",

    "DEPENDS_ON",
    "BLOCKS",
    "BLOCKED_BY",
    "AFFECTS",

    "ASSIGNED_TO",
    "OWNS",
    "REPORTED_BY",
    "CREATED_BY",

    "PARTICIPATED_IN",
    "DISCUSSED_IN",

    "IMPLEMENTS",
    "IMPLEMENTED_BY",

    "RELATED_TO",

    "MADE_DECISION",
    "RAISED",

    "HAS_ACTION_ITEM",

    "CHANGED",
    "REVERSED_BY"
]


# ------------------------------------------------------------
# EVENT TYPES
# ------------------------------------------------------------
# Events capture changes through time.
#
# This is extremely important for the Temporal KG and for
# later TabPFN feature engineering.
#
# Example:
#
#   Open -> Blocked
#
# becomes:
#
#   event_type = STATUS_CHANGE
#   from_value = Open
#   to_value   = Blocked
#
# ------------------------------------------------------------

EVENT_TYPES = [
    "ISSUE_CREATED",
    "STATUS_CHANGE",
    "ASSIGNEE_CHANGE",
    "PRIORITY_CHANGE",
    "RESOLUTION_CHANGE",

    "COMPONENT_CHANGE",
    "VERSION_CHANGE",
    "LINK_CHANGE",

    "SUMMARY_CHANGE",
    "DESCRIPTION_CHANGE",

    "STORY_POINT_CHANGE",

    "BLOCKED",
    "UNBLOCKED",

    "REQUIREMENT_CREATED",
    "REQUIREMENT_CHANGED",

    "SCOPE_CHANGE",

    "DEPENDENCY_ADDED",
    "DEPENDENCY_REMOVED",

    "DECISION_MADE",
    "DECISION_REVERSED",

    "ACTION_CREATED",
    "ACTION_COMPLETED",

    "MILESTONE_CHANGE"
]


# ------------------------------------------------------------
# TABPFN-DOWNSTREAM INFORMATION CHECKLIST
# ------------------------------------------------------------
# IMPORTANT:
#
# These are NOT features calculated during extraction.
#
# This list documents the information that our extraction
# pipeline must preserve so that feature engineering can later
# calculate useful project-risk predictors.
#
# ------------------------------------------------------------

TABPFN_INFORMATION_REQUIREMENTS = {

    # Requirement / scope information
    "requirements": [
        "requirement_id",
        "requirement_text",
        "requirement_version",
        "requirement_owner",
        "requirement_created_time",
        "requirement_updated_time",
        "requirement_changes",
        "requirement_dependencies"
    ],

    # Jira workflow information
    "issues": [
        "issue_id",
        "issue_key",
        "issue_type",
        "status",
        "status_history",
        "priority",
        "resolution",
        "assignee",
        "reporter",
        "creator",
        "created_at",
        "updated_at",
        "due_at",
        "resolved_at"
    ],

    # Blocking information
    "blockers": [
        "blocked_entity",
        "blocker_entity",
        "blocked_at",
        "unblocked_at",
        "blocking_reason",
        "blocking_status"
    ],

    # Dependency information
    "dependencies": [
        "source_entity",
        "target_entity",
        "dependency_type",
        "dependency_status",
        "created_at",
        "removed_at"
    ],

    # Ownership / team information
    "ownership": [
        "assignee",
        "owner",
        "team",
        "assignment_changes",
        "cross_team_relationships"
    ],

    # Decisions
    "decisions": [
        "decision_text",
        "decision_maker",
        "decision_time",
        "decision_target",
        "decision_status",
        "reversed_decision"
    ],

    # Action items
    "actions": [
        "action_text",
        "action_owner",
        "created_at",
        "due_at",
        "completed_at",
        "action_status"
    ],

    # Planning / schedule
    "planning": [
        "milestone",
        "sprint",
        "planned_start",
        "planned_end",
        "due_date",
        "story_points",
        "fix_version",
        "affected_version"
    ],

    # Change history
    "changes": [
        "change_type",
        "changed_at",
        "changed_by",
        "old_value",
        "new_value"
    ],

    # Source traceability
    "provenance": [
        "source_type",
        "source_id",
        "instance_id",
        "chunk_id",
        "evidence_text"
    ]
}


print("=" * 70)
print("MASTER EXTRACTION SCHEMA CONFIGURATION")
print("=" * 70)

print("\nEntity types:")
print(len(ENTITY_TYPES))

for item in ENTITY_TYPES:
    print(" -", item)


print("\nRelationship types:")
print(len(RELATIONSHIP_TYPES))

for item in RELATIONSHIP_TYPES:
    print(" -", item)


print("\nEvent types:")
print(len(EVENT_TYPES))

for item in EVENT_TYPES:
    print(" -", item)


print("\nTabPFN information groups:")
print(len(TABPFN_INFORMATION_REQUIREMENTS))

for group in TABPFN_INFORMATION_REQUIREMENTS:
    print(" -", group)


print("\nSUCCESS: Master extraction vocabulary defined.")
```

```python
# ============================================================
# STEP 11: CREATE MASTER JSON RECORD TEMPLATE
# ============================================================
# Purpose:
# Define the common JSON structure used for Jira, PRD and
# Meeting extraction outputs.
#
# Every source uses the SAME high-level structure.
#
# This makes downstream Entity Resolution and Temporal KG
# construction much easier.
# ============================================================

from copy import deepcopy


def create_master_record(
    repository_project,
    project_key,
    instance_id,
    source_type,
    source_id
):
    """
    Create an empty extraction record.

    Parameters
    ----------
    repository_project:
        GC1 folder/project identifier.
        Example: 24_DOCS

    project_key:
        Original Jira project key.
        Example: DOCS

    instance_id:
        Synthetic project instance.
        Example: DOCS-I01

    source_type:
        JIRA, PRD or MEETING

    source_id:
        Exact source artefact identifier.
        Example:
            DOCS-15004
            DOCS-I01-PRD-V1
            DOCS-I01-M002

    Returns
    -------
    Python dictionary following the common GC1 extraction
    schema.
    """

    record = {

        # ====================================================
        # PROJECT / INSTANCE IDENTITY
        # ====================================================

        "repository_project": repository_project,

        "project_key": project_key,

        "instance_id": instance_id,


        # ====================================================
        # SOURCE INFORMATION
        # ====================================================

        "source": {

            "source_type": source_type,

            "source_id": source_id,

            # Source timestamp when applicable.
            "source_timestamp": None
        },


        # ====================================================
        # ENTITIES
        # ====================================================
        # Each extracted entity receives a LOCAL ID.
        #
        # Example:
        #
        # {
        #   "local_id": "E0001",
        #   "type": "ISSUE",
        #   "name": "DOCS-8052",
        #   "normalized_name": null,
        #   "attributes": {},
        #   "extraction_method": "structured",
        #   "confidence": 1.0
        # }
        #
        # normalized_name remains empty for now because
        # normalization/entity resolution belongs downstream.
        # ====================================================

        "entities": [],


        # ====================================================
        # RELATIONSHIPS
        # ====================================================
        #
        # Example:
        #
        # {
        #   "source_entity": "E0001",
        #   "type": "DEPENDS_ON",
        #   "target_entity": "E0002",
        #   "timestamp": null,
        #   "extraction_method": "structured",
        #   "confidence": 1.0
        # }
        # ====================================================

        "relationships": [],


        # ====================================================
        # EVENTS
        # ====================================================
        # Used for temporal changes.
        #
        # Example:
        #
        # {
        #   "event_type": "STATUS_CHANGE",
        #   "entity": "E0001",
        #   "event_time": "...",
        #   "actor": "...",
        #   "from_value": "Open",
        #   "to_value": "Blocked"
        # }
        # ====================================================

        "events": [],


        # ====================================================
        # DECISIONS
        # ====================================================
        # Mainly extracted from meetings/PRDs.
        #
        # Keep separate because decisions are particularly
        # important for decision-risk modelling.
        # ====================================================

        "decisions": [],


        # ====================================================
        # ACTION ITEMS
        # ====================================================
        # Example:
        #
        # Priya:
        # "I'll chase the owner again on Monday."
        #
        # We want to preserve:
        #   action
        #   owner
        #   created time
        #   due time
        #   completion status
        # ====================================================

        "action_items": [],


        # ====================================================
        # BLOCKERS
        # ====================================================
        # Explicit representation makes later blocker-duration
        # and blocker-count features easier to construct.
        # ====================================================

        "blockers": [],


        # ====================================================
        # DEPENDENCIES
        # ====================================================
        # Explicit dependency representation supports graph
        # construction and dependency-complexity features.
        # ====================================================

        "dependencies": [],


        # ====================================================
        # TEMPORAL INFORMATION
        # ====================================================

        "temporal": {

            "created_at": None,
            "updated_at": None,
            "due_at": None,
            "resolved_at": None,

            "blocked_at": None,
            "unblocked_at": None,

            "valid_from": None,
            "valid_to": None
        },


        # ====================================================
        # SOURCE-SPECIFIC ATTRIBUTES
        # ====================================================
        # This section prevents useful structured information
        # from being discarded merely because it does not fit
        # another section.
        #
        # For Jira this may include:
        #   priority
        #   resolution
        #   components
        #   versions
        #   labels
        #   story points
        #
        # For PRD:
        #   version
        #   status
        #
        # For Meeting:
        #   title
        #   participants
        # ====================================================

        "attributes": {},


        # ====================================================
        # PROVENANCE / EVIDENCE
        # ====================================================
        # Essential for GraphRAG.
        #
        # We preserve where each record came from so that
        # future explanations can trace back to evidence.
        # ====================================================

        "provenance": {

            "source_type": source_type,

            "source_id": source_id,

            "instance_id": instance_id,

            "chunk_id": None,

            "evidence_text": None,

            "file_name": None
        }
    }

    return record
```

```python
# ============================================================
# STEP 13: INSPECT INSTANCE / TIMELINE MANIFESTS
# ============================================================
# Purpose:
# Understand how the synthetic instance DOCS-I01 was generated
# from the underlying Jira project.
#
# WHY THIS MATTERS:
#
# The original Jira dataset contains 300 tickets covering a
# potentially large historical time range.
#
# However, DOCS-I01 represents a specific synthetic project
# instance / time period.
#
# We must NOT automatically attach all 300 Jira tickets to
# DOCS-I01.
#
# Doing that could mix future Jira information with an earlier
# PRD/meeting timeline, which would make:
#
#   - Temporal KG incorrect
#   - temporal snapshots incorrect
#   - TabPFN features incorrect
#   - risk prediction evaluation unreliable
#
# Therefore we inspect the generation manifests BEFORE
# constructing Jira extraction records.
# ============================================================


# ------------------------------------------------------------
# Define manifest paths.
# ------------------------------------------------------------

repository_manifest_file = (
    text_folder /
    "repository_instance_manifest.json"
)

generation_manifest_file = (
    instance_folder /
    "generation_manifest.json"
)

snapshot_manifest_file = (
    instance_folder /
    "snapshot_manifest.json"
)


# ------------------------------------------------------------
# Helper function for loading and displaying JSON.
# ------------------------------------------------------------

def inspect_json_file(file_path, title):
    """
    Load a JSON file and display its complete content.

    This is only an inspection helper.
    It does NOT modify the file.
    """

    print("\n" + "=" * 80)
    print(title)
    print("=" * 80)

    print("\nFile:")
    print(file_path)

    print("\nExists:")
    print(file_path.exists())

    if not file_path.exists():

        print("\nWARNING: File not found.")
        return None


    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as file:

        data = json.load(file)


    print("\nTop-level type:")
    print(type(data).__name__)


    if isinstance(data, dict):

        print("\nTop-level keys:")

        for key in data.keys():
            print(" -", key)


    print("\nComplete content:\n")

    print(
        json.dumps(
            data,
            indent=2,
            ensure_ascii=False
        )
    )

    return data
```

```python
# ============================================================
# STEP 14: INSPECT REPOSITORY-INSTANCE MANIFEST
# ============================================================
# Purpose:
# Understand how the original Jira repository/project was
# divided into synthetic project instances such as:
#
#     DOCS-I01
#     DOCS-I02
#
# We are particularly looking for:
#
#   - instance IDs
#   - Jira ticket IDs/keys
#   - date ranges
#   - snapshot boundaries
#   - source ticket mappings
# ============================================================


repository_manifest = inspect_json_file(

    repository_manifest_file,

    "REPOSITORY INSTANCE MANIFEST"
)
```

```python
# ============================================================
# STEP 15: INSPECT INSTANCE GENERATION MANIFEST
# ============================================================
# Purpose:
# Understand exactly what information was used to generate:
#
#     DOCS-I01 PRD
#     DOCS-I01 Meeting 1
#     DOCS-I01 Meeting 2
#
# This may contain:
#
#   - project information
#   - source Jira ticket keys
#   - generation dates
#   - timeline boundaries
#   - snapshot information
#   - source evidence mappings
#
# We will use this information to preserve correct provenance.
# ============================================================


generation_manifest = inspect_json_file(

    generation_manifest_file,

    "DOCS-I01 GENERATION MANIFEST"
)
```

```python
# ============================================================
# STEP 16: INSPECT INSTANCE SNAPSHOT MANIFEST
# ============================================================
# Purpose:
# Understand the temporal snapshot used for DOCS-I01.
#
# This is especially important for our research because the
# Temporal KG must preserve what was known at a particular
# point in time.
#
# Later, TabPFN features should also be generated using only
# information available at the corresponding project snapshot.
#
# We are looking for fields such as:
#
#   snapshot_date
#   start_date
#   end_date
#   cutoff_date
#   included_ticket_keys
#   selected_ticket_keys
#   source_events
#
# Actual field names may differ. We inspect before assuming.
# ============================================================


snapshot_manifest = inspect_json_file(

    snapshot_manifest_file,

    "DOCS-I01 SNAPSHOT MANIFEST"
)
```

```python
# ============================================================
# STEP 17: BUILD THE ALIGNED JIRA INSTANCE
# ============================================================
# Purpose:
# Identify Jira tickets that contain activity relevant to the
# DOCS-I01 time window.
#
# IMPORTANT:
# We do NOT filter only by ticket creation date.
#
# A ticket may have been created before the instance started
# but still have:
#
#   - status changes
#   - comments
#   - assignments
#   - link changes
#   - resolution
#   - other changelog events
#
# during the DOCS-I01 timeline.
#
# Such tickets must still be included.
#
# Instance:
#   DOCS-I01
#
# Timeline:
#   2016-07-05 -> 2017-03-13
#
# The manifest tells us that the expected instance contains
# approximately:
#
#   167 unique tickets
#   1821 Jira events
#
# We use those ONLY for validation.
# They are NOT extraction features.
# ============================================================

from datetime import datetime


# ------------------------------------------------------------
# Read timeline directly from snapshot manifest.
#
# This avoids hard-coding dates.
# ------------------------------------------------------------

INSTANCE_START = datetime.fromisoformat(
    snapshot_manifest["instance_start"]
)

INSTANCE_END = datetime.fromisoformat(
    snapshot_manifest["instance_end"]
)


print("Instance:", PILOT_INSTANCE)
print("Start   :", INSTANCE_START)
print("End     :", INSTANCE_END)
```

```python
# ============================================================
# STEP 18: TIMESTAMP NORMALIZATION HELPER
# ============================================================
# Purpose:
# Jira timestamps and manifest timestamps may use slightly
# different ISO formats.
#
# This helper converts them into Python datetime objects so
# that temporal comparisons are reliable.
#
# It returns None for missing/unparseable timestamps rather
# than crashing the complete extraction pipeline.
# ============================================================

from datetime import datetime


def parse_timestamp(value):

    # Missing timestamp
    if value is None:
        return None

    # Already converted
    if isinstance(value, datetime):
        return value

    # Convert to string and remove surrounding spaces
    value = str(value).strip()

    if not value:
        return None

    try:

        # ----------------------------------------------------
        # Jira often uses:
        #
        # +0000
        #
        # Python ISO parsing works more consistently with:
        #
        # +00:00
        # ----------------------------------------------------

        if (
            len(value) >= 5
            and value[-5] in ["+", "-"]
            and value[-3] != ":"
        ):

            value = (
                value[:-2]
                + ":"
                + value[-2:]
            )

        return datetime.fromisoformat(value)

    except Exception:

        return None


# ------------------------------------------------------------
# Quick test
# ------------------------------------------------------------

test_dates = [
    "2021-12-29T00:48:12.000+0000",
    "2016-07-05T00:00:00+00:00",
    None
]


print("Timestamp parser test:\n")

for value in test_dates:

    print(
        value,
        "->",
        parse_timestamp(value)
    )
```

```python
# ============================================================
# STEP 19: LOAD ALL 300 CURATED JIRA TICKETS
# ============================================================
# Purpose:
# Load jira_tickets.jsonl into memory so that we can identify
# which tickets contain activity inside DOCS-I01.
#
# Each line of the JSONL file represents one Jira ticket.
# ============================================================


all_jira_tickets = []


with open(
    jira_file,
    "r",
    encoding="utf-8"
) as file:

    for line in file:

        if not line.strip():
            continue

        ticket = json.loads(line)

        all_jira_tickets.append(ticket)


print("=" * 70)
print("JIRA DATASET LOADED")
print("=" * 70)

print("\nTickets loaded:", len(all_jira_tickets))


if len(all_jira_tickets) == 300:

    print("SUCCESS: All 300 Jira tickets loaded.")

else:

    print(
        "WARNING: Expected 300 tickets but found",
        len(all_jira_tickets)
    )
```

```python
# ============================================================
# STEP 20: COLLECT TEMPORAL ACTIVITY FOR EACH JIRA TICKET
# ============================================================
# Purpose:
# Collect timestamps representing actual Jira activity.
#
# Sources examined:
#
#   1. Ticket creation
#   2. Status history
#   3. Comments
#   4. Full changelog
#   5. Resolution
#
# We deliberately DO NOT use updated_at by itself for instance
# membership because it only gives the most recent update and
# may occur long after the period we are studying.
# ============================================================


def collect_ticket_activity_times(ticket):

    activity_times = []


    # --------------------------------------------------------
    # Ticket creation
    # --------------------------------------------------------

    created = parse_timestamp(
        ticket.get("created_at")
    )

    if created:
        activity_times.append(created)


    # --------------------------------------------------------
    # Status history
    # --------------------------------------------------------

    for status_event in (
        ticket.get("status_history") or []
    ):

        timestamp = parse_timestamp(
            status_event.get("changed_at")
        )

        if timestamp:
            activity_times.append(timestamp)


    # --------------------------------------------------------
    # Comments
    # --------------------------------------------------------

    for comment in (
        ticket.get("comments") or []
    ):

        timestamp = parse_timestamp(
            comment.get("created_at")
        )

        if timestamp:
            activity_times.append(timestamp)


    # --------------------------------------------------------
    # Complete Jira changelog
    #
    # This is particularly valuable because it contains:
    #
    #   assignment changes
    #   priority changes
    #   links
    #   components
    #   versions
    #   story points
    #   status changes
    #   etc.
    # --------------------------------------------------------

    for history in (
        ticket.get("full_changelog") or []
    ):

        timestamp = parse_timestamp(
            history.get("created")
        )

        if timestamp:
            activity_times.append(timestamp)


    # --------------------------------------------------------
    # Resolution
    # --------------------------------------------------------

    resolved = parse_timestamp(
        ticket.get("resolved_at")
    )

    if resolved:
        activity_times.append(resolved)


    # Remove duplicates and sort chronologically
    activity_times = sorted(
        set(activity_times)
    )


    return activity_times
```

```python
# ============================================================
# STEP 21: SELECT JIRA TICKETS ACTIVE IN DOCS-I01
# ============================================================
# Purpose:
# Include a ticket if at least ONE genuine Jira activity event
# occurred during the instance time window.
#
# This is better than filtering only on ticket creation date.
# ============================================================


instance_jira_tickets = []


for ticket in all_jira_tickets:

    activity_times = (
        collect_ticket_activity_times(ticket)
    )


    # --------------------------------------------------------
    # Keep activity occurring inside the instance window.
    # --------------------------------------------------------

    activity_in_window = [

        timestamp

        for timestamp in activity_times

        if (
            INSTANCE_START
            <= timestamp
            <= INSTANCE_END
        )
    ]


    # --------------------------------------------------------
    # Include ticket if at least one event occurred.
    # --------------------------------------------------------

    if activity_in_window:

        instance_jira_tickets.append(
            {
                "ticket": ticket,
                "activity_times":
                    activity_in_window
            }
        )


print("=" * 70)
print("DOCS-I01 JIRA ALIGNMENT")
print("=" * 70)

print(
    "\nRepository tickets:",
    len(all_jira_tickets)
)

print(
    "Tickets with activity in instance:",
    len(instance_jira_tickets)
)

print(
    "Manifest expected unique tickets:",
    generation_manifest[
        "instance_ticket_count"
    ]
)
```

```python
# ============================================================
# STEP 22: VALIDATE INSTANCE TICKET ALIGNMENT
# ============================================================
# Purpose:
# Compare our independently reconstructed Jira instance with
# the hidden generation metadata.
#
# IMPORTANT:
#
# The manifest is used ONLY here as a validation check.
#
# It will NOT become:
#
#   - extraction evidence
#   - KG data
#   - TabPFN features
#
# ============================================================


actual_ticket_count = len(
    instance_jira_tickets
)

expected_ticket_count = (
    generation_manifest[
        "instance_ticket_count"
    ]
)


print("=" * 70)
print("INSTANCE ALIGNMENT VALIDATION")
print("=" * 70)


print(
    "\nCalculated active tickets:",
    actual_ticket_count
)

print(
    "Expected manifest tickets:",
    expected_ticket_count
)


difference = (
    actual_ticket_count
    - expected_ticket_count
)


print(
    "Difference:",
    difference
)


if actual_ticket_count == expected_ticket_count:

    print(
        "\nSUCCESS:"
        "\nOur Jira temporal filtering exactly reproduces "
        "the generated instance."
    )

else:

    print(
        "\nNOT YET EXACT:"
        "\nThe instance-generation logic uses additional "
        "selection criteria."
    )

    print(
        "\nDo NOT continue to extraction until we inspect "
        "this difference."
    )
```

```python
# ============================================================
# STEP 23: CREATE SOURCE-SCOPED LOCAL ENTITY IDs
# ============================================================
# Purpose:
# Every entity extracted from Jira, PRD or Meeting needs a
# temporary unique identifier so that relationships/events can
# refer to it unambiguously.
#
# Examples:
#
#   JIRA_DOCS-8052_E001
#   PRD_DOCS-I01-PRD-V1_E001
#   MEETING_DOCS-I01-M002_E001
#
# IMPORTANT:
# These are NOT final Knowledge Graph IDs.
#
# They are temporary extraction IDs only.
#
# The next Entity Resolution stage will determine whether
# entities from different sources refer to the same real
# entity and assign/construct canonical identities.
#
# These IDs are also NOT TabPFN predictive features.
# ============================================================


def make_local_entity_id(
    source_type,
    source_id,
    index
):
    """
    Create a source-scoped temporary entity ID.

    Parameters
    ----------
    source_type:
        JIRA, PRD or MEETING

    source_id:
        Exact source artefact ID.
        Example: DOCS-8052

    index:
        Entity number inside that source record.

    Returns
    -------
    Example:
        JIRA_DOCS-8052_E001
    """

    # Replace spaces with underscores just in case
    safe_source_type = str(
        source_type
    ).replace(" ", "_")

    safe_source_id = str(
        source_id
    ).replace(" ", "_")

    return (
        f"{safe_source_type}_"
        f"{safe_source_id}_"
        f"E{index:03d}"
    )


# ------------------------------------------------------------
# Quick tests
# ------------------------------------------------------------

print(
    make_local_entity_id(
        "JIRA",
        "DOCS-8052",
        1
    )
)

print(
    make_local_entity_id(
        "PRD",
        "DOCS-I01-PRD-V1",
        5
    )
)

print(
    make_local_entity_id(
        "MEETING",
        "DOCS-I01-M002",
        12
    )
)
```

```python
# ============================================================
# STEP 24: ENTITY CREATION HELPER
# ============================================================
# Purpose:
# Add an entity to an extraction record.
#
# The function:
#
#   1. Checks whether the same entity already exists locally.
#   2. Reuses its ID if it already exists.
#   3. Otherwise creates a new source-scoped local ID.
#
# Example:
#
# If DOCS-8052 appears five times in the same source record,
# we should not create five separate ISSUE entities.
#
# NOTE:
# This is ONLY local deduplication.
#
# Cross-document/cross-source matching belongs to the
# Entity Resolution stage.
# ============================================================


def add_entity(
    record,
    entity_type,
    name,
    attributes=None,
    extraction_method="structured",
    confidence=1.0
):

    # --------------------------------------------------------
    # Ignore missing/empty names.
    # --------------------------------------------------------

    if name is None:
        return None

    name = str(name).strip()

    if not name:
        return None


    if attributes is None:
        attributes = {}


    # --------------------------------------------------------
    # Check whether this exact type + name already exists
    # inside the current source record.
    # --------------------------------------------------------

    for entity in record["entities"]:

        if (
            entity["type"] == entity_type
            and entity["name"] == name
        ):

            return entity["local_id"]


    # --------------------------------------------------------
    # Obtain source information from the master record.
    # --------------------------------------------------------

    source_type = record[
        "source"
    ]["source_type"]

    source_id = record[
        "source"
    ]["source_id"]


    # --------------------------------------------------------
    # Create unique source-scoped temporary ID.
    # --------------------------------------------------------

    local_id = make_local_entity_id(

        source_type,

        source_id,

        len(record["entities"]) + 1
    )


    # --------------------------------------------------------
    # Construct entity object.
    # --------------------------------------------------------

    entity = {

        "local_id":
            local_id,

        "type":
            entity_type,

        "name":
            name,

        # Entity Resolution will handle normalization later.
        "normalized_name":
            None,

        "attributes":
            attributes,

        # Tells us how this entity was obtained.
        #
        # Examples:
        #   structured
        #   regex
        #   gliner
        #   rule
        "extraction_method":
            extraction_method,

        # Structured Jira information gets confidence 1.0.
        # GLiNER entities later will retain model confidence.
        "confidence":
            confidence
    }


    record["entities"].append(
        entity
    )


    return local_id
```

```python
# ============================================================
# STEP 25: RELATIONSHIP CREATION HELPER
# ============================================================
# Purpose:
# Represent connections between extracted entities.
#
# Examples:
#
#   PROJECT --HAS_ISSUE--> ISSUE
#
#   ISSUE --ASSIGNED_TO--> PERSON
#
#   ISSUE --DEPENDS_ON--> ISSUE
#
# These relationships will later become candidate edges in the
# Temporal Knowledge Graph.
# ============================================================


def add_relationship(
    record,
    source_entity,
    relationship_type,
    target_entity,
    timestamp=None,
    extraction_method="structured",
    confidence=1.0,
    attributes=None
):

    # Cannot create relationship without both ends.
    if (
        source_entity is None
        or target_entity is None
    ):
        return


    if attributes is None:
        attributes = {}


    relationship = {

        "source_entity":
            source_entity,

        "type":
            relationship_type,

        "target_entity":
            target_entity,

        "timestamp":
            timestamp,

        "attributes":
            attributes,

        "extraction_method":
            extraction_method,

        "confidence":
            confidence
    }


    # --------------------------------------------------------
    # Avoid exact duplicate relationships within this record.
    # --------------------------------------------------------

    if relationship not in record[
        "relationships"
    ]:

        record[
            "relationships"
        ].append(
            relationship
        )
```

```python
# ============================================================
# STEP 26: TEMPORAL EVENT CREATION HELPER
# ============================================================
# Purpose:
# Represent changes/events with timestamps.
#
# Example:
#
#   Open -> Blocked
#
# becomes:
#
#   STATUS_CHANGE
#   from_value = Open
#   to_value   = Blocked
#   event_time = timestamp
#
# Later this can support features such as:
#
#   blocker duration
#   number of status changes
#   assignment changes
#   priority changes
#   issue activity
#   workflow volatility
#   time to resolution
#
# IMPORTANT:
# We preserve the raw event here.
# We DO NOT calculate TabPFN features yet.
# ============================================================


def add_event(
    record,
    event_type,
    entity,
    event_time,
    actor=None,
    from_value=None,
    to_value=None,
    attributes=None,
    extraction_method="structured"
):

    if attributes is None:
        attributes = {}


    event = {

        "event_type":
            event_type,

        "entity":
            entity,

        "event_time":
            event_time,

        "actor":
            actor,

        "from_value":
            from_value,

        "to_value":
            to_value,

        "attributes":
            attributes,

        "extraction_method":
            extraction_method
    }


    record["events"].append(
        event
    )
```

```python
# ============================================================
# STEP 27: MAP JIRA CHANGELOG FIELDS TO EVENT TYPES
# ============================================================
# Purpose:
# Jira projects may use slightly different fields/capitalization.
#
# We map them into a consistent event vocabulary that the
# Temporal KG and feature-engineering stages can understand.
# ============================================================


JIRA_CHANGE_EVENT_MAP = {

    "status":
        "STATUS_CHANGE",

    "assignee":
        "ASSIGNEE_CHANGE",

    "priority":
        "PRIORITY_CHANGE",

    "resolution":
        "RESOLUTION_CHANGE",

    "component":
        "COMPONENT_CHANGE",

    "fix version":
        "VERSION_CHANGE",

    "version":
        "VERSION_CHANGE",

    "affected version":
        "VERSION_CHANGE",

    "link":
        "LINK_CHANGE",

    "summary":
        "SUMMARY_CHANGE",

    "description":
        "DESCRIPTION_CHANGE",

    "story points":
        "STORY_POINT_CHANGE"
}


print("=" * 60)
print("JIRA CHANGELOG EVENT MAPPING")
print("=" * 60)


for field_name, event_type in (
    JIRA_CHANGE_EVENT_MAP.items()
):

    print(
        f"{field_name:20s} -> {event_type}"
    )
```

```python
# ============================================================
# STEP 28: STRUCTURED JIRA EXTRACTION FUNCTION
# ============================================================
# Purpose:
# Convert ONE Jira ticket into the common Master Extraction
# JSON structure.
#
# We directly preserve reliable Jira information including:
#
#   Project
#   Issue
#   People
#   Components
#   Versions
#   Relationships
#   Dependencies
#   Blockers
#   Status changes
#   Changelog events
#   Comments
#   Temporal information
#   Provenance
#
# NO GLiNER is used here.
#
# GLiNER will later be used mainly for unstructured text in:
#
#   PRDs
#   Meeting transcripts
#   selected Jira text where useful
# ============================================================


def extract_jira_ticket(
    ticket,
    repository_project,
    instance_id,
    instance_start,
    instance_end,
    source_file_name
):

    # ========================================================
    # A. CREATE EMPTY MASTER RECORD
    # ========================================================

    record = create_master_record(

        repository_project=
            repository_project,

        project_key=
            ticket.get("project_key"),

        instance_id=
            instance_id,

        source_type=
            "JIRA",

        source_id=
            ticket.get("issue_key")
    )


    # ========================================================
    # B. PROJECT ENTITY
    # ========================================================

    project_entity = add_entity(

        record,

        "PROJECT",

        ticket.get("project_key"),

        attributes={

            "project_id":
                ticket.get("project_id"),

            "project_name":
                ticket.get("project_name"),

            "source_repository":
                ticket.get("source_repository")
        }
    )


    # ========================================================
    # C. ISSUE ENTITY
    # ========================================================

    issue_entity = add_entity(

        record,

        "ISSUE",

        ticket.get("issue_key"),

        attributes={

            "issue_id":
                ticket.get("issue_id"),

            "issue_type":
                ticket.get("issue_type"),

            "summary":
                ticket.get("summary"),

            "description":
                ticket.get("description")
        }
    )


    # Project -> Issue relationship
    add_relationship(

        record,

        project_entity,

        "HAS_ISSUE",

        issue_entity
    )


    # ========================================================
    # D. PEOPLE
    # ========================================================

    person_fields = [

        ("assignee", "ASSIGNED_TO"),

        ("reporter", "REPORTED_BY"),

        ("creator", "CREATED_BY")
    ]


    for field_name, relation_type in (
        person_fields
    ):

        person_name = ticket.get(
            field_name
        )


        if person_name:

            person_entity = add_entity(

                record,

                "PERSON",

                person_name
            )


            add_relationship(

                record,

                issue_entity,

                relation_type,

                person_entity
            )


    # ========================================================
    # E. COMPONENTS
    # ========================================================

    for component in (
        ticket.get("components") or []
    ):

        component_name = (
            component.get("name")
        )


        component_entity = add_entity(

            record,

            "COMPONENT",

            component_name,

            attributes={

                "component_id":
                    component.get("id")
            }
        )


        add_relationship(

            record,

            issue_entity,

            "HAS_COMPONENT",

            component_entity
        )


    # ========================================================
    # F. ISSUE RELATIONSHIPS
    # ========================================================

    relation_sources = [

        ("blocks", "BLOCKS"),

        ("blocked_by", "BLOCKED_BY"),

        ("depends_on", "DEPENDS_ON"),

        ("other_issue_links", "RELATED_TO")
    ]


    for field_name, relation_type in (
        relation_sources
    ):

        for linked in (
            ticket.get(field_name) or []
        ):


            # ------------------------------------------------
            # Links may be dictionaries or simple strings.
            # ------------------------------------------------

            if isinstance(linked, dict):

                linked_key = linked.get(
                    "issue_key"
                )

            else:

                linked_key = linked


            if not linked_key:
                continue


            linked_entity = add_entity(

                record,

                "ISSUE",

                linked_key
            )


            add_relationship(

                record,

                issue_entity,

                relation_type,

                linked_entity,

                attributes=(
                    linked
                    if isinstance(linked, dict)
                    else {}
                )
            )


            # ------------------------------------------------
            # Keep explicit dependency representation.
            #
            # This makes later dependency feature generation
            # easier without having to reconstruct everything
            # from generic relationships.
            # ------------------------------------------------

            if relation_type in [

                "BLOCKS",
                "BLOCKED_BY",
                "DEPENDS_ON"

            ]:

                record[
                    "dependencies"
                ].append({

                    "source_entity":
                        issue_entity,

                    "target_entity":
                        linked_entity,

                    "dependency_type":
                        relation_type,

                    "dependency_status":
                        None,

                    "created_at":
                        None,

                    "removed_at":
                        None
                })


    # ========================================================
    # G. TEMPORAL FIELDS
    # ========================================================

    record["temporal"] = {

        "created_at":
            ticket.get("created_at"),

        "updated_at":
            ticket.get("updated_at"),

        "due_at":
            ticket.get("due_at"),

        "resolved_at":
            ticket.get("resolved_at"),

        "blocked_at":
            ticket.get("blocked_at"),

        "unblocked_at":
            ticket.get("unblocked_at"),

        # These represent the selected project-instance window.
        "valid_from":
            instance_start.isoformat(),

        "valid_to":
            instance_end.isoformat()
    }


    # ========================================================
    # H. BLOCKER INFORMATION
    # ========================================================

    if ticket.get("blocked_at"):

        record["blockers"].append({

            "blocked_entity":
                issue_entity,

            "blocker_entity":
                None,

            "blocked_at":
                ticket.get("blocked_at"),

            "unblocked_at":
                ticket.get("unblocked_at"),

            "blocking_reason":
                None,

            "blocking_status":
                (
                    "RESOLVED"
                    if ticket.get("unblocked_at")
                    else "ACTIVE"
                )
        })


    # ========================================================
    # I. STATUS HISTORY
    # ========================================================
    #
    # This is temporarily preserved here.
    #
    # The same status changes may also exist inside the full
    # changelog. We will detect/deduplicate that after testing.
    # ========================================================

    for status_event in (
        ticket.get("status_history") or []
    ):

        event_time = parse_timestamp(

            status_event.get(
                "changed_at"
            )
        )


        if (
            event_time
            and instance_start
            <= event_time
            <= instance_end
        ):

            add_event(

                record,

                event_type=
                    "STATUS_CHANGE",

                entity=
                    issue_entity,

                event_time=
                    status_event.get(
                        "changed_at"
                    ),

                actor=
                    status_event.get(
                        "changed_by"
                    ),

                from_value=
                    status_event.get(
                        "from_status"
                    ),

                to_value=
                    status_event.get(
                        "to_status"
                    ),

                attributes={
                    "source_structure":
                        "status_history"
                }
            )


    # ========================================================
    # J. FULL CHANGELOG
    # ========================================================

    for history in (
        ticket.get("full_changelog") or []
    ):

        history_time = parse_timestamp(
            history.get("created")
        )


        # Only preserve events belonging to DOCS-I01.
        if not (
            history_time
            and instance_start
            <= history_time
            <= instance_end
        ):

            continue


        for item in (
            history.get("items") or []
        ):

            field_name = str(
                item.get("field") or ""
            ).strip()


            mapped_event_type = (
                JIRA_CHANGE_EVENT_MAP.get(

                    field_name.lower(),

                    "OTHER_CHANGE"
                )
            )


            author = (
                history
                .get("author", {})
                .get("displayName")
            )


            add_event(

                record,

                event_type=
                    mapped_event_type,

                entity=
                    issue_entity,

                event_time=
                    history.get("created"),

                actor=
                    author,

                from_value=
                    item.get("fromString"),

                to_value=
                    item.get("toString"),

                attributes={

                    "jira_field":
                        field_name,

                    "field_type":
                        item.get(
                            "fieldtype"
                        ),

                    "source_structure":
                        "full_changelog"
                }
            )


    # ========================================================
    # K. COMMENTS INSIDE INSTANCE WINDOW
    # ========================================================

    comments_in_window = []


    for comment in (
        ticket.get("comments") or []
    ):

        comment_time = parse_timestamp(

            comment.get(
                "created_at"
            )
        )


        if (
            comment_time
            and instance_start
            <= comment_time
            <= instance_end
        ):

            comments_in_window.append(
                comment
            )


    # ========================================================
    # L. STRUCTURED ATTRIBUTES
    # ========================================================
    #
    # Keep useful fields even if we do not immediately use
    # them. This helps prevent information loss before the
    # Temporal KG / TabPFN stages.
    # ========================================================

    record["attributes"] = {

        "issue_type":
            ticket.get("issue_type"),

        "summary":
            ticket.get("summary"),

        "description":
            ticket.get("description"),

        "related_requirement":
            ticket.get(
                "related_requirement"
            ),

        "epic_feature":
            ticket.get(
                "epic_feature"
            ),

        "current_status":
            ticket.get("status"),

        "priority":
            ticket.get("priority"),

        "resolution":
            ticket.get("resolution"),

        "team":
            ticket.get("team"),

        "labels":
            ticket.get("labels") or [],

        "fix_versions":
            ticket.get(
                "fix_versions"
            ) or [],

        "affected_versions":
            ticket.get(
                "affected_versions"
            ) or [],

        "comments":
            comments_in_window
    }


    # ========================================================
    # M. PROVENANCE
    # ========================================================
    #
    # GraphRAG will eventually need to trace extracted
    # information back to source evidence.
    # ========================================================

    evidence_parts = []


    if ticket.get("summary"):

        evidence_parts.append(
            ticket["summary"]
        )


    if ticket.get("description"):

        evidence_parts.append(
            ticket["description"]
        )


    record["provenance"] = {

        "source_type":
            "JIRA",

        "source_id":
            ticket.get("issue_key"),

        "instance_id":
            instance_id,

        "chunk_id":
            None,

        "evidence_text":
            "\n".join(
                evidence_parts
            ),

        "file_name":
            source_file_name
    }


    return record
```

```python
# ============================================================
# STEP 29: TEST STRUCTURED EXTRACTION ON ONE JIRA TICKET
# ============================================================
# Purpose:
# Test the complete structured Jira extractor on ONE ticket
# before scaling to all 167 tickets in DOCS-I01.
#
# This protects us from propagating extraction bugs across the
# complete dataset.
# ============================================================


sample_instance_ticket = (
    instance_jira_tickets[0]["ticket"]
)


sample_jira_record = extract_jira_ticket(

    ticket=
        sample_instance_ticket,

    repository_project=
        PILOT_PROJECT,

    instance_id=
        PILOT_INSTANCE,

    instance_start=
        INSTANCE_START,

    instance_end=
        INSTANCE_END,

    source_file_name=
        jira_file.name
)


print("=" * 70)
print("SAMPLE EXTRACTED JIRA RECORD")
print("=" * 70)


print(
    json.dumps(
        sample_jira_record,
        indent=2,
        ensure_ascii=False
    )
)
```

```python
# ============================================================
# STEP 30: SUMMARIZE SAMPLE JIRA EXTRACTION
# ============================================================
# Purpose:
# Produce a compact summary of the extraction result.
#
# We specifically inspect:
#
#   entity types
#   local entity IDs
#   relationships
#   event counts
#   blockers
#   dependencies
#
# We will also check whether STATUS_CHANGE events are duplicated
# because Jira provides both:
#
#   status_history
#
# and
#
#   full_changelog
#
# We must remove duplicate temporal events before later
# TabPFN feature engineering.
# ============================================================


print("=" * 70)
print("JIRA EXTRACTION TEST SUMMARY")
print("=" * 70)


# ------------------------------------------------------------
# Source
# ------------------------------------------------------------

print(
    "\nSource:",
    sample_jira_record[
        "source"
    ]["source_id"]
)


# ------------------------------------------------------------
# Basic counts
# ------------------------------------------------------------

print(
    "\nEntities:",
    len(
        sample_jira_record[
            "entities"
        ]
    )
)


print(
    "Relationships:",
    len(
        sample_jira_record[
            "relationships"
        ]
    )
)


print(
    "Events:",
    len(
        sample_jira_record[
            "events"
        ]
    )
)


print(
    "Blockers:",
    len(
        sample_jira_record[
            "blockers"
        ]
    )
)


print(
    "Dependencies:",
    len(
        sample_jira_record[
            "dependencies"
        ]
    )
)


# ------------------------------------------------------------
# Entities
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("ENTITIES")
print("-" * 70)


for entity in sample_jira_record[
    "entities"
]:

    print(
        entity["local_id"],
        "|",
        entity["type"],
        "|",
        entity["name"]
    )


# ------------------------------------------------------------
# Relationships
# ------------------------------------------------------------

print("\n" + "-" * 70)
print("RELATIONSHIPS")
print("-" * 70)


for relationship in sample_jira_record[
    "relationships"
]:

    print(
        relationship[
            "source_entity"
        ],
        "--",
        relationship[
            "type"
        ],
        "-->",
        relationship[
            "target_entity"
        ]
    )


# ------------------------------------------------------------
# Event counts
# ------------------------------------------------------------

event_counts = {}


for event in sample_jira_record[
    "events"
]:

    event_type = event[
        "event_type"
    ]

    event_counts[event_type] = (
        event_counts.get(
            event_type,
            0
        )
        + 1
    )


print("\n" + "-" * 70)
print("EVENT COUNTS")
print("-" * 70)


for event_type, count in sorted(
    event_counts.items()
):

    print(
        f"{event_type:25s}",
        count
    )


# ------------------------------------------------------------
# Identify possible duplicate STATUS_CHANGE events.
#
# We compare:
#   event time
#   old status
#   new status
#
# If the same transition appears in both status_history and
# full_changelog, we will remove one copy in the next step.
# ------------------------------------------------------------

status_event_keys = {}


for event in sample_jira_record[
    "events"
]:

    if event[
        "event_type"
    ] != "STATUS_CHANGE":

        continue


    key = (

        str(event["event_time"]),

        str(event["from_value"]),

        str(event["to_value"])
    )


    status_event_keys.setdefault(
        key,
        []
    ).append(
        event
    )


duplicate_status_groups = {

    key: events

    for key, events
    in status_event_keys.items()

    if len(events) > 1
}


print("\n" + "-" * 70)
print("DUPLICATE STATUS CHECK")
print("-" * 70)


print(
    "Duplicate status transition groups:",
    len(duplicate_status_groups)
)


for key, events in (
    duplicate_status_groups.items()
):

    print("\nDuplicate transition:")
    print(" Time :", key[0])
    print(" From :", key[1])
    print(" To   :", key[2])

    print(" Sources:")

    for event in events:

        print(
            "  -",
            event[
                "attributes"
            ].get(
                "source_structure"
            )
        )
```

```python
# ============================================================
# STEP 31: EXTRACT ALL JIRA TICKETS FOR THE PILOT INSTANCE
# ============================================================
# Purpose:
# Apply the structured Jira extraction function to all 167
# tickets belonging to DOCS-I01.
#
# IMPORTANT:
# We keep the results in memory first.
#
# We DO NOT write the final JSONL yet because we first want
# to perform dataset-wide quality checks.
# ============================================================


pilot_jira_records = []


for item in instance_jira_tickets:

    ticket = item["ticket"]

    extracted_record = extract_jira_ticket(

        ticket=ticket,

        repository_project=PILOT_PROJECT,

        instance_id=PILOT_INSTANCE,

        instance_start=INSTANCE_START,

        instance_end=INSTANCE_END,

        source_file_name=jira_file.name
    )

    pilot_jira_records.append(
        extracted_record
    )


print("=" * 70)
print("PILOT JIRA EXTRACTION")
print("=" * 70)

print(
    "\nInput Jira tickets :",
    len(instance_jira_tickets)
)

print(
    "Extracted records  :",
    len(pilot_jira_records)
)


if (
    len(pilot_jira_records)
    ==
    len(instance_jira_tickets)
):

    print(
        "\nSUCCESS: Every pilot Jira ticket "
        "produced one extraction record."
    )

else:

    print(
        "\nWARNING: Input and output record "
        "counts do not match."
    )
```

```python
# ============================================================
# STEP 32: SUMMARIZE ALL PILOT JIRA EXTRACTIONS
# ============================================================
# Purpose:
# Count the entities, relationships, events, blockers and
# dependencies extracted across the complete DOCS-I01 Jira
# instance.
#
# This gives us a sanity check before saving the dataset.
# ============================================================

from collections import Counter


entity_type_counts = Counter()
relationship_type_counts = Counter()
event_type_counts = Counter()


total_entities = 0
total_relationships = 0
total_events = 0
total_blockers = 0
total_dependencies = 0


for record in pilot_jira_records:

    # --------------------------------------------------------
    # Entities
    # --------------------------------------------------------

    total_entities += len(
        record["entities"]
    )

    for entity in record["entities"]:

        entity_type_counts[
            entity["type"]
        ] += 1


    # --------------------------------------------------------
    # Relationships
    # --------------------------------------------------------

    total_relationships += len(
        record["relationships"]
    )

    for relationship in record[
        "relationships"
    ]:

        relationship_type_counts[
            relationship["type"]
        ] += 1


    # --------------------------------------------------------
    # Events
    # --------------------------------------------------------

    total_events += len(
        record["events"]
    )

    for event in record["events"]:

        event_type_counts[
            event["event_type"]
        ] += 1


    # --------------------------------------------------------
    # Blockers / dependencies
    # --------------------------------------------------------

    total_blockers += len(
        record["blockers"]
    )

    total_dependencies += len(
        record["dependencies"]
    )


print("=" * 70)
print("DOCS-I01 JIRA EXTRACTION SUMMARY")
print("=" * 70)


print("\nRecords       :", len(pilot_jira_records))
print("Entities      :", total_entities)
print("Relationships :", total_relationships)
print("Events        :", total_events)
print("Blockers      :", total_blockers)
print("Dependencies  :", total_dependencies)


print("\n" + "-" * 70)
print("ENTITY TYPES")
print("-" * 70)

for entity_type, count in (
    entity_type_counts.most_common()
):

    print(
        f"{entity_type:25s}",
        count
    )


print("\n" + "-" * 70)
print("RELATIONSHIP TYPES")
print("-" * 70)

for relation_type, count in (
    relationship_type_counts.most_common()
):

    print(
        f"{relation_type:25s}",
        count
    )


print("\n" + "-" * 70)
print("EVENT TYPES")
print("-" * 70)

for event_type, count in (
    event_type_counts.most_common()
):

    print(
        f"{event_type:25s}",
        count
    )
```

```python
# ============================================================
# STEP 33: CHECK DUPLICATE STATUS EVENTS ACROSS ALL TICKETS
# ============================================================
# Purpose:
# Determine whether STATUS_CHANGE events are duplicated because
# the same transition occurs in both:
#
#     status_history
#
# and
#
#     full_changelog
#
# Duplicate events would later inflate features such as:
#
#     status_change_count
#     workflow_volatility
#     transition_frequency
#
# Therefore we detect them before saving the extraction output.
# ============================================================


duplicate_status_records = []


for record in pilot_jira_records:

    seen_status_events = {}


    for event in record["events"]:

        if event["event_type"] != "STATUS_CHANGE":
            continue


        # ----------------------------------------------------
        # Normalize the event identity using:
        #
        # timestamp + old status + new status
        # ----------------------------------------------------

        event_time = parse_timestamp(
            event["event_time"]
        )


        key = (

            event_time.isoformat()
            if event_time
            else str(event["event_time"]),

            str(event["from_value"]),

            str(event["to_value"])
        )


        seen_status_events.setdefault(
            key,
            []
        ).append(event)


    duplicates = {

        key: events

        for key, events
        in seen_status_events.items()

        if len(events) > 1
    }


    if duplicates:

        duplicate_status_records.append({

            "source_id":
                record["source"]["source_id"],

            "duplicates":
                duplicates
        })


print("=" * 70)
print("DATASET-WIDE DUPLICATE STATUS CHECK")
print("=" * 70)


print(
    "\nRecords containing duplicate status transitions:",
    len(duplicate_status_records)
)


duplicate_transition_count = sum(

    len(item["duplicates"])

    for item in duplicate_status_records
)


print(
    "Duplicate transition groups:",
    duplicate_transition_count
)


# ------------------------------------------------------------
# Display only the first few examples if duplicates exist.
# ------------------------------------------------------------

if duplicate_status_records:

    print("\nExample duplicates:\n")


    for item in duplicate_status_records[:5]:

        print(
            "Ticket:",
            item["source_id"]
        )


        for key, events in (
            item["duplicates"].items()
        ):

            print(
                " ",
                key[0],
                "|",
                key[1],
                "->",
                key[2]
            )

            print(
                "  Sources:",
                [
                    event[
                        "attributes"
                    ].get(
                        "source_structure"
                    )

                    for event in events
                ]
            )


        print()


else:

    print(
        "\nSUCCESS: No duplicate STATUS_CHANGE "
        "events detected."
    )
```

```python
# ============================================================
# STEP 34: INSPECT UNMAPPED JIRA CHANGELOG FIELDS
# ============================================================
# Purpose:
# Find Jira changelog fields currently classified as
# OTHER_CHANGE.
#
# Some of these may contain useful information for:
#
#   scope change
#   sprint movement
#   due-date change
#   estimate change
#   ownership change
#   workflow changes
#
# If useful fields exist, we will give them explicit event
# types before creating the final JSONL.
# ============================================================


unmapped_field_counts = Counter()


for record in pilot_jira_records:

    for event in record["events"]:

        if event["event_type"] != "OTHER_CHANGE":
            continue


        jira_field = (
            event[
                "attributes"
            ].get(
                "jira_field"
            )
        )


        unmapped_field_counts[
            jira_field
        ] += 1


print("=" * 70)
print("UNMAPPED JIRA CHANGELOG FIELDS")
print("=" * 70)


print(
    "\nUnique unmapped fields:",
    len(unmapped_field_counts)
)


for field_name, count in (
    unmapped_field_counts.most_common()
):

    print(
        f"{str(field_name):35s}",
        count
    )
```

```python
# ============================================================
# STEP 36: EXPAND JIRA CHANGELOG EVENT MAPPING
# ============================================================
# Purpose:
# The pilot analysis found several useful Jira changelog fields
# that were previously classified as OTHER_CHANGE.
#
# We now assign explicit event types so that useful temporal
# information is not lost before:
#
#   Temporal KG
#       ↓
#   Feature Engineering
#       ↓
#   TabPFN
#
# Particularly important additions:
#
#   Sprint       -> schedule / sprint movement
#   Epic Link    -> scope / hierarchy
#   Epic Child   -> scope growth
#   Estimate     -> effort change
#   Actual Time  -> actual effort
#   Parent       -> hierarchy change
#   Labels       -> metadata change
#
# We preserve Rank and Workflow too, even though they may not
# ultimately become TabPFN features.
# ============================================================


JIRA_CHANGE_EVENT_MAP = {

    # --------------------------------------------------------
    # Workflow / state
    # --------------------------------------------------------

    "status":
        "STATUS_CHANGE",

    "resolution":
        "RESOLUTION_CHANGE",

    "workflow":
        "WORKFLOW_CHANGE",

    "epic status":
        "EPIC_STATUS_CHANGE",


    # --------------------------------------------------------
    # Ownership
    # --------------------------------------------------------

    "assignee":
        "ASSIGNEE_CHANGE",


    # --------------------------------------------------------
    # Priority / ordering
    # --------------------------------------------------------

    "priority":
        "PRIORITY_CHANGE",

    "rank":
        "RANK_CHANGE",


    # --------------------------------------------------------
    # Scope / hierarchy
    # --------------------------------------------------------

    "epic link":
        "EPIC_LINK_CHANGE",

    "epic child":
        "EPIC_CHILD_CHANGE",

    "epic name":
        "EPIC_NAME_CHANGE",

    "parent":
        "PARENT_CHANGE",

    "issuetype":
        "ISSUE_TYPE_CHANGE",

    "project":
        "PROJECT_CHANGE",

    "key":
        "ISSUE_KEY_CHANGE",


    # --------------------------------------------------------
    # Schedule / planning
    # --------------------------------------------------------

    "sprint":
        "SPRINT_CHANGE",

    "fix version":
        "VERSION_CHANGE",

    "version":
        "VERSION_CHANGE",

    "affected version":
        "VERSION_CHANGE",


    # --------------------------------------------------------
    # Effort / estimation
    # --------------------------------------------------------

    "story points":
        "STORY_POINT_CHANGE",

    "estimate":
        "ESTIMATE_CHANGE",

    "actual time":
        "ACTUAL_TIME_CHANGE",


    # --------------------------------------------------------
    # Graph relationships
    # --------------------------------------------------------

    "link":
        "LINK_CHANGE",

    "remoteissuelink":
        "REMOTE_LINK_CHANGE",


    # --------------------------------------------------------
    # Content / requirements / metadata
    # --------------------------------------------------------

    "summary":
        "SUMMARY_CHANGE",

    "description":
        "DESCRIPTION_CHANGE",

    "component":
        "COMPONENT_CHANGE",

    "labels":
        "LABEL_CHANGE",

    "operating system":
        "OPERATING_SYSTEM_CHANGE",

    "case":
        "CASE_CHANGE",

    "call to action":
        "CALL_TO_ACTION_CHANGE",


    # --------------------------------------------------------
    # Evidence / artefact changes
    # --------------------------------------------------------

    "comment":
        "COMMENT_CHANGE",

    "attachment":
        "ATTACHMENT_CHANGE"
}


print("=" * 70)
print("EXPANDED JIRA EVENT MAPPING")
print("=" * 70)

print("\nMapped fields:", len(JIRA_CHANGE_EVENT_MAP))


for field_name, event_type in (
    JIRA_CHANGE_EVENT_MAP.items()
):

    print(
        f"{field_name:25s} -> {event_type}"
    )
```

```python
# ============================================================
# STEP 39: FINAL DUPLICATE STATUS-EVENT CHECK
# ============================================================
# Purpose:
# Ensure status transitions are not counted twice because
# Jira contains both status_history and full_changelog.
#
# Duplicate status events must be fixed before saving because
# they could inflate temporal ML features.
# ============================================================


duplicate_status_groups = []


for record in pilot_jira_records:

    seen = {}


    for event in record["events"]:

        if event["event_type"] != "STATUS_CHANGE":
            continue


        parsed_time = parse_timestamp(
            event["event_time"]
        )


        key = (

            parsed_time.isoformat()
            if parsed_time
            else str(event["event_time"]),

            str(event["from_value"]),

            str(event["to_value"])
        )


        seen.setdefault(
            key,
            []
        ).append(event)


    for key, events in seen.items():

        if len(events) > 1:

            duplicate_status_groups.append({

                "ticket":
                    record[
                        "source"
                    ]["source_id"],

                "key":
                    key,

                "events":
                    events
            })


print("=" * 70)
print("FINAL DUPLICATE STATUS CHECK")
print("=" * 70)


print(
    "\nDuplicate transition groups:",
    len(duplicate_status_groups)
)


if duplicate_status_groups:

    print("\nFirst 5 examples:")


    for duplicate in (
        duplicate_status_groups[:5]
    ):

        print(
            "\nTicket:",
            duplicate["ticket"]
        )

        print(
            "Transition:",
            duplicate["key"]
        )

        print(
            "Sources:",
            [
                event[
                    "attributes"
                ].get(
                    "source_structure"
                )

                for event in duplicate[
                    "events"
                ]
            ]
        )


else:

    print(
        "\nSUCCESS: No duplicated status "
        "transitions detected."
    )
```

```python
# ============================================================
# STEP 40: DEDUPLICATE STATUS CHANGE EVENTS
# ============================================================
# Purpose:
# Jira stores the same status transition in two structures:
#
#     1. status_history
#     2. full_changelog
#
# Example:
#
#     Open -> In Progress
#
# may therefore appear twice with the exact same timestamp.
#
# Keeping both would incorrectly double:
#
#     status_change_count
#     workflow volatility
#     transition frequency
#
# and could distort downstream Temporal KG / TabPFN features.
#
# Strategy:
# For identical STATUS_CHANGE events defined by:
#
#     timestamp
#     from_status
#     to_status
#
# keep only ONE event.
#
# When both sources exist, prefer:
#
#     full_changelog
#
# because it belongs to the richer general Jira event stream.
# ============================================================


def deduplicate_status_events(record):

    # --------------------------------------------------------
    # Events that are NOT status changes are always preserved.
    # --------------------------------------------------------

    non_status_events = [

        event

        for event in record["events"]

        if event["event_type"] != "STATUS_CHANGE"
    ]


    # --------------------------------------------------------
    # Collect status events by their real transition identity.
    # --------------------------------------------------------

    status_groups = {}


    for event in record["events"]:

        if event["event_type"] != "STATUS_CHANGE":
            continue


        parsed_time = parse_timestamp(
            event["event_time"]
        )


        normalized_time = (

            parsed_time.isoformat()

            if parsed_time

            else str(event["event_time"])
        )


        key = (

            normalized_time,

            str(event["from_value"]),

            str(event["to_value"])
        )


        status_groups.setdefault(
            key,
            []
        ).append(event)


    # --------------------------------------------------------
    # Select exactly ONE event from each transition group.
    # --------------------------------------------------------

    deduplicated_status_events = []


    for key, events in status_groups.items():

        # Default to first available event.
        selected_event = events[0]


        # Prefer full_changelog if present.
        for event in events:

            source_structure = (
                event
                .get("attributes", {})
                .get("source_structure")
            )


            if (
                source_structure
                == "full_changelog"
            ):

                selected_event = event
                break


        deduplicated_status_events.append(
            selected_event
        )


    # --------------------------------------------------------
    # Recombine status + non-status events.
    # --------------------------------------------------------

    record["events"] = (

        non_status_events
        +
        deduplicated_status_events
    )


    # --------------------------------------------------------
    # Sort events chronologically where possible.
    #
    # This makes the JSON easier for the Temporal KG stage.
    # --------------------------------------------------------

    record["events"].sort(

        key=lambda event: (

            parse_timestamp(
                event.get("event_time")
            )
            or datetime.min.replace(
                tzinfo=INSTANCE_START.tzinfo
            )
        )
    )


    return record
```

```python
# ============================================================
# STEP 41: APPLY STATUS DEDUPLICATION TO PILOT DATASET
# ============================================================
# Purpose:
# Remove duplicated Jira status transitions from every
# extracted DOCS-I01 Jira record.
# ============================================================


events_before = sum(

    len(record["events"])

    for record in pilot_jira_records
)


status_before = sum(

    1

    for record in pilot_jira_records

    for event in record["events"]

    if event["event_type"] == "STATUS_CHANGE"
)


# ------------------------------------------------------------
# Apply deduplication.
# ------------------------------------------------------------

pilot_jira_records = [

    deduplicate_status_events(record)

    for record in pilot_jira_records
]


events_after = sum(

    len(record["events"])

    for record in pilot_jira_records
)


status_after = sum(

    1

    for record in pilot_jira_records

    for event in record["events"]

    if event["event_type"] == "STATUS_CHANGE"
)


print("=" * 70)
print("STATUS EVENT DEDUPLICATION")
print("=" * 70)


print("\nTotal events before :", events_before)
print("Total events after  :", events_after)

print(
    "Events removed      :",
    events_before - events_after
)


print("\nSTATUS_CHANGE before:", status_before)
print("STATUS_CHANGE after :", status_after)

print(
    "Duplicate status events removed:",
    status_before - status_after
)
```

```python
# ============================================================
# STEP 42: VERIFY STATUS DEDUPLICATION
# ============================================================
# Purpose:
# Confirm that no duplicate status transitions remain after
# Step 41.
# ============================================================


remaining_duplicates = []


for record in pilot_jira_records:

    seen = set()


    for event in record["events"]:

        if event["event_type"] != "STATUS_CHANGE":
            continue


        parsed_time = parse_timestamp(
            event["event_time"]
        )


        key = (

            parsed_time.isoformat()
            if parsed_time
            else str(event["event_time"]),

            str(event["from_value"]),

            str(event["to_value"])
        )


        if key in seen:

            remaining_duplicates.append({

                "ticket":
                    record[
                        "source"
                    ]["source_id"],

                "transition":
                    key
            })


        seen.add(key)


print("=" * 70)
print("POST-DEDUPLICATION VALIDATION")
print("=" * 70)


print(
    "\nRemaining duplicate status events:",
    len(remaining_duplicates)
)


if len(remaining_duplicates) == 0:

    print(
        "\nSUCCESS:"
        "\nAll duplicate status transitions "
        "have been removed."
    )

else:

    print(
        "\nWARNING:"
        "\nDuplicate status transitions still exist."
    )
```

```python
# ============================================================
# STEP 43: FINAL CLEANED JIRA EXTRACTION SUMMARY
# ============================================================
# Purpose:
# Produce the final statistics for the cleaned DOCS-I01 Jira
# extraction before saving it.
# ============================================================

from collections import Counter


final_entity_counts = Counter()
final_relationship_counts = Counter()
final_event_counts = Counter()


for record in pilot_jira_records:

    for entity in record["entities"]:

        final_entity_counts[
            entity["type"]
        ] += 1


    for relationship in record[
        "relationships"
    ]:

        final_relationship_counts[
            relationship["type"]
        ] += 1


    for event in record["events"]:

        final_event_counts[
            event["event_type"]
        ] += 1


print("=" * 70)
print("FINAL CLEANED DOCS-I01 JIRA EXTRACTION")
print("=" * 70)


print(
    "\nRecords:",
    len(pilot_jira_records)
)


print(
    "Entities:",
    sum(final_entity_counts.values())
)


print(
    "Relationships:",
    sum(final_relationship_counts.values())
)


print(
    "Events:",
    sum(final_event_counts.values())
)


print(
    "Blockers:",
    sum(
        len(record["blockers"])
        for record in pilot_jira_records
    )
)


print(
    "Dependencies:",
    sum(
        len(record["dependencies"])
        for record in pilot_jira_records
    )
)


print("\nEVENT TYPES")
print("-" * 70)


for event_type, count in (
    final_event_counts.most_common()
):

    print(
        f"{event_type:30s}",
        count
    )
```

```python
# ============================================================
# STEP 44: SAVE CLEANED PILOT JIRA EXTRACTION
# ============================================================
# Purpose:
# Save the validated structured Jira extraction for DOCS-I01.
#
# Output format:
#     JSONL
#
# One line = one Jira source record.
#
# JSONL is useful because:
#
#   - easy to process incrementally
#   - scalable to larger datasets
#   - convenient for Entity Resolution
#   - easy to load into graph-processing pipelines
#
# IMPORTANT:
# This file contains extraction results only.
#
# Hidden generation manifests are NOT copied into it.
# ============================================================


pilot_output_folder = (

    EXTRACTION_ROOT
    / PILOT_PROJECT
    / PILOT_INSTANCE
)


pilot_output_folder.mkdir(

    parents=True,

    exist_ok=True
)


jira_output_file = (

    pilot_output_folder
    / "jira_extracted.jsonl"
)


with open(
    jira_output_file,
    "w",
    encoding="utf-8"
) as file:

    for record in pilot_jira_records:

        file.write(

            json.dumps(
                record,
                ensure_ascii=False
            )

            + "\n"
        )


print("=" * 70)
print("JIRA EXTRACTION SAVED")
print("=" * 70)


print("\nOutput file:")
print(jira_output_file)


print(
    "\nRecords written:",
    len(pilot_jira_records)
)


print(
    "File exists:",
    jira_output_file.exists()
)


print(
    "File size:",
    round(
        jira_output_file.stat().st_size
        / 1024
        / 1024,
        2
    ),
    "MB"
)
```

```python
# ============================================================
# STEP 45: VERIFY SAVED JIRA JSONL
# ============================================================
# Purpose:
# Reload the saved JSONL file and confirm:
#
#   - every line is valid JSON
#   - record count is still 167
#   - important sections are present
#
# This is the final validation before moving to PRD extraction.
# ============================================================


reloaded_jira_records = []


with open(
    jira_output_file,
    "r",
    encoding="utf-8"
) as file:

    for line_number, line in enumerate(
        file,
        start=1
    ):

        if not line.strip():
            continue


        try:

            record = json.loads(line)

            reloaded_jira_records.append(
                record
            )


        except json.JSONDecodeError as error:

            print(
                "Invalid JSON on line",
                line_number
            )

            raise error


print("=" * 70)
print("SAVED JIRA FILE VALIDATION")
print("=" * 70)


print(
    "\nRecords reloaded:",
    len(reloaded_jira_records)
)


print(
    "Expected records:",
    len(pilot_jira_records)
)


required_sections = [

    "entities",
    "relationships",
    "events",
    "blockers",
    "dependencies",
    "temporal",
    "attributes",
    "provenance"
]


missing_sections = []


for record in reloaded_jira_records:

    for section in required_sections:

        if section not in record:

            missing_sections.append(
                (
                    record[
                        "source"
                    ]["source_id"],
                    section
                )
            )


print(
    "\nMissing required sections:",
    len(missing_sections)
)


if (
    len(reloaded_jira_records)
    ==
    len(pilot_jira_records)
    and
    len(missing_sections) == 0
):

    print(
        "\nSUCCESS:"
        "\nPilot Jira extraction JSONL is valid "
        "and ready for downstream use."
    )

else:

    print(
        "\nCHECK REQUIRED."
    )
```

```python
# ============================================================
# STEP 46: LOAD THE PILOT PRD
# ============================================================
# Purpose:
# Load the final generated PRD for DOCS-I01.
#
# IMPORTANT:
# We use ONLY:
#
#     DOCS-I01-PRD-V1.txt
#
# We do NOT use:
#
#     *.prompt.txt
#
# because prompt files contain generation instructions rather
# than enterprise project evidence.
#
# The raw PRD text is retained because it will later provide
# evidence/provenance for extracted facts.
# ============================================================


prd_file = prd_files[0]


with open(
    prd_file,
    "r",
    encoding="utf-8"
) as file:

    prd_text = file.read()


print("=" * 70)
print("PRD LOADED")
print("=" * 70)

print("\nFile       :", prd_file.name)
print("Characters :", len(prd_text))

print("\nFirst 1000 characters:\n")
print(prd_text[:1000])
```

```python
# ============================================================
# STEP 47: PARSE PRD DOCUMENT METADATA
# ============================================================
# Purpose:
# Extract high-confidence structured metadata directly from
# the PRD header.
#
# We extract:
#
#   document title
#   version
#   date
#   status
#
# These values are useful for:
#
#   temporal alignment
#   requirement-version tracking
#   provenance
#   future PRD-change analysis
# ============================================================

import re


def extract_markdown_field(text, field_name):
    """
    Extract a Markdown metadata field such as:

        **Version:** 1

    Returns None when the field is absent.
    """

    pattern = rf"\*\*{re.escape(field_name)}:\*\*\s*(.+)"

    match = re.search(
        pattern,
        text,
        flags=re.IGNORECASE
    )

    if match:
        return match.group(1).strip()

    return None


# ------------------------------------------------------------
# Extract first Markdown heading as document title.
# ------------------------------------------------------------

title_match = re.search(
    r"^#\s+(.+)$",
    prd_text,
    flags=re.MULTILINE
)


prd_title = (
    title_match.group(1).strip()
    if title_match
    else None
)


prd_version = extract_markdown_field(
    prd_text,
    "Version"
)

prd_date = extract_markdown_field(
    prd_text,
    "Date"
)

prd_status = extract_markdown_field(
    prd_text,
    "Status"
)


print("=" * 70)
print("PRD METADATA")
print("=" * 70)

print("\nTitle   :", prd_title)
print("Version :", prd_version)
print("Date    :", prd_date)
print("Status  :", prd_status)
```

```python
# ============================================================
# STEP 48: SPLIT PRD INTO MARKDOWN SECTIONS
# ============================================================
# Purpose:
# Break the PRD into logical sections based on Markdown
# headings.
#
# Examples:
#
#   Objectives
#   Scope
#   Stakeholders and Ownership
#   Current Work Areas
#   Functional Requirements
#   Non-Functional Requirements
#   Milestones and Targets
#   Dependencies
#   Assumptions
#   Known Risks and Uncertainties
#   Open Questions
#   Acceptance Criteria
#
# WHY:
# Section-aware extraction improves precision because we know
# the semantic role of the text before applying GLiNER/rules.
# ============================================================


def split_markdown_sections(text):

    sections = []

    current_heading = "DOCUMENT_HEADER"
    current_level = 0
    current_lines = []


    for line in text.splitlines():

        heading_match = re.match(
            r"^(#{2,6})\s+(.+)$",
            line
        )


        if heading_match:

            # Save previous section
            if current_lines:

                sections.append({

                    "heading":
                        current_heading,

                    "level":
                        current_level,

                    "text":
                        "\n".join(
                            current_lines
                        ).strip()
                })


            current_level = len(
                heading_match.group(1)
            )

            current_heading = (
                heading_match.group(2).strip()
            )

            current_lines = []

        else:

            current_lines.append(line)


    # Save final section
    if current_lines:

        sections.append({

            "heading":
                current_heading,

            "level":
                current_level,

            "text":
                "\n".join(
                    current_lines
                ).strip()
        })


    return sections


prd_sections = split_markdown_sections(
    prd_text
)


print("=" * 70)
print("PRD SECTIONS")
print("=" * 70)

print(
    "\nSections found:",
    len(prd_sections)
)


for index, section in enumerate(
    prd_sections,
    start=1
):

    print(
        f"{index:02d}.",
        section["heading"],
        "|",
        len(section["text"]),
        "characters"
    )
```

```python
# ============================================================
# STEP 49: EXTRACT EXPLICIT PRD REQUIREMENTS
# ============================================================
# Purpose:
# Extract requirements with explicit requirement IDs.
#
# Example:
#
#   **REQ-006** Documentation shall describe...
#
# or:
#
#   **REQ-013 (Accuracy)** Procedures shall be verifiable...
#
# We preserve:
#
#   requirement_id
#   optional requirement label
#   requirement text
#   PRD version
#   PRD date
#
# These become REQUIREMENT entities and later support:
#
#   requirement count
#   requirement volatility
#   requirement dependencies
#   scope growth
#   requirement version changes
# ============================================================


requirement_pattern = re.compile(

    r"\*\*(REQ-\d+)"
    r"(?:\s*\(([^)]+)\))?"
    r"\*\*\s*(.+)",

    flags=re.IGNORECASE
)


prd_requirements = []


for line in prd_text.splitlines():

    match = requirement_pattern.search(
        line.strip()
    )


    if not match:
        continue


    requirement_id = (
        match.group(1).upper()
    )

    requirement_label = (
        match.group(2).strip()
        if match.group(2)
        else None
    )

    requirement_text = (
        match.group(3).strip()
    )


    prd_requirements.append({

        "requirement_id":
            requirement_id,

        "label":
            requirement_label,

        "text":
            requirement_text,

        "prd_version":
            prd_version,

        "prd_date":
            prd_date
    })


print("=" * 70)
print("EXPLICIT PRD REQUIREMENTS")
print("=" * 70)

print(
    "\nRequirements found:",
    len(prd_requirements)
)


for requirement in prd_requirements:

    print(
        "\n",
        requirement[
            "requirement_id"
        ],
        "|",
        requirement[
            "label"
        ]
    )

    print(
        requirement[
            "text"
        ]
    )
```

```python
# ============================================================
# STEP 50: IDENTIFY IMPORTANT PRD SECTIONS
# ============================================================
# Purpose:
# Locate sections that contain high-value structured project
# information.
#
# We deliberately use section structure where possible rather
# than asking a language model to infer everything.
# ============================================================


IMPORTANT_SECTION_KEYWORDS = {

    "requirements": [
        "functional requirements",
        "non-functional requirements"
    ],

    "milestones": [
        "milestones",
        "targets"
    ],

    "dependencies": [
        "dependencies"
    ],

    "stakeholders": [
        "stakeholders",
        "ownership"
    ],

    "risks": [
        "known risks",
        "uncertainties"
    ],

    "open_questions": [
        "open questions"
    ],

    "acceptance_criteria": [
        "acceptance criteria"
    ],

    "scope": [
        "scope"
    ]
}


important_sections = {
    key: []
    for key in IMPORTANT_SECTION_KEYWORDS
}


for section in prd_sections:

    heading_lower = (
        section["heading"].lower()
    )


    for category, keywords in (
        IMPORTANT_SECTION_KEYWORDS.items()
    ):

        if any(
            keyword in heading_lower
            for keyword in keywords
        ):

            important_sections[
                category
            ].append(section)


print("=" * 70)
print("IMPORTANT PRD SECTION CHECK")
print("=" * 70)


for category, sections in (
    important_sections.items()
):

    print(
        f"\n{category.upper():20s}:",
        len(sections)
    )

    for section in sections:

        print(
            "  -",
            section["heading"]
        )
```

```python
# ============================================================
# STEP 51: EXTRACT PRD MILESTONES AND TARGETS
# ============================================================
# Purpose:
# Extract structured milestone information from the PRD's
# "Milestones and Targets" Markdown table.
#
# We preserve:
#
#   milestone name
#   target
#   notes
#   PRD version
#   PRD date
#
# These facts may later support:
#
#   milestone count
#   schedule changes
#   delayed/undated work
#   sprint planning
#   schedule-risk feature engineering
# ============================================================


# ------------------------------------------------------------
# Get the milestone section found in Step 50.
# ------------------------------------------------------------

milestone_sections = important_sections[
    "milestones"
]


prd_milestones = []


for section in milestone_sections:

    lines = [
        line.strip()
        for line in section["text"].splitlines()
        if line.strip()
    ]


    for line in lines:

        # Markdown table rows begin/end with |
        if not line.startswith("|"):
            continue


        columns = [
            column.strip()
            for column in line.strip("|").split("|")
        ]


        # Ignore malformed rows
        if len(columns) < 3:
            continue


        # Ignore table header
        if columns[0].lower() == "milestone":
            continue


        # Ignore separator row such as:
        # |---|---|---|
        if all(
            set(column) <= {"-", ":", " "}
            for column in columns
        ):
            continue


        prd_milestones.append({

            "milestone":
                columns[0],

            "target":
                columns[1],

            "notes":
                columns[2],

            "prd_version":
                prd_version,

            "prd_date":
                prd_date
        })


print("=" * 70)
print("PRD MILESTONES")
print("=" * 70)


print(
    "\nMilestones found:",
    len(prd_milestones)
)


for milestone in prd_milestones:

    print("\nMilestone:", milestone["milestone"])
    print("Target   :", milestone["target"])
    print("Notes    :", milestone["notes"])
```

```python
# ============================================================
# STEP 52: EXTRACT EXPLICIT PRD DEPENDENCY STATEMENTS
# ============================================================
# Purpose:
# Extract every explicit bullet under the PRD Dependencies
# section.
#
# At this stage we preserve the complete dependency statement.
#
# Later:
#   GLiNER identifies entities inside the statement.
#   Relation extraction connects the source/target entities.
#
# This avoids prematurely guessing dependency endpoints.
# ============================================================


dependency_sections = important_sections[
    "dependencies"
]


prd_dependency_statements = []


for section in dependency_sections:

    for line in section["text"].splitlines():

        clean_line = line.strip()


        # Match Markdown bullet points.
        if re.match(
            r"^[-*]\s+",
            clean_line
        ):

            dependency_text = re.sub(
                r"^[-*]\s+",
                "",
                clean_line
            ).strip()


            prd_dependency_statements.append({

                "text":
                    dependency_text,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


print("=" * 70)
print("PRD DEPENDENCY STATEMENTS")
print("=" * 70)


print(
    "\nDependencies found:",
    len(prd_dependency_statements)
)


for index, dependency in enumerate(
    prd_dependency_statements,
    start=1
):

    print(
        f"\n{index}.",
        dependency["text"]
    )
```

```python
# ============================================================
# STEP 53: EXTRACT PRD STAKEHOLDER / OWNERSHIP INFORMATION
# ============================================================
# Purpose:
# Extract stakeholder groups from the structured PRD section.
#
# Example:
#
#   Documentation team:
#       writers assigned per component...
#
# This information later supports:
#
#   TEAM entities
#   ownership relationships
#   cross-team dependency analysis
#   ownership-gap features
# ============================================================


stakeholder_sections = important_sections[
    "stakeholders"
]


prd_stakeholders = []


for section in stakeholder_sections:

    for line in section["text"].splitlines():

        clean_line = line.strip()


        # ----------------------------------------------------
        # Expected structure:
        #
        # - **Documentation team:** description
        # ----------------------------------------------------

        match = re.match(
            r"^[-*]\s+\*\*(.+?):\*\*\s*(.+)$",
            clean_line
        )


        if not match:
            continue


        stakeholder_name = (
            match.group(1).strip()
        )

        stakeholder_description = (
            match.group(2).strip()
        )


        prd_stakeholders.append({

            "name":
                stakeholder_name,

            "description":
                stakeholder_description,

            "prd_version":
                prd_version,

            "prd_date":
                prd_date
        })


print("=" * 70)
print("PRD STAKEHOLDERS / OWNERSHIP")
print("=" * 70)


print(
    "\nStakeholder groups found:",
    len(prd_stakeholders)
)


for stakeholder in prd_stakeholders:

    print(
        "\nStakeholder:",
        stakeholder["name"]
    )

    print(
        "Description:",
        stakeholder["description"]
    )
```

```python
# ============================================================
# STEP 54: EXTRACT PRD SCOPE ITEMS
# ============================================================
# Purpose:
# Preserve explicit project scope.
#
# We distinguish:
#
#     IN_SCOPE
#     OUT_OF_SCOPE
#
# This is important because future PRD versions/events can be
# compared against this baseline to derive:
#
#     scope additions
#     scope removals
#     scope growth
#     scope volatility
#
# We do NOT calculate those features during extraction.
# ============================================================


prd_scope_items = []


for section in prd_sections:

    heading_lower = (
        section["heading"]
        .lower()
        .strip()
    )


    if heading_lower == "in scope":

        scope_type = "IN_SCOPE"


    elif heading_lower.startswith(
        "out of scope"
    ):

        scope_type = "OUT_OF_SCOPE"


    else:

        continue


    for line in section["text"].splitlines():

        clean_line = line.strip()


        if re.match(
            r"^[-*]\s+",
            clean_line
        ):

            scope_text = re.sub(
                r"^[-*]\s+",
                "",
                clean_line
            ).strip()


            prd_scope_items.append({

                "scope_type":
                    scope_type,

                "text":
                    scope_text,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


print("=" * 70)
print("PRD SCOPE ITEMS")
print("=" * 70)


print(
    "\nScope items found:",
    len(prd_scope_items)
)


for item in prd_scope_items:

    print(
        "\n",
        item["scope_type"],
        "|",
        item["text"]
    )
```

```python
# ============================================================
# STEP 55: EXTRACT PRD ACCEPTANCE CRITERIA
# ============================================================
# Purpose:
# Preserve explicit acceptance criteria.
#
# Acceptance criteria are useful for linking requirements to
# intended completion/verification conditions.
#
# They may later contribute to:
#
#   requirement state
#   completion evidence
#   milestone validation
#   GraphRAG explanations
# ============================================================


acceptance_sections = important_sections[
    "acceptance_criteria"
]


prd_acceptance_criteria = []


for section in acceptance_sections:

    for line in section["text"].splitlines():

        clean_line = line.strip()


        if re.match(
            r"^[-*]\s+",
            clean_line
        ):

            criterion = re.sub(
                r"^[-*]\s+",
                "",
                clean_line
            ).strip()


            prd_acceptance_criteria.append({

                "text":
                    criterion,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


print("=" * 70)
print("PRD ACCEPTANCE CRITERIA")
print("=" * 70)


print(
    "\nAcceptance criteria found:",
    len(prd_acceptance_criteria)
)


for index, criterion in enumerate(
    prd_acceptance_criteria,
    start=1
):

    print(
        f"\n{index}.",
        criterion["text"]
    )
```

```python
# ============================================================
# STEP 56: EXTRACT EXPLICIT PRD RISK / UNCERTAINTY EVIDENCE
# ============================================================
# Purpose:
# Preserve risk/uncertainty statements explicitly written in
# the PRD.
#
# IMPORTANT:
# These are SOURCE FACTS / EVIDENCE.
#
# We are NOT assigning:
#
#     schedule_risk = 1
#     scope_risk = 1
#
# and we are NOT generating the TabPFN prediction target.
#
# This information is preserved for:
#
#   Temporal KG evidence
#   GraphRAG
#   later feature engineering where appropriate
# ============================================================


risk_sections = important_sections[
    "risks"
]


prd_risk_evidence = []


for section in risk_sections:

    for line in section["text"].splitlines():

        clean_line = line.strip()


        if not re.match(
            r"^[-*]\s+",
            clean_line
        ):
            continue


        risk_text = re.sub(
            r"^[-*]\s+",
            "",
            clean_line
        ).strip()


        # ----------------------------------------------------
        # Some risks begin with a bold short label:
        #
        # **Carry-over of scheduled work.** Description...
        #
        # Preserve label separately when available.
        # ----------------------------------------------------

        label_match = re.match(
            r"^\*\*(.+?)\*\*\s*(.*)$",
            risk_text
        )


        if label_match:

            risk_label = (
                label_match.group(1)
                .strip()
                .rstrip(".")
            )

            risk_description = (
                label_match.group(2)
                .strip()
            )

        else:

            risk_label = None
            risk_description = risk_text


        prd_risk_evidence.append({

            "label":
                risk_label,

            "text":
                risk_description,

            "prd_version":
                prd_version,

            "prd_date":
                prd_date
        })


print("=" * 70)
print("PRD RISK / UNCERTAINTY EVIDENCE")
print("=" * 70)


print(
    "\nRisk statements found:",
    len(prd_risk_evidence)
)


for risk in prd_risk_evidence:

    print(
        "\nLabel:",
        risk["label"]
    )

    print(
        "Evidence:",
        risk["text"]
    )
```

```python
# ============================================================
# STEP 57: EXTRACT PRD OPEN QUESTIONS
# ============================================================
# Purpose:
# Preserve explicitly unresolved project questions.
#
# These may later support:
#
#   unresolved-decision counts
#   ownership gaps
#   dependency uncertainty
#   GraphRAG evidence
#
# Again, we preserve facts rather than calculating risk here.
# ============================================================


open_question_sections = (
    important_sections[
        "open_questions"
    ]
)


prd_open_questions = []


for section in open_question_sections:

    for line in section["text"].splitlines():

        clean_line = line.strip()


        # Numbered Markdown item:
        # 1. Question...
        match = re.match(
            r"^\d+\.\s+(.+)$",
            clean_line
        )


        if not match:
            continue


        question_text = (
            match.group(1).strip()
        )


        prd_open_questions.append({

            "text":
                question_text,

            "prd_version":
                prd_version,

            "prd_date":
                prd_date
        })


print("=" * 70)
print("PRD OPEN QUESTIONS")
print("=" * 70)


print(
    "\nOpen questions found:",
    len(prd_open_questions)
)


for index, question in enumerate(
    prd_open_questions,
    start=1
):

    print(
        f"\n{index}.",
        question["text"]
    )
```

```python
# ============================================================
# STEP 58: DETERMINISTIC PRD EXTRACTION SUMMARY
# ============================================================
# Purpose:
# Summarize everything extracted from explicit PRD structure
# before introducing GLiNER.
#
# This gives us a high-confidence deterministic baseline.
# ============================================================


print("=" * 70)
print("DETERMINISTIC PRD EXTRACTION SUMMARY")
print("=" * 70)


print(
    "\nRequirements       :",
    len(prd_requirements)
)

print(
    "Milestones         :",
    len(prd_milestones)
)

print(
    "Dependencies       :",
    len(prd_dependency_statements)
)

print(
    "Stakeholder groups :",
    len(prd_stakeholders)
)

print(
    "Scope items        :",
    len(prd_scope_items)
)

print(
    "Acceptance criteria:",
    len(prd_acceptance_criteria)
)

print(
    "Risk evidence      :",
    len(prd_risk_evidence)
)

print(
    "Open questions     :",
    len(prd_open_questions)
)


print("\nPRD metadata:")
print(" Version :", prd_version)
print(" Date    :", prd_date)
print(" Status  :", prd_status)
```

```python
# ============================================================
# STEP 59: INSTALL GLINER
# ============================================================
# Purpose:
# Install the lightweight open-source GLiNER model.
#
# GLiNER will be used for semantic entity extraction from
# unstructured PRD and Meeting text.
#
# IMPORTANT:
# GLiNER is NOT replacing our deterministic extraction.
#
# We already extracted high-confidence structured information:
#
#   requirements
#   milestones
#   dependencies
#   scope
#   stakeholders
#   acceptance criteria
#   explicit risk evidence
#   open questions
#
# GLiNER will COMPLEMENT this by finding semantic entities such
# as:
#
#   features
#   systems
#   components
#   teams
#   people
#   project artefacts
#   contextual dependencies
#   action-related concepts
#
# Once the model is downloaded, no hosted API/token quota is
# required for inference.
# ============================================================

!pip install -q gliner
```

```python
# ============================================================
# STEP 60: LOAD GLINER MODEL
# ============================================================
# Purpose:
# Load the local semantic entity-extraction model.
#
# The first execution downloads the model weights.
# After loading, inference is performed locally in the
# Colab runtime rather than through a hosted LLM API.
# ============================================================

from gliner import GLiNER


GLINER_MODEL_NAME = "urchade/gliner_medium-v2.1"


print("Loading GLiNER model:")
print(GLINER_MODEL_NAME)


gliner_model = GLiNER.from_pretrained(
    GLINER_MODEL_NAME
)


print("\nSUCCESS: GLiNER model loaded.")
```

```python
# ============================================================
# STEP 61: DEFINE GC1 GLINER ENTITY LABELS
# ============================================================
# Purpose:
# Define the semantic entity types that GLiNER should identify
# from unstructured enterprise-project text.
#
# IMPORTANT DESIGN CHOICE:
#
# We do NOT ask GLiNER to rediscover structured entities such
# as REQ-001 when deterministic parsing already gives us a
# higher-confidence result.
#
# We mainly use GLiNER for concepts that can appear naturally
# inside sentences.
# ============================================================


GLINER_ENTITY_LABELS = [

    # Product/project functionality
    "feature",

    # Technical elements
    "system",
    "component",

    # Human/organizational entities
    "person",
    "team",

    # Work/planning concepts
    "milestone",

    # Contextual project concepts
    "dependency",
    "blocker",
    "action item",

    # Decision concepts
    "decision"
]


print("=" * 70)
print("GC1 GLINER LABELS")
print("=" * 70)


for label in GLINER_ENTITY_LABELS:
    print("-", label)


print(
    "\nNumber of semantic labels:",
    len(GLINER_ENTITY_LABELS)
)
```

```python
# ============================================================
# STEP 62: CREATE PRD GLINER CHUNKS
# ============================================================
# Purpose:
# Convert PRD sections into manageable semantic-extraction
# chunks.
#
# WHY SECTION-AWARE CHUNKING?
#
# A phrase appearing under:
#
#     Dependencies
#
# has different context from the same phrase appearing under:
#
#     Scope
#
# We therefore preserve the section heading with every chunk.
#
# We also limit chunk size to avoid giving the small model
# unnecessarily large text blocks.
# ============================================================


MAX_CHUNK_CHARS = 1800


def create_prd_chunks(
    sections,
    max_chars=1800
):
    """
    Create section-aware PRD text chunks.

    Each chunk retains:
        chunk_id
        section heading
        text

    Long sections are split by paragraphs.
    """

    chunks = []

    chunk_counter = 1


    for section in sections:

        heading = section["heading"]

        text = section["text"].strip()


        # Ignore empty parent headings such as:
        # "3. Scope"
        if not text:
            continue


        # ----------------------------------------------------
        # Split section into paragraphs.
        # ----------------------------------------------------

        paragraphs = [

            paragraph.strip()

            for paragraph
            in re.split(
                r"\n\s*\n",
                text
            )

            if paragraph.strip()
        ]


        current_text = ""


        for paragraph in paragraphs:

            # Include heading as contextual information.
            candidate = (

                current_text
                + "\n"
                + paragraph

                if current_text

                else paragraph
            )


            # ------------------------------------------------
            # If candidate becomes too large, save the current
            # chunk and start another one.
            # ------------------------------------------------

            if (
                len(candidate) > max_chars
                and current_text
            ):

                chunks.append({

                    "chunk_id":
                        f"PRD_CHUNK_{chunk_counter:03d}",

                    "section":
                        heading,

                    "text":
                        (
                            f"Section: {heading}\n"
                            f"{current_text}"
                        )
                })


                chunk_counter += 1

                current_text = paragraph


            else:

                current_text = candidate


        # Save final chunk from section.
        if current_text:

            chunks.append({

                "chunk_id":
                    f"PRD_CHUNK_{chunk_counter:03d}",

                "section":
                    heading,

                "text":
                    (
                        f"Section: {heading}\n"
                        f"{current_text}"
                    )
            })


            chunk_counter += 1


    return chunks


prd_chunks = create_prd_chunks(
    prd_sections,
    MAX_CHUNK_CHARS
)


print("=" * 70)
print("PRD GLINER CHUNKS")
print("=" * 70)


print(
    "\nChunks created:",
    len(prd_chunks)
)


for chunk in prd_chunks:

    print(
        chunk["chunk_id"],
        "|",
        chunk["section"],
        "|",
        len(chunk["text"]),
        "characters"
    )
```

```python
# ============================================================
# STEP 63: TEST GLINER ON ONE PRD CHUNK
# ============================================================
# Purpose:
# Test semantic entity extraction before processing every PRD
# section.
#
# We choose a reasonably informative chunk rather than blindly
# assuming the first chunk contains useful entities.
# ============================================================


# ------------------------------------------------------------
# Prefer a Current Work Area or Dependencies chunk.
# ------------------------------------------------------------

test_chunk = None


for chunk in prd_chunks:

    heading_lower = (
        chunk["section"].lower()
    )


    if (
        "access control" in heading_lower
        or
        "dependencies" in heading_lower
    ):

        test_chunk = chunk
        break


# Fallback
if test_chunk is None:
    test_chunk = prd_chunks[0]


print("=" * 70)
print("GLINER TEST CHUNK")
print("=" * 70)

print("\nChunk ID :", test_chunk["chunk_id"])
print("Section  :", test_chunk["section"])

print("\nText:\n")
print(test_chunk["text"])


# ------------------------------------------------------------
# Run semantic extraction.
# ------------------------------------------------------------

test_gliner_entities = (
    gliner_model.predict_entities(

        test_chunk["text"],

        GLINER_ENTITY_LABELS,

        threshold=0.40
    )
)


print("\n" + "=" * 70)
print("GLINER TEST ENTITIES")
print("=" * 70)


print(
    "\nEntities found:",
    len(test_gliner_entities)
)


for entity in test_gliner_entities:

    print(
        f"{entity['label']:15s}",
        "|",
        f"{entity['score']:.3f}",
        "|",
        entity["text"]
    )
```

```python
# ============================================================
# STEP 64: REFINE GLINER LABEL SET
# ============================================================
# Purpose:
# After testing GLiNER on real GC1 PRD text, we found that it
# works well for semantic/named concepts such as:
#
#   Engineering
#   mongos
#   mongod
#   Access control
#
# But concepts such as:
#
#   dependency
#   blocker
#   decision
#   action item
#
# are better represented as relationships/events rather than
# standalone semantic entities.
#
# Explicit Jira IDs such as DOCS-6963 will be extracted
# deterministically using regex.
# ============================================================

GLINER_ENTITY_LABELS = [
    "feature",
    "system",
    "component",
    "team",
    "person",
    "milestone"
]


print("Refined GLiNER labels:")

for label in GLINER_ENTITY_LABELS:
    print("-", label)
```

```python
# ============================================================
# STEP 65: EXTRACT EXPLICIT JIRA ISSUE REFERENCES - CORRECTED
# ============================================================
# Purpose:
# Extract Jira-style issue/ticket references from PRD text.
#
# Examples:
#
#     DOCS-6963
#     SERVER-1234
#     PRODUCT-165
#     INT-1680
#
# IMPORTANT:
#
# Requirement identifiers such as:
#
#     REQ-001
#     REQ-002
#
# have a similar textual pattern, but they are REQUIREMENTS,
# not Jira issues.
#
# Therefore REQ-* identifiers are explicitly excluded here.
#
# They were already extracted correctly in Step 49 as
# REQUIREMENT entities.
# ============================================================


# ------------------------------------------------------------
# General Jira-style identifier pattern.
# ------------------------------------------------------------

ISSUE_KEY_PATTERN = re.compile(
    r"\b[A-Z][A-Z0-9_]+-\d+\b"
)


prd_issue_references = []


for chunk in prd_chunks:

    matches = ISSUE_KEY_PATTERN.findall(
        chunk["text"]
    )


    for issue_key in matches:

        # ----------------------------------------------------
        # EXCLUDE PRD requirement IDs.
        #
        # REQ-001 etc. belong to REQUIREMENT extraction,
        # not ISSUE extraction.
        # ----------------------------------------------------

        if issue_key.upper().startswith("REQ-"):
            continue


        prd_issue_references.append({

            "issue_key":
                issue_key,

            "chunk_id":
                chunk["chunk_id"],

            "section":
                chunk["section"],

            "extraction_method":
                "regex",

            "confidence":
                1.0
        })


# ------------------------------------------------------------
# Unique issue keys are useful for validation.
#
# We still retain prd_issue_references separately because the
# same Jira issue may appear in multiple sections/chunks and
# those occurrences provide useful provenance.
# ------------------------------------------------------------

unique_prd_issue_keys = sorted(
    set(
        item["issue_key"]
        for item in prd_issue_references
    )
)


print("=" * 70)
print("PRD EXPLICIT JIRA ISSUE REFERENCES - CORRECTED")
print("=" * 70)


print(
    "\nTotal Jira issue occurrences:",
    len(prd_issue_references)
)


print(
    "Unique Jira issue keys:",
    len(unique_prd_issue_keys)
)


for issue_key in unique_prd_issue_keys:

    print("-", issue_key)


# ------------------------------------------------------------
# Safety check:
# No REQ-* identifier should remain.
# ------------------------------------------------------------

req_ids_wrongly_included = [

    key

    for key in unique_prd_issue_keys

    if key.upper().startswith("REQ-")
]


print(
    "\nREQ-* identifiers incorrectly classified as issues:",
    len(req_ids_wrongly_included)
)


if len(req_ids_wrongly_included) == 0:

    print(
        "SUCCESS: Requirements and Jira issues "
        "are correctly separated."
    )
```

```python
# ============================================================
# STEP 66: DEFINE GENERIC PERSON-ROLE FILTER
# ============================================================
# Purpose:
# Prevent generic organizational roles from becoming PERSON
# entities.
#
# Example false positives:
#
#   writer
#   reviewer
#   owner
#   user
#
# These words may still be important in relationships or
# ownership statements, but they should not become named
# PERSON nodes unless an actual person's name is present.
# ============================================================


GENERIC_PERSON_TERMS = {
    "writer",
    "reviewer",
    "owner",
    "user",
    "author",
    "developer",
    "engineer",
    "manager",
    "contributor"
}


def is_valid_gliner_entity(entity):

    text = str(
        entity["text"]
    ).strip()

    label = str(
        entity["label"]
    ).lower().strip()


    # Reject empty entities.
    if not text:
        return False


    # Reject generic PERSON roles.
    if (
        label == "person"
        and text.lower()
        in GENERIC_PERSON_TERMS
    ):
        return False


    return True
```

```python
# ============================================================
# STEP 67: RETEST GLINER WITH REFINED CONFIGURATION
# ============================================================
# Purpose:
# Confirm that the refined entity-label set still identifies
# useful semantic entities while reducing noisy extraction.
# ============================================================


refined_test_predictions = (
    gliner_model.predict_entities(

        test_chunk["text"],

        GLINER_ENTITY_LABELS,

        threshold=0.40
    )
)


refined_test_predictions = [

    entity

    for entity in refined_test_predictions

    if is_valid_gliner_entity(entity)
]


print("=" * 70)
print("REFINED GLINER TEST")
print("=" * 70)


print(
    "\nEntities retained:",
    len(refined_test_predictions)
)


for entity in refined_test_predictions:

    print(
        f"{entity['label']:15s}",
        "|",
        f"{entity['score']:.3f}",
        "|",
        entity["text"]
    )
```

```python
# ============================================================
# STEP 68: RUN GLINER ACROSS ALL PRD CHUNKS
# ============================================================
# Purpose:
# Apply the refined GLiNER configuration to all 21 PRD chunks.
#
# For every extracted semantic entity we preserve:
#
#   text
#   entity label/type
#   confidence score
#   PRD section
#   chunk ID
#
# This provenance is important because later we need to know
# exactly where a model-extracted entity came from.
#
# IMPORTANT:
# We are NOT performing Entity Resolution here.
#
# Therefore:
#
#   "Engineering"
#   "engineering"
#
# may remain separate extracted mentions.
#
# The downstream Entity Resolution stage will determine
# whether they represent the same canonical entity.
# ============================================================


prd_gliner_entities = []


for chunk in prd_chunks:

    # --------------------------------------------------------
    # Run GLiNER on this PRD chunk.
    # --------------------------------------------------------

    predictions = gliner_model.predict_entities(

        chunk["text"],

        GLINER_ENTITY_LABELS,

        threshold=0.40
    )


    # --------------------------------------------------------
    # Apply our quality filter.
    #
    # Example:
    # Generic PERSON terms such as "writer" are rejected.
    # --------------------------------------------------------

    predictions = [

        entity

        for entity in predictions

        if is_valid_gliner_entity(entity)
    ]


    # --------------------------------------------------------
    # Preserve every accepted entity occurrence together with
    # its evidence location.
    # --------------------------------------------------------

    for prediction in predictions:

        prd_gliner_entities.append({

            "text":
                prediction["text"],

            "label":
                prediction["label"],

            "confidence":
                float(
                    prediction["score"]
                ),

            "chunk_id":
                chunk["chunk_id"],

            "section":
                chunk["section"],

            "extraction_method":
                "gliner"
        })


print("=" * 70)
print("FULL PRD GLINER EXTRACTION")
print("=" * 70)


print(
    "\nPRD chunks processed:",
    len(prd_chunks)
)


print(
    "Semantic entity occurrences:",
    len(prd_gliner_entities)
)


print(
    "\nSUCCESS: GLiNER processed the complete PRD."
)
```

```python
# ============================================================
# STEP 69: INSPECT FULL PRD GLINER OUTPUT
# ============================================================
# Purpose:
# Evaluate the semantic extraction before merging it with the
# deterministic PRD extraction.
#
# We inspect:
#
#   1. Number of entities by type
#   2. Lowest-confidence predictions
#   3. Highest-confidence predictions
#
# This helps detect obvious noise before constructing the
# final PRD JSON.
# ============================================================

from collections import Counter


prd_gliner_label_counts = Counter(

    entity["label"]

    for entity in prd_gliner_entities
)


print("=" * 70)
print("PRD GLINER ENTITY DISTRIBUTION")
print("=" * 70)


print(
    "\nTotal entity occurrences:",
    len(prd_gliner_entities)
)


print("\nENTITY TYPES")
print("-" * 70)


for label, count in (
    prd_gliner_label_counts.most_common()
):

    print(
        f"{label:20s}",
        count
    )


# ------------------------------------------------------------
# Sort all predictions by confidence.
# ------------------------------------------------------------

sorted_prd_gliner = sorted(

    prd_gliner_entities,

    key=lambda entity:
        entity["confidence"]
)


# ------------------------------------------------------------
# Lowest confidence predictions
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("LOWEST-CONFIDENCE EXAMPLES")
print("=" * 70)


for entity in sorted_prd_gliner[:20]:

    print(
        f"{entity['confidence']:.3f}",
        "|",
        f"{entity['label']:12s}",
        "|",
        entity["text"],
        "|",
        entity["section"]
    )


# ------------------------------------------------------------
# Highest confidence predictions
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("HIGHEST-CONFIDENCE EXAMPLES")
print("=" * 70)


for entity in sorted_prd_gliner[-20:]:

    print(
        f"{entity['confidence']:.3f}",
        "|",
        f"{entity['label']:12s}",
        "|",
        entity["text"],
        "|",
        entity["section"]
    )
```

```python
# ============================================================
# STEP 70: INSPECT UNIQUE GLINER SEMANTIC ENTITIES
# ============================================================
# Purpose:
# Show unique text/type combinations produced by GLiNER.
#
# NOTE:
# This is ONLY an inspection view.
#
# We are NOT performing full Entity Resolution.
#
# For example:
#
#   Engineering
#   engineering
#
# may remain separate because canonicalization belongs to the
# next project stage.
# ============================================================


unique_gliner_entities = {}


for entity in prd_gliner_entities:

    key = (

        entity["label"],

        entity["text"]
    )


    # --------------------------------------------------------
    # If the same exact entity occurs multiple times, keep its
    # highest observed confidence for this inspection summary.
    # --------------------------------------------------------

    if (
        key not in unique_gliner_entities
        or
        entity["confidence"]
        >
        unique_gliner_entities[
            key
        ]["confidence"]
    ):

        unique_gliner_entities[
            key
        ] = entity


print("=" * 70)
print("UNIQUE PRD GLINER ENTITIES")
print("=" * 70)


print(
    "\nUnique type/text combinations:",
    len(unique_gliner_entities)
)


# ------------------------------------------------------------
# Group by semantic type.
# ------------------------------------------------------------

for label in GLINER_ENTITY_LABELS:

    entities_for_label = [

        entity

        for (
            entity_label,
            entity_text
        ), entity

        in unique_gliner_entities.items()

        if entity_label == label
    ]


    if not entities_for_label:
        continue


    entities_for_label = sorted(

        entities_for_label,

        key=lambda entity:
            -entity["confidence"]
    )


    print(
        "\n" + "-" * 70
    )

    print(
        label.upper(),
        f"({len(entities_for_label)})"
    )

    print(
        "-" * 70
    )


    for entity in entities_for_label:

        print(
            f"{entity['confidence']:.3f}",
            "|",
            entity["text"]
        )
```

```python
# ============================================================
# STEP 71: FILTER PRD GLINER ENTITIES
# ============================================================
# Purpose:
# Clean GLiNER predictions before adding them to the final
# extraction JSON.
#
# Why filtering is necessary:
#
# GLiNER correctly finds many useful entities:
#     mongos
#     mongod
#     MongoDB
#     Cloud Manager
#     Server engineering
#
# But it can also misclassify:
#     REQ-009     -> milestone
#     2016-07-12  -> milestone
#     writer      -> person
#     epic        -> system
#
# Therefore we combine:
#
#     confidence threshold
#             +
#     deterministic validation rules
#
# This improves precision before Entity Resolution / KG.
# ============================================================


GLINER_ACCEPT_THRESHOLD = 0.55


# ------------------------------------------------------------
# Generic terms that should not become standalone KG entities.
# ------------------------------------------------------------

GENERIC_ENTITY_TERMS = {

    # Generic people/roles
    "writer",
    "writers",
    "single writer",
    "reviewer",
    "reviewers",
    "owner",
    "someone with product knowledge",

    # Generic organizational words
    "team",

    # Generic planning words
    "sprint",
    "several sprints",
    "sprint cadence",

    # Generic Jira concepts
    "epic"
}


# ------------------------------------------------------------
# Date detector.
#
# Dates belong in temporal information, not MILESTONE nodes.
# ------------------------------------------------------------

DATE_PATTERN = re.compile(
    r"^\d{4}-\d{2}-\d{2}$"
)


def accept_gliner_entity(entity):
    """
    Decide whether a GLiNER prediction should be retained.

    This performs extraction-quality filtering only.
    It does NOT perform Entity Resolution.
    """

    text = str(
        entity["text"]
    ).strip()

    label = str(
        entity["label"]
    ).lower().strip()

    confidence = float(
        entity["confidence"]
    )


    # --------------------------------------------------------
    # 1. Minimum confidence
    # --------------------------------------------------------

    if confidence < GLINER_ACCEPT_THRESHOLD:
        return False


    # --------------------------------------------------------
    # 2. Reject generic terms
    # --------------------------------------------------------

    if text.lower() in GENERIC_ENTITY_TERMS:
        return False


    # --------------------------------------------------------
    # 3. REQ-* identifiers are REQUIREMENTS.
    #
    # They were already deterministically extracted.
    # Never accept a GLiNER reinterpretation of them.
    # --------------------------------------------------------

    if re.fullmatch(
        r"REQ-\d+",
        text,
        flags=re.IGNORECASE
    ):
        return False


    # --------------------------------------------------------
    # 4. Dates are temporal values, not milestone entities.
    # --------------------------------------------------------

    if DATE_PATTERN.fullmatch(text):
        return False


    return True


# ------------------------------------------------------------
# Apply filter to all PRD predictions.
# ------------------------------------------------------------

prd_gliner_filtered = [

    entity

    for entity in prd_gliner_entities

    if accept_gliner_entity(entity)
]


print("=" * 70)
print("FILTERED PRD GLINER OUTPUT")
print("=" * 70)

print(
    "\nOriginal occurrences:",
    len(prd_gliner_entities)
)

print(
    "Retained occurrences:",
    len(prd_gliner_filtered)
)

print(
    "Removed occurrences :",
    len(prd_gliner_entities)
    - len(prd_gliner_filtered)
)
```

```python
# ============================================================
# STEP 72: INSPECT FILTERED UNIQUE GLINER ENTITIES
# ============================================================
# Purpose:
# Verify the final semantic entities that will be allowed into
# the PRD extraction JSON.
#
# Exact repeated mentions are summarized here for inspection.
#
# IMPORTANT:
# Case variants such as:
#
#     Engineering
#     engineering
#
# are NOT merged here.
#
# Entity Resolution handles that downstream.
# ============================================================


filtered_unique = {}


for entity in prd_gliner_filtered:

    key = (
        entity["label"],
        entity["text"]
    )


    if (
        key not in filtered_unique
        or
        entity["confidence"]
        >
        filtered_unique[key]["confidence"]
    ):

        filtered_unique[key] = entity


print("=" * 70)
print("FILTERED UNIQUE PRD ENTITIES")
print("=" * 70)


print(
    "\nUnique entities:",
    len(filtered_unique)
)


for label in GLINER_ENTITY_LABELS:

    group = [

        entity

        for (
            entity_label,
            entity_text
        ), entity
        in filtered_unique.items()

        if entity_label == label
    ]


    if not group:
        continue


    group = sorted(
        group,
        key=lambda x: -x["confidence"]
    )


    print("\n" + "-" * 70)

    print(
        label.upper(),
        f"({len(group)})"
    )

    print("-" * 70)


    for entity in group:

        print(
            f"{entity['confidence']:.3f}",
            "|",
            entity["text"]
        )
```

```python
# ============================================================
# STEP 73A: ADD PRD-SPECIFIC RELATIONSHIP TYPES
# ============================================================
# Purpose:
# Add the additional relationship types introduced during
# PRD extraction to our master relationship vocabulary.
#
# HAS_MILESTONE:
#     Project -> Milestone
#
# HAS_STAKEHOLDER:
#     Project -> Stakeholder/Team
#
# REFERENCES_ISSUE:
#     PRD/Project -> Jira issue explicitly mentioned in PRD
#
# MENTIONS:
#     Project -> Semantic entity detected by GLiNER
#
# We check whether each type already exists before adding it
# so rerunning this cell will not create duplicates.
# ============================================================

NEW_PRD_RELATIONSHIP_TYPES = [
    "HAS_MILESTONE",
    "HAS_STAKEHOLDER",
    "REFERENCES_ISSUE",
    "MENTIONS"
]

for relationship_type in NEW_PRD_RELATIONSHIP_TYPES:

    if relationship_type not in RELATIONSHIP_TYPES:
        RELATIONSHIP_TYPES.append(
            relationship_type
        )


print("PRD relationship types added successfully.")

print("\nCurrent relationship vocabulary:")

for relationship_type in RELATIONSHIP_TYPES:
    print("-", relationship_type)
```

```python
# ============================================================
# STEP 73: MAP GLINER LABELS TO MASTER ENTITY TYPES
# ============================================================
# Purpose:
# GLiNER produces lowercase semantic labels such as:
#
#     feature
#     system
#     component
#     team
#
# Our Master Extraction Schema uses standardized uppercase
# entity types:
#
#     FEATURE
#     SYSTEM
#     COMPONENT
#     TEAM
#
# This mapping keeps Jira, PRD and Meeting outputs consistent.
# ============================================================

GLINER_TO_MASTER_TYPE = {
    "feature": "FEATURE",
    "system": "SYSTEM",
    "component": "COMPONENT",
    "team": "TEAM",
    "person": "PERSON",
    "milestone": "MILESTONE"
}


print("=" * 70)
print("GLINER -> MASTER ENTITY TYPE MAPPING")
print("=" * 70)

for source_type, master_type in GLINER_TO_MASTER_TYPE.items():
    print(f"{source_type:15s} -> {master_type}")
```

```python
# ============================================================
# STEP 74: CREATE BASE PRD MASTER RECORD
# ============================================================
# Purpose:
# Create the common extraction record for:
#
#     DOCS-I01-PRD-V1
#
# Everything extracted deterministically and by GLiNER will
# now be added to this record.
# ============================================================


PRD_SOURCE_ID = prd_file.stem
# Example:
# DOCS-I01-PRD-V1


prd_record = create_master_record(
    repository_project=PILOT_PROJECT,
    project_key="DOCS",
    instance_id=PILOT_INSTANCE,
    source_type="PRD",
    source_id=PRD_SOURCE_ID
)


# PRD source timestamp
prd_record["source"]["source_timestamp"] = prd_date


# ------------------------------------------------------------
# PRD-level temporal information
# ------------------------------------------------------------

prd_record["temporal"]["created_at"] = prd_date
prd_record["temporal"]["valid_from"] = prd_date


# ------------------------------------------------------------
# PRD-level attributes
# ------------------------------------------------------------

prd_record["attributes"].update({

    "title": prd_title,
    "version": prd_version,
    "status": prd_status,

    # Preserve structured project information.
    "scope_items": prd_scope_items,
    "acceptance_criteria": prd_acceptance_criteria,
    "risk_evidence": prd_risk_evidence,
    "open_questions": prd_open_questions
})


# ------------------------------------------------------------
# PRD-level provenance
# ------------------------------------------------------------

prd_record["provenance"] = {

    "source_type": "PRD",
    "source_id": PRD_SOURCE_ID,
    "instance_id": PILOT_INSTANCE,

    # Whole PRD record does not correspond to one chunk.
    "chunk_id": None,

    # Preserve original document as evidence.
    "evidence_text": prd_text,

    "file_name": prd_file.name
}


print(
    json.dumps(
        prd_record,
        indent=2,
        ensure_ascii=False
    )[:3000]
)
```

```python
# ============================================================
# STEP 75: ADD PROJECT AND REQUIREMENT ENTITIES
# ============================================================
# Purpose:
# Convert the deterministically parsed requirements into
# high-confidence REQUIREMENT entities.
#
# Every requirement is linked:
#
#     PROJECT --HAS_REQUIREMENT--> REQUIREMENT
#
# This information will later support:
#
#     requirement counts
#     requirement volatility
#     scope evolution
#     requirement dependencies
#     Temporal KG traversal
# ============================================================


# ------------------------------------------------------------
# Project entity
# ------------------------------------------------------------

prd_project_entity = add_entity(
    prd_record,
    "PROJECT",
    "DOCS",
    attributes={
        "project_name": "Documentation",
        "instance_id": PILOT_INSTANCE
    },
    extraction_method="structured",
    confidence=1.0
)


# ------------------------------------------------------------
# Requirement entities
# ------------------------------------------------------------

requirement_entity_ids = {}


for requirement in prd_requirements:

    requirement_id = requirement[
        "requirement_id"
    ]


    # Determine functional/non-functional type.
    requirement_number = int(
        requirement_id.split("-")[1]
    )


    requirement_type = (
        "FUNCTIONAL"
        if requirement_number <= 12
        else "NON_FUNCTIONAL"
    )


    entity_id = add_entity(
        prd_record,
        "REQUIREMENT",
        requirement_id,

        attributes={
            "text":
                requirement["text"],

            "label":
                requirement["label"],

            "requirement_type":
                requirement_type,

            "prd_version":
                prd_version,

            "prd_date":
                prd_date
        },

        extraction_method="structured",
        confidence=1.0
    )


    requirement_entity_ids[
        requirement_id
    ] = entity_id


    add_relationship(
        prd_record,
        prd_project_entity,
        "HAS_REQUIREMENT",
        entity_id,
        timestamp=prd_date,
        extraction_method="structured",
        confidence=1.0
    )
```

```python
# ============================================================
# STEP 76: ADD DETERMINISTIC MILESTONE ENTITIES
# ============================================================
# Purpose:
# Convert the 5 explicitly parsed PRD milestones into
# high-confidence MILESTONE entities.
#
# We use deterministic milestones rather than GLiNER's
# milestone predictions because the PRD contains an explicit
# milestone table.
# ============================================================


milestone_entity_ids = {}


for milestone in prd_milestones:

    milestone_name = milestone[
        "milestone"
    ]


    entity_id = add_entity(
        prd_record,
        "MILESTONE",
        milestone_name,

        attributes={
            "target":
                milestone["target"],

            "notes":
                milestone["notes"],

            "prd_version":
                prd_version,

            "prd_date":
                prd_date
        },

        extraction_method="structured",
        confidence=1.0
    )


    milestone_entity_ids[
        milestone_name
    ] = entity_id


    add_relationship(
        prd_record,
        prd_project_entity,
        "HAS_MILESTONE",
        entity_id,
        timestamp=prd_date,
        extraction_method="structured",
        confidence=1.0
    )
```

```python
# ============================================================
# STEP 77: ADD STAKEHOLDER / TEAM ENTITIES
# ============================================================
# Purpose:
# Convert explicit stakeholder groups into TEAM entities.
#
# These are deterministic because they come from the PRD's
# Stakeholders and Ownership section.
#
# This is preferable to relying only on GLiNER.
# ============================================================


stakeholder_entity_ids = {}


for stakeholder in prd_stakeholders:

    stakeholder_name = stakeholder[
        "name"
    ]


    entity_id = add_entity(
        prd_record,
        "TEAM",
        stakeholder_name,

        attributes={
            "description":
                stakeholder["description"],

            "prd_version":
                prd_version
        },

        extraction_method="structured",
        confidence=1.0
    )


    stakeholder_entity_ids[
        stakeholder_name
    ] = entity_id


    add_relationship(
        prd_record,
        prd_project_entity,
        "HAS_STAKEHOLDER",
        entity_id,
        timestamp=prd_date,
        extraction_method="structured",
        confidence=1.0
    )
```

```python
# ============================================================
# STEP 78: ADD EXPLICIT JIRA ISSUE REFERENCES
# ============================================================
# Purpose:
# Add Jira issue IDs explicitly mentioned inside the PRD.
#
# This is extremely valuable for Entity Resolution because:
#
#     PRD says DOCS-6963
#
# and Jira extraction also contains:
#
#     DOCS-6963
#
# The downstream resolver can connect these source-local
# entities into one canonical ISSUE.
# ============================================================


prd_issue_entity_ids = {}


for issue_key in unique_prd_issue_keys:

    issue_entity = add_entity(
        prd_record,
        "ISSUE",
        issue_key,
        attributes={},
        extraction_method="regex",
        confidence=1.0
    )


    prd_issue_entity_ids[
        issue_key
    ] = issue_entity


    add_relationship(
        prd_record,
        prd_project_entity,
        "REFERENCES_ISSUE",
        issue_entity,
        timestamp=prd_date,
        extraction_method="regex",
        confidence=1.0
    )
```

```python
# ============================================================
# STEP 79: ADD FILTERED GLINER SEMANTIC ENTITIES
# ============================================================
# Purpose:
# Add semantic entities that were not reliably available from
# explicit PRD structure.
#
# Examples:
#
#     mongos
#     mongod
#     MongoDB
#     CRUD
#     sharding
#     Cloud Manager
#
# IMPORTANT:
# We preserve:
#
#     extraction_method = gliner
#     confidence
#
# so downstream users can distinguish model-derived entities
# from deterministic entities.
# ============================================================


for prediction in prd_gliner_filtered:

    master_type = (
        GLINER_TO_MASTER_TYPE[
            prediction["label"]
        ]
    )


    entity_id = add_entity(
        prd_record,
        master_type,
        prediction["text"],

        attributes={
            "section":
                prediction["section"],

            "chunk_id":
                prediction["chunk_id"]
        },

        extraction_method="gliner",
        confidence=prediction[
            "confidence"
        ]
    )


    # --------------------------------------------------------
    # Generic project association.
    #
    # We deliberately do NOT infer stronger semantic
    # relationships merely because two entities occur in the
    # same paragraph.
    # --------------------------------------------------------

    add_relationship(
        prd_record,
        prd_project_entity,
        "MENTIONS",
        entity_id,
        timestamp=prd_date,
        extraction_method="gliner",
        confidence=prediction[
            "confidence"
        ]
    )
```

```python
# ============================================================
# STEP 80: PRESERVE PRD DEPENDENCY STATEMENTS
# ============================================================
# Purpose:
# Preserve all 4 explicit PRD dependency statements.
#
# Some dependency statements contain obvious entities, while
# others describe general dependencies such as reviewer
# availability.
#
# Rather than inventing source/target nodes, we preserve the
# raw dependency fact and its evidence.
#
# Entity Resolution / KG construction can later connect the
# appropriate canonical entities.
# ============================================================


for index, dependency in enumerate(
    prd_dependency_statements,
    start=1
):

    prd_record[
        "dependencies"
    ].append({

        "dependency_id":
            f"{PRD_SOURCE_ID}_DEP_{index:03d}",

        "text":
            dependency["text"],

        "source_entity":
            None,

        "target_entity":
            None,

        "dependency_type":
            "EXPLICIT_PRD_DEPENDENCY",

        "dependency_status":
            None,

        "created_at":
            prd_date,

        "removed_at":
            None,

        "extraction_method":
            "structured",

        "confidence":
            1.0
    })
```

```python
# ============================================================
# STEP 81: PRESERVE RISK EVIDENCE AND OPEN QUESTIONS
# ============================================================
# Purpose:
# Keep explicit PRD uncertainty/risk statements and unresolved
# questions without turning them into TabPFN target labels.
#
# These may later support:
#
#     evidence retrieval
#     decision uncertainty
#     ownership-gap analysis
#     GraphRAG explanations
#
# But TabPFN remains responsible for predictive risk output.
# ============================================================


prd_record["attributes"][
    "risk_evidence"
] = prd_risk_evidence


prd_record["attributes"][
    "open_questions"
] = prd_open_questions
```

```python
# ============================================================
# STEP 82: CREATE REQUIREMENT TEMPORAL EVENTS
# ============================================================
# Purpose:
# Represent the existence/introduction of requirements in PRD
# Version 1 as temporal events.
#
# If later PRD versions exist, the same design can represent:
#
#     REQUIREMENT_CREATED
#     REQUIREMENT_CHANGED
#
# which can support requirement-volatility features.
# ============================================================


for requirement_id, entity_id in (
    requirement_entity_ids.items()
):

    add_event(
        prd_record,
        event_type="REQUIREMENT_CREATED",
        entity=entity_id,
        event_time=prd_date,
        actor=None,
        from_value=None,
        to_value=requirement_id,

        attributes={
            "prd_version":
                prd_version
        },

        extraction_method="structured"
    )
```

```python
# ============================================================
# STEP 83: VALIDATE COMPLETE PRD EXTRACTION
# ============================================================
# Purpose:
# Inspect the final combined deterministic + GLiNER PRD
# extraction before saving it.
# ============================================================

from collections import Counter


prd_entity_counts = Counter(
    entity["type"]
    for entity in prd_record["entities"]
)


prd_relation_counts = Counter(
    relation["type"]
    for relation in prd_record["relationships"]
)


prd_event_counts = Counter(
    event["event_type"]
    for event in prd_record["events"]
)


print("=" * 70)
print("FINAL PRD EXTRACTION SUMMARY")
print("=" * 70)


print(
    "\nSource:",
    prd_record["source"]["source_id"]
)

print(
    "Entities:",
    len(prd_record["entities"])
)

print(
    "Relationships:",
    len(prd_record["relationships"])
)

print(
    "Events:",
    len(prd_record["events"])
)

print(
    "Dependencies:",
    len(prd_record["dependencies"])
)


print("\nENTITY TYPES")
print("-" * 70)

for entity_type, count in (
    prd_entity_counts.most_common()
):

    print(
        f"{entity_type:25s}",
        count
    )


print("\nRELATIONSHIP TYPES")
print("-" * 70)

for relation_type, count in (
    prd_relation_counts.most_common()
):

    print(
        f"{relation_type:25s}",
        count
    )


print("\nEVENT TYPES")
print("-" * 70)

for event_type, count in (
    prd_event_counts.most_common()
):

    print(
        f"{event_type:25s}",
        count
    )


print("\nStructured information")
print("-" * 70)

print(
    "Requirements        :",
    len(prd_requirements)
)

print(
    "Milestones          :",
    len(prd_milestones)
)

print(
    "Stakeholders        :",
    len(prd_stakeholders)
)

print(
    "Scope items         :",
    len(prd_scope_items)
)

print(
    "Acceptance criteria :",
    len(prd_acceptance_criteria)
)

print(
    "Risk evidence       :",
    len(prd_risk_evidence)
)

print(
    "Open questions      :",
    len(prd_open_questions)
)
```

```python
# ============================================================
# STEP 84: SAVE PILOT PRD EXTRACTION
# ============================================================
# Purpose:
# Save the completed deterministic + GLiNER PRD extraction.
#
# Output:
#
# GC1_Entity_Extraction/
# └── 24_DOCS/
#     └── DOCS-I01/
#         ├── jira_extracted.jsonl
#         └── prd_extracted.jsonl
#
# We use JSONL for consistency with the Jira extraction and
# with the downstream Entity Resolution pipeline.
#
# This PRD currently produces ONE source-level record, so the
# JSONL file contains one line.
# ============================================================


prd_output_file = (
    pilot_output_folder
    / "prd_extracted.jsonl"
)


with open(
    prd_output_file,
    "w",
    encoding="utf-8"
) as file:

    file.write(
        json.dumps(
            prd_record,
            ensure_ascii=False
        )
        + "\n"
    )


print("=" * 70)
print("PRD EXTRACTION SAVED")
print("=" * 70)

print("\nOutput file:")
print(prd_output_file)

print(
    "\nFile exists:",
    prd_output_file.exists()
)

print(
    "File size:",
    round(
        prd_output_file.stat().st_size / 1024,
        2
    ),
    "KB"
)
```

```python
# ============================================================
# STEP 85: VALIDATE SAVED PRD EXTRACTION
# ============================================================
# Purpose:
# Reload the saved PRD JSONL and verify that:
#
#   - it is valid JSON
#   - exactly one PRD record exists
#   - all mandatory master-schema sections exist
#   - all 16 requirements survived
#   - all 4 dependency statements survived
#
# This ensures the saved handoff artifact matches what we
# validated in memory.
# ============================================================


reloaded_prd_records = []


with open(
    prd_output_file,
    "r",
    encoding="utf-8"
) as file:

    for line_number, line in enumerate(
        file,
        start=1
    ):

        if not line.strip():
            continue

        try:

            reloaded_prd_records.append(
                json.loads(line)
            )

        except json.JSONDecodeError as error:

            print(
                "Invalid JSON on line:",
                line_number
            )

            raise error


# ------------------------------------------------------------
# Required master-schema sections
# ------------------------------------------------------------

required_sections = [
    "entities",
    "relationships",
    "events",
    "decisions",
    "action_items",
    "blockers",
    "dependencies",
    "temporal",
    "attributes",
    "provenance"
]


missing_sections = []


for record in reloaded_prd_records:

    for section in required_sections:

        if section not in record:

            missing_sections.append(
                section
            )


# ------------------------------------------------------------
# Count requirements after reload
# ------------------------------------------------------------

reloaded_requirement_count = sum(

    1

    for record in reloaded_prd_records

    for entity in record["entities"]

    if entity["type"] == "REQUIREMENT"
)


# ------------------------------------------------------------
# Count dependencies
# ------------------------------------------------------------

reloaded_dependency_count = sum(

    len(record["dependencies"])

    for record in reloaded_prd_records
)


print("=" * 70)
print("SAVED PRD FILE VALIDATION")
print("=" * 70)


print(
    "\nRecords reloaded:",
    len(reloaded_prd_records)
)

print(
    "Missing required sections:",
    len(missing_sections)
)

print(
    "Requirements:",
    reloaded_requirement_count
)

print(
    "Dependencies:",
    reloaded_dependency_count
)


if (
    len(reloaded_prd_records) == 1
    and len(missing_sections) == 0
    and reloaded_requirement_count == 16
    and reloaded_dependency_count == 4
):

    print(
        "\nSUCCESS:"
        "\nPilot PRD extraction JSONL is valid "
        "and ready for downstream use."
    )

else:

    print(
        "\nCHECK REQUIRED:"
        "\nOne or more PRD validation checks failed."
    )
```

```python
# ============================================================
# STEP 86: LOAD PILOT MEETING TRANSCRIPTS
# ============================================================
# Purpose:
# Load the two final meeting transcripts belonging to DOCS-I01.
#
# Expected:
#
#     DOCS-I01-M001.txt
#     DOCS-I01-M002.txt
#
# IMPORTANT:
# .prompt.txt files were already excluded earlier and must
# never enter the extraction pipeline.
# ============================================================


meeting_documents = []


for meeting_file in meeting_files:

    with open(
        meeting_file,
        "r",
        encoding="utf-8"
    ) as file:

        meeting_text = file.read()


    meeting_documents.append({

        "file_path":
            meeting_file,

        "file_name":
            meeting_file.name,

        "text":
            meeting_text
    })


print("=" * 70)
print("MEETING DOCUMENTS LOADED")
print("=" * 70)

print(
    "\nMeetings loaded:",
    len(meeting_documents)
)


for meeting in meeting_documents:

    print(
        "\n",
        meeting["file_name"],
        "|",
        len(meeting["text"]),
        "characters"
    )
```

```python
# ============================================================
# STEP 87: PARSE MEETING HEADER METADATA
# ============================================================
# Purpose:
# Extract high-confidence metadata directly from each meeting.
#
# Fields:
#
#     meeting_id
#     meeting_title
#     meeting_datetime
#     participants
#
# Participants are preserved exactly as written at this stage.
# Entity Resolution can later normalize names/roles.
# ============================================================


def extract_header_value(
    text,
    field_name
):
    """
    Extract a single-line meeting header field.

    Example:
        Meeting ID: DOCS-I01-M002
    """

    pattern = (
        rf"^{re.escape(field_name)}:\s*(.+)$"
    )


    match = re.search(
        pattern,
        text,
        flags=(
            re.IGNORECASE
            | re.MULTILINE
        )
    )


    if match:
        return match.group(1).strip()

    return None


parsed_meetings = []


for meeting in meeting_documents:

    text = meeting["text"]


    meeting_id = extract_header_value(
        text,
        "Meeting ID"
    )


    meeting_title = extract_header_value(
        text,
        "Meeting title"
    )


    meeting_datetime = extract_header_value(
        text,
        "Date/time"
    )


    participants_raw = extract_header_value(
        text,
        "Participants"
    )


    # --------------------------------------------------------
    # Participants are comma-separated in our generated format.
    # --------------------------------------------------------

    participants = []


    if participants_raw:

        participants = [

            participant.strip()

            for participant
            in participants_raw.split(",")

            if participant.strip()
        ]


    parsed_meetings.append({

        "file_name":
            meeting["file_name"],

        "text":
            text,

        "meeting_id":
            meeting_id,

        "meeting_title":
            meeting_title,

        "meeting_datetime":
            meeting_datetime,

        "participants_raw":
            participants_raw,

        "participants":
            participants
    })


print("=" * 70)
print("MEETING METADATA")
print("=" * 70)


for meeting in parsed_meetings:

    print("\n" + "-" * 70)

    print(
        "Meeting ID    :",
        meeting["meeting_id"]
    )

    print(
        "Title         :",
        meeting["meeting_title"]
    )

    print(
        "Date/time     :",
        meeting["meeting_datetime"]
    )

    print(
        "Participants  :",
        len(meeting["participants"])
    )


    for participant in meeting["participants"]:

        print(
            "   -",
            participant
        )
```

```python
# ============================================================
# STEP 88: PARSE MEETING UTTERANCES
# ============================================================
# Purpose:
# Convert the transcript into individual speaker utterances.
#
# Example:
#
#     [00:12] Marcus: Yes. There's no comment...
#
# becomes:
#
#     speaker_timestamp = 00:12
#     speaker           = Marcus
#     text              = Yes. There's no comment...
#
# WHY THIS MATTERS:
#
# Later we can associate:
#
#     decisions
#     action items
#     blockers
#     dependencies
#
# with the person who actually said them and their position
# within the meeting.
#
# This is valuable for Temporal KG provenance.
# ============================================================


UTTERANCE_PATTERN = re.compile(

    r"^\[(\d{2}:\d{2})\]\s+"
    r"([^:]+):\s*"
    r"(.*)$"
)


def parse_meeting_utterances(
    meeting_text
):

    utterances = []


    for line in meeting_text.splitlines():

        line = line.strip()


        match = UTTERANCE_PATTERN.match(
            line
        )


        if not match:
            continue


        utterances.append({

            "speaker_timestamp":
                match.group(1).strip(),

            "speaker":
                match.group(2).strip(),

            "text":
                match.group(3).strip()
        })


    return utterances


for meeting in parsed_meetings:

    meeting["utterances"] = (
        parse_meeting_utterances(
            meeting["text"]
        )
    )


print("=" * 70)
print("MEETING UTTERANCE PARSING")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\n",
        meeting["meeting_id"],
        "| Utterances:",
        len(meeting["utterances"])
    )


    print("\nFirst 3 utterances:")


    for utterance in (
        meeting["utterances"][:3]
    ):

        print(
            f"[{utterance['speaker_timestamp']}]",
            utterance["speaker"],
            "->",
            utterance["text"]
        )
```

```python
# ============================================================
# STEP 89: EXTRACT EXPLICIT MEETING REFERENCES
# ============================================================
# Purpose:
# Deterministically identify explicit Jira issue IDs and PRD
# requirement IDs appearing in each meeting.
#
# This gives Entity Resolution strong cross-source anchors:
#
#     Jira DOCS-8052
#            ↕
#     Meeting DOCS-8052
#
# and:
#
#     PRD REQ-003
#            ↕
#     Meeting REQ-003
# ============================================================


REQUIREMENT_ID_PATTERN = re.compile(
    r"\bREQ-\d+\b",
    flags=re.IGNORECASE
)


MEETING_ISSUE_PATTERN = re.compile(
    r"\b[A-Z][A-Z0-9_]+-\d+\b"
)


for meeting in parsed_meetings:

    meeting_issue_occurrences = []

    meeting_requirement_occurrences = []


    for utterance_index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        text = utterance["text"]


        # ----------------------------------------------------
        # Requirements
        # ----------------------------------------------------

        requirement_ids = (
            REQUIREMENT_ID_PATTERN.findall(
                text
            )
        )


        for requirement_id in requirement_ids:

            meeting_requirement_occurrences.append({

                "requirement_id":
                    requirement_id.upper(),

                "utterance_index":
                    utterance_index,

                "speaker":
                    utterance["speaker"],

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    text
            })


        # ----------------------------------------------------
        # Jira-style identifiers
        # ----------------------------------------------------

        issue_keys = (
            MEETING_ISSUE_PATTERN.findall(
                text
            )
        )


        for issue_key in issue_keys:

            # REQ-* belongs to REQUIREMENT, not ISSUE.
            if issue_key.upper().startswith(
                "REQ-"
            ):
                continue


            meeting_issue_occurrences.append({

                "issue_key":
                    issue_key,

                "utterance_index":
                    utterance_index,

                "speaker":
                    utterance["speaker"],

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    text
            })


    meeting[
        "issue_occurrences"
    ] = meeting_issue_occurrences


    meeting[
        "requirement_occurrences"
    ] = meeting_requirement_occurrences


print("=" * 70)
print("MEETING EXPLICIT REFERENCES")
print("=" * 70)


for meeting in parsed_meetings:

    unique_issues = sorted(
        set(
            item["issue_key"]

            for item
            in meeting[
                "issue_occurrences"
            ]
        )
    )


    unique_requirements = sorted(
        set(
            item["requirement_id"]

            for item
            in meeting[
                "requirement_occurrences"
            ]
        )
    )


    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    print(
        "Issue occurrences:",
        len(
            meeting[
                "issue_occurrences"
            ]
        )
    )


    print(
        "Unique issues:",
        len(unique_issues)
    )


    for issue in unique_issues:
        print("  ISSUE:", issue)


    print(
        "Requirement occurrences:",
        len(
            meeting[
                "requirement_occurrences"
            ]
        )
    )


    print(
        "Unique requirements:",
        len(unique_requirements)
    )


    for requirement in unique_requirements:
        print(
            "  REQUIREMENT:",
            requirement
        )
```

```python
# ============================================================
# STEP 90: VALIDATE DETERMINISTIC MEETING PARSING
# ============================================================
# Purpose:
# Verify that both meetings were parsed correctly before
# semantic extraction.
#
# We check:
#
#     meeting metadata
#     participants
#     utterances
#     issue references
#     requirement references
# ============================================================


print("=" * 70)
print("DETERMINISTIC MEETING PARSING SUMMARY")
print("=" * 70)


for meeting in parsed_meetings:

    unique_issues = set(
        item["issue_key"]

        for item
        in meeting["issue_occurrences"]
    )


    unique_requirements = set(
        item["requirement_id"]

        for item
        in meeting[
            "requirement_occurrences"
        ]
    )


    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )

    print(
        "Title:",
        meeting["meeting_title"]
    )

    print(
        "Date:",
        meeting["meeting_datetime"]
    )

    print(
        "Participants:",
        len(meeting["participants"])
    )

    print(
        "Utterances:",
        len(meeting["utterances"])
    )

    print(
        "Unique Jira issues:",
        len(unique_issues)
    )

    print(
        "Unique requirements:",
        len(unique_requirements)
    )
```

```python
# ============================================================
# STEP 91: DETECT MEETING ACTION-ITEM CANDIDATES
# ============================================================
# Purpose:
# Identify utterances that may describe future work,
# commitments, assignments or follow-up actions.
#
# Examples:
#
#   "I'll verify it..."
#   "I'll ping them again..."
#   "Dana, find out where INT-1680 stands."
#   "Marcus: page lists ... by Wednesday"
#
# IMPORTANT:
# This step creates CANDIDATES only.
#
# We do not immediately turn every candidate into a final
# ACTION_ITEM because simple language rules can create false
# positives.
#
# Later we validate and structure:
#
#   action owner
#   action text
#   target entity
#   due time
#   status
#   evidence
# ============================================================


ACTION_PATTERNS = [

    # First-person commitments
    r"\bI['’]?ll\b",
    r"\bI will\b",
    r"\bI need to\b",
    r"\bI['’]?m going to\b",

    # Explicit requested actions
    r"\bneed to\b",
    r"\baction is to\b",

    # Planning language
    r"\bwill do\b",
    r"\bwill verify\b",
    r"\bwill check\b",
    r"\bwill find out\b",
    r"\bwill reply\b",
    r"\bwill assign\b",
    r"\bwill reopen\b",
    r"\bwill publish\b"
]


action_regex = re.compile(
    "|".join(ACTION_PATTERNS),
    flags=re.IGNORECASE
)


for meeting in parsed_meetings:

    action_candidates = []


    for index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        if action_regex.search(
            utterance["text"]
        ):

            action_candidates.append({

                "utterance_index":
                    index,

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "speaker":
                    utterance["speaker"],

                "text":
                    utterance["text"],

                "candidate_type":
                    "ACTION_ITEM"
            })


    meeting[
        "action_candidates"
    ] = action_candidates


print("=" * 70)
print("ACTION-ITEM CANDIDATES")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )

    print(
        "Candidates:",
        len(
            meeting[
                "action_candidates"
            ]
        )
    )


    for candidate in meeting[
        "action_candidates"
    ]:

        print(
            f"\n[{candidate['speaker_timestamp']}]",
            candidate["speaker"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 92: DETECT MEETING DECISION CANDIDATES
# ============================================================
# Purpose:
# Identify utterances that may contain:
#
#   decisions
#   approvals
#   deferrals
#   explicit choices
#   reversals
#
# Examples:
#
#   "It stays out of scope."
#   "I'll reopen it as low priority."
#   "I'm not deciding that here."
#   "We'll decide at the start of next sprint."
#
# IMPORTANT:
# A deferred decision is NOT the same as a completed decision.
#
# Therefore this step only finds candidate evidence.
# ============================================================


DECISION_PATTERNS = [

    r"\bdecid(?:e|ed|ing)\b",
    r"\bwe['’]?ll decide\b",
    r"\bnot deciding\b",

    r"\bagreed\b",
    r"\bagree\b",

    r"\bstays?\b",
    r"\bwill remain\b",

    r"\breopen\b",
    r"\bmove\b.*\bnext sprint\b",

    r"\bleave\b.*\bout\b",

    r"\bout of scope\b",

    r"\bwe will\b",
    r"\bwe['’]?ll\b"
]


decision_regex = re.compile(
    "|".join(DECISION_PATTERNS),
    flags=re.IGNORECASE
)


for meeting in parsed_meetings:

    decision_candidates = []


    for index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        if decision_regex.search(
            utterance["text"]
        ):

            decision_candidates.append({

                "utterance_index":
                    index,

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "speaker":
                    utterance["speaker"],

                "text":
                    utterance["text"],

                "candidate_type":
                    "DECISION"
            })


    meeting[
        "decision_candidates"
    ] = decision_candidates


print("=" * 70)
print("DECISION CANDIDATES")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )

    print(
        "Candidates:",
        len(
            meeting[
                "decision_candidates"
            ]
        )
    )


    for candidate in meeting[
        "decision_candidates"
    ]:

        print(
            f"\n[{candidate['speaker_timestamp']}]",
            candidate["speaker"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 93: DETECT MEETING DEPENDENCY CANDIDATES
# ============================================================
# Purpose:
# Identify statements describing dependencies or prerequisites.
#
# Examples:
#
#   "depends on INT-1680"
#   "waiting on..."
#   "subject to..."
#   "gates the tutorial"
#
# We preserve the complete utterance because later we need to
# identify the dependency endpoints and its status.
# ============================================================


DEPENDENCY_PATTERNS = [

    r"\bdepends?\s+on\b",
    r"\bdependency\b",
    r"\bwaiting\s+on\b",
    r"\bwaits?\s+on\b",
    r"\bsubject\s+to\b",
    r"\bprerequisite\b",
    r"\bgates?\b",
    r"\bwaiting\s+for\b"
]


dependency_regex = re.compile(
    "|".join(DEPENDENCY_PATTERNS),
    flags=re.IGNORECASE
)


for meeting in parsed_meetings:

    dependency_candidates = []


    for index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        if dependency_regex.search(
            utterance["text"]
        ):

            dependency_candidates.append({

                "utterance_index":
                    index,

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "speaker":
                    utterance["speaker"],

                "text":
                    utterance["text"],

                "candidate_type":
                    "DEPENDENCY"
            })


    meeting[
        "dependency_candidates"
    ] = dependency_candidates


print("=" * 70)
print("DEPENDENCY CANDIDATES")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )

    print(
        "Candidates:",
        len(
            meeting[
                "dependency_candidates"
            ]
        )
    )


    for candidate in meeting[
        "dependency_candidates"
    ]:

        print(
            f"\n[{candidate['speaker_timestamp']}]",
            candidate["speaker"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 94: DETECT BLOCKER CANDIDATES WITH NEGATION AWARENESS
# ============================================================
# Purpose:
# Identify blocker-related statements while distinguishing:
#
#     POSITIVE blocker evidence
#
# from:
#
#     NEGATED blocker evidence
#
# Example from our meeting:
#
#     "It's a dependency I can't confirm,
#      not a blocker today."
#
# A naive keyword rule would incorrectly create a blocker.
#
# We therefore mark the candidate as:
#
#     POSITIVE
#     NEGATED
#
# and preserve the evidence text.
# ============================================================


BLOCKER_PATTERN = re.compile(

    r"\bblock(?:ed|er|ing)?\b"
    r"|\bunblocks?\b",

    flags=re.IGNORECASE
)


BLOCKER_NEGATION_PATTERNS = [

    r"\bnot\s+(?:a\s+)?blocker\b",

    r"\bnot\s+blocked\b",

    r"\bisn['’]?t\s+blocking\b",

    r"\bis\s+not\s+blocking\b",

    r"\bdoesn['’]?t\s+block\b",

    r"\bnot\s+blocking\b"
]


blocker_negation_regex = re.compile(

    "|".join(
        BLOCKER_NEGATION_PATTERNS
    ),

    flags=re.IGNORECASE
)


for meeting in parsed_meetings:

    blocker_candidates = []


    for index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        text = utterance["text"]


        if not BLOCKER_PATTERN.search(text):
            continue


        is_negated = bool(
            blocker_negation_regex.search(
                text
            )
        )


        blocker_candidates.append({

            "utterance_index":
                index,

            "speaker_timestamp":
                utterance[
                    "speaker_timestamp"
                ],

            "speaker":
                utterance["speaker"],

            "text":
                text,

            "candidate_type":
                "BLOCKER",

            "polarity":
                (
                    "NEGATED"
                    if is_negated
                    else "POSITIVE"
                )
        })


    meeting[
        "blocker_candidates"
    ] = blocker_candidates


print("=" * 70)
print("BLOCKER CANDIDATES")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    print(
        "Candidates:",
        len(
            meeting[
                "blocker_candidates"
            ]
        )
    )


    for candidate in meeting[
        "blocker_candidates"
    ]:

        print(
            f"\n[{candidate['speaker_timestamp']}]",
            candidate["speaker"],
            "|",
            candidate["polarity"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 95: MEETING CANDIDATE EXTRACTION SUMMARY
# ============================================================
# Purpose:
# Summarize all candidate information extracted from both
# meetings before final structuring.
#
# We want to know whether our conservative rules are producing
# reasonable candidate volumes.
# ============================================================


print("=" * 70)
print("MEETING CANDIDATE EXTRACTION SUMMARY")
print("=" * 70)


for meeting in parsed_meetings:

    positive_blockers = sum(

        1

        for candidate in meeting[
            "blocker_candidates"
        ]

        if candidate[
            "polarity"
        ] == "POSITIVE"
    )


    negated_blockers = sum(

        1

        for candidate in meeting[
            "blocker_candidates"
        ]

        if candidate[
            "polarity"
        ] == "NEGATED"
    )


    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    print(
        "Action candidates     :",
        len(
            meeting[
                "action_candidates"
            ]
        )
    )


    print(
        "Decision candidates   :",
        len(
            meeting[
                "decision_candidates"
            ]
        )
    )


    print(
        "Dependency candidates :",
        len(
            meeting[
                "dependency_candidates"
            ]
        )
    )


    print(
        "Positive blockers     :",
        positive_blockers
    )


    print(
        "Negated blockers      :",
        negated_blockers
    )
```

```python
# ============================================================
# STEP 96: REFINE BLOCKER CANDIDATES
# ============================================================
# Purpose:
# Convert raw blocker candidates into more precise states.
#
# Raw keyword detection is NOT enough.
#
# Examples:
#
#   "DOCS-8052 is Blocked"
#       -> ACTIVE
#
#   "not a blocker today"
#       -> NEGATED
#
#   "Is that blocking you?"
#       -> QUESTION
#
#   "That unblocks you"
#       -> UNBLOCKING
#
#   "If X blocks Y..."
#       -> CONDITIONAL
#
# Only ACTIVE statements should later become current
# blocker facts.
# ============================================================


def classify_blocker_candidate(candidate):

    text = candidate["text"].strip()
    text_lower = text.lower()


    # --------------------------------------------------------
    # 1. Explicit negation
    # --------------------------------------------------------

    if blocker_negation_regex.search(text):
        return "NEGATED"


    # --------------------------------------------------------
    # 2. Question rather than assertion
    # --------------------------------------------------------

    if (
        text.endswith("?")
        or
        re.search(
            r"\b(is|are|does|do|did|will|would|could|can)\b.*\bblock",
            text_lower
        )
        and "?" in text
    ):
        return "QUESTION"


    # --------------------------------------------------------
    # 3. Unblocking statement
    # --------------------------------------------------------

    if re.search(
        r"\bunblocks?\b|\bunblocked\b",
        text_lower
    ):
        return "UNBLOCKING"


    # --------------------------------------------------------
    # 4. Conditional blocker statement
    # --------------------------------------------------------

    if re.search(
        r"\bif\b.*\bblock",
        text_lower
    ):
        return "CONDITIONAL"


    # --------------------------------------------------------
    # 5. Otherwise treat explicit blocker assertion as ACTIVE.
    # --------------------------------------------------------

    return "ACTIVE"


for meeting in parsed_meetings:

    for candidate in meeting[
        "blocker_candidates"
    ]:

        candidate["blocker_state"] = (
            classify_blocker_candidate(
                candidate
            )
        )


print("=" * 70)
print("REFINED BLOCKER EVIDENCE")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    for candidate in meeting[
        "blocker_candidates"
    ]:

        print(
            f"[{candidate['speaker_timestamp']}]",
            candidate["blocker_state"],
            "|",
            candidate["speaker"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 97: REFINE DEPENDENCY CANDIDATES
# ============================================================
# Purpose:
# Characterize dependency statements more accurately.
#
# Possible states:
#
#   CONFIRMED
#   UNCERTAIN
#   NEGATED
#   QUESTION
#   CONDITIONAL
#
# IMPORTANT:
# Some utterances contain multiple clauses and may contain
# both negated and positive/uncertain dependency evidence.
#
# Therefore the original evidence text is ALWAYS preserved.
# ============================================================


def classify_dependency_candidate(candidate):

    text = candidate["text"].strip()
    lower = text.lower()


    # --------------------------------------------------------
    # Question
    # --------------------------------------------------------

    if "?" in text and re.search(
        r"\bdepend|\bwaiting|\bgate",
        lower
    ):
        return "QUESTION"


    # --------------------------------------------------------
    # Explicit uncertainty
    # --------------------------------------------------------

    uncertainty_patterns = [

        r"\bcan't confirm\b",
        r"\bcannot confirm\b",
        r"\bdon't know\b",
        r"\bdo not know\b",
        r"\bnot sure\b",
        r"\bunclear\b",
        r"\bcan't tell\b"
    ]


    if any(
        re.search(pattern, lower)
        for pattern in uncertainty_patterns
    ):
        return "UNCERTAIN"


    # --------------------------------------------------------
    # Explicit negation
    # --------------------------------------------------------

    if re.search(
        r"\bdoesn['’]?t depend\b"
        r"|\bdoes not depend\b"
        r"|\bnot dependent\b",
        lower
    ):
        return "NEGATED"


    # --------------------------------------------------------
    # Conditional
    # --------------------------------------------------------

    if re.search(
        r"\bif\b.*\bdepend",
        lower
    ):
        return "CONDITIONAL"


    return "CONFIRMED"


for meeting in parsed_meetings:

    for candidate in meeting[
        "dependency_candidates"
    ]:

        candidate[
            "dependency_state"
        ] = classify_dependency_candidate(
            candidate
        )


print("=" * 70)
print("REFINED DEPENDENCY EVIDENCE")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    for candidate in meeting[
        "dependency_candidates"
    ]:

        print(
            f"[{candidate['speaker_timestamp']}]",
            candidate["dependency_state"],
            "|",
            candidate["speaker"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 98: REFINE DECISION CANDIDATES
# ============================================================
# Purpose:
# Distinguish actual decisions from proposals, tentative
# choices and explicitly deferred decisions.
#
# This prevents statements such as:
#
#   "I'm not deciding that here"
#
# from becoming DECISION_MADE.
#
# Later this distinction can support:
#
#   decision count
#   unresolved decision count
#   decision reversal detection
#   decision latency
# ============================================================


def classify_decision_candidate(candidate):

    text = candidate["text"].strip()
    lower = text.lower()


    # --------------------------------------------------------
    # Explicitly deferred
    # --------------------------------------------------------

    deferred_patterns = [

        r"\bnot deciding\b",
        r"\brather not decide\b",
        r"\bwe['’]?ll decide\b",
        r"\bwe will decide\b",
        r"\bdecide then\b",
        r"\bopen question\b"
    ]


    if any(
        re.search(pattern, lower)
        for pattern in deferred_patterns
    ):
        return "DEFERRED"


    # --------------------------------------------------------
    # Tentative/non-final choice
    # --------------------------------------------------------

    tentative_patterns = [

        r"\btentatively\b",
        r"\bnot a firm commitment\b",
        r"\bi['’]?d lean\b",
        r"\bi would lean\b",
        r"\bi think\b",
        r"\bi['’]?d rather\b"
    ]


    if any(
        re.search(pattern, lower)
        for pattern in tentative_patterns
    ):
        return "TENTATIVE"


    # --------------------------------------------------------
    # Proposal/request
    # --------------------------------------------------------

    proposed_patterns = [

        r"\bi['’]?d like to\b",
        r"\bpropos",
        r"\bshould we\b",
        r"\bcan we\b"
    ]


    if any(
        re.search(pattern, lower)
        for pattern in proposed_patterns
    ):
        return "PROPOSED"


    # --------------------------------------------------------
    # Historical decision
    # --------------------------------------------------------

    historical_patterns = [

        r"\blast time we agreed\b",
        r"\bwe agreed\b"
    ]


    if any(
        re.search(pattern, lower)
        for pattern in historical_patterns
    ):
        return "HISTORICAL"


    # --------------------------------------------------------
    # Remaining candidate = explicit/current decision.
    # --------------------------------------------------------

    return "MADE"


for meeting in parsed_meetings:

    for candidate in meeting[
        "decision_candidates"
    ]:

        candidate[
            "decision_state"
        ] = classify_decision_candidate(
            candidate
        )


print("=" * 70)
print("REFINED DECISION EVIDENCE")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    for candidate in meeting[
        "decision_candidates"
    ]:

        print(
            f"[{candidate['speaker_timestamp']}]",
            candidate["decision_state"],
            "|",
            candidate["speaker"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 99: FILTER FINAL ACTION ITEMS
# ============================================================
# Purpose:
# Remove weak/false action candidates while preserving genuine
# commitments and assignments.
#
# Strong actions include:
#
#   I'll check...
#   I'll verify...
#   I'll assign...
#   I need to...
#   action is to...
#
# Meeting recap statements are also preserved because they
# explicitly summarize assigned work.
# ============================================================


STRONG_ACTION_PATTERNS = [

    r"\bI['’]?ll\s+(?:check|verify|ask|find|find out|ping|write|bring|assign|reopen|reply|add|put|publish|flag|set|go through|aim|scope)",

    r"\bI will\s+",

    r"\bI need to\s+",

    r"\baction is to\s+",

    r"\bwill do\b",

    r"\brecap\b",

    r"\bto recap actions\b"
]


strong_action_regex = re.compile(
    "|".join(
        STRONG_ACTION_PATTERNS
    ),
    flags=re.IGNORECASE
)


for meeting in parsed_meetings:

    final_action_candidates = []


    for candidate in meeting[
        "action_candidates"
    ]:

        if strong_action_regex.search(
            candidate["text"]
        ):

            final_action_candidates.append(
                candidate
            )


    meeting[
        "final_action_candidates"
    ] = final_action_candidates


print("=" * 70)
print("FILTERED ACTION ITEMS")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )

    print(
        "Raw candidates:",
        len(
            meeting["action_candidates"]
        )
    )

    print(
        "Retained:",
        len(
            meeting[
                "final_action_candidates"
            ]
        )
    )


    for candidate in meeting[
        "final_action_candidates"
    ]:

        print(
            f"\n[{candidate['speaker_timestamp']}]",
            candidate["speaker"],
            "->",
            candidate["text"]
        )
```

```python
# ============================================================
# STEP 100: FINAL RULE-BASED MEETING EXTRACTION SUMMARY
# ============================================================
# Purpose:
# Summarize the cleaned meeting evidence before adding GLiNER.
#
# At this point we have:
#
#   explicit Jira/REQ references
#   action items
#   decision states
#   dependency states
#   blocker states
#
# GLiNER will next add semantic entities such as systems,
# features, components and teams.
# ============================================================


from collections import Counter


print("=" * 70)
print("FINAL RULE-BASED MEETING SUMMARY")
print("=" * 70)


for meeting in parsed_meetings:

    decision_states = Counter(

        candidate["decision_state"]

        for candidate in meeting[
            "decision_candidates"
        ]
    )


    dependency_states = Counter(

        candidate["dependency_state"]

        for candidate in meeting[
            "dependency_candidates"
        ]
    )


    blocker_states = Counter(

        candidate["blocker_state"]

        for candidate in meeting[
            "blocker_candidates"
        ]
    )


    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    print(
        "Final action items:",
        len(
            meeting[
                "final_action_candidates"
            ]
        )
    )


    print(
        "Decision states:",
        dict(decision_states)
    )


    print(
        "Dependency states:",
        dict(dependency_states)
    )


    print(
        "Blocker states:",
        dict(blocker_states)
    )
```

```python
# ============================================================
# STEP 101: CREATE SPEAKER-AWARE MEETING GLINER CHUNKS
# ============================================================
# Purpose:
# Prepare each meeting utterance for semantic entity extraction.
#
# We preserve:
#
#   meeting ID
#   utterance index
#   speaker
#   speaker timestamp
#   utterance text
#
# GLiNER will extract semantic entities from the utterance text.
#
# We do NOT combine the complete meeting into one large chunk,
# because speaker-level provenance is important for:
#
#   decisions
#   actions
#   ownership
#   evidence tracing
# ============================================================


meeting_gliner_chunks = []


for meeting in parsed_meetings:

    for index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        meeting_gliner_chunks.append({

            "meeting_id":
                meeting["meeting_id"],

            "utterance_index":
                index,

            "speaker":
                utterance["speaker"],

            "speaker_timestamp":
                utterance[
                    "speaker_timestamp"
                ],

            "text":
                utterance["text"]
        })


print("=" * 70)
print("MEETING GLINER CHUNKS")
print("=" * 70)


print(
    "\nTotal utterances prepared:",
    len(meeting_gliner_chunks)
)


for meeting in parsed_meetings:

    count = sum(

        1

        for chunk in meeting_gliner_chunks

        if chunk["meeting_id"]
        == meeting["meeting_id"]
    )


    print(
        meeting["meeting_id"],
        ":",
        count
    )
```

```python
# ============================================================
# STEP 102: RUN GLINER ON MEETING UTTERANCES
# ============================================================
# Purpose:
# Extract semantic entities from all 109 speaker utterances.
#
# We reuse the PRD configuration:
#
#   feature
#   system
#   component
#   team
#   person
#   milestone
#
# and:
#
#   threshold = 0.40 during model inference
#
# followed by our stricter acceptance filter:
#
#   confidence >= 0.55
#   + deterministic false-positive filtering
#
# This preserves consistency between PRD and Meeting
# extraction.
# ============================================================


meeting_gliner_entities = []


for chunk in meeting_gliner_chunks:

    predictions = (
        gliner_model.predict_entities(

            chunk["text"],

            GLINER_ENTITY_LABELS,

            threshold=0.40
        )
    )


    for prediction in predictions:

        candidate = {

            "text":
                prediction["text"],

            "label":
                prediction["label"],

            "confidence":
                float(
                    prediction["score"]
                ),

            "meeting_id":
                chunk["meeting_id"],

            "utterance_index":
                chunk[
                    "utterance_index"
                ],

            "speaker":
                chunk["speaker"],

            "speaker_timestamp":
                chunk[
                    "speaker_timestamp"
                ],

            "evidence_text":
                chunk["text"],

            "extraction_method":
                "gliner"
        }


        # Reuse our PRD quality filter.
        if accept_gliner_entity(
            candidate
        ):

            meeting_gliner_entities.append(
                candidate
            )


print("=" * 70)
print("MEETING GLINER EXTRACTION COMPLETE")
print("=" * 70)


print(
    "\nRetained semantic entity occurrences:",
    len(meeting_gliner_entities)
)


for meeting in parsed_meetings:

    count = sum(

        1

        for entity in meeting_gliner_entities

        if entity["meeting_id"]
        == meeting["meeting_id"]
    )


    print(
        meeting["meeting_id"],
        ":",
        count
    )
```

```python
# ============================================================
# STEP 103: INSPECT MEETING GLINER OUTPUT
# ============================================================
# Purpose:
# Quality-check model-generated meeting entities before they
# enter the extraction JSON.
#
# We inspect:
#
#   counts by semantic type
#   lowest-confidence retained examples
#   highest-confidence examples
# ============================================================

from collections import Counter


meeting_gliner_counts = Counter(

    entity["label"]

    for entity in meeting_gliner_entities
)


print("=" * 70)
print("MEETING GLINER ENTITY DISTRIBUTION")
print("=" * 70)


print(
    "\nTotal occurrences:",
    len(meeting_gliner_entities)
)


print("\nENTITY TYPES")
print("-" * 70)


for label, count in (
    meeting_gliner_counts.most_common()
):

    print(
        f"{label:20s}",
        count
    )


sorted_meeting_gliner = sorted(

    meeting_gliner_entities,

    key=lambda entity:
        entity["confidence"]
)


print("\n" + "=" * 70)
print("LOWEST-CONFIDENCE RETAINED EXAMPLES")
print("=" * 70)


for entity in sorted_meeting_gliner[:20]:

    print(
        f"{entity['confidence']:.3f}",
        "|",
        f"{entity['label']:12s}",
        "|",
        entity["text"],
        "|",
        entity["meeting_id"]
    )


print("\n" + "=" * 70)
print("HIGHEST-CONFIDENCE EXAMPLES")
print("=" * 70)


for entity in sorted_meeting_gliner[-20:]:

    print(
        f"{entity['confidence']:.3f}",
        "|",
        f"{entity['label']:12s}",
        "|",
        entity["text"],
        "|",
        entity["meeting_id"]
    )
```

```python
# ============================================================
# STEP 104: FINAL MEETING GLINER SAFETY FILTER
# ============================================================
# Purpose:
# Remove model predictions that conflict with deterministic
# identifiers.
#
# Examples:
#
#   DOCS-8052 must be ISSUE
#   REQ-003   must be REQUIREMENT
#
# They must never become:
#
#   FEATURE
#   SYSTEM
#   PERSON
#   etc.
#
# Deterministic identifiers always take precedence over
# model-generated semantic classification.
# ============================================================


def is_structured_identifier(text):

    text = str(text).strip()


    # PRD requirement
    if re.fullmatch(
        r"REQ-\d+",
        text,
        flags=re.IGNORECASE
    ):
        return True


    # Jira-style issue identifier
    if re.fullmatch(
        r"[A-Z][A-Z0-9_]+-\d+",
        text
    ):
        return True


    return False


meeting_gliner_entities_clean = [

    entity

    for entity in meeting_gliner_entities

    if not is_structured_identifier(
        entity["text"]
    )
]


print("=" * 70)
print("FINAL MEETING GLINER FILTER")
print("=" * 70)


print(
    "\nBefore:",
    len(meeting_gliner_entities)
)

print(
    "After :",
    len(
        meeting_gliner_entities_clean
    )
)

print(
    "Removed structured-ID conflicts:",
    len(meeting_gliner_entities)
    -
    len(meeting_gliner_entities_clean)
)
```

```python
# ============================================================
# STEP 105: CREATE BASE MEETING MASTER RECORDS
# ============================================================
# Purpose:
# Create one Master Extraction record for each meeting:
#
#     DOCS-I01-M001
#     DOCS-I01-M002
#
# The records will later receive:
#
#   meeting entity
#   participants
#   Jira issues
#   requirements
#   GLiNER semantic entities
#   actions
#   decisions
#   dependencies
#   blockers
#   temporal/provenance evidence
# ============================================================


meeting_records = []


for meeting in parsed_meetings:

    record = create_master_record(

        repository_project=
            PILOT_PROJECT,

        project_key=
            "DOCS",

        instance_id=
            PILOT_INSTANCE,

        source_type=
            "MEETING",

        source_id=
            meeting["meeting_id"]
    )


    # --------------------------------------------------------
    # Meeting timestamp
    # --------------------------------------------------------

    record[
        "source"
    ][
        "source_timestamp"
    ] = meeting[
        "meeting_datetime"
    ]


    # --------------------------------------------------------
    # Temporal information
    # --------------------------------------------------------

    record[
        "temporal"
    ][
        "created_at"
    ] = meeting[
        "meeting_datetime"
    ]


    record[
        "temporal"
    ][
        "valid_from"
    ] = meeting[
        "meeting_datetime"
    ]


    # --------------------------------------------------------
    # Meeting attributes
    # --------------------------------------------------------

    record[
        "attributes"
    ].update({

        "meeting_title":
            meeting[
                "meeting_title"
            ],

        "participants":
            meeting[
                "participants"
            ],

        "utterance_count":
            len(
                meeting[
                    "utterances"
                ]
            )
    })


    # --------------------------------------------------------
    # Preserve full meeting text as source-level evidence.
    # --------------------------------------------------------

    record[
        "provenance"
    ] = {

        "source_type":
            "MEETING",

        "source_id":
            meeting[
                "meeting_id"
            ],

        "instance_id":
            PILOT_INSTANCE,

        "chunk_id":
            None,

        "evidence_text":
            meeting["text"],

        "file_name":
            meeting[
                "file_name"
            ]
    }


    meeting_records.append({

        "meeting":
            meeting,

        "record":
            record
    })


print("=" * 70)
print("BASE MEETING RECORDS CREATED")
print("=" * 70)


print(
    "\nRecords:",
    len(meeting_records)
)


for item in meeting_records:

    print(
        "-",
        item[
            "record"
        ][
            "source"
        ][
            "source_id"
        ]
    )
```

```python
# ============================================================
# STEP 106: APPLY MEETING-SPECIFIC GLINER QUALITY FILTER
# ============================================================
# Purpose:
# Meeting transcripts are conversational, so GLiNER produces
# more false positives than it did on the structured PRD.
#
# Examples observed:
#
#     "I"         -> PERSON       (wrong)
#     "we"        -> PERSON       (wrong)
#     "6963"      -> MILESTONE    (wrong)
#     "2014"      -> MILESTONE    (wrong)
#     "DOCS-6826" -> COMPONENT    (wrong)
#     "INT-1680"  -> SYSTEM       (wrong)
#
# We already know meeting participants and speakers from the
# transcript structure, so PERSON entities will be created
# deterministically later.
#
# For meeting GLiNER extraction we retain mainly:
#
#     FEATURE
#     SYSTEM
#     COMPONENT
#     TEAM
#
# Deterministic extraction takes precedence for:
#
#     PERSON
#     ISSUE
#     REQUIREMENT
#     dates
# ============================================================


MEETING_ALLOWED_GLINER_LABELS = {
    "feature",
    "system",
    "component",
    "team"
}


# Generic conversational/project terms that should not become
# standalone semantic KG entities.
MEETING_GENERIC_TERMS = {
    "i",
    "we",
    "me",
    "us",
    "you",
    "they",
    "them",
    "he",
    "she",

    "daemon",
    "parameter section",
    "minor priority",
    "product dependency"
}


def accept_meeting_gliner_entity(entity):

    text = str(
        entity["text"]
    ).strip()

    label = str(
        entity["label"]
    ).lower().strip()

    confidence = float(
        entity["confidence"]
    )


    # --------------------------------------------------------
    # 1. Only retain semantic types useful for meetings.
    # --------------------------------------------------------

    if label not in MEETING_ALLOWED_GLINER_LABELS:
        return False


    # --------------------------------------------------------
    # 2. Keep our conservative confidence threshold.
    # --------------------------------------------------------

    if confidence < 0.55:
        return False


    # --------------------------------------------------------
    # 3. Reject generic/conversational terms.
    # --------------------------------------------------------

    if text.lower() in MEETING_GENERIC_TERMS:
        return False


    # --------------------------------------------------------
    # 4. Reject structured Jira/requirement identifiers.
    #
    # These are already extracted deterministically.
    # --------------------------------------------------------

    if is_structured_identifier(text):
        return False


    # --------------------------------------------------------
    # 5. Reject bare numbers / ticket-number fragments.
    #
    # Examples:
    #     6963
    #     8052
    # --------------------------------------------------------

    if re.fullmatch(
        r"\d+",
        text
    ):
        return False


    # --------------------------------------------------------
    # 6. Reject year-like values.
    #
    # Dates/years belong in temporal information.
    # --------------------------------------------------------

    if re.fullmatch(
        r"(19|20)\d{2}",
        text
    ):
        return False


    return True


meeting_gliner_final = [

    entity

    for entity in meeting_gliner_entities

    if accept_meeting_gliner_entity(
        entity
    )
]


print("=" * 70)
print("MEETING-SPECIFIC GLINER FILTER")
print("=" * 70)

print(
    "\nOriginal retained:",
    len(meeting_gliner_entities)
)

print(
    "Final retained   :",
    len(meeting_gliner_final)
)

print(
    "Removed          :",
    len(meeting_gliner_entities)
    - len(meeting_gliner_final)
)
```

```python
# ============================================================
# STEP 107: INSPECT FINAL MEETING GLINER ENTITIES
# ============================================================
# Purpose:
# Verify that the final meeting semantic entities are clean
# enough to enter the extraction JSON.
#
# Exact duplicate mentions are summarized for inspection only.
#
# Entity Resolution will later handle aliases such as:
#
#     Cloud Manager
#     Cloud
#
# or capitalization variants.
# ============================================================

from collections import Counter


final_meeting_gliner_counts = Counter(

    entity["label"]

    for entity in meeting_gliner_final
)


print("=" * 70)
print("FINAL MEETING SEMANTIC ENTITY DISTRIBUTION")
print("=" * 70)


print(
    "\nTotal retained occurrences:",
    len(meeting_gliner_final)
)


for label, count in (
    final_meeting_gliner_counts.most_common()
):

    print(
        f"{label:20s}",
        count
    )


# ------------------------------------------------------------
# Build unique label/text view.
# ------------------------------------------------------------

unique_meeting_semantic = {}


for entity in meeting_gliner_final:

    key = (
        entity["label"],
        entity["text"]
    )


    if (
        key not in unique_meeting_semantic
        or
        entity["confidence"]
        >
        unique_meeting_semantic[
            key
        ]["confidence"]
    ):

        unique_meeting_semantic[
            key
        ] = entity


print(
    "\nUnique semantic entities:",
    len(unique_meeting_semantic)
)


for label in [
    "feature",
    "system",
    "component",
    "team"
]:

    group = [

        entity

        for (
            entity_label,
            entity_text
        ), entity

        in unique_meeting_semantic.items()

        if entity_label == label
    ]


    if not group:
        continue


    group = sorted(
        group,
        key=lambda x:
            -x["confidence"]
    )


    print("\n" + "-" * 70)

    print(
        label.upper(),
        f"({len(group)})"
    )

    print("-" * 70)


    for entity in group:

        print(
            f"{entity['confidence']:.3f}",
            "|",
            entity["text"]
        )
```

```python
# ============================================================
# STEP 108: CHECK MEETING PARTICIPANT / SPEAKER CONSISTENCY
# ============================================================
# Purpose:
# Compare the participant list in the meeting header with the
# actual speakers appearing in the transcript.
#
# If a speaker appears in the transcript but not the header,
# we still preserve that person rather than silently dropping
# them.
# ============================================================


for meeting in parsed_meetings:

    header_participants = set(
        meeting["participants"]
    )


    transcript_speakers = set(

        utterance["speaker"]

        for utterance
        in meeting["utterances"]
    )


    speakers_not_in_header = sorted(

        transcript_speakers
        - header_participants
    )


    participants_not_speaking = sorted(

        header_participants
        - transcript_speakers
    )


    print("=" * 70)

    print(
        "Meeting:",
        meeting["meeting_id"]
    )


    print(
        "\nHeader participants:",
        sorted(header_participants)
    )


    print(
        "Transcript speakers:",
        sorted(transcript_speakers)
    )


    print(
        "\nSpeakers not in header:",
        speakers_not_in_header
    )


    print(
        "Header participants who never spoke:",
        participants_not_speaking
    )
```

```python
# ============================================================
# STEP 109: NORMALIZE MEETING PARTICIPANTS INTO NAME + ROLE
# ============================================================
# Purpose:
# Meeting headers contain participants in this form:
#
#     Alex (Docs Lead)
#     Dana (Server Engineering Liaison)
#
# Transcript utterances use only:
#
#     Alex
#     Dana
#
# We therefore split each header participant into:
#
#     name
#     role
#
# This is NOT Entity Resolution.
#
# We are simply parsing the structured format already present
# in the meeting header.
#
# Example output:
#
# {
#     "name": "Alex",
#     "role": "Docs Lead"
# }
#
# This information can later support:
#
#     PERSON entities
#     ownership
#     team relationships
#     cross-team features
# ============================================================


PARTICIPANT_PATTERN = re.compile(
    r"^\s*(.*?)\s*(?:\((.*?)\))?\s*$"
)


for meeting in parsed_meetings:

    parsed_participants = []


    for participant in meeting["participants"]:

        match = PARTICIPANT_PATTERN.match(
            participant
        )


        if not match:
            continue


        person_name = (
            match.group(1).strip()
        )


        person_role = (
            match.group(2).strip()
            if match.group(2)
            else None
        )


        parsed_participants.append({

            "name":
                person_name,

            "role":
                person_role,

            "raw":
                participant
        })


    meeting[
        "parsed_participants"
    ] = parsed_participants


print("=" * 70)
print("PARSED MEETING PARTICIPANTS")
print("=" * 70)


for meeting in parsed_meetings:

    print(
        "\nMeeting:",
        meeting["meeting_id"]
    )


    for participant in meeting[
        "parsed_participants"
    ]:

        print(
            participant["name"],
            "| Role:",
            participant["role"]
        )
```

```python
# ============================================================
# STEP 110: RECHECK PARTICIPANT / SPEAKER CONSISTENCY
# ============================================================
# Purpose:
# Compare normalized participant NAMES with transcript speakers.
#
# After removing role descriptions, the two sets should match.
# ============================================================


for meeting in parsed_meetings:

    participant_names = set(

        participant["name"]

        for participant in meeting[
            "parsed_participants"
        ]
    )


    transcript_speakers = set(

        utterance["speaker"]

        for utterance in meeting[
            "utterances"
        ]
    )


    speakers_not_in_header = sorted(
        transcript_speakers
        - participant_names
    )


    participants_not_speaking = sorted(
        participant_names
        - transcript_speakers
    )


    print("=" * 70)

    print(
        "Meeting:",
        meeting["meeting_id"]
    )


    print(
        "\nNormalized participants:",
        sorted(participant_names)
    )


    print(
        "Transcript speakers:",
        sorted(transcript_speakers)
    )


    print(
        "\nSpeakers not in header:",
        speakers_not_in_header
    )


    print(
        "Participants who never spoke:",
        participants_not_speaking
    )


    if (
        not speakers_not_in_header
        and
        not participants_not_speaking
    ):

        print(
            "\nSUCCESS: Participant names and "
            "transcript speakers match."
        )
```

```python
# ============================================================
# STEP 111: ADD MEETING RELATIONSHIP TYPES
# ============================================================
# Purpose:
# Extend the controlled relationship vocabulary with
# relationships required by meeting extraction.
# ============================================================


NEW_MEETING_RELATIONSHIP_TYPES = [

    "HAS_MEETING",

    "PARTICIPATED_IN",

    "REFERENCES_REQUIREMENT",

    "REFERENCES_ISSUE",

    "MENTIONS"
]


for relationship_type in (
    NEW_MEETING_RELATIONSHIP_TYPES
):

    if (
        relationship_type
        not in RELATIONSHIP_TYPES
    ):

        RELATIONSHIP_TYPES.append(
            relationship_type
        )


print(
    "Meeting relationship vocabulary updated."
)
```

```python
# ============================================================
# STEP 112: ADD PROJECT, MEETING AND PARTICIPANT ENTITIES
# ============================================================
# Purpose:
# Create deterministic entities for:
#
#     PROJECT
#     MEETING
#     PERSON
#
# Participant names come from structured meeting metadata,
# therefore confidence = 1.0.
#
# Participant roles are retained as PERSON attributes.
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]


    # --------------------------------------------------------
    # Project
    # --------------------------------------------------------

    project_entity = add_entity(

        record,

        "PROJECT",

        "DOCS",

        attributes={
            "project_name":
                "Documentation",

            "instance_id":
                PILOT_INSTANCE
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # --------------------------------------------------------
    # Meeting
    # --------------------------------------------------------

    meeting_entity = add_entity(

        record,

        "MEETING",

        meeting["meeting_id"],

        attributes={

            "title":
                meeting[
                    "meeting_title"
                ],

            "meeting_datetime":
                meeting[
                    "meeting_datetime"
                ]
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    add_relationship(

        record,

        project_entity,

        "HAS_MEETING",

        meeting_entity,

        timestamp=
            meeting[
                "meeting_datetime"
            ]
    )


    # Save IDs for later cells.
    item[
        "project_entity"
    ] = project_entity

    item[
        "meeting_entity"
    ] = meeting_entity


    # --------------------------------------------------------
    # Participants
    # --------------------------------------------------------

    participant_entity_ids = {}


    for participant in meeting[
        "parsed_participants"
    ]:

        person_entity = add_entity(

            record,

            "PERSON",

            participant["name"],

            attributes={
                "role":
                    participant["role"]
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        participant_entity_ids[
            participant["name"]
        ] = person_entity


        add_relationship(

            record,

            person_entity,

            "PARTICIPATED_IN",

            meeting_entity,

            timestamp=
                meeting[
                    "meeting_datetime"
                ]
        )


    item[
        "participant_entity_ids"
    ] = participant_entity_ids
```

```python
# ============================================================
# STEP 113: ADD EXPLICIT ISSUE / REQUIREMENT REFERENCES
# ============================================================
# Purpose:
# Convert deterministic cross-source references into entities.
#
# Example:
#
# Meeting:
#     DOCS-8052
#
# Jira:
#     DOCS-8052
#
# These remain separate source-local entities now.
# Entity Resolution later connects them canonically.
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]

    meeting_entity = item[
        "meeting_entity"
    ]


    # --------------------------------------------------------
    # Unique Jira issues
    # --------------------------------------------------------

    unique_issues = sorted(
        set(
            occurrence["issue_key"]

            for occurrence
            in meeting[
                "issue_occurrences"
            ]
        )
    )


    issue_entity_ids = {}


    for issue_key in unique_issues:

        entity_id = add_entity(

            record,

            "ISSUE",

            issue_key,

            extraction_method=
                "regex",

            confidence=1.0
        )


        issue_entity_ids[
            issue_key
        ] = entity_id


        add_relationship(

            record,

            meeting_entity,

            "REFERENCES_ISSUE",

            entity_id,

            timestamp=
                meeting[
                    "meeting_datetime"
                ],

            extraction_method=
                "regex",

            confidence=1.0
        )


    item[
        "issue_entity_ids"
    ] = issue_entity_ids


    # --------------------------------------------------------
    # Unique requirements
    # --------------------------------------------------------

    unique_requirements = sorted(
        set(
            occurrence[
                "requirement_id"
            ]

            for occurrence
            in meeting[
                "requirement_occurrences"
            ]
        )
    )


    requirement_entity_ids = {}


    for requirement_id in (
        unique_requirements
    ):

        entity_id = add_entity(

            record,

            "REQUIREMENT",

            requirement_id,

            extraction_method=
                "regex",

            confidence=1.0
        )


        requirement_entity_ids[
            requirement_id
        ] = entity_id


        add_relationship(

            record,

            meeting_entity,

            "REFERENCES_REQUIREMENT",

            entity_id,

            timestamp=
                meeting[
                    "meeting_datetime"
                ],

            extraction_method=
                "regex",

            confidence=1.0
        )


    item[
        "requirement_entity_ids"
    ] = requirement_entity_ids
```

```python
# ============================================================
# STEP 114: ADD CLEAN MEETING GLINER ENTITIES
# ============================================================
# Purpose:
# Add only the final filtered semantic entities from Step 106.
#
# PERSON is NOT taken from GLiNER.
# Participant/speaker PERSON entities were created
# deterministically in Step 112.
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]

    meeting_entity = item[
        "meeting_entity"
    ]


    relevant_predictions = [

        entity

        for entity
        in meeting_gliner_final

        if entity["meeting_id"]
        == meeting["meeting_id"]
    ]


    for prediction in (
        relevant_predictions
    ):

        master_type = (
            GLINER_TO_MASTER_TYPE[
                prediction["label"]
            ]
        )


        semantic_entity = add_entity(

            record,

            master_type,

            prediction["text"],

            attributes={

                "utterance_index":
                    prediction[
                        "utterance_index"
                    ],

                "speaker":
                    prediction[
                        "speaker"
                    ],

                "speaker_timestamp":
                    prediction[
                        "speaker_timestamp"
                    ]
            },

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ]
        )


        add_relationship(

            record,

            meeting_entity,

            "MENTIONS",

            semantic_entity,

            timestamp=
                meeting[
                    "meeting_datetime"
                ],

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ],

            attributes={

                "speaker":
                    prediction[
                        "speaker"
                    ],

                "speaker_timestamp":
                    prediction[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    prediction[
                        "evidence_text"
                    ]
            }
        )
```

```python
# ============================================================
# STEP 115: ADD MEETING ACTION ITEMS
# ============================================================
# Purpose:
# Preserve cleaned action/commitment statements with:
#
#     owner/speaker
#     meeting timestamp
#     speaker timestamp
#     evidence text
#
# IMPORTANT:
# Recap utterances may repeat actions stated earlier.
#
# We preserve them as evidence but mark whether the statement
# is a recap so downstream logic can avoid blindly counting
# every occurrence as an independent action.
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]


    for index, candidate in enumerate(
        meeting[
            "final_action_candidates"
        ],
        start=1
    ):

        text_lower = (
            candidate["text"].lower()
        )


        is_recap = (
            "recap" in text_lower
        )


        action_item = {

            "action_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_ACTION_{index:03d}"
                ),

            "action_text":
                candidate["text"],

            "action_owner":
                candidate["speaker"],

            "meeting_time":
                meeting[
                    "meeting_datetime"
                ],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "due_at":
                None,

            "completed_at":
                None,

            "action_status":
                "OPEN",

            "is_recap":
                is_recap,

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        }


        record[
            "action_items"
        ].append(
            action_item
        )
```

```python
# ============================================================
# STEP 116: ADD MEETING DECISION EVIDENCE
# ============================================================
# Purpose:
# Preserve decision-related statements together with their
# state:
#
#     MADE
#     DEFERRED
#     PROPOSED
#     TENTATIVE
#     HISTORICAL
#
# This distinction is important for later:
#
#     unresolved decision count
#     decision latency
#     decision reversal detection
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]


    for index, candidate in enumerate(
        meeting[
            "decision_candidates"
        ],
        start=1
    ):

        record[
            "decisions"
        ].append({

            "decision_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_DECISION_{index:03d}"
                ),

            "decision_text":
                candidate["text"],

            "decision_maker":
                candidate["speaker"],

            "decision_time":
                meeting[
                    "meeting_datetime"
                ],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "decision_state":
                candidate[
                    "decision_state"
                ],

            "decision_target":
                None,

            "reversed_decision":
                None,

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        })
```

```python
# ============================================================
# STEP 117: ADD MEETING DEPENDENCY EVIDENCE
# ============================================================
# Purpose:
# Preserve dependency statements and their interpreted state.
#
# States include:
#
#     CONFIRMED
#     UNCERTAIN
#     NEGATED
#     QUESTION
#     CONDITIONAL
#
# We do not invent source/target endpoints when the utterance
# does not make them sufficiently explicit.
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]


    for index, candidate in enumerate(
        meeting[
            "dependency_candidates"
        ],
        start=1
    ):

        record[
            "dependencies"
        ].append({

            "dependency_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_DEP_{index:03d}"
                ),

            "text":
                candidate["text"],

            "source_entity":
                None,

            "target_entity":
                None,

            "dependency_type":
                "MEETING_DEPENDENCY",

            "dependency_status":
                candidate[
                    "dependency_state"
                ],

            "created_at":
                meeting[
                    "meeting_datetime"
                ],

            "speaker":
                candidate["speaker"],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        })
```

```python
# ============================================================
# STEP 118: ADD BLOCKER EVIDENCE
# ============================================================
# Purpose:
# Preserve blocker-related evidence with precise state.
#
# States:
#
#     ACTIVE
#     NEGATED
#     QUESTION
#     UNBLOCKING
#     CONDITIONAL
#
# This prevents statements such as:
#
#     "not a blocker today"
#
# from becoming active blocker facts.
#
# Recap statements are marked separately so repeated evidence
# is not mistaken for multiple independent blockers later.
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]


    for index, candidate in enumerate(
        meeting[
            "blocker_candidates"
        ],
        start=1
    ):

        text_lower = (
            candidate["text"].lower()
        )


        record[
            "blockers"
        ].append({

            "blocker_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_BLOCKER_{index:03d}"
                ),

            "blocked_entity":
                None,

            "blocker_entity":
                None,

            "blocker_state":
                candidate[
                    "blocker_state"
                ],

            "is_active_assertion":
                (
                    candidate[
                        "blocker_state"
                    ]
                    == "ACTIVE"
                ),

            "is_recap":
                "recap" in text_lower,

            "meeting_time":
                meeting[
                    "meeting_datetime"
                ],

            "speaker":
                candidate["speaker"],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "blocking_reason":
                None,

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        })
```

```python
# ============================================================
# STEP 119: CREATE MEETING TEMPORAL EVENTS
# ============================================================
# Purpose:
# Add conservative temporal events from meeting evidence.
#
# For now:
#
#     MADE decisions -> DECISION_MADE event
#
# Other states remain preserved in the decisions array but do
# not become "made decision" events.
# ============================================================


for item in meeting_records:

    meeting = item["meeting"]
    record = item["record"]

    meeting_entity = item[
        "meeting_entity"
    ]


    for decision in record[
        "decisions"
    ]:

        if (
            decision[
                "decision_state"
            ]
            != "MADE"
        ):
            continue


        add_event(

            record,

            event_type=
                "DECISION_MADE",

            entity=
                meeting_entity,

            event_time=
                meeting[
                    "meeting_datetime"
                ],

            actor=
                decision[
                    "decision_maker"
                ],

            from_value=
                None,

            to_value=
                decision[
                    "decision_text"
                ],

            attributes={

                "speaker_timestamp":
                    decision[
                        "speaker_timestamp"
                    ],

                "decision_id":
                    decision[
                        "decision_id"
                    ]
            },

            extraction_method=
                "rule"
        )
```

```python
# ============================================================
# STEP 120: FINAL MEETING EXTRACTION SUMMARY
# ============================================================
# Purpose:
# Validate both complete meeting records before saving them.
# ============================================================


print("=" * 70)
print("FINAL MEETING EXTRACTION SUMMARY")
print("=" * 70)


for item in meeting_records:

    record = item["record"]


    entity_counts = Counter(
        entity["type"]
        for entity in record["entities"]
    )


    relationship_counts = Counter(
        relationship["type"]
        for relationship
        in record["relationships"]
    )


    decision_state_counts = Counter(
        decision["decision_state"]
        for decision
        in record["decisions"]
    )


    dependency_state_counts = Counter(
        dependency[
            "dependency_status"
        ]
        for dependency
        in record["dependencies"]
    )


    blocker_state_counts = Counter(
        blocker["blocker_state"]
        for blocker
        in record["blockers"]
    )


    print("\n" + "=" * 70)

    print(
        "Meeting:",
        record["source"]["source_id"]
    )


    print(
        "Entities      :",
        len(record["entities"])
    )

    print(
        "Relationships :",
        len(record["relationships"])
    )

    print(
        "Events        :",
        len(record["events"])
    )

    print(
        "Actions       :",
        len(record["action_items"])
    )

    print(
        "Decisions     :",
        len(record["decisions"])
    )

    print(
        "Dependencies  :",
        len(record["dependencies"])
    )

    print(
        "Blocker evidence:",
        len(record["blockers"])
    )


    print(
        "\nEntity types:",
        dict(entity_counts)
    )


    print(
        "Decision states:",
        dict(decision_state_counts)
    )


    print(
        "Dependency states:",
        dict(dependency_state_counts)
    )


    print(
        "Blocker states:",
        dict(blocker_state_counts)
    )
```

```python
# ============================================================
# STEP 121: SAVE PILOT MEETING EXTRACTION
# ============================================================
# Purpose:
# Save the completed extraction records for both DOCS-I01
# meetings.
#
# Output:
#
# GC1_Entity_Extraction/
# └── 24_DOCS/
#     └── DOCS-I01/
#         ├── jira_extracted.jsonl
#         ├── prd_extracted.jsonl
#         └── meetings_extracted.jsonl
#
# meetings_extracted.jsonl contains:
#
#     Line 1 -> DOCS-I01-M001
#     Line 2 -> DOCS-I01-M002
#
# Each record preserves:
#
#     entities
#     relationships
#     events
#     decisions
#     action items
#     dependencies
#     blocker evidence
#     temporal information
#     provenance
# ============================================================


meetings_output_file = (
    pilot_output_folder
    / "meetings_extracted.jsonl"
)


with open(
    meetings_output_file,
    "w",
    encoding="utf-8"
) as file:

    for item in meeting_records:

        record = item["record"]

        file.write(
            json.dumps(
                record,
                ensure_ascii=False
            )
            + "\n"
        )


print("=" * 70)
print("MEETING EXTRACTION SAVED")
print("=" * 70)


print("\nOutput file:")
print(meetings_output_file)


print(
    "\nRecords written:",
    len(meeting_records)
)


print(
    "File exists:",
    meetings_output_file.exists()
)


print(
    "File size:",
    round(
        meetings_output_file.stat().st_size
        / 1024,
        2
    ),
    "KB"
)
```

```python
# ============================================================
# STEP 122: VALIDATE SAVED MEETING EXTRACTION
# ============================================================
# Purpose:
# Reload meetings_extracted.jsonl and verify:
#
#   1. JSON is valid
#   2. Exactly 2 meeting records exist
#   3. Required schema sections exist
#   4. Meeting IDs are correct
#   5. PERSON / ISSUE / REQUIREMENT information survived
#   6. Action / decision / dependency / blocker evidence
#      survived
#
# This is the final validation of the meeting extraction.
# ============================================================


reloaded_meeting_records = []


with open(
    meetings_output_file,
    "r",
    encoding="utf-8"
) as file:

    for line_number, line in enumerate(
        file,
        start=1
    ):

        if not line.strip():
            continue


        try:

            record = json.loads(line)

            reloaded_meeting_records.append(
                record
            )


        except json.JSONDecodeError as error:

            print(
                "Invalid JSON on line:",
                line_number
            )

            raise error


# ------------------------------------------------------------
# Mandatory sections expected in every master record.
# ------------------------------------------------------------

required_sections = [
    "entities",
    "relationships",
    "events",
    "decisions",
    "action_items",
    "blockers",
    "dependencies",
    "temporal",
    "attributes",
    "provenance"
]


missing_sections = []


for record in reloaded_meeting_records:

    for section in required_sections:

        if section not in record:

            missing_sections.append(
                (
                    record[
                        "source"
                    ][
                        "source_id"
                    ],
                    section
                )
            )


# ------------------------------------------------------------
# Extract meeting IDs after reload.
# ------------------------------------------------------------

reloaded_meeting_ids = [

    record[
        "source"
    ][
        "source_id"
    ]

    for record
    in reloaded_meeting_records
]


expected_meeting_ids = sorted([

    meeting[
        "meeting_id"
    ]

    for meeting
    in parsed_meetings
])


# ------------------------------------------------------------
# Calculate total preserved information.
# ------------------------------------------------------------

total_entities = sum(
    len(record["entities"])
    for record in reloaded_meeting_records
)


total_relationships = sum(
    len(record["relationships"])
    for record in reloaded_meeting_records
)


total_events = sum(
    len(record["events"])
    for record in reloaded_meeting_records
)


total_actions = sum(
    len(record["action_items"])
    for record in reloaded_meeting_records
)


total_decisions = sum(
    len(record["decisions"])
    for record in reloaded_meeting_records
)


total_dependencies = sum(
    len(record["dependencies"])
    for record in reloaded_meeting_records
)


total_blocker_evidence = sum(
    len(record["blockers"])
    for record in reloaded_meeting_records
)


print("=" * 70)
print("SAVED MEETING FILE VALIDATION")
print("=" * 70)


print(
    "\nRecords reloaded:",
    len(reloaded_meeting_records)
)


print(
    "Expected records:",
    len(parsed_meetings)
)


print(
    "\nMeeting IDs:"
)

for meeting_id in reloaded_meeting_ids:
    print(" -", meeting_id)


print(
    "\nMissing required sections:",
    len(missing_sections)
)


print("\nPreserved information:")

print(
    " Entities         :",
    total_entities
)

print(
    " Relationships    :",
    total_relationships
)

print(
    " Events           :",
    total_events
)

print(
    " Action evidence  :",
    total_actions
)

print(
    " Decision evidence:",
    total_decisions
)

print(
    " Dependencies     :",
    total_dependencies
)

print(
    " Blocker evidence :",
    total_blocker_evidence
)


# ------------------------------------------------------------
# Final validation.
# ------------------------------------------------------------

records_ok = (
    len(reloaded_meeting_records)
    ==
    len(parsed_meetings)
)


ids_ok = (
    sorted(reloaded_meeting_ids)
    ==
    expected_meeting_ids
)


sections_ok = (
    len(missing_sections) == 0
)


if (
    records_ok
    and ids_ok
    and sections_ok
):

    print(
        "\nSUCCESS:"
        "\nPilot meeting extraction JSONL is valid "
        "and ready for downstream use."
    )

else:

    print(
        "\nCHECK REQUIRED:"
        "\nOne or more meeting validation checks failed."
    )
```

```python
# ============================================================
# STEP 123: LOAD ALL PILOT EXTRACTION OUTPUTS
# ============================================================
# Purpose:
# Load the final validated extraction outputs from:
#
#     Jira
#     PRD
#     Meetings
#
# into one place for cross-source validation.
#
# We are NOT merging entities here.
#
# Entity Resolution belongs to the next stage.
# ============================================================


def load_jsonl(file_path):
    """
    Load a JSONL file into a list of Python dictionaries.
    """

    records = []

    with open(
        file_path,
        "r",
        encoding="utf-8"
    ) as file:

        for line in file:

            if line.strip():

                records.append(
                    json.loads(line)
                )

    return records


jira_final = load_jsonl(
    jira_output_file
)

prd_final = load_jsonl(
    prd_output_file
)

meetings_final = load_jsonl(
    meetings_output_file
)


print("=" * 70)
print("CROSS-SOURCE PILOT DATA LOADED")
print("=" * 70)

print("\nJira records    :", len(jira_final))
print("PRD records     :", len(prd_final))
print("Meeting records :", len(meetings_final))

print(
    "\nTotal source records:",
    len(jira_final)
    + len(prd_final)
    + len(meetings_final)
)
```

```python
# ============================================================
# STEP 124: VALIDATE PROJECT / INSTANCE CONSISTENCY
# ============================================================
# Purpose:
# Ensure all extracted records belong to the same pilot
# project instance.
#
# This prevents accidental mixing of:
#
#     DOCS-I01
#     DOCS-I02
#     other repositories
#
# which would corrupt Entity Resolution and the Temporal KG.
# ============================================================


all_final_records = (
    jira_final
    + prd_final
    + meetings_final
)


repository_projects = sorted(
    set(
        record["repository_project"]
        for record in all_final_records
    )
)


project_keys = sorted(
    set(
        record["project_key"]
        for record in all_final_records
    )
)


instance_ids = sorted(
    set(
        record["instance_id"]
        for record in all_final_records
    )
)


print("=" * 70)
print("PROJECT / INSTANCE CONSISTENCY")
print("=" * 70)

print(
    "\nRepository projects:",
    repository_projects
)

print(
    "Project keys:",
    project_keys
)

print(
    "Instance IDs:",
    instance_ids
)


if (
    repository_projects == ["24_DOCS"]
    and
    project_keys == ["DOCS"]
    and
    instance_ids == ["DOCS-I01"]
):

    print(
        "\nSUCCESS: All extracted records belong "
        "to the correct pilot instance."
    )

else:

    print(
        "\nCHECK REQUIRED: Mixed project/instance "
        "identifiers detected."
    )
```

```python
# ============================================================
# STEP 125: CHECK CROSS-SOURCE ENTITY ANCHORS
# ============================================================
# Purpose:
# Identify exact entity names that occur across multiple
# source types.
#
# These are ideal inputs for the next Entity Resolution stage.
#
# IMPORTANT:
# We do NOT merge them here.
#
# We only demonstrate that extraction preserved the anchors.
# ============================================================

from collections import defaultdict


entity_sources = defaultdict(
    lambda: defaultdict(list)
)


for record in all_final_records:

    source_type = (
        record[
            "source"
        ][
            "source_type"
        ]
    )


    source_id = (
        record[
            "source"
        ][
            "source_id"
        ]
    )


    for entity in record["entities"]:

        entity_name = entity[
            "name"
        ]


        entity_sources[
            entity_name
        ][
            source_type
        ].append({

            "source_id":
                source_id,

            "local_id":
                entity[
                    "local_id"
                ],

            "type":
                entity[
                    "type"
                ]
        })


# ------------------------------------------------------------
# Keep entities appearing in at least two different sources.
# ------------------------------------------------------------

cross_source_entities = {

    name: sources

    for name, sources
    in entity_sources.items()

    if len(sources) >= 2
}


print("=" * 70)
print("CROSS-SOURCE ENTITY ANCHORS")
print("=" * 70)


print(
    "\nEntities appearing across >= 2 source types:",
    len(cross_source_entities)
)


# Display a manageable sample.
for name, sources in list(
    sorted(
        cross_source_entities.items()
    )
)[:50]:

    print("\nENTITY:", name)

    for source_type, occurrences in (
        sources.items()
    ):

        print(
            " ",
            source_type,
            "->",
            len(occurrences),
            "occurrence(s)"
        )
```

```python
# ============================================================
# STEP 126: VERIFY IMPORTANT CROSS-SOURCE IDENTIFIERS
# ============================================================
# Purpose:
# Verify that known identifiers survived extraction in the
# correct source types.
#
# This is a direct handoff-quality check for Entity Resolution.
# ============================================================


TEST_ANCHORS = [
    "DOCS-6963",
    "REQ-003",
    "REQ-007",
    "REQ-011"
]


print("=" * 70)
print("IMPORTANT ENTITY ANCHOR CHECK")
print("=" * 70)


for anchor in TEST_ANCHORS:

    print("\nANCHOR:", anchor)


    if anchor not in entity_sources:

        print("  NOT FOUND")
        continue


    for source_type, occurrences in (
        entity_sources[
            anchor
        ].items()
    ):

        print(
            " ",
            source_type,
            "->"
        )


        for occurrence in occurrences:

            print(
                "    ",
                occurrence[
                    "source_id"
                ],
                "|",
                occurrence[
                    "local_id"
                ],
                "|",
                occurrence[
                    "type"
                ]
            )
```

```python
# ============================================================
# STEP 127: VALIDATE TEMPORAL EVENT BOUNDARIES - CORRECTED
# ============================================================
# Purpose:
# Ensure timestamped extraction events fall inside the
# DOCS-I01 project-instance timeline.
#
# PROBLEM FIXED:
#
# Some source timestamps are:
#
#     2016-07-10
#
# while Jira timestamps may be:
#
#     2016-07-10T12:30:00+00:00
#
# The first is timezone-naive and the second is timezone-aware.
# Python cannot directly compare them.
#
# We therefore normalize ALL timestamps to UTC before
# performing temporal comparisons.
#
# This changes only the validation comparison.
# It does NOT modify the saved extraction data.
# ============================================================

from datetime import datetime, timezone


def parse_timestamp_utc(value):
    """
    Convert different GC1 timestamp formats into
    timezone-aware UTC datetime objects.

    Examples supported:

        2016-07-10

        2016-07-10T00:00:00+00:00

        2016-07-10T00:00:00.000+0000

    Missing/unparseable values return None.
    """

    if value is None:
        return None


    # --------------------------------------------------------
    # Already a datetime object
    # --------------------------------------------------------

    if isinstance(value, datetime):

        dt = value


    else:

        value = str(value).strip()

        if not value:
            return None


        try:

            # ------------------------------------------------
            # Convert Jira timezone:
            #
            #     +0000
            #
            # into:
            #
            #     +00:00
            # ------------------------------------------------

            if (
                len(value) >= 5
                and value[-5] in ["+", "-"]
                and value[-3] != ":"
            ):

                value = (
                    value[:-2]
                    + ":"
                    + value[-2:]
                )


            dt = datetime.fromisoformat(
                value
            )


        except Exception:

            return None


    # --------------------------------------------------------
    # If timestamp has NO timezone:
    #
    #     2016-07-10
    #
    # treat it as UTC for our GC1 timeline.
    # --------------------------------------------------------

    if dt.tzinfo is None:

        dt = dt.replace(
            tzinfo=timezone.utc
        )


    # --------------------------------------------------------
    # Convert any aware timezone to UTC.
    # --------------------------------------------------------

    else:

        dt = dt.astimezone(
            timezone.utc
        )


    return dt


# ============================================================
# NORMALIZE INSTANCE BOUNDARIES
# ============================================================

INSTANCE_START_UTC = parse_timestamp_utc(
    INSTANCE_START
)

INSTANCE_END_UTC = parse_timestamp_utc(
    INSTANCE_END
)


print("=" * 70)
print("TEMPORAL BOUNDARY VALIDATION")
print("=" * 70)


print(
    "\nInstance start:",
    INSTANCE_START_UTC
)

print(
    "Instance end  :",
    INSTANCE_END_UTC
)


# ============================================================
# CHECK ALL EXTRACTED EVENTS
# ============================================================

events_checked = 0

events_outside_window = []

unparseable_event_times = []


for record in all_final_records:

    for event in record["events"]:

        raw_event_time = event.get(
            "event_time"
        )


        event_time = parse_timestamp_utc(
            raw_event_time
        )


        # ----------------------------------------------------
        # Missing timestamp:
        # nothing to compare.
        # ----------------------------------------------------

        if raw_event_time is None:
            continue


        # ----------------------------------------------------
        # Timestamp exists but could not be parsed.
        # Preserve separately for validation.
        # ----------------------------------------------------

        if event_time is None:

            unparseable_event_times.append({

                "source_type":
                    record[
                        "source"
                    ][
                        "source_type"
                    ],

                "source_id":
                    record[
                        "source"
                    ][
                        "source_id"
                    ],

                "event_type":
                    event[
                        "event_type"
                    ],

                "event_time":
                    raw_event_time
            })

            continue


        events_checked += 1


        # ----------------------------------------------------
        # Compare normalized UTC timestamps.
        # ----------------------------------------------------

        if not (
            INSTANCE_START_UTC
            <= event_time
            <= INSTANCE_END_UTC
        ):

            events_outside_window.append({

                "source_type":
                    record[
                        "source"
                    ][
                        "source_type"
                    ],

                "source_id":
                    record[
                        "source"
                    ][
                        "source_id"
                    ],

                "event_type":
                    event[
                        "event_type"
                    ],

                "event_time":
                    raw_event_time
            })


# ============================================================
# RESULTS
# ============================================================

print(
    "\nTimestamped events checked:",
    events_checked
)


print(
    "Unparseable event timestamps:",
    len(
        unparseable_event_times
    )
)


print(
    "Events outside instance window:",
    len(
        events_outside_window
    )
)


# ------------------------------------------------------------
# Display problems only if they exist.
# ------------------------------------------------------------

if unparseable_event_times:

    print(
        "\nFirst unparseable timestamps:"
    )

    for item in (
        unparseable_event_times[:10]
    ):

        print(item)


if events_outside_window:

    print(
        "\nFirst events outside window:"
    )

    for item in (
        events_outside_window[:20]
    ):

        print(item)


# ------------------------------------------------------------
# Final validation
# ------------------------------------------------------------

if (
    len(unparseable_event_times) == 0
    and
    len(events_outside_window) == 0
):

    print(
        "\nSUCCESS:"
        "\nAll timestamped extraction events are "
        "valid and inside DOCS-I01."
    )

else:

    print(
        "\nCHECK REQUIRED:"
        "\nOne or more temporal validation "
        "issues were detected."
    )
```

```python
# ============================================================
# STEP 128: TABPFN DOWNSTREAM INFORMATION CHECK
# ============================================================
# Purpose:
# Verify that extraction preserved the raw information needed
# for later Temporal KG snapshot feature engineering.
#
# TabPFN will NOT directly consume these JSON entities.
#
# Later:
#
#     extraction
#         ↓
#     entity resolution
#         ↓
#     Temporal KG
#         ↓
#     weekly snapshots
#         ↓
#     numerical/categorical feature table
#         ↓
#     TabPFN
#
# We therefore check whether important raw evidence categories
# are present now.
# ============================================================


all_events = [

    event

    for record in all_final_records

    for event in record["events"]
]


all_relationships = [

    relationship

    for record in all_final_records

    for relationship in record[
        "relationships"
    ]
]


all_dependencies = [

    dependency

    for record in all_final_records

    for dependency in record[
        "dependencies"
    ]
]


all_blockers = [

    blocker

    for record in all_final_records

    for blocker in record[
        "blockers"
    ]
]


all_actions = [

    action

    for record in all_final_records

    for action in record[
        "action_items"
    ]
]


all_decisions = [

    decision

    for record in all_final_records

    for decision in record[
        "decisions"
    ]
]


event_types_present = set(
    event["event_type"]
    for event in all_events
)


relationship_types_present = set(
    relationship["type"]
    for relationship in all_relationships
)


checks = {

    # Requirement volatility / changes
    "requirements":
        any(
            entity["type"]
            == "REQUIREMENT"

            for record
            in all_final_records

            for entity
            in record["entities"]
        ),

    # Blocker age / blocker activity
    "blocker_information":
        len(all_blockers) > 0,

    # Dependency complexity
    "dependency_information":
        (
            len(all_dependencies) > 0
            or
            "DEPENDS_ON"
            in relationship_types_present
        ),

    # Scope growth
    "scope_change_information":
        (
            "EPIC_CHILD_CHANGE"
            in event_types_present
            or
            len(
                prd_scope_items
            ) > 0
        ),

    # Decision analysis
    "decision_information":
        len(all_decisions) > 0,

    # Ownership
    "ownership_information":
        (
            "ASSIGNED_TO"
            in relationship_types_present
            or
            "ASSIGNEE_CHANGE"
            in event_types_present
        ),

    # Sprint / schedule movement
    "sprint_information":
        "SPRINT_CHANGE"
        in event_types_present,

    # Effort volatility
    "effort_information":
        (
            "STORY_POINT_CHANGE"
            in event_types_present
            or
            "ESTIMATE_CHANGE"
            in event_types_present
            or
            "ACTUAL_TIME_CHANGE"
            in event_types_present
        ),

    # Action items
    "action_information":
        len(all_actions) > 0,

    # Status/workflow dynamics
    "workflow_information":
        (
            "STATUS_CHANGE"
            in event_types_present
        )
}


print("=" * 70)
print("TABPFN DOWNSTREAM INFORMATION CHECK")
print("=" * 70)


for name, available in checks.items():

    print(
        f"{name:30s}",
        available
    )


if all(checks.values()):

    print(
        "\nSUCCESS:"
        "\nThe pilot extraction preserves all major "
        "raw information categories required for later "
        "Temporal KG -> feature engineering -> TabPFN."
    )

else:

    print(
        "\nSome information categories are missing."
    )
```

```python
# ============================================================
# STEP 129: FINAL PILOT EXTRACTION HANDOFF SUMMARY
# ============================================================
# Purpose:
# Produce one final summary of the complete extraction output
# that will eventually be handed to Entity Resolution.
# ============================================================


source_counts = Counter(
    record[
        "source"
    ][
        "source_type"
    ]

    for record
    in all_final_records
)


entity_counts = Counter(

    entity["type"]

    for record
    in all_final_records

    for entity
    in record["entities"]
)


relationship_counts = Counter(

    relationship["type"]

    for record
    in all_final_records

    for relationship
    in record["relationships"]
)


event_counts = Counter(

    event["event_type"]

    for record
    in all_final_records

    for event
    in record["events"]
)


print("=" * 70)
print("GC1 PILOT EXTRACTION - FINAL HANDOFF SUMMARY")
print("=" * 70)


print("\nProject :", PILOT_PROJECT)
print("Instance:", PILOT_INSTANCE)


print("\nSOURCE RECORDS")
print("-" * 70)

for source_type, count in (
    source_counts.items()
):

    print(
        f"{source_type:20s}",
        count
    )


print(
    "\nTotal source records:",
    len(all_final_records)
)


print("\nENTITY OCCURRENCES")
print("-" * 70)

for entity_type, count in (
    entity_counts.most_common()
):

    print(
        f"{entity_type:25s}",
        count
    )


print("\nRELATIONSHIPS")
print("-" * 70)

for relation_type, count in (
    relationship_counts.most_common()
):

    print(
        f"{relation_type:25s}",
        count
    )


print("\nEVENTS")
print("-" * 70)

for event_type, count in (
    event_counts.most_common()
):

    print(
        f"{event_type:30s}",
        count
    )


print("\nOTHER EVIDENCE")
print("-" * 70)

print(
    "Dependencies:",
    len(all_dependencies)
)

print(
    "Blocker evidence:",
    len(all_blockers)
)

print(
    "Action evidence:",
    len(all_actions)
)

print(
    "Decision evidence:",
    len(all_decisions)
)


print("\n" + "=" * 70)

print(
    "PILOT EXTRACTION COMPLETE"
)

print("=" * 70)
```

```python
# ============================================================
# STEP 130: DISCOVER ALL GC1 REPOSITORIES
# ============================================================
# Purpose:
# Automatically discover every Jira repository/project that
# has corresponding generated PRD/Meeting data.
#
# We previously developed the pipeline manually using:
#
#     24_DOCS
#
# Now we prepare to scale the SAME validated extraction logic
# across the complete GC1 dataset.
#
# IMPORTANT:
# This step only discovers repositories.
# It does NOT modify or extract any data.
# ============================================================


# ------------------------------------------------------------
# Discover Jira repository folders.
# ------------------------------------------------------------

all_jira_repositories = sorted([
    folder
    for folder in JIRA_ROOT.iterdir()
    if folder.is_dir()
])


# ------------------------------------------------------------
# Discover generated PRD/Meeting repository folders.
# ------------------------------------------------------------

all_generated_repositories = sorted([
    folder
    for folder in PRD_MEETING_ROOT.iterdir()
    if folder.is_dir()
])


jira_repository_names = {
    folder.name
    for folder in all_jira_repositories
}


generated_repository_names = {
    folder.name
    for folder in all_generated_repositories
}


# ------------------------------------------------------------
# Repositories existing in BOTH datasets.
#
# These are candidates for automated extraction.
# ------------------------------------------------------------

common_repository_names = sorted(
    jira_repository_names
    &
    generated_repository_names
)


# ------------------------------------------------------------
# Detect mismatches.
# ------------------------------------------------------------

jira_only = sorted(
    jira_repository_names
    -
    generated_repository_names
)


generated_only = sorted(
    generated_repository_names
    -
    jira_repository_names
)


print("=" * 70)
print("GC1 MULTI-PROJECT DISCOVERY")
print("=" * 70)


print(
    "\nJira repositories:",
    len(jira_repository_names)
)


print(
    "Generated-data repositories:",
    len(generated_repository_names)
)


print(
    "Repositories available in BOTH:",
    len(common_repository_names)
)


print("\nCommon repositories:")

for repository in common_repository_names:
    print(" -", repository)


print("\nOnly in Jira:")
print(jira_only)


print("\nOnly in generated data:")
print(generated_only)


# ------------------------------------------------------------
# Basic validation.
# ------------------------------------------------------------

if (
    len(common_repository_names) == 30
    and
    len(jira_only) == 0
    and
    len(generated_only) == 0
):

    print(
        "\nSUCCESS:"
        "\nAll 30 GC1 repositories are aligned "
        "between Jira and generated data."
    )

else:

    print(
        "\nCHECK:"
        "\nRepository counts or folder alignment "
        "differ from the expected 30-project setup."
    )
```

```python
# ============================================================
# STEP 131: DISCOVER VALID INSTANCES FOR ALL REPOSITORIES
# ============================================================
# Purpose:
# Read each repository's instance-selection manifest and
# identify the instances that were actually selected for the
# synthetic GC1 dataset.
#
# Example:
#
#     24_DOCS
#         ↓
#     selected_instances
#         ↓
#     DOCS-I01
#
# IMPORTANT:
#
# repository_instance_manifest.json is HIDDEN CONTROL METADATA.
#
# We use it ONLY for:
#
#     instance discovery
#     timeline alignment
#     validation
#
# It must NOT become:
#
#     extracted entities
#     KG evidence
#     TabPFN features
#     GraphRAG evidence
# ============================================================


repository_instance_inventory = []


for repository_name in common_repository_names:

    generated_repository_folder = (
        PRD_MEETING_ROOT
        / repository_name
    )


    manifest_file = (
        generated_repository_folder
        / "repository_instance_manifest.json"
    )


    # --------------------------------------------------------
    # Handle missing manifest safely.
    # --------------------------------------------------------

    if not manifest_file.exists():

        repository_instance_inventory.append({

            "repository_folder":
                repository_name,

            "repository_key":
                None,

            "project_name":
                None,

            "instance_id":
                None,

            "instance_start":
                None,

            "instance_end":
                None,

            "expected_tickets":
                None,

            "expected_events":
                None,

            "status":
                "MISSING_MANIFEST"
        })

        continue


    # --------------------------------------------------------
    # Load control manifest.
    # --------------------------------------------------------

    with open(
        manifest_file,
        "r",
        encoding="utf-8"
    ) as file:

        manifest = json.load(file)


    repository_key = manifest.get(
        "repository_key"
    )


    project_name = manifest.get(
        "project_name"
    )


    selected_instances = (
        manifest.get(
            "selected_instances"
        )
        or []
    )


    # --------------------------------------------------------
    # Handle repository with no selected instances.
    # --------------------------------------------------------

    if not selected_instances:

        repository_instance_inventory.append({

            "repository_folder":
                repository_name,

            "repository_key":
                repository_key,

            "project_name":
                project_name,

            "instance_id":
                None,

            "instance_start":
                None,

            "instance_end":
                None,

            "expected_tickets":
                None,

            "expected_events":
                None,

            "status":
                "NO_SELECTED_INSTANCE"
        })

        continue


    # --------------------------------------------------------
    # Add one inventory row per selected instance.
    # --------------------------------------------------------

    for instance in selected_instances:

        instance_id = instance.get(
            "instance_id"
        )


        instance_folder = (
            generated_repository_folder
            / instance_id
        )


        repository_instance_inventory.append({

            "repository_folder":
                repository_name,

            "repository_key":
                repository_key,

            "project_name":
                project_name,

            "instance_id":
                instance_id,

            "instance_start":
                instance.get("start"),

            "instance_end":
                instance.get("end"),

            "expected_tickets":
                instance.get("tickets"),

            "expected_events":
                instance.get("events"),

            "instance_folder_exists":
                instance_folder.exists(),

            "status":
                (
                    "READY"
                    if instance_folder.exists()
                    else "MISSING_INSTANCE_FOLDER"
                )
        })


print("=" * 70)
print("GC1 INSTANCE INVENTORY")
print("=" * 70)


print(
    "\nInventory rows:",
    len(repository_instance_inventory)
)


for row in repository_instance_inventory:

    print(
        "\n",
        row["repository_folder"],
        "|",
        row["repository_key"],
        "|",
        row["instance_id"],
        "|",
        row["status"],
        "| tickets:",
        row["expected_tickets"]
    )
```

```python
# ============================================================
# STEP 132: MULTI-PROJECT EXTRACTION READINESS SUMMARY
# ============================================================
# Purpose:
# Summarize exactly how many repositories and valid project
# instances will be processed.
#
# We stop here if important folders/manifests are missing.
# ============================================================

from collections import Counter


inventory_status_counts = Counter(

    row["status"]

    for row in repository_instance_inventory
)


ready_instances = [

    row

    for row in repository_instance_inventory

    if row["status"] == "READY"
]


print("=" * 70)
print("GC1 MULTI-PROJECT EXTRACTION READINESS")
print("=" * 70)


print(
    "\nRepositories discovered:",
    len(common_repository_names)
)


print(
    "Selected instances ready:",
    len(ready_instances)
)


print("\nInventory status:")

for status, count in (
    inventory_status_counts.items()
):

    print(
        f"{status:30s}",
        count
    )


print("\nREADY INSTANCES")
print("-" * 70)


for row in ready_instances:

    print(
        f"{row['repository_folder']:20s}",
        "|",
        f"{row['instance_id']:20s}",
        "| tickets:",
        row["expected_tickets"],
        "|",
        row["instance_start"],
        "→",
        row["instance_end"]
    )


# ------------------------------------------------------------
# Expected workload.
# ------------------------------------------------------------

expected_total_instance_tickets = sum(

    row["expected_tickets"] or 0

    for row in ready_instances
)


print(
    "\nExpected Jira ticket-instance records:",
    expected_total_instance_tickets
)


# ------------------------------------------------------------
# Do not start extraction if there are structural errors.
# ------------------------------------------------------------

bad_instances = [

    row

    for row in repository_instance_inventory

    if row["status"] != "READY"
]


if not bad_instances:

    print(
        "\nSUCCESS:"
        "\nAll selected instances are structurally ready "
        "for automated extraction."
    )

else:

    print(
        "\nCHECK REQUIRED:"
        "\nSome repositories/instances are not ready."
    )


    for row in bad_instances:

        print(
            " -",
            row["repository_folder"],
            "|",
            row["instance_id"],
            "|",
            row["status"]
        )
```

```python
# ============================================================
# STEP 133: PRODUCTION EXTRACTION CONFIGURATION
# ============================================================
# Purpose:
# Freeze the configuration validated on DOCS-I01 so the same
# methodology can be applied consistently to all 30 projects.
#
# IMPORTANT:
# We are NOT changing the extraction methodology here.
# We are packaging the validated pilot settings for reuse.
# ============================================================


# ------------------------------------------------------------
# GLiNER configuration validated during the pilot.
# ------------------------------------------------------------

PRODUCTION_GLINER_MODEL = GLINER_MODEL_NAME

PRODUCTION_GLINER_INFERENCE_THRESHOLD = 0.40
PRODUCTION_GLINER_ACCEPT_THRESHOLD = 0.55


# ------------------------------------------------------------
# Semantic entity labels.
# ------------------------------------------------------------

PRODUCTION_GLINER_LABELS = [
    "feature",
    "system",
    "component",
    "team",
    "person",
    "milestone"
]


# ------------------------------------------------------------
# Meeting-specific GLiNER labels.
#
# PERSON is deliberately excluded because meeting participants
# and speakers are available deterministically.
# ------------------------------------------------------------

PRODUCTION_MEETING_GLINER_LABELS = {
    "feature",
    "system",
    "component",
    "team"
}


# ------------------------------------------------------------
# Output filenames.
# ------------------------------------------------------------

JIRA_OUTPUT_NAME = "jira_extracted.jsonl"
PRD_OUTPUT_NAME = "prd_extracted.jsonl"
MEETING_OUTPUT_NAME = "meetings_extracted.jsonl"
VALIDATION_OUTPUT_NAME = "validation_report.json"


print("=" * 70)
print("PRODUCTION CONFIGURATION")
print("=" * 70)

print("\nReady instances :", len(ready_instances))

print(
    "Expected Jira records:",
    expected_total_instance_tickets
)

print(
    "GLiNER inference threshold:",
    PRODUCTION_GLINER_INFERENCE_THRESHOLD
)

print(
    "GLiNER acceptance threshold:",
    PRODUCTION_GLINER_ACCEPT_THRESHOLD
)

print("\nSUCCESS: Production configuration frozen.")
```

```python
# ============================================================
# STEP 134: GENERIC INSTANCE INPUT DISCOVERY
# ============================================================
# Purpose:
# Given one inventory row, discover all files required by the
# extraction pipeline.
#
# This function works for:
#
#     JBEAP-I01
#     JBIDE-I01
#     SAK-I01
#     ...
#     QTCREATORBUG-I01
#
# It does NOT assume DOCS-specific filenames.
# ============================================================


def discover_instance_inputs(
    inventory_row
):

    repository_folder = (
        inventory_row[
            "repository_folder"
        ]
    )

    instance_id = (
        inventory_row[
            "instance_id"
        ]
    )


    # --------------------------------------------------------
    # Repository folders.
    # --------------------------------------------------------

    jira_repository_folder = (
        JIRA_ROOT
        / repository_folder
    )


    generated_repository_folder = (
        PRD_MEETING_ROOT
        / repository_folder
    )


    instance_folder = (
        generated_repository_folder
        / instance_id
    )


    # --------------------------------------------------------
    # Jira input.
    # --------------------------------------------------------

    jira_file = (
        jira_repository_folder
        / "jira_tickets.jsonl"
    )


    # --------------------------------------------------------
    # PRD / Meeting directories.
    # --------------------------------------------------------

    prd_directory = (
        instance_folder
        / "prds"
    )


    meeting_directory = (
        instance_folder
        / "meetings"
    )


    # --------------------------------------------------------
    # Final generated PRDs only.
    # Exclude *.prompt.txt.
    # --------------------------------------------------------

    prd_files = sorted([

        file

        for file in (
            prd_directory.glob("*.txt")
            if prd_directory.exists()
            else []
        )

        if not file.name.endswith(
            ".prompt.txt"
        )
    ])


    # --------------------------------------------------------
    # Final generated meetings only.
    # --------------------------------------------------------

    meeting_files = sorted([

        file

        for file in (
            meeting_directory.glob("*.txt")
            if meeting_directory.exists()
            else []
        )

        if not file.name.endswith(
            ".prompt.txt"
        )
    ])


    # --------------------------------------------------------
    # Snapshot manifest used only for alignment/control.
    # --------------------------------------------------------

    snapshot_manifest_file = (
        instance_folder
        / "snapshot_manifest.json"
    )


    # --------------------------------------------------------
    # Output directory.
    # --------------------------------------------------------

    output_folder = (
        EXTRACTION_ROOT
        / repository_folder
        / instance_id
    )


    return {

        "repository_folder":
            repository_folder,

        "repository_key":
            inventory_row[
                "repository_key"
            ],

        "project_name":
            inventory_row[
                "project_name"
            ],

        "instance_id":
            instance_id,

        "instance_start":
            inventory_row[
                "instance_start"
            ],

        "instance_end":
            inventory_row[
                "instance_end"
            ],

        "expected_tickets":
            inventory_row[
                "expected_tickets"
            ],

        "expected_events":
            inventory_row[
                "expected_events"
            ],

        "jira_file":
            jira_file,

        "prd_files":
            prd_files,

        "meeting_files":
            meeting_files,

        "snapshot_manifest_file":
            snapshot_manifest_file,

        "output_folder":
            output_folder
    }
```

```python
# ============================================================
# STEP 135: VALIDATE INPUT FILES FOR ALL 30 INSTANCES
# ============================================================
# Purpose:
# Check every selected instance BEFORE running extraction.
#
# Expected per instance:
#
#     Jira file       = present
#     PRD files       = 1
#     Meeting files   = 2
#     Snapshot file   = present
#
# This catches structural problems early.
# ============================================================


production_inputs = []

input_validation_rows = []


for inventory_row in ready_instances:

    inputs = discover_instance_inputs(
        inventory_row
    )


    production_inputs.append(
        inputs
    )


    jira_ok = (
        inputs[
            "jira_file"
        ].exists()
    )


    prd_ok = (
        len(
            inputs[
                "prd_files"
            ]
        )
        == 1
    )


    meetings_ok = (
        len(
            inputs[
                "meeting_files"
            ]
        )
        == 2
    )


    snapshot_ok = (
        inputs[
            "snapshot_manifest_file"
        ].exists()
    )


    all_ok = (
        jira_ok
        and prd_ok
        and meetings_ok
        and snapshot_ok
    )


    input_validation_rows.append({

        "repository":
            inputs[
                "repository_folder"
            ],

        "instance":
            inputs[
                "instance_id"
            ],

        "jira":
            jira_ok,

        "prd_count":
            len(
                inputs[
                    "prd_files"
                ]
            ),

        "meeting_count":
            len(
                inputs[
                    "meeting_files"
                ]
            ),

        "snapshot":
            snapshot_ok,

        "ready":
            all_ok
    })


print("=" * 70)
print("ALL-INSTANCE INPUT VALIDATION")
print("=" * 70)


for row in input_validation_rows:

    print(
        f"{row['repository']:20s}",
        "|",
        f"{row['instance']:20s}",
        "| Jira:",
        row["jira"],
        "| PRD:",
        row["prd_count"],
        "| Meetings:",
        row["meeting_count"],
        "| Snapshot:",
        row["snapshot"],
        "| READY:",
        row["ready"]
    )


ready_input_count = sum(

    1

    for row in input_validation_rows

    if row["ready"]
)


print(
    "\nInstances passing input validation:",
    ready_input_count,
    "/",
    len(input_validation_rows)
)


if (
    ready_input_count
    ==
    len(input_validation_rows)
):

    print(
        "\nSUCCESS:"
        "\nAll production instances have the "
        "required extraction inputs."
    )

else:

    print(
        "\nSTOP:"
        "\nOne or more instances are missing "
        "required source files."
    )
```

```python
# ============================================================
# STEP 135A: INSPECT JBEAP-I01 DOCUMENT STRUCTURE
# ============================================================
# Purpose:
# JBEAP-I01 contains more generated artefacts than our DOCS
# pilot:
#
#     4 PRDs
#     6 Meetings
#
# Before changing the production pipeline, inspect their exact
# filenames and metadata.
#
# IMPORTANT:
# We do NOT delete or ignore anything yet.
# We first determine whether these are:
#
#     multiple PRD versions
#     multiple scheduled meetings
#     duplicate files
#     or another generation structure
# ============================================================


jbeap_inputs = next(

    inputs

    for inputs in production_inputs

    if inputs["repository_folder"] == "01_JBEAP"
)


print("=" * 70)
print("JBEAP-I01 DOCUMENT STRUCTURE")
print("=" * 70)


print("\nPRD FILES")
print("-" * 70)

for file in jbeap_inputs["prd_files"]:

    print(
        file.name,
        "|",
        round(
            file.stat().st_size / 1024,
            2
        ),
        "KB"
    )


print("\nMEETING FILES")
print("-" * 70)

for file in jbeap_inputs["meeting_files"]:

    print(
        file.name,
        "|",
        round(
            file.stat().st_size / 1024,
            2
        ),
        "KB"
    )


# ------------------------------------------------------------
# Also inspect corresponding metadata filenames.
# ------------------------------------------------------------

instance_folder = (
    PRD_MEETING_ROOT
    / "01_JBEAP"
    / "JBEAP-I01"
)


prd_meta_files = sorted(
    (
        instance_folder
        / "prds"
    ).glob("*.meta.json")
)


meeting_meta_files = sorted(
    (
        instance_folder
        / "meetings"
    ).glob("*.meta.json")
)


print("\nPRD METADATA FILES")
print("-" * 70)

for file in prd_meta_files:
    print(file.name)


print("\nMEETING METADATA FILES")
print("-" * 70)

for file in meeting_meta_files:
    print(file.name)
```

```python
# ============================================================
# STEP 135B: INSPECT JBEAP GENERATION SCHEDULE
# ============================================================
# Purpose:
# Determine whether the 4 PRDs and 6 meetings are intentional
# generated artefacts belonging to JBEAP-I01.
#
# The generation manifest is CONTROL METADATA ONLY.
#
# We use it for:
#     artefact discovery / validation
#
# We do NOT feed it into:
#     extraction
#     GLiNER
#     KG
#     TabPFN
# ============================================================


jbeap_generation_manifest_file = (
    PRD_MEETING_ROOT
    / "01_JBEAP"
    / "JBEAP-I01"
    / "generation_manifest.json"
)


with open(
    jbeap_generation_manifest_file,
    "r",
    encoding="utf-8"
) as file:

    jbeap_generation_manifest = json.load(
        file
    )


print("=" * 70)
print("JBEAP-I01 GENERATION MANIFEST")
print("=" * 70)


print("\nTop-level keys:")

for key in jbeap_generation_manifest.keys():
    print(" -", key)


# ------------------------------------------------------------
# Print fields related to generated artefacts/schedules.
# We don't assume exact field names.
# ------------------------------------------------------------

print("\nARTEFACT / SCHEDULE RELATED CONTENT")
print("-" * 70)


for key, value in (
    jbeap_generation_manifest.items()
):

    key_lower = key.lower()


    if any(
        word in key_lower

        for word in [
            "prd",
            "meeting",
            "schedule",
            "artifact",
            "artefact",
            "document"
        ]
    ):

        print("\nKEY:", key)

        print(
            json.dumps(
                value,
                indent=2,
                ensure_ascii=False
            )
        )
```

```python
# ============================================================
# STEP 135C: CORRECT PRODUCTION ARTEFACT DISCOVERY
# ============================================================
# Purpose:
# Select ONLY the PRDs and meetings officially listed in each
# instance's generation_manifest.json.
#
# WHY:
# Some folders may contain extra generated files from previous
# generation runs.
#
# Example:
#
# JBEAP-I01 folder physically contains:
#
#   PRD V1, V2, V3, V4
#   M001 ... M006
#
# But its official generation manifest contains only:
#
#   PRD V1
#   M001
#   M002
#
# Therefore folder contents alone are NOT authoritative.
#
# The generation manifest is used ONLY as control metadata for
# selecting the correct source artefacts.
#
# It is NOT passed to GLiNER and does NOT become:
#
#   extraction evidence
#   KG evidence
#   TabPFN features
# ============================================================


def discover_instance_inputs(
    inventory_row
):

    repository_folder = (
        inventory_row["repository_folder"]
    )

    instance_id = (
        inventory_row["instance_id"]
    )


    # --------------------------------------------------------
    # Main folders
    # --------------------------------------------------------

    jira_repository_folder = (
        JIRA_ROOT
        / repository_folder
    )

    generated_repository_folder = (
        PRD_MEETING_ROOT
        / repository_folder
    )

    instance_folder = (
        generated_repository_folder
        / instance_id
    )


    # --------------------------------------------------------
    # Jira input
    # --------------------------------------------------------

    jira_file = (
        jira_repository_folder
        / "jira_tickets.jsonl"
    )


    # --------------------------------------------------------
    # Control manifests
    # --------------------------------------------------------

    generation_manifest_file = (
        instance_folder
        / "generation_manifest.json"
    )

    snapshot_manifest_file = (
        instance_folder
        / "snapshot_manifest.json"
    )


    # --------------------------------------------------------
    # Load generation manifest.
    # --------------------------------------------------------

    if not generation_manifest_file.exists():

        return {

            "repository_folder":
                repository_folder,

            "repository_key":
                inventory_row["repository_key"],

            "project_name":
                inventory_row["project_name"],

            "instance_id":
                instance_id,

            "instance_start":
                inventory_row["instance_start"],

            "instance_end":
                inventory_row["instance_end"],

            "expected_tickets":
                inventory_row["expected_tickets"],

            "expected_events":
                inventory_row["expected_events"],

            "jira_file":
                jira_file,

            "prd_files":
                [],

            "meeting_files":
                [],

            "generation_manifest_file":
                generation_manifest_file,

            "snapshot_manifest_file":
                snapshot_manifest_file,

            "output_folder":
                (
                    EXTRACTION_ROOT
                    / repository_folder
                    / instance_id
                ),

            "discovery_error":
                "MISSING_GENERATION_MANIFEST"
        }


    with open(
        generation_manifest_file,
        "r",
        encoding="utf-8"
    ) as file:

        generation_manifest = json.load(
            file
        )


    # ========================================================
    # OFFICIAL PRD FILES
    # ========================================================

    official_prd_files = []


    for scheduled_prd in (
        generation_manifest.get(
            "prd_schedule"
        )
        or []
    ):

        # Prefer document_id rather than trusting the absolute
        # output path embedded in control metadata.
        document_id = scheduled_prd.get(
            "document_id"
        )


        if document_id:

            prd_path = (
                instance_folder
                / "prds"
                / f"{document_id}.txt"
            )

            official_prd_files.append(
                prd_path
            )


    # ========================================================
    # OFFICIAL MEETING FILES
    # ========================================================

    official_meeting_files = []


    for scheduled_meeting in (
        generation_manifest.get(
            "meeting_schedule"
        )
        or []
    ):

        meeting_id = scheduled_meeting.get(
            "meeting_id"
        )


        if meeting_id:

            meeting_path = (
                instance_folder
                / "meetings"
                / f"{meeting_id}.txt"
            )

            official_meeting_files.append(
                meeting_path
            )


    # --------------------------------------------------------
    # Remove accidental duplicates while preserving order.
    # --------------------------------------------------------

    official_prd_files = list(
        dict.fromkeys(
            official_prd_files
        )
    )


    official_meeting_files = list(
        dict.fromkeys(
            official_meeting_files
        )
    )


    # --------------------------------------------------------
    # Output folder
    # --------------------------------------------------------

    output_folder = (
        EXTRACTION_ROOT
        / repository_folder
        / instance_id
    )


    return {

        "repository_folder":
            repository_folder,

        "repository_key":
            inventory_row[
                "repository_key"
            ],

        "project_name":
            inventory_row[
                "project_name"
            ],

        "instance_id":
            instance_id,

        "instance_start":
            inventory_row[
                "instance_start"
            ],

        "instance_end":
            inventory_row[
                "instance_end"
            ],

        "expected_tickets":
            inventory_row[
                "expected_tickets"
            ],

        "expected_events":
            inventory_row[
                "expected_events"
            ],

        "jira_file":
            jira_file,

        # IMPORTANT:
        # These now come from the official generation schedule.
        "prd_files":
            official_prd_files,

        "meeting_files":
            official_meeting_files,

        "generation_manifest_file":
            generation_manifest_file,

        "snapshot_manifest_file":
            snapshot_manifest_file,

        "output_folder":
            output_folder,

        "discovery_error":
            None
    }


print(
    "SUCCESS: Production artefact discovery "
    "now follows generation_manifest schedules."
)
```

```python
# ============================================================
# STEP 135D: REBUILD PRODUCTION INPUT INVENTORY
# ============================================================
# Purpose:
# Re-run source discovery for all 30 instances using the
# corrected generation-manifest-based selection logic.
# ============================================================


production_inputs = [

    discover_instance_inputs(
        inventory_row
    )

    for inventory_row
    in ready_instances
]


print("=" * 70)
print("PRODUCTION INPUT INVENTORY REBUILT")
print("=" * 70)

print(
    "\nInstances:",
    len(production_inputs)
)
```

```python
# ============================================================
# STEP 135E: REVALIDATE OFFICIAL PRODUCTION INPUTS
# ============================================================
# Purpose:
# Validate the official artefacts selected from each
# generation manifest.
#
# Expected GC1 generation contract:
#
#     1 official PRD
#     2 official meetings
#
# Extra files physically present in a folder are ignored when
# they are not part of the official generation schedule.
# ============================================================


input_validation_rows = []


for inputs in production_inputs:

    jira_ok = (
        inputs["jira_file"].exists()
    )


    generation_manifest_ok = (
        inputs[
            "generation_manifest_file"
        ].exists()
    )


    snapshot_ok = (
        inputs[
            "snapshot_manifest_file"
        ].exists()
    )


    prd_count = len(
        inputs["prd_files"]
    )


    meeting_count = len(
        inputs["meeting_files"]
    )


    prd_files_exist = all(
        file.exists()
        for file in inputs[
            "prd_files"
        ]
    )


    meeting_files_exist = all(
        file.exists()
        for file in inputs[
            "meeting_files"
        ]
    )


    # --------------------------------------------------------
    # For the current GC1 generation contract we expect:
    #
    #     1 scheduled PRD
    #     2 scheduled meetings
    #
    # This is checked against the manifest-selected artefacts,
    # NOT against every file physically present in the folder.
    # --------------------------------------------------------

    prd_schedule_ok = (
        prd_count == 1
    )


    meeting_schedule_ok = (
        meeting_count == 2
    )


    all_ok = all([

        jira_ok,
        generation_manifest_ok,
        snapshot_ok,
        prd_files_exist,
        meeting_files_exist,
        prd_schedule_ok,
        meeting_schedule_ok,
        inputs[
            "discovery_error"
        ] is None
    ])


    input_validation_rows.append({

        "repository":
            inputs[
                "repository_folder"
            ],

        "instance":
            inputs[
                "instance_id"
            ],

        "jira":
            jira_ok,

        "prd_count":
            prd_count,

        "prd_files_exist":
            prd_files_exist,

        "meeting_count":
            meeting_count,

        "meeting_files_exist":
            meeting_files_exist,

        "snapshot":
            snapshot_ok,

        "generation_manifest":
            generation_manifest_ok,

        "ready":
            all_ok
    })


print("=" * 70)
print("OFFICIAL ALL-INSTANCE INPUT VALIDATION")
print("=" * 70)


for row in input_validation_rows:

    print(

        f"{row['repository']:20s}",

        "|",

        f"{row['instance']:20s}",

        "| Jira:",
        row["jira"],

        "| PRD:",
        row["prd_count"],

        "| Meetings:",
        row["meeting_count"],

        "| READY:",
        row["ready"]
    )


ready_input_count = sum(

    1

    for row in input_validation_rows

    if row["ready"]
)


print(
    "\nInstances passing validation:",
    ready_input_count,
    "/",
    len(input_validation_rows)
)


if (
    ready_input_count
    ==
    len(input_validation_rows)
):

    print(
        "\nSUCCESS:"
        "\nAll 30 selected instances have the "
        "correct official source artefacts."
    )

else:

    print(
        "\nSTOP:"
        "\nAt least one official instance still "
        "has an input problem."
    )
```

```python
# ============================================================
# STEP 136: SELECT SECOND PROJECT FOR PRODUCTION TEST
# ============================================================
# Purpose:
# Test the generalized multi-project pipeline on a repository
# other than DOCS before launching all 30.
#
# Selected:
#
#     01_JBEAP
#         ↓
#     JBEAP-I01
#
# Expected aligned Jira tickets:
#     131
# ============================================================


PRODUCTION_TEST_REPOSITORY = (
    "01_JBEAP"
)


test_inputs = next(

    inputs

    for inputs in production_inputs

    if inputs[
        "repository_folder"
    ]
    ==
    PRODUCTION_TEST_REPOSITORY
)


print("=" * 70)
print("SECOND-PROJECT PRODUCTION TEST")
print("=" * 70)


print(
    "\nRepository:",
    test_inputs[
        "repository_folder"
    ]
)


print(
    "Project key:",
    test_inputs[
        "repository_key"
    ]
)


print(
    "Instance:",
    test_inputs[
        "instance_id"
    ]
)


print(
    "Timeline:",
    test_inputs[
        "instance_start"
    ],
    "→",
    test_inputs[
        "instance_end"
    ]
)


print(
    "Expected Jira tickets:",
    test_inputs[
        "expected_tickets"
    ]
)


print(
    "\nJira file:",
    test_inputs[
        "jira_file"
    ]
)


print(
    "\nPRD:"
)

for file in test_inputs[
    "prd_files"
]:
    print(" -", file.name)


print(
    "\nMeetings:"
)

for file in test_inputs[
    "meeting_files"
]:
    print(" -", file.name)
```

```python
# ============================================================
# STEP 137: TEST GENERIC JIRA ALIGNMENT ON JBEAP-I01
# ============================================================
# Purpose:
# Verify that our generic temporal-alignment logic reproduces
# the expected JBEAP-I01 ticket count.
#
# Expected:
#
#     131 tickets
#
# IMPORTANT:
# We still do NOT extract/save JBEAP yet.
#
# First we validate alignment.
# ============================================================


# ------------------------------------------------------------
# Load repository Jira tickets.
# ------------------------------------------------------------

test_all_jira_tickets = []


with open(
    test_inputs[
        "jira_file"
    ],
    "r",
    encoding="utf-8"
) as file:

    for line in file:

        if line.strip():

            test_all_jira_tickets.append(
                json.loads(line)
            )


# ------------------------------------------------------------
# Normalize instance boundaries.
# ------------------------------------------------------------

test_instance_start = (
    parse_timestamp_utc(
        test_inputs[
            "instance_start"
        ]
    )
)


test_instance_end = (
    parse_timestamp_utc(
        test_inputs[
            "instance_end"
        ]
    )
)


# ------------------------------------------------------------
# Identify tickets with activity inside the instance window.
# ------------------------------------------------------------

test_instance_jira_tickets = []


for ticket in test_all_jira_tickets:

    activity_times = (
        collect_ticket_activity_times(
            ticket
        )
    )


    # Convert every activity timestamp to UTC-safe datetime.
    normalized_activity_times = [

        parse_timestamp_utc(
            timestamp
        )

        for timestamp
        in activity_times
    ]


    activity_in_window = [

        timestamp

        for timestamp
        in normalized_activity_times

        if (
            timestamp is not None
            and
            test_instance_start
            <= timestamp
            <= test_instance_end
        )
    ]


    if activity_in_window:

        test_instance_jira_tickets.append({

            "ticket":
                ticket,

            "activity_times":
                activity_in_window
        })


calculated_count = len(
    test_instance_jira_tickets
)


expected_count = test_inputs[
    "expected_tickets"
]


print("=" * 70)
print("JBEAP-I01 ALIGNMENT TEST")
print("=" * 70)


print(
    "\nRepository Jira tickets:",
    len(test_all_jira_tickets)
)


print(
    "Calculated instance tickets:",
    calculated_count
)


print(
    "Expected instance tickets:",
    expected_count
)


print(
    "Difference:",
    calculated_count
    - expected_count
)


if calculated_count == expected_count:

    print(
        "\nSUCCESS:"
        "\nGeneric Jira alignment works on JBEAP-I01."
    )

else:

    print(
        "\nSTOP:"
        "\nGeneric alignment does not reproduce "
        "the expected JBEAP-I01 ticket count."
    )
```

```python
# ============================================================
# STEP 138: AUDIT JBEAP JIRA CHANGELOG FIELDS
# ============================================================
# Purpose:
# Before applying the DOCS-validated Jira extractor to another
# repository, inspect all changelog field names appearing in
# the aligned JBEAP-I01 tickets.
#
# WHY:
# Different Jira repositories may use custom fields that were
# not present in DOCS.
#
# We want to identify those BEFORE production extraction so
# useful information is not silently collapsed into:
#
#     OTHER_CHANGE
#
# This is a validation/audit step only.
# ============================================================

from collections import Counter


jbeap_changelog_fields = Counter()


for item in test_instance_jira_tickets:

    ticket = item["ticket"]


    for history in (
        ticket.get("full_changelog") or []
    ):

        history_time = parse_timestamp_utc(
            history.get("created")
        )


        # Only inspect events belonging to JBEAP-I01.
        if not (
            history_time
            and
            test_instance_start
            <= history_time
            <= test_instance_end
        ):
            continue


        for change_item in (
            history.get("items") or []
        ):

            field_name = str(
                change_item.get("field") or ""
            ).strip()


            if field_name:

                jbeap_changelog_fields[
                    field_name
                ] += 1


print("=" * 70)
print("JBEAP-I01 CHANGELOG FIELD AUDIT")
print("=" * 70)


print(
    "\nUnique changelog fields:",
    len(jbeap_changelog_fields)
)


for field_name, count in (
    jbeap_changelog_fields.most_common()
):

    mapped_event = (
        JIRA_CHANGE_EVENT_MAP.get(
            field_name.lower()
        )
    )


    print(
        f"{field_name:35s}",
        f"{count:5d}",
        "|",
        mapped_event
        if mapped_event
        else "UNMAPPED"
    )
```

```python
# ============================================================
# STEP 139: IDENTIFY NEW JIRA FIELD TYPES
# ============================================================
# Purpose:
# Produce a compact list of JBEAP fields that are not covered
# by our existing production event vocabulary.
#
# If this list is empty, the DOCS event mapping generalizes
# directly to JBEAP.
#
# If it is not empty, we inspect the new fields before
# productionizing Jira extraction.
# ============================================================


jbeap_unmapped_fields = {

    field_name: count

    for field_name, count
    in jbeap_changelog_fields.items()

    if (
        field_name.lower()
        not in JIRA_CHANGE_EVENT_MAP
    )
}


print("=" * 70)
print("JBEAP-I01 UNMAPPED CHANGELOG FIELDS")
print("=" * 70)


print(
    "\nUnmapped fields:",
    len(jbeap_unmapped_fields)
)


for field_name, count in sorted(
    jbeap_unmapped_fields.items(),
    key=lambda item: -item[1]
):

    print(
        f"{field_name:35s}",
        count
    )


if not jbeap_unmapped_fields:

    print(
        "\nSUCCESS:"
        "\nThe current Jira event vocabulary covers "
        "all JBEAP-I01 changelog fields."
    )
```

```python
# ============================================================
# STEP 140: EXPAND JIRA EVENT VOCABULARY FOR JBEAP
# ============================================================
# Purpose:
# JBEAP-I01 introduced 24 changelog fields that did not occur
# in the DOCS pilot.
#
# We assign explicit event types instead of collapsing them
# into OTHER_CHANGE.
#
# This preserves potentially useful information for:
#
#     Temporal KG
#     release/schedule features
#     effort features
#     blocker features
#     ownership features
#     cross-team/dependency features
# ============================================================


JBEAP_EVENT_MAP = {

    # --------------------------------------------------------
    # Release / schedule
    # --------------------------------------------------------

    "cdw release":
        "RELEASE_CHANGE",

    "target release":
        "TARGET_RELEASE_CHANGE",

    "fix build":
        "BUILD_TARGET_CHANGE",


    # --------------------------------------------------------
    # Workflow acknowledgement / coordination
    # --------------------------------------------------------

    "cdw devel_ack":
        "DEVELOPMENT_ACK_CHANGE",

    "cdw pm_ack":
        "PM_ACK_CHANGE",

    "cdw qa_ack":
        "QA_ACK_CHANGE",

    "cdw docs_ack":
        "DOCS_ACK_CHANGE",


    # --------------------------------------------------------
    # Development artefacts / links
    # --------------------------------------------------------

    "git pull request":
        "PULL_REQUEST_CHANGE",

    "bugzilla update":
        "BUGZILLA_UPDATE",

    "bugzilla references":
        "BUGZILLA_REFERENCE_CHANGE",

    "help desk ticket reference":
        "HELPDESK_REFERENCE_CHANGE",


    # --------------------------------------------------------
    # Ownership / responsibility
    # --------------------------------------------------------

    "tester":
        "TESTER_CHANGE",

    "need info from":
        "INFO_REQUEST_OWNER_CHANGE",


    # --------------------------------------------------------
    # Effort / time
    # --------------------------------------------------------

    "timeestimate":
        "ESTIMATE_CHANGE",

    "timeoriginalestimate":
        "ORIGINAL_ESTIMATE_CHANGE",

    "timespent":
        "ACTUAL_TIME_CHANGE",

    "worklogid":
        "WORKLOG_CHANGE",


    # --------------------------------------------------------
    # Blocker information
    # --------------------------------------------------------

    "cdw blocker":
        "BLOCKER_FIELD_CHANGE",


    # --------------------------------------------------------
    # Scope / impact
    # --------------------------------------------------------

    "affects":
        "AFFECTS_CHANGE",

    "affects testing":
        "TEST_IMPACT_CHANGE",


    # --------------------------------------------------------
    # Documentation / release-note information
    # --------------------------------------------------------

    "release notes docs status":
        "RELEASE_NOTE_STATUS_CHANGE",

    "release notes text":
        "RELEASE_NOTE_TEXT_CHANGE",

    "steps to reproduce":
        "REPRODUCTION_STEPS_CHANGE",


    # --------------------------------------------------------
    # Environment/context
    # --------------------------------------------------------

    "environment":
        "ENVIRONMENT_CHANGE"
}


# ------------------------------------------------------------
# Extend the existing production mapping.
# ------------------------------------------------------------

JIRA_CHANGE_EVENT_MAP.update(
    JBEAP_EVENT_MAP
)


print("=" * 70)
print("JIRA EVENT VOCABULARY UPDATED")
print("=" * 70)


print(
    "\nJBEAP fields added:",
    len(JBEAP_EVENT_MAP)
)


print(
    "Total mapped Jira fields:",
    len(JIRA_CHANGE_EVENT_MAP)
)


for field_name, event_type in (
    JBEAP_EVENT_MAP.items()
):

    print(
        f"{field_name:35s}",
        "->",
        event_type
    )
```

```python
# ============================================================
# STEP 141: RECHECK JBEAP CHANGELOG COVERAGE
# ============================================================
# Purpose:
# Verify that every changelog field observed in JBEAP-I01 now
# has an explicit event type.
# ============================================================


jbeap_still_unmapped = {

    field_name: count

    for field_name, count
    in jbeap_changelog_fields.items()

    if (
        field_name.lower()
        not in JIRA_CHANGE_EVENT_MAP
    )
}


print("=" * 70)
print("JBEAP CHANGELOG COVERAGE RECHECK")
print("=" * 70)


print(
    "\nStill unmapped:",
    len(jbeap_still_unmapped)
)


for field_name, count in (
    jbeap_still_unmapped.items()
):

    print(
        field_name,
        count
    )


if not jbeap_still_unmapped:

    print(
        "\nSUCCESS:"
        "\nAll JBEAP-I01 changelog fields now have "
        "explicit event mappings."
    )
```

```python
# ============================================================
# STEP 142: GLOBAL JIRA CHANGELOG FIELD AUDIT
# ============================================================
# Purpose:
# Scan the aligned Jira activity of ALL 30 project instances
# and identify any changelog fields not yet covered by our
# production event vocabulary.
#
# This prevents us from repeatedly discovering new custom
# fields while processing projects one by one.
#
# IMPORTANT:
# This is an audit only.
# No extraction output is written.
# ============================================================


global_changelog_field_counts = Counter()

global_unmapped_field_counts = Counter()

global_unmapped_field_projects = {}


for index, inputs in enumerate(
    production_inputs,
    start=1
):

    repository = inputs[
        "repository_folder"
    ]


    instance_start = (
        parse_timestamp_utc(
            inputs["instance_start"]
        )
    )


    instance_end = (
        parse_timestamp_utc(
            inputs["instance_end"]
        )
    )


    # --------------------------------------------------------
    # Load the repository's 300 Jira tickets.
    # --------------------------------------------------------

    repository_tickets = []


    with open(
        inputs["jira_file"],
        "r",
        encoding="utf-8"
    ) as file:

        for line in file:

            if line.strip():

                repository_tickets.append(
                    json.loads(line)
                )


    # --------------------------------------------------------
    # Inspect changelog events occurring inside the selected
    # instance timeline.
    # --------------------------------------------------------

    for ticket in repository_tickets:

        for history in (
            ticket.get(
                "full_changelog"
            )
            or []
        ):

            history_time = (
                parse_timestamp_utc(
                    history.get("created")
                )
            )


            if not (
                history_time
                and
                instance_start
                <= history_time
                <= instance_end
            ):

                continue


            for change_item in (
                history.get("items")
                or []
            ):

                field_name = str(
                    change_item.get(
                        "field"
                    )
                    or ""
                ).strip()


                if not field_name:
                    continue


                global_changelog_field_counts[
                    field_name
                ] += 1


                if (
                    field_name.lower()
                    not in JIRA_CHANGE_EVENT_MAP
                ):

                    global_unmapped_field_counts[
                        field_name
                    ] += 1


                    global_unmapped_field_projects.setdefault(
                        field_name,
                        set()
                    ).add(
                        repository
                    )


    print(
        f"[{index:02d}/30]",
        repository,
        "audited"
    )


print("\n" + "=" * 70)
print("GLOBAL JIRA CHANGELOG FIELD AUDIT")
print("=" * 70)


print(
    "\nUnique changelog fields across GC1:",
    len(
        global_changelog_field_counts
    )
)


print(
    "Unique fields still unmapped:",
    len(
        global_unmapped_field_counts
    )
)


print("\nUNMAPPED FIELDS")
print("-" * 70)


for field_name, count in (
    global_unmapped_field_counts
    .most_common()
):

    projects = sorted(
        global_unmapped_field_projects[
            field_name
        ]
    )


    print(
        f"{field_name:40s}",
        f"{count:6d}",
        "| projects:",
        ", ".join(projects)
    )


if not global_unmapped_field_counts:

    print(
        "\nSUCCESS:"
        "\nThe production event vocabulary covers "
        "every Jira changelog field observed across "
        "all 30 aligned GC1 instances."
    )
```

```python
# ============================================================
# STEP 144: PRODUCTION JIRA EVENT-TYPE RESOLVER
# ============================================================
# Purpose:
# Standardize known Jira fields while safely preserving
# repository-specific custom fields.
#
# Known field:
#
#     Sprint
#        ↓
#     SPRINT_CHANGE
#
# Unknown/custom field:
#
#     User Impact Score
#        ↓
#     CUSTOM_FIELD_CHANGE
#
# BUT the original field name and old/new values remain in:
#
#     attributes["jira_field"]
#     from_value
#     to_value
#
# Therefore NO source information is discarded.
# ============================================================


def resolve_jira_event_type(
    field_name
):

    normalized_field = str(
        field_name or ""
    ).strip().lower()


    if normalized_field in (
        JIRA_CHANGE_EVENT_MAP
    ):

        return (
            JIRA_CHANGE_EVENT_MAP[
                normalized_field
            ]
        )


    return "CUSTOM_FIELD_CHANGE"


print("=" * 70)
print("PRODUCTION EVENT RESOLVER TEST")
print("=" * 70)


test_fields = [
    "status",
    "Sprint",
    "timeestimate",
    "fixVersions",
    "User Impact Score",
    "Hadoop Flags"
]


for field in test_fields:

    print(
        f"{field:25s}",
        "->",
        resolve_jira_event_type(
            field
        )
    )
```

```python
# ============================================================
# STEP 145: FINAL PRODUCTION JIRA EXTRACTION FUNCTION
# ============================================================
# Purpose:
# Convert ONE Jira ticket into the common GC1 Master Extraction
# JSON structure.
#
# This is the PRODUCTION version of the Jira extractor.
#
# It incorporates everything validated during:
#
#   DOCS-I01 pilot
#   JBEAP-I01 audit
#   Global 30-project Jira field audit
#
# ------------------------------------------------------------
# WHAT THIS EXTRACTOR PRESERVES
# ------------------------------------------------------------
#
# ENTITIES:
#   PROJECT
#   ISSUE
#   PERSON
#   COMPONENT
#   linked ISSUE entities
#
# RELATIONSHIPS:
#   HAS_ISSUE
#   ASSIGNED_TO
#   REPORTED_BY
#   CREATED_BY
#   HAS_COMPONENT
#   DEPENDS_ON
#   BLOCKS
#   BLOCKED_BY
#   RELATED_TO
#
# TEMPORAL EVENTS:
#   status changes
#   sprint changes
#   assignment changes
#   priority changes
#   effort changes
#   version changes
#   workflow changes
#   project-specific custom-field changes
#   etc.
#
# BLOCKERS / DEPENDENCIES
#
# ATTRIBUTES:
#   summary
#   description
#   priority
#   labels
#   versions
#   comments
#   etc.
#
# PROVENANCE:
#   source file
#   source issue
#   source evidence
#
# ------------------------------------------------------------
# IMPORTANT PRODUCTION CHANGE
# ------------------------------------------------------------
#
# We NO LONGER use:
#
#     unknown Jira field -> OTHER_CHANGE
#
# Instead:
#
#     known Jira field
#          ↓
#     standardized event type
#
#     repository-specific field
#          ↓
#     CUSTOM_FIELD_CHANGE
#
# The original Jira field name and old/new values are still
# retained, so no source information is discarded.
# ============================================================


def extract_jira_ticket(
    ticket,
    repository_project,
    instance_id,
    instance_start,
    instance_end,
    source_file_name
):

    # ========================================================
    # A. NORMALIZE INSTANCE BOUNDARIES
    # ========================================================
    #
    # Production datasets contain a mixture of timezone-aware
    # and timezone-naive timestamps.
    #
    # Convert the boundaries to UTC-safe datetime objects.
    # ========================================================

    instance_start_utc = parse_timestamp_utc(
        instance_start
    )

    instance_end_utc = parse_timestamp_utc(
        instance_end
    )


    # ========================================================
    # B. CREATE EMPTY MASTER RECORD
    # ========================================================

    record = create_master_record(

        repository_project=
            repository_project,

        project_key=
            ticket.get("project_key"),

        instance_id=
            instance_id,

        source_type=
            "JIRA",

        source_id=
            ticket.get("issue_key")
    )


    # ========================================================
    # C. PROJECT ENTITY
    # ========================================================

    project_entity = add_entity(

        record,

        "PROJECT",

        ticket.get("project_key"),

        attributes={

            "project_id":
                ticket.get("project_id"),

            "project_name":
                ticket.get("project_name"),

            "source_repository":
                ticket.get("source_repository")
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # D. ISSUE ENTITY
    # ========================================================

    issue_entity = add_entity(

        record,

        "ISSUE",

        ticket.get("issue_key"),

        attributes={

            "issue_id":
                ticket.get("issue_id"),

            "issue_type":
                ticket.get("issue_type"),

            "summary":
                ticket.get("summary"),

            "description":
                ticket.get("description")
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # --------------------------------------------------------
    # PROJECT -> ISSUE
    # --------------------------------------------------------

    add_relationship(

        record,

        project_entity,

        "HAS_ISSUE",

        issue_entity,

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # E. PEOPLE
    # ========================================================
    #
    # Preserve structured Jira identities exactly as provided.
    #
    # Entity Resolution will later determine whether the same
    # person appears across multiple Jira tickets/documents.
    # ========================================================

    person_fields = [

        (
            "assignee",
            "ASSIGNED_TO"
        ),

        (
            "reporter",
            "REPORTED_BY"
        ),

        (
            "creator",
            "CREATED_BY"
        )
    ]


    for (
        field_name,
        relationship_type
    ) in person_fields:

        person_name = ticket.get(
            field_name
        )


        if not person_name:
            continue


        person_entity = add_entity(

            record,

            "PERSON",

            person_name,

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            issue_entity,

            relationship_type,

            person_entity,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # ========================================================
    # F. COMPONENTS
    # ========================================================

    for component in (
        ticket.get("components")
        or []
    ):

        # Components are normally dictionaries.
        if isinstance(
            component,
            dict
        ):

            component_name = (
                component.get("name")
            )

            component_id = (
                component.get("id")
            )

        else:

            # Safe fallback for repositories where component
            # values may be stored as strings.
            component_name = str(
                component
            )

            component_id = None


        if not component_name:
            continue


        component_entity = add_entity(

            record,

            "COMPONENT",

            component_name,

            attributes={

                "component_id":
                    component_id
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            issue_entity,

            "HAS_COMPONENT",

            component_entity,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # ========================================================
    # G. ISSUE RELATIONSHIPS / DEPENDENCIES
    # ========================================================

    relation_sources = [

        (
            "blocks",
            "BLOCKS"
        ),

        (
            "blocked_by",
            "BLOCKED_BY"
        ),

        (
            "depends_on",
            "DEPENDS_ON"
        ),

        (
            "other_issue_links",
            "RELATED_TO"
        )
    ]


    for (
        field_name,
        relationship_type
    ) in relation_sources:

        linked_items = (
            ticket.get(field_name)
            or []
        )


        for linked in linked_items:

            # ------------------------------------------------
            # Linked issues may be represented as:
            #
            #     dictionary
            #
            # OR
            #
            #     simple issue-key string
            # ------------------------------------------------

            if isinstance(
                linked,
                dict
            ):

                linked_key = (
                    linked.get("issue_key")
                    or
                    linked.get("key")
                )

                linked_attributes = linked

            else:

                linked_key = linked

                linked_attributes = {}


            if not linked_key:
                continue


            linked_entity = add_entity(

                record,

                "ISSUE",

                linked_key,

                extraction_method=
                    "structured",

                confidence=1.0
            )


            add_relationship(

                record,

                issue_entity,

                relationship_type,

                linked_entity,

                attributes=
                    linked_attributes,

                extraction_method=
                    "structured",

                confidence=1.0
            )


            # ------------------------------------------------
            # Preserve explicit dependency representation.
            # ------------------------------------------------

            if relationship_type in [

                "BLOCKS",
                "BLOCKED_BY",
                "DEPENDS_ON"

            ]:

                record[
                    "dependencies"
                ].append({

                    "source_entity":
                        issue_entity,

                    "target_entity":
                        linked_entity,

                    "dependency_type":
                        relationship_type,

                    "dependency_status":
                        None,

                    "created_at":
                        None,

                    "removed_at":
                        None,

                    "extraction_method":
                        "structured",

                    "confidence":
                        1.0
                })


    # ========================================================
    # H. TEMPORAL FIELDS
    # ========================================================

    record["temporal"] = {

        "created_at":
            ticket.get("created_at"),

        "updated_at":
            ticket.get("updated_at"),

        "due_at":
            ticket.get("due_at"),

        "resolved_at":
            ticket.get("resolved_at"),

        "blocked_at":
            ticket.get("blocked_at"),

        "unblocked_at":
            ticket.get("unblocked_at"),

        # Project-instance validity window
        "valid_from":
            (
                instance_start_utc.isoformat()
                if instance_start_utc
                else None
            ),

        "valid_to":
            (
                instance_end_utc.isoformat()
                if instance_end_utc
                else None
            )
    }


    # ========================================================
    # I. EXPLICIT BLOCKER INFORMATION
    # ========================================================

    if ticket.get("blocked_at"):

        record[
            "blockers"
        ].append({

            "blocked_entity":
                issue_entity,

            "blocker_entity":
                None,

            "blocked_at":
                ticket.get("blocked_at"),

            "unblocked_at":
                ticket.get(
                    "unblocked_at"
                ),

            "blocking_reason":
                None,

            "blocking_status":
                (
                    "RESOLVED"

                    if ticket.get(
                        "unblocked_at"
                    )

                    else "ACTIVE"
                ),

            "extraction_method":
                "structured",

            "confidence":
                1.0
        })


    # ========================================================
    # J. STATUS HISTORY
    # ========================================================
    #
    # We initially preserve status_history.
    #
    # Jira may duplicate these same transitions inside
    # full_changelog.
    #
    # The validated deduplicate_status_events() function will
    # remove those duplicates after extraction.
    # ========================================================

    for status_event in (
        ticket.get(
            "status_history"
        )
        or []
    ):

        event_time = (
            parse_timestamp_utc(

                status_event.get(
                    "changed_at"
                )
            )
        )


        if not (
            event_time
            and
            instance_start_utc
            <= event_time
            <= instance_end_utc
        ):

            continue


        add_event(

            record,

            event_type=
                "STATUS_CHANGE",

            entity=
                issue_entity,

            event_time=
                status_event.get(
                    "changed_at"
                ),

            actor=
                status_event.get(
                    "changed_by"
                ),

            from_value=
                status_event.get(
                    "from_status"
                ),

            to_value=
                status_event.get(
                    "to_status"
                ),

            attributes={

                "source_structure":
                    "status_history"
            },

            extraction_method=
                "structured"
        )


    # ========================================================
    # K. FULL JIRA CHANGELOG
    # ========================================================
    #
    # This is where the production resolver is important.
    #
    # Every field becomes either:
    #
    #     standardized event type
    #
    # OR
    #
    #     CUSTOM_FIELD_CHANGE
    #
    # Nothing becomes OTHER_CHANGE.
    # ========================================================

    for history in (
        ticket.get(
            "full_changelog"
        )
        or []
    ):

        history_time = (
            parse_timestamp_utc(
                history.get("created")
            )
        )


        # ----------------------------------------------------
        # Only preserve changelog events belonging to the
        # selected project-instance timeline.
        # ----------------------------------------------------

        if not (
            history_time
            and
            instance_start_utc
            <= history_time
            <= instance_end_utc
        ):

            continue


        # ----------------------------------------------------
        # Extract author safely.
        # ----------------------------------------------------

        history_author = (
            history.get("author")
            or {}
        )


        if isinstance(
            history_author,
            dict
        ):

            author = (

                history_author.get(
                    "displayName"
                )

                or

                history_author.get(
                    "name"
                )

                or

                history_author.get(
                    "key"
                )
            )

        else:

            author = str(
                history_author
            )


        # ----------------------------------------------------
        # Process every changed field.
        # ----------------------------------------------------

        for change_item in (
            history.get("items")
            or []
        ):

            field_name = str(
                change_item.get(
                    "field"
                )
                or ""
            ).strip()


            if not field_name:
                continue


            # ================================================
            # PRODUCTION EVENT-TYPE RESOLUTION
            # ================================================
            #
            # Example:
            #
            # Sprint
            #     -> SPRINT_CHANGE
            #
            # timeestimate
            #     -> ESTIMATE_CHANGE
            #
            # User Impact Score
            #     -> CUSTOM_FIELD_CHANGE
            #
            # Original field name remains preserved below.
            # ================================================

            mapped_event_type = (
                resolve_jira_event_type(
                    field_name
                )
            )


            # ------------------------------------------------
            # Determine whether this is a custom-field fallback.
            # ------------------------------------------------

            is_custom_fallback = (
                mapped_event_type
                ==
                "CUSTOM_FIELD_CHANGE"
            )


            add_event(

                record,

                event_type=
                    mapped_event_type,

                entity=
                    issue_entity,

                event_time=
                    history.get(
                        "created"
                    ),

                actor=
                    author,

                from_value=
                    change_item.get(
                        "fromString"
                    ),

                to_value=
                    change_item.get(
                        "toString"
                    ),

                attributes={

                    # ----------------------------------------
                    # CRITICAL:
                    # Preserve original Jira field name.
                    # ----------------------------------------

                    "jira_field":
                        field_name,

                    "field_type":
                        change_item.get(
                            "fieldtype"
                        ),

                    "field_id":
                        change_item.get(
                            "fieldId"
                        ),

                    "source_structure":
                        "full_changelog",

                    # Allows downstream code to distinguish
                    # standardized events from fallback custom
                    # field events.
                    "custom_field_fallback":
                        is_custom_fallback
                },

                extraction_method=
                    "structured"
            )


    # ========================================================
    # L. COMMENTS INSIDE INSTANCE WINDOW
    # ========================================================

    comments_in_window = []


    for comment in (
        ticket.get("comments")
        or []
    ):

        # ----------------------------------------------------
        # Different repositories may use slightly different
        # comment timestamp keys.
        # ----------------------------------------------------

        raw_comment_time = (

            comment.get("created_at")

            or

            comment.get("created")
        )


        comment_time = (
            parse_timestamp_utc(
                raw_comment_time
            )
        )


        if (
            comment_time
            and
            instance_start_utc
            <= comment_time
            <= instance_end_utc
        ):

            comments_in_window.append(
                comment
            )


    # ========================================================
    # M. STRUCTURED ISSUE ATTRIBUTES
    # ========================================================
    #
    # Preserve potentially useful raw information rather than
    # prematurely deciding which fields TabPFN will consume.
    #
    # Feature engineering happens AFTER:
    #
    #     Entity Resolution
    #         ↓
    #     Temporal KG
    #         ↓
    #     Weekly snapshots
    # ========================================================

    record[
        "attributes"
    ] = {

        "issue_type":
            ticket.get(
                "issue_type"
            ),

        "summary":
            ticket.get(
                "summary"
            ),

        "description":
            ticket.get(
                "description"
            ),

        "related_requirement":
            ticket.get(
                "related_requirement"
            ),

        "epic_feature":
            ticket.get(
                "epic_feature"
            ),

        "current_status":
            ticket.get(
                "status"
            ),

        "priority":
            ticket.get(
                "priority"
            ),

        "resolution":
            ticket.get(
                "resolution"
            ),

        "team":
            ticket.get(
                "team"
            ),

        "labels":
            (
                ticket.get("labels")
                or []
            ),

        "fix_versions":
            (
                ticket.get(
                    "fix_versions"
                )
                or []
            ),

        "affected_versions":
            (
                ticket.get(
                    "affected_versions"
                )
                or []
            ),

        "comments":
            comments_in_window
    }


    # ========================================================
    # N. PROVENANCE
    # ========================================================
    #
    # Preserve source evidence for future GraphRAG/explanation.
    # ========================================================

    evidence_parts = []


    if ticket.get("summary"):

        evidence_parts.append(
            str(
                ticket["summary"]
            )
        )


    if ticket.get("description"):

        evidence_parts.append(
            str(
                ticket["description"]
            )
        )


    record[
        "provenance"
    ] = {

        "source_type":
            "JIRA",

        "source_id":
            ticket.get(
                "issue_key"
            ),

        "instance_id":
            instance_id,

        "chunk_id":
            None,

        "evidence_text":
            "\n".join(
                evidence_parts
            ),

        "file_name":
            source_file_name
    }


    # ========================================================
    # O. RETURN EXTRACTION RECORD
    # ========================================================

    return record
```

```python
# ============================================================
# STEP 146: VALIDATE GLOBAL PRODUCTION EVENT RESOLUTION
# ============================================================
# Purpose:
# Verify that every observed Jira changelog field across all
# 30 projects now resolves to either:
#
#     a standardized event type
#
# OR
#
#     CUSTOM_FIELD_CHANGE
#
# Nothing should become:
#
#     OTHER_CHANGE
#     None
#
# This freezes the production Jira event strategy.
# ============================================================


resolution_counts = Counter()

custom_field_counts = Counter()


for field_name, count in (
    global_changelog_field_counts.items()
):

    event_type = (
        resolve_jira_event_type(
            field_name
        )
    )


    resolution_counts[
        event_type
    ] += count


    if (
        event_type
        == "CUSTOM_FIELD_CHANGE"
    ):

        custom_field_counts[
            field_name
        ] += count


invalid_resolutions = [

    field_name

    for field_name
    in global_changelog_field_counts

    if (
        resolve_jira_event_type(
            field_name
        )
        in [
            None,
            "",
            "OTHER_CHANGE"
        ]
    )
]


print("=" * 70)
print("GLOBAL PRODUCTION EVENT RESOLUTION")
print("=" * 70)


print(
    "\nObserved Jira field types:",
    len(
        global_changelog_field_counts
    )
)


print(
    "Invalid/unresolved fields:",
    len(invalid_resolutions)
)


print(
    "Fields using CUSTOM_FIELD_CHANGE:",
    len(custom_field_counts)
)


print(
    "\nCUSTOM FIELD OCCURRENCES:",
    sum(
        custom_field_counts.values()
    )
)


print("\nTop custom fields:")
print("-" * 70)


for field_name, count in (
    custom_field_counts.most_common(20)
):

    print(
        f"{field_name:40s}",
        count
    )


if not invalid_resolutions:

    print(
        "\nSUCCESS:"
        "\nEvery Jira changelog field across all "
        "30 GC1 instances has a safe production "
        "event representation."
    )
```

```python
# ============================================================
# STEP 147: PRODUCTION JIRA EXTRACTION TEST - JBEAP-I01
# ============================================================
# Purpose:
# Run the FINAL production Jira extractor on all 131 aligned
# JBEAP-I01 tickets.
#
# This is our second-project production test.
#
# IMPORTANT:
# We keep the records in memory first.
# Nothing is saved yet.
#
# After extraction we will:
#
#   1. deduplicate STATUS_CHANGE events
#   2. validate record count
#   3. inspect event distribution
#   4. verify CUSTOM_FIELD_CHANGE preservation
#   5. validate timestamps
#
# Only then will we save JBEAP Jira output.
# ============================================================


jbeap_jira_records = []


for item in test_instance_jira_tickets:

    ticket = item["ticket"]


    extracted_record = extract_jira_ticket(

        ticket=ticket,

        repository_project=
            test_inputs["repository_folder"],

        instance_id=
            test_inputs["instance_id"],

        instance_start=
            test_instance_start,

        instance_end=
            test_instance_end,

        source_file_name=
            test_inputs["jira_file"].name
    )


    jbeap_jira_records.append(
        extracted_record
    )


print("=" * 70)
print("JBEAP PRODUCTION JIRA EXTRACTION")
print("=" * 70)


print(
    "\nInput aligned tickets:",
    len(test_instance_jira_tickets)
)


print(
    "Extracted records     :",
    len(jbeap_jira_records)
)


if (
    len(jbeap_jira_records)
    ==
    test_inputs["expected_tickets"]
):

    print(
        "\nSUCCESS:"
        "\nEvery JBEAP-I01 Jira ticket produced "
        "one extraction record."
    )

else:

    print(
        "\nCHECK REQUIRED:"
        "\nJBEAP extraction count does not match "
        "the expected instance count."
    )
```

```python
# ============================================================
# STEP 148: DEDUPLICATE JBEAP STATUS EVENTS
# ============================================================
# Purpose:
# Jira may represent identical status transitions in both:
#
#     status_history
#     full_changelog
#
# The DOCS pilot demonstrated this clearly.
#
# Apply the same validated deduplication strategy to JBEAP.
# ============================================================


events_before = sum(

    len(record["events"])

    for record in jbeap_jira_records
)


status_before = sum(

    1

    for record in jbeap_jira_records

    for event in record["events"]

    if event["event_type"]
    == "STATUS_CHANGE"
)


jbeap_jira_records = [

    deduplicate_status_events(
        record
    )

    for record
    in jbeap_jira_records
]


events_after = sum(

    len(record["events"])

    for record in jbeap_jira_records
)


status_after = sum(

    1

    for record in jbeap_jira_records

    for event in record["events"]

    if event["event_type"]
    == "STATUS_CHANGE"
)


print("=" * 70)
print("JBEAP STATUS DEDUPLICATION")
print("=" * 70)


print(
    "\nEvents before:",
    events_before
)

print(
    "Events after :",
    events_after
)

print(
    "Events removed:",
    events_before
    - events_after
)


print(
    "\nSTATUS_CHANGE before:",
    status_before
)

print(
    "STATUS_CHANGE after :",
    status_after
)

print(
    "Duplicate status events removed:",
    status_before
    - status_after
)
```

```python
# ============================================================
# STEP 149: VERIFY JBEAP STATUS DEDUPLICATION
# ============================================================
# Purpose:
# Confirm that no exact duplicate status transitions remain.
# ============================================================


remaining_jbeap_status_duplicates = []


for record in jbeap_jira_records:

    seen = set()


    for event in record["events"]:

        if (
            event["event_type"]
            != "STATUS_CHANGE"
        ):
            continue


        event_time = (
            parse_timestamp_utc(
                event.get(
                    "event_time"
                )
            )
        )


        key = (

            (
                event_time.isoformat()
                if event_time
                else str(
                    event.get(
                        "event_time"
                    )
                )
            ),

            str(
                event.get(
                    "from_value"
                )
            ),

            str(
                event.get(
                    "to_value"
                )
            )
        )


        if key in seen:

            remaining_jbeap_status_duplicates.append({

                "source_id":
                    record[
                        "source"
                    ][
                        "source_id"
                    ],

                "transition":
                    key
            })


        seen.add(key)


print("=" * 70)
print("JBEAP STATUS DUPLICATE VALIDATION")
print("=" * 70)


print(
    "\nRemaining duplicate status events:",
    len(
        remaining_jbeap_status_duplicates
    )
)


if not remaining_jbeap_status_duplicates:

    print(
        "\nSUCCESS:"
        "\nNo duplicate JBEAP status transitions remain."
    )

else:

    print(
        "\nCHECK REQUIRED."
    )


    for item in (
        remaining_jbeap_status_duplicates[:10]
    ):

        print(item)
```

```python
# ============================================================
# STEP 150: JBEAP PRODUCTION EVENT DISTRIBUTION
# ============================================================
# Purpose:
# Inspect the final event types produced by the production
# extractor.
#
# We specifically check:
#
#     standardized events
#     CUSTOM_FIELD_CHANGE
#     OTHER_CHANGE
#
# OTHER_CHANGE must remain zero.
# ============================================================


jbeap_event_counts = Counter(

    event["event_type"]

    for record in jbeap_jira_records

    for event in record["events"]
)


print("=" * 70)
print("JBEAP PRODUCTION EVENT DISTRIBUTION")
print("=" * 70)


print(
    "\nTotal events:",
    sum(
        jbeap_event_counts.values()
    )
)


for event_type, count in (
    jbeap_event_counts.most_common()
):

    print(
        f"{event_type:35s}",
        count
    )


print(
    "\nCUSTOM_FIELD_CHANGE:",
    jbeap_event_counts.get(
        "CUSTOM_FIELD_CHANGE",
        0
    )
)


print(
    "OTHER_CHANGE:",
    jbeap_event_counts.get(
        "OTHER_CHANGE",
        0
    )
)
```

```python
# ============================================================
# STEP 152: VALIDATE JBEAP EVENT TIMESTAMPS
# ============================================================
# Purpose:
# Ensure every timestamped JBEAP extraction event remains
# inside JBEAP-I01:
#
# 2015-11-24
#       ↓
# 2016-08-01
# ============================================================


jbeap_events_checked = 0

jbeap_unparseable_times = []

jbeap_outside_events = []


for record in jbeap_jira_records:

    for event in record["events"]:

        raw_time = event.get(
            "event_time"
        )


        if raw_time is None:
            continue


        event_time = (
            parse_timestamp_utc(
                raw_time
            )
        )


        if event_time is None:

            jbeap_unparseable_times.append({

                "source_id":
                    record[
                        "source"
                    ][
                        "source_id"
                    ],

                "event_type":
                    event[
                        "event_type"
                    ],

                "event_time":
                    raw_time
            })

            continue


        jbeap_events_checked += 1


        if not (
            test_instance_start
            <= event_time
            <= test_instance_end
        ):

            jbeap_outside_events.append({

                "source_id":
                    record[
                        "source"
                    ][
                        "source_id"
                    ],

                "event_type":
                    event[
                        "event_type"
                    ],

                "event_time":
                    raw_time
            })


print("=" * 70)
print("JBEAP TEMPORAL VALIDATION")
print("=" * 70)


print(
    "\nTimestamped events checked:",
    jbeap_events_checked
)

print(
    "Unparseable timestamps:",
    len(
        jbeap_unparseable_times
    )
)

print(
    "Events outside instance:",
    len(
        jbeap_outside_events
    )
)


if (
    not jbeap_unparseable_times
    and
    not jbeap_outside_events
):

    print(
        "\nSUCCESS:"
        "\nAll JBEAP Jira event timestamps are "
        "valid and inside JBEAP-I01."
    )
```

```python
# ============================================================
# STEP 154: GENERIC PRODUCTION PRD PARSER
# ============================================================
# Purpose:
# Package the validated DOCS PRD parsing logic into reusable
# functions that work across all 30 project instances.
#
# This stage performs DETERMINISTIC extraction only:
#
#   metadata
#   Markdown sections
#   explicit REQ-* identifiers
#   important section discovery
#
# We DO NOT assume:
#
#   exactly 16 requirements
#   identical section counts
#   DOCS-specific content
# ============================================================


def extract_prd_metadata(prd_text):
    """
    Extract PRD title, version, date and status.
    """

    # --------------------------------------------------------
    # First Markdown H1 heading
    # --------------------------------------------------------

    title_match = re.search(
        r"^#\s+(.+)$",
        prd_text,
        flags=re.MULTILINE
    )

    title = (
        title_match.group(1).strip()
        if title_match
        else None
    )


    def get_field(field_name):

        pattern = (
            rf"\*\*{re.escape(field_name)}:\*\*"
            rf"\s*(.+)"
        )

        match = re.search(
            pattern,
            prd_text,
            flags=re.IGNORECASE
        )

        return (
            match.group(1).strip()
            if match
            else None
        )


    return {

        "title":
            title,

        "version":
            get_field("Version"),

        "date":
            get_field("Date"),

        "status":
            get_field("Status")
    }


def split_prd_sections(prd_text):
    """
    Split PRD Markdown into heading-aware sections.
    """

    sections = []

    current_heading = "DOCUMENT_HEADER"
    current_level = 0
    current_lines = []


    for line in prd_text.splitlines():

        heading_match = re.match(
            r"^(#{2,6})\s+(.+)$",
            line
        )


        if heading_match:

            if current_lines:

                sections.append({

                    "heading":
                        current_heading,

                    "level":
                        current_level,

                    "text":
                        "\n".join(
                            current_lines
                        ).strip()
                })


            current_level = len(
                heading_match.group(1)
            )

            current_heading = (
                heading_match.group(2).strip()
            )

            current_lines = []


        else:

            current_lines.append(
                line
            )


    if current_lines:

        sections.append({

            "heading":
                current_heading,

            "level":
                current_level,

            "text":
                "\n".join(
                    current_lines
                ).strip()
        })


    return sections


def extract_explicit_requirements(
    prd_text,
    prd_version,
    prd_date
):
    """
    Extract explicit REQ-* requirements.
    """

    requirement_pattern = re.compile(

        r"\*\*(REQ-\d+)"
        r"(?:\s*\(([^)]+)\))?"
        r"\*\*\s*(.+)",

        flags=re.IGNORECASE
    )


    requirements = []


    for line in prd_text.splitlines():

        match = requirement_pattern.search(
            line.strip()
        )


        if not match:
            continue


        requirements.append({

            "requirement_id":
                match.group(1).upper(),

            "label":
                (
                    match.group(2).strip()
                    if match.group(2)
                    else None
                ),

            "text":
                match.group(3).strip(),

            "prd_version":
                prd_version,

            "prd_date":
                prd_date
        })


    return requirements


print(
    "SUCCESS: Generic PRD parser functions defined."
)
```

```python
# ============================================================
# STEP 155: GENERIC PRD SECTION DISCOVERY
# ============================================================
# Purpose:
# Identify high-value PRD sections across repositories even
# when headings contain numbering or slight wording changes.
# ============================================================


PRODUCTION_PRD_SECTION_KEYWORDS = {

    "requirements": [
        "functional requirements",
        "non-functional requirements",
        "requirements"
    ],

    "milestones": [
        "milestones",
        "targets"
    ],

    "dependencies": [
        "dependencies",
        "dependency"
    ],

    "stakeholders": [
        "stakeholders",
        "ownership"
    ],

    "risks": [
        "known risks",
        "risks",
        "uncertainties"
    ],

    "open_questions": [
        "open questions",
        "unresolved questions"
    ],

    "acceptance_criteria": [
        "acceptance criteria"
    ],

    "scope": [
        "scope"
    ]
}


def discover_prd_sections(
    sections
):

    discovered = {

        category: []

        for category
        in PRODUCTION_PRD_SECTION_KEYWORDS
    }


    for section in sections:

        heading_lower = (
            section["heading"]
            .lower()
            .strip()
        )


        for category, keywords in (
            PRODUCTION_PRD_SECTION_KEYWORDS.items()
        ):

            if any(
                keyword in heading_lower
                for keyword in keywords
            ):

                discovered[
                    category
                ].append(
                    section
                )


    return discovered


print(
    "SUCCESS: Generic PRD section discovery defined."
)
```

```python
# ============================================================
# STEP 156: GENERIC DETERMINISTIC PRD DETAIL EXTRACTION
# ============================================================


def extract_prd_milestones(
    milestone_sections,
    prd_version,
    prd_date
):

    milestones = []


    for section in milestone_sections:

        lines = [
            line.strip()

            for line
            in section["text"].splitlines()

            if line.strip()
        ]


        for line in lines:

            if not line.startswith("|"):
                continue


            columns = [

                column.strip()

                for column
                in line.strip("|").split("|")
            ]


            if len(columns) < 3:
                continue


            if columns[0].lower() == "milestone":
                continue


            if all(
                set(column)
                <= {"-", ":", " "}

                for column in columns
            ):
                continue


            milestones.append({

                "milestone":
                    columns[0],

                "target":
                    columns[1],

                "notes":
                    columns[2],

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return milestones



def extract_prd_bullets(
    sections,
    prd_version,
    prd_date
):
    """
    Generic bullet extractor for dependency/acceptance sections.
    """

    items = []


    for section in sections:

        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            if not re.match(
                r"^[-*]\s+",
                clean
            ):
                continue


            text = re.sub(
                r"^[-*]\s+",
                "",
                clean
            ).strip()


            items.append({

                "text":
                    text,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return items



def extract_prd_stakeholders(
    stakeholder_sections,
    prd_version,
    prd_date
):

    stakeholders = []


    for section in stakeholder_sections:

        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            match = re.match(
                r"^[-*]\s+\*\*(.+?):\*\*\s*(.+)$",
                clean
            )


            if not match:
                continue


            stakeholders.append({

                "name":
                    match.group(1).strip(),

                "description":
                    match.group(2).strip(),

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return stakeholders



def extract_prd_scope(
    sections,
    prd_version,
    prd_date
):

    scope_items = []


    for section in sections:

        heading = (
            section["heading"]
            .lower()
            .strip()
        )


        if (
            heading == "in scope"
            or heading.endswith(
                "in scope"
            )
        ):

            scope_type = "IN_SCOPE"


        elif (
            "out of scope"
            in heading
        ):

            scope_type = "OUT_OF_SCOPE"


        else:

            continue


        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            if not re.match(
                r"^[-*]\s+",
                clean
            ):
                continue


            scope_items.append({

                "scope_type":
                    scope_type,

                "text":
                    re.sub(
                        r"^[-*]\s+",
                        "",
                        clean
                    ).strip(),

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return scope_items



def extract_prd_risks(
    risk_sections,
    prd_version,
    prd_date
):

    risks = []


    for section in risk_sections:

        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            if not re.match(
                r"^[-*]\s+",
                clean
            ):
                continue


            risk_text = re.sub(
                r"^[-*]\s+",
                "",
                clean
            ).strip()


            label_match = re.match(
                r"^\*\*(.+?)\*\*\s*(.*)$",
                risk_text
            )


            if label_match:

                label = (
                    label_match
                    .group(1)
                    .strip()
                    .rstrip(".")
                )

                text = (
                    label_match
                    .group(2)
                    .strip()
                )

            else:

                label = None
                text = risk_text


            risks.append({

                "label":
                    label,

                "text":
                    text,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return risks



def extract_prd_open_questions(
    question_sections,
    prd_version,
    prd_date
):

    questions = []


    for section in question_sections:

        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            match = re.match(
                r"^\d+\.\s+(.+)$",
                clean
            )


            if not match:
                continue


            questions.append({

                "text":
                    match.group(1).strip(),

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return questions


print(
    "SUCCESS: Generic deterministic PRD extractors defined."
)
```

```python
# ============================================================
# STEP 157: GENERIC PRD ISSUE-REFERENCE EXTRACTION
# ============================================================
# Purpose:
# Extract explicit Jira-style issue references while excluding
# REQ-* requirement identifiers.
# ============================================================


def extract_prd_issue_references(
    prd_chunks
):

    occurrences = []


    for chunk in prd_chunks:

        matches = (
            ISSUE_KEY_PATTERN.findall(
                chunk["text"]
            )
        )


        for issue_key in matches:

            if (
                issue_key.upper()
                .startswith("REQ-")
            ):
                continue


            occurrences.append({

                "issue_key":
                    issue_key,

                "chunk_id":
                    chunk["chunk_id"],

                "section":
                    chunk["section"],

                "extraction_method":
                    "regex",

                "confidence":
                    1.0
            })


    return occurrences


print(
    "SUCCESS: Generic PRD issue-reference extractor defined."
)
```

```python
# ============================================================
# STEP 158: TEST GENERIC PRD PARSER ON JBEAP-I01
# ============================================================
# Purpose:
# Verify that the generalized deterministic PRD parser works
# on a project other than DOCS before adding GLiNER.
# ============================================================


jbeap_prd_file = (
    test_inputs["prd_files"][0]
)


with open(
    jbeap_prd_file,
    "r",
    encoding="utf-8"
) as file:

    jbeap_prd_text = file.read()


# ------------------------------------------------------------
# Metadata
# ------------------------------------------------------------

jbeap_prd_metadata = (
    extract_prd_metadata(
        jbeap_prd_text
    )
)


# ------------------------------------------------------------
# Sections
# ------------------------------------------------------------

jbeap_prd_sections = (
    split_prd_sections(
        jbeap_prd_text
    )
)


jbeap_important_sections = (
    discover_prd_sections(
        jbeap_prd_sections
    )
)


# ------------------------------------------------------------
# Requirements
# ------------------------------------------------------------

jbeap_prd_requirements = (
    extract_explicit_requirements(

        jbeap_prd_text,

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


# ------------------------------------------------------------
# Milestones
# ------------------------------------------------------------

jbeap_prd_milestones = (
    extract_prd_milestones(

        jbeap_important_sections[
            "milestones"
        ],

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


# ------------------------------------------------------------
# Dependencies
# ------------------------------------------------------------

jbeap_prd_dependencies = (
    extract_prd_bullets(

        jbeap_important_sections[
            "dependencies"
        ],

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


# ------------------------------------------------------------
# Stakeholders
# ------------------------------------------------------------

jbeap_prd_stakeholders = (
    extract_prd_stakeholders(

        jbeap_important_sections[
            "stakeholders"
        ],

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


# ------------------------------------------------------------
# Scope
# ------------------------------------------------------------

jbeap_prd_scope = (
    extract_prd_scope(

        jbeap_prd_sections,

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


# ------------------------------------------------------------
# Acceptance criteria
# ------------------------------------------------------------

jbeap_prd_acceptance = (
    extract_prd_bullets(

        jbeap_important_sections[
            "acceptance_criteria"
        ],

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


# ------------------------------------------------------------
# Risks
# ------------------------------------------------------------

jbeap_prd_risks = (
    extract_prd_risks(

        jbeap_important_sections[
            "risks"
        ],

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


# ------------------------------------------------------------
# Open questions
# ------------------------------------------------------------

jbeap_prd_questions = (
    extract_prd_open_questions(

        jbeap_important_sections[
            "open_questions"
        ],

        jbeap_prd_metadata[
            "version"
        ],

        jbeap_prd_metadata[
            "date"
        ]
    )
)


print("=" * 70)
print("JBEAP DETERMINISTIC PRD TEST")
print("=" * 70)


print("\nFile:", jbeap_prd_file.name)

print(
    "Title:",
    jbeap_prd_metadata[
        "title"
    ]
)

print(
    "Version:",
    jbeap_prd_metadata[
        "version"
    ]
)

print(
    "Date:",
    jbeap_prd_metadata[
        "date"
    ]
)

print(
    "Status:",
    jbeap_prd_metadata[
        "status"
    ]
)


print(
    "\nSections:",
    len(
        jbeap_prd_sections
    )
)

print(
    "Requirements:",
    len(
        jbeap_prd_requirements
    )
)

print(
    "Milestones:",
    len(
        jbeap_prd_milestones
    )
)

print(
    "Dependencies:",
    len(
        jbeap_prd_dependencies
    )
)

print(
    "Stakeholders:",
    len(
        jbeap_prd_stakeholders
    )
)

print(
    "Scope items:",
    len(
        jbeap_prd_scope
    )
)

print(
    "Acceptance criteria:",
    len(
        jbeap_prd_acceptance
    )
)

print(
    "Risk evidence:",
    len(
        jbeap_prd_risks
    )
)

print(
    "Open questions:",
    len(
        jbeap_prd_questions
    )
)


print("\nIMPORTANT SECTIONS")
print("-" * 70)


for category, sections in (
    jbeap_important_sections.items()
):

    print(
        f"{category:20s}",
        len(sections),
        [
            section["heading"]
            for section in sections
        ]
    )
```

```python
# ============================================================
# STEP 159: INSPECT JBEAP PRD FORMATTING
# ============================================================
# Purpose:
# Our DOCS parser expected specific Markdown formats.
#
# JBEAP successfully exposed the correct sections, but some
# extractors returned zero results.
#
# We now inspect the RAW text of:
#
#   document header
#   milestones
#   ownership
#   risks
#   open questions
#
# This allows us to make the production parser genuinely
# format-tolerant rather than hard-coding assumptions.
# ============================================================


TARGET_CATEGORIES = [
    "milestones",
    "stakeholders",
    "risks",
    "open_questions"
]


print("=" * 70)
print("JBEAP PRD FORMAT INSPECTION")
print("=" * 70)


# ------------------------------------------------------------
# Document header
# ------------------------------------------------------------

print("\n" + "=" * 70)
print("DOCUMENT HEADER")
print("=" * 70)

print(
    jbeap_prd_text[:1200]
)


# ------------------------------------------------------------
# Important sections
# ------------------------------------------------------------

for category in TARGET_CATEGORIES:

    print("\n" + "=" * 70)

    print(
        category.upper()
    )

    print("=" * 70)


    sections = (
        jbeap_important_sections[
            category
        ]
    )


    if not sections:

        print(
            "NO SECTION FOUND"
        )

        continue


    for section in sections:

        print(
            "\nHEADING:",
            section["heading"]
        )

        print("-" * 70)

        print(
            section["text"]
        )
```

```python
# ============================================================
# STEP 160: VALIDATE JBEAP REQUIREMENT IDENTIFIERS
# ============================================================
# Purpose:
# Inspect all explicit requirement IDs found in JBEAP.
#
# We check:
#
#   number of occurrences
#   number of unique IDs
#   duplicate IDs
#   first/last requirement
#
# This ensures our generic requirement parser is not
# double-counting requirements.
# ============================================================

from collections import Counter


jbeap_requirement_ids = [

    requirement[
        "requirement_id"
    ]

    for requirement
    in jbeap_prd_requirements
]


requirement_id_counts = Counter(
    jbeap_requirement_ids
)


duplicate_requirement_ids = {

    requirement_id: count

    for requirement_id, count
    in requirement_id_counts.items()

    if count > 1
}


print("=" * 70)
print("JBEAP REQUIREMENT VALIDATION")
print("=" * 70)


print(
    "\nRequirement occurrences:",
    len(jbeap_requirement_ids)
)


print(
    "Unique requirement IDs:",
    len(
        set(
            jbeap_requirement_ids
        )
    )
)


print(
    "Duplicate requirement IDs:",
    len(
        duplicate_requirement_ids
    )
)


if duplicate_requirement_ids:

    print(
        "\nDuplicates:"
    )

    for requirement_id, count in (
        duplicate_requirement_ids.items()
    ):

        print(
            requirement_id,
            count
        )


print("\nREQUIREMENTS")
print("-" * 70)


for requirement in (
    jbeap_prd_requirements
):

    print(
        requirement[
            "requirement_id"
        ],
        "|",
        requirement[
            "label"
        ],
        "|",
        requirement[
            "text"
        ][:150]
    )
```

```python
# ============================================================
# STEP 161: PRODUCTION PRD METADATA PARSER
# ============================================================
# Purpose:
# Support metadata formats observed across different GC1 PRDs.
#
# DOCS example:
#     **Version:** 1
#     **Date:** 2016-07-10
#     **Status:** Initial draft for planning
#
# JBEAP example:
#     **Project key:** JBEAP-I01
#     **PRD version:** 1 (initial)
#     **Document date:** 2015-11-29
#     **Release stream:** EAP 7.0.0 (...)
#
# We preserve the complete version string and also derive the
# numeric version where possible.
# ============================================================


def extract_prd_metadata(prd_text):

    # --------------------------------------------------------
    # Document title
    # --------------------------------------------------------

    title_match = re.search(
        r"^#\s+(.+)$",
        prd_text,
        flags=re.MULTILINE
    )

    title = (
        title_match.group(1).strip()
        if title_match
        else None
    )


    # --------------------------------------------------------
    # Generic Markdown metadata reader
    # --------------------------------------------------------

    def get_any_field(field_names):

        for field_name in field_names:

            pattern = (
                rf"^\*\*{re.escape(field_name)}:\*\*"
                rf"\s*(.+)$"
            )

            match = re.search(
                pattern,
                prd_text,
                flags=(
                    re.IGNORECASE
                    | re.MULTILINE
                )
            )

            if match:
                return match.group(1).strip()

        return None


    # --------------------------------------------------------
    # Accept alternative field names.
    # --------------------------------------------------------

    version_raw = get_any_field([
        "Version",
        "PRD version"
    ])


    document_date = get_any_field([
        "Date",
        "Document date"
    ])


    explicit_status = get_any_field([
        "Status",
        "Document status",
        "PRD status"
    ])


    project_key = get_any_field([
        "Project key"
    ])


    release_stream = get_any_field([
        "Release stream"
    ])


    # --------------------------------------------------------
    # Extract numeric version when possible.
    #
    # Example:
    #
    #     1 (initial)
    #
    # becomes:
    #
    #     version = 1
    #     version_raw = 1 (initial)
    # --------------------------------------------------------

    version = None


    if version_raw:

        version_match = re.search(
            r"\d+",
            version_raw
        )

        if version_match:
            version = version_match.group(0)


    # --------------------------------------------------------
    # Optional status hint from version qualifier.
    #
    # IMPORTANT:
    # This is not treated as a formal source Status field.
    # It is stored separately as a version qualifier.
    # --------------------------------------------------------

    version_qualifier = None


    if version_raw:

        qualifier_match = re.search(
            r"\(([^)]+)\)",
            version_raw
        )

        if qualifier_match:

            version_qualifier = (
                qualifier_match
                .group(1)
                .strip()
            )


    return {

        "title":
            title,

        "version":
            version,

        "version_raw":
            version_raw,

        "version_qualifier":
            version_qualifier,

        "date":
            document_date,

        "status":
            explicit_status,

        "project_key_metadata":
            project_key,

        "release_stream":
            release_stream
    }


print(
    "SUCCESS: Format-tolerant PRD metadata parser defined."
)
```

```python
# ============================================================
# STEP 162: FORMAT-TOLERANT PRD MILESTONE PARSER
# ============================================================
# Purpose:
# Support both:
#
# DOCS:
#   | Milestone | Target | Notes |
#
# JBEAP:
#   | Milestone | Status / Expectation |
#
# Additional columns are preserved when present.
# ============================================================


def extract_prd_milestones(
    milestone_sections,
    prd_version,
    prd_date
):

    milestones = []


    for section in milestone_sections:

        lines = [

            line.strip()

            for line
            in section["text"].splitlines()

            if line.strip()
        ]


        # ----------------------------------------------------
        # Find table rows.
        # ----------------------------------------------------

        table_rows = [

            line

            for line in lines

            if line.startswith("|")
        ]


        if not table_rows:
            continue


        # ----------------------------------------------------
        # First valid table row is normally the header.
        # ----------------------------------------------------

        header_columns = None


        for line in table_rows:

            columns = [

                column.strip()

                for column
                in line.strip("|").split("|")
            ]


            # Separator row
            if all(
                set(column)
                <= {"-", ":", " "}

                for column in columns
            ):
                continue


            if header_columns is None:

                header_columns = columns

                continue


            # ------------------------------------------------
            # Data row
            # ------------------------------------------------

            if len(columns) < 2:
                continue


            milestone_name = (
                columns[0]
            )


            # ------------------------------------------------
            # Flexible interpretation:
            #
            # 2 columns:
            #   milestone + status/expectation
            #
            # 3+ columns:
            #   milestone + target + notes
            # ------------------------------------------------

            if len(columns) == 2:

                target = None
                notes = columns[1]

            else:

                target = columns[1]

                notes = " | ".join(
                    columns[2:]
                )


            milestones.append({

                "milestone":
                    milestone_name,

                "target":
                    target,

                "notes":
                    notes,

                "table_headers":
                    header_columns,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return milestones


print(
    "SUCCESS: Flexible milestone parser defined."
)
```

```python
# ============================================================
# STEP 163: FORMAT-TOLERANT STAKEHOLDER / OWNERSHIP PARSER
# ============================================================
# Purpose:
# Support both structured stakeholder labels and plain
# ownership bullet statements.
#
# Structured example:
#
#   - **Documentation team:** owns documentation...
#
# Plain example:
#
#   - A QA team verifying fixes on milestone builds.
#
# For plain bullets:
#
#   name = None
#   description = full source statement
#
# GLiNER can later identify semantic TEAM/COMPONENT entities
# inside the statement.
# ============================================================


def extract_prd_stakeholders(
    stakeholder_sections,
    prd_version,
    prd_date
):

    stakeholders = []


    for section in stakeholder_sections:

        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            if not re.match(
                r"^[-*]\s+",
                clean
            ):
                continue


            bullet_text = re.sub(
                r"^[-*]\s+",
                "",
                clean
            ).strip()


            # ------------------------------------------------
            # Structured form:
            #
            # **Team:** description
            # ------------------------------------------------

            structured_match = re.match(
                r"^\*\*(.+?):\*\*\s*(.+)$",
                bullet_text
            )


            if structured_match:

                stakeholder_name = (
                    structured_match
                    .group(1)
                    .strip()
                )

                description = (
                    structured_match
                    .group(2)
                    .strip()
                )

                format_type = (
                    "NAMED_STAKEHOLDER"
                )


            else:

                stakeholder_name = None

                description = (
                    bullet_text
                )

                format_type = (
                    "OWNERSHIP_STATEMENT"
                )


            stakeholders.append({

                "name":
                    stakeholder_name,

                "description":
                    description,

                "format_type":
                    format_type,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return stakeholders


print(
    "SUCCESS: Flexible ownership parser defined."
)
```

```python
# ============================================================
# STEP 164: FORMAT-TOLERANT RISK PARSER
# ============================================================
# Purpose:
# Support:
#
# DOCS:
#     - **Risk label.** Description
#
# JBEAP:
#     1. **Messaging performance (REQ-011):** Description
#
# Risk statements remain SOURCE EVIDENCE.
# They are NOT TabPFN target labels.
# ============================================================


def extract_prd_risks(
    risk_sections,
    prd_version,
    prd_date
):

    risks = []


    for section in risk_sections:

        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            # ------------------------------------------------
            # Accept:
            #
            # - item
            # * item
            # 1. item
            # ------------------------------------------------

            item_match = re.match(
                r"^(?:[-*]\s+|\d+\.\s+)(.+)$",
                clean
            )


            if not item_match:
                continue


            risk_text = (
                item_match
                .group(1)
                .strip()
            )


            # ------------------------------------------------
            # Extract optional bold label.
            #
            # Handles:
            #
            # **Messaging performance (REQ-011):**
            # **Carry-over of scheduled work.**
            # ------------------------------------------------

            label_match = re.match(
                r"^\*\*(.+?)\*\*\s*(.*)$",
                risk_text
            )


            if label_match:

                label = (
                    label_match
                    .group(1)
                    .strip()
                    .rstrip(":")
                    .rstrip(".")
                )

                text = (
                    label_match
                    .group(2)
                    .strip()
                )


            else:

                label = None
                text = risk_text


            risks.append({

                "label":
                    label,

                "text":
                    text,

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return risks


print(
    "SUCCESS: Flexible risk parser defined."
)
```

```python
# ============================================================
# STEP 165: FORMAT-TOLERANT OPEN-QUESTION PARSER
# ============================================================
# Purpose:
# Support both:
#
#     1. Question...
#
# and:
#
#     - Question...
# ============================================================


def extract_prd_open_questions(
    question_sections,
    prd_version,
    prd_date
):

    questions = []


    for section in question_sections:

        for line in section[
            "text"
        ].splitlines():

            clean = line.strip()


            match = re.match(
                r"^(?:[-*]\s+|\d+\.\s+)(.+)$",
                clean
            )


            if not match:
                continue


            questions.append({

                "text":
                    match.group(1).strip(),

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            })


    return questions


print(
    "SUCCESS: Flexible open-question parser defined."
)
```

```python
# ============================================================
# STEP 167: GENERIC PRODUCTION PRD CHUNKER
# ============================================================
# Purpose:
# Convert any parsed PRD into section-aware chunks for GLiNER.
#
# We preserve:
#   chunk_id
#   section
#   text
#
# Empty parent sections are ignored.
# Long sections are split into manageable chunks.
# ============================================================


PRODUCTION_PRD_MAX_CHARS = 1800


def create_production_prd_chunks(
    sections,
    max_chars=PRODUCTION_PRD_MAX_CHARS
):

    chunks = []

    chunk_counter = 1


    for section in sections:

        heading = section["heading"]

        text = (
            section["text"]
            .strip()
        )


        if not text:
            continue


        paragraphs = [

            paragraph.strip()

            for paragraph
            in re.split(
                r"\n\s*\n",
                text
            )

            if paragraph.strip()
        ]


        current_text = ""


        for paragraph in paragraphs:

            candidate = (

                current_text
                + "\n"
                + paragraph

                if current_text

                else paragraph
            )


            if (
                len(candidate) > max_chars
                and current_text
            ):

                chunks.append({

                    "chunk_id":
                        f"PRD_CHUNK_{chunk_counter:03d}",

                    "section":
                        heading,

                    "text":
                        (
                            f"Section: {heading}\n"
                            f"{current_text}"
                        )
                })


                chunk_counter += 1
                current_text = paragraph


            else:

                current_text = candidate


        if current_text:

            chunks.append({

                "chunk_id":
                    f"PRD_CHUNK_{chunk_counter:03d}",

                "section":
                    heading,

                "text":
                    (
                        f"Section: {heading}\n"
                        f"{current_text}"
                    )
            })


            chunk_counter += 1


    return chunks


print(
    "SUCCESS: Generic production PRD chunker defined."
)
```

```python
# ============================================================
# STEP 168: GENERIC PRODUCTION PRD GLINER EXTRACTOR
# ============================================================
# Purpose:
# Run GLiNER across any PRD's section-aware chunks.
#
# Model inference:
#     threshold = 0.40
#
# Final acceptance:
#     confidence >= 0.55
#     + deterministic quality filters
#
# Structured IDs such as REQ-001 and JIRA-123 are NOT allowed
# to override deterministic extraction.
# ============================================================


def extract_production_prd_gliner(
    prd_chunks
):

    accepted_entities = []


    for chunk in prd_chunks:

        predictions = (
            gliner_model.predict_entities(

                chunk["text"],

                PRODUCTION_GLINER_LABELS,

                threshold=
                    PRODUCTION_GLINER_INFERENCE_THRESHOLD
            )
        )


        for prediction in predictions:

            candidate = {

                "text":
                    prediction["text"],

                "label":
                    prediction["label"],

                "confidence":
                    float(
                        prediction["score"]
                    ),

                "chunk_id":
                    chunk["chunk_id"],

                "section":
                    chunk["section"],

                "extraction_method":
                    "gliner"
            }


            # ------------------------------------------------
            # Reuse the validated semantic filter.
            # ------------------------------------------------

            if not accept_gliner_entity(
                candidate
            ):
                continue


            # ------------------------------------------------
            # Deterministic structured identifiers take
            # precedence over GLiNER.
            # ------------------------------------------------

            if is_structured_identifier(
                candidate["text"]
            ):
                continue


            accepted_entities.append(
                candidate
            )


    return accepted_entities


print(
    "SUCCESS: Generic production PRD GLiNER extractor defined."
)
```

```python
# ============================================================
# STEP 169: COMPLETE PRODUCTION PRD EXTRACTOR
# ============================================================
# Purpose:
# Convert ONE official PRD into one complete GC1 Master
# Extraction record.
#
# This function combines:
#
#   deterministic parsing
#   regex identifiers
#   GLiNER semantic extraction
#   temporal information
#   provenance
#
# It does NOT perform Entity Resolution.
# ============================================================


def extract_production_prd(
    prd_file,
    repository_project,
    project_key,
    project_name,
    instance_id
):

    # ========================================================
    # A. LOAD PRD
    # ========================================================

    with open(
        prd_file,
        "r",
        encoding="utf-8"
    ) as file:

        prd_text = file.read()


    # ========================================================
    # B. METADATA
    # ========================================================

    metadata = extract_prd_metadata(
        prd_text
    )


    prd_version = metadata[
        "version"
    ]

    prd_date = metadata[
        "date"
    ]


    # ========================================================
    # C. SECTIONS
    # ========================================================

    sections = split_prd_sections(
        prd_text
    )


    important_sections = (
        discover_prd_sections(
            sections
        )
    )


    # ========================================================
    # D. DETERMINISTIC EXTRACTION
    # ========================================================

    requirements = (
        extract_explicit_requirements(

            prd_text,
            prd_version,
            prd_date
        )
    )


    milestones = (
        extract_prd_milestones(

            important_sections[
                "milestones"
            ],

            prd_version,
            prd_date
        )
    )


    dependencies = (
        extract_prd_bullets(

            important_sections[
                "dependencies"
            ],

            prd_version,
            prd_date
        )
    )


    stakeholders = (
        extract_prd_stakeholders(

            important_sections[
                "stakeholders"
            ],

            prd_version,
            prd_date
        )
    )


    scope_items = (
        extract_prd_scope(

            sections,

            prd_version,
            prd_date
        )
    )


    acceptance_criteria = (
        extract_prd_bullets(

            important_sections[
                "acceptance_criteria"
            ],

            prd_version,
            prd_date
        )
    )


    risk_evidence = (
        extract_prd_risks(

            important_sections[
                "risks"
            ],

            prd_version,
            prd_date
        )
    )


    open_questions = (
        extract_prd_open_questions(

            important_sections[
                "open_questions"
            ],

            prd_version,
            prd_date
        )
    )


    # ========================================================
    # E. GLINER CHUNKS
    # ========================================================

    prd_chunks = (
        create_production_prd_chunks(
            sections
        )
    )


    # ========================================================
    # F. EXPLICIT JIRA ISSUE REFERENCES
    # ========================================================

    issue_occurrences = (
        extract_prd_issue_references(
            prd_chunks
        )
    )


    unique_issue_keys = sorted(
        set(
            item["issue_key"]

            for item
            in issue_occurrences
        )
    )


    # ========================================================
    # G. GLINER SEMANTIC ENTITIES
    # ========================================================

    gliner_entities = (
        extract_production_prd_gliner(
            prd_chunks
        )
    )


    # ========================================================
    # H. CREATE MASTER RECORD
    # ========================================================

    source_id = prd_file.stem


    record = create_master_record(

        repository_project=
            repository_project,

        project_key=
            project_key,

        instance_id=
            instance_id,

        source_type=
            "PRD",

        source_id=
            source_id
    )


    record[
        "source"
    ][
        "source_timestamp"
    ] = prd_date


    record[
        "temporal"
    ][
        "created_at"
    ] = prd_date


    record[
        "temporal"
    ][
        "valid_from"
    ] = prd_date


    # ========================================================
    # I. PRD ATTRIBUTES
    # ========================================================

    record[
        "attributes"
    ].update({

        "title":
            metadata["title"],

        "version":
            metadata["version"],

        "version_raw":
            metadata["version_raw"],

        "version_qualifier":
            metadata[
                "version_qualifier"
            ],

        "status":
            metadata["status"],

        "project_key_metadata":
            metadata[
                "project_key_metadata"
            ],

        "release_stream":
            metadata[
                "release_stream"
            ],

        "scope_items":
            scope_items,

        "acceptance_criteria":
            acceptance_criteria,

        "risk_evidence":
            risk_evidence,

        "open_questions":
            open_questions
    })


    # ========================================================
    # J. PROJECT ENTITY
    # ========================================================

    project_entity = add_entity(

        record,

        "PROJECT",

        project_key,

        attributes={

            "project_name":
                project_name,

            "instance_id":
                instance_id
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # K. REQUIREMENT ENTITIES
    # ========================================================

    requirement_entity_ids = {}


    for requirement in requirements:

        requirement_id = (
            requirement[
                "requirement_id"
            ]
        )


        requirement_entity = add_entity(

            record,

            "REQUIREMENT",

            requirement_id,

            attributes={

                "text":
                    requirement[
                        "text"
                    ],

                "label":
                    requirement[
                        "label"
                    ],

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        requirement_entity_ids[
            requirement_id
        ] = requirement_entity


        add_relationship(

            record,

            project_entity,

            "HAS_REQUIREMENT",

            requirement_entity,

            timestamp=
                prd_date,

            extraction_method=
                "structured",

            confidence=1.0
        )


        # ----------------------------------------------------
        # Requirement temporal event.
        # ----------------------------------------------------

        add_event(

            record,

            event_type=
                "REQUIREMENT_CREATED",

            entity=
                requirement_entity,

            event_time=
                prd_date,

            actor=
                None,

            from_value=
                None,

            to_value=
                requirement_id,

            attributes={

                "prd_version":
                    prd_version
            },

            extraction_method=
                "structured"
        )


    # ========================================================
    # L. MILESTONE ENTITIES
    # ========================================================

    for milestone in milestones:

        milestone_entity = add_entity(

            record,

            "MILESTONE",

            milestone[
                "milestone"
            ],

            attributes={

                "target":
                    milestone[
                        "target"
                    ],

                "notes":
                    milestone[
                        "notes"
                    ],

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            project_entity,

            "HAS_MILESTONE",

            milestone_entity,

            timestamp=
                prd_date,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # ========================================================
    # M. NAMED STAKEHOLDER ENTITIES
    # ========================================================
    #
    # Only create a TEAM entity when the source explicitly
    # provides a stakeholder name.
    #
    # Plain ownership statements remain preserved in
    # attributes and GLiNER may identify semantic entities.
    # ========================================================

    for stakeholder in stakeholders:

        stakeholder_name = (
            stakeholder.get("name")
        )


        if not stakeholder_name:
            continue


        stakeholder_entity = (
            add_entity(

                record,

                "TEAM",

                stakeholder_name,

                attributes={

                    "description":
                        stakeholder[
                            "description"
                        ]
                },

                extraction_method=
                    "structured",

                confidence=1.0
            )
        )


        add_relationship(

            record,

            project_entity,

            "HAS_STAKEHOLDER",

            stakeholder_entity,

            timestamp=
                prd_date,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # Preserve ALL ownership statements.
    record[
        "attributes"
    ][
        "stakeholder_ownership"
    ] = stakeholders


    # ========================================================
    # N. EXPLICIT JIRA ISSUE REFERENCES
    # ========================================================

    for issue_key in unique_issue_keys:

        issue_entity = add_entity(

            record,

            "ISSUE",

            issue_key,

            extraction_method=
                "regex",

            confidence=1.0
        )


        add_relationship(

            record,

            project_entity,

            "REFERENCES_ISSUE",

            issue_entity,

            timestamp=
                prd_date,

            extraction_method=
                "regex",

            confidence=1.0
        )


    # ========================================================
    # O. GLINER SEMANTIC ENTITIES
    # ========================================================

    for prediction in gliner_entities:

        master_type = (
            GLINER_TO_MASTER_TYPE[
                prediction["label"]
            ]
        )


        semantic_entity = add_entity(

            record,

            master_type,

            prediction["text"],

            attributes={

                "section":
                    prediction[
                        "section"
                    ],

                "chunk_id":
                    prediction[
                        "chunk_id"
                    ]
            },

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ]
        )


        add_relationship(

            record,

            project_entity,

            "MENTIONS",

            semantic_entity,

            timestamp=
                prd_date,

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ]
        )


    # ========================================================
    # P. DEPENDENCY EVIDENCE
    # ========================================================

    for index, dependency in enumerate(
        dependencies,
        start=1
    ):

        record[
            "dependencies"
        ].append({

            "dependency_id":
                (
                    f"{source_id}"
                    f"_DEP_{index:03d}"
                ),

            "text":
                dependency[
                    "text"
                ],

            "source_entity":
                None,

            "target_entity":
                None,

            "dependency_type":
                "EXPLICIT_PRD_DEPENDENCY",

            "dependency_status":
                None,

            "created_at":
                prd_date,

            "removed_at":
                None,

            "extraction_method":
                "structured",

            "confidence":
                1.0
        })


    # ========================================================
    # Q. EXTRACTION DIAGNOSTICS
    # ========================================================

    record[
        "attributes"
    ][
        "extraction_diagnostics"
    ] = {

        "section_count":
            len(sections),

        "chunk_count":
            len(prd_chunks),

        "requirement_count":
            len(requirements),

        "milestone_count":
            len(milestones),

        "dependency_count":
            len(dependencies),

        "stakeholder_statement_count":
            len(stakeholders),

        "scope_item_count":
            len(scope_items),

        "acceptance_criteria_count":
            len(
                acceptance_criteria
            ),

        "risk_evidence_count":
            len(risk_evidence),

        "open_question_count":
            len(open_questions),

        "explicit_issue_reference_count":
            len(unique_issue_keys),

        "gliner_occurrence_count":
            len(gliner_entities)
    }


    # ========================================================
    # R. PROVENANCE
    # ========================================================

    record[
        "provenance"
    ] = {

        "source_type":
            "PRD",

        "source_id":
            source_id,

        "instance_id":
            instance_id,

        "chunk_id":
            None,

        "evidence_text":
            prd_text,

        "file_name":
            prd_file.name
    }


    return record
```

```python
# ============================================================
# STEP 169: COMPLETE PRODUCTION PRD EXTRACTOR
# ============================================================
# Purpose:
# Convert ONE official PRD into one complete GC1 Master
# Extraction record.
#
# This function combines:
#
#   deterministic parsing
#   regex identifiers
#   GLiNER semantic extraction
#   temporal information
#   provenance
#
# It does NOT perform Entity Resolution.
# ============================================================


def extract_production_prd(
    prd_file,
    repository_project,
    project_key,
    project_name,
    instance_id
):

    # ========================================================
    # A. LOAD PRD
    # ========================================================

    with open(
        prd_file,
        "r",
        encoding="utf-8"
    ) as file:

        prd_text = file.read()


    # ========================================================
    # B. METADATA
    # ========================================================

    metadata = extract_prd_metadata(
        prd_text
    )


    prd_version = metadata[
        "version"
    ]

    prd_date = metadata[
        "date"
    ]


    # ========================================================
    # C. SECTIONS
    # ========================================================

    sections = split_prd_sections(
        prd_text
    )


    important_sections = (
        discover_prd_sections(
            sections
        )
    )


    # ========================================================
    # D. DETERMINISTIC EXTRACTION
    # ========================================================

    requirements = (
        extract_explicit_requirements(

            prd_text,
            prd_version,
            prd_date
        )
    )


    milestones = (
        extract_prd_milestones(

            important_sections[
                "milestones"
            ],

            prd_version,
            prd_date
        )
    )


    dependencies = (
        extract_prd_bullets(

            important_sections[
                "dependencies"
            ],

            prd_version,
            prd_date
        )
    )


    stakeholders = (
        extract_prd_stakeholders(

            important_sections[
                "stakeholders"
            ],

            prd_version,
            prd_date
        )
    )


    scope_items = (
        extract_prd_scope(

            sections,

            prd_version,
            prd_date
        )
    )


    acceptance_criteria = (
        extract_prd_bullets(

            important_sections[
                "acceptance_criteria"
            ],

            prd_version,
            prd_date
        )
    )


    risk_evidence = (
        extract_prd_risks(

            important_sections[
                "risks"
            ],

            prd_version,
            prd_date
        )
    )


    open_questions = (
        extract_prd_open_questions(

            important_sections[
                "open_questions"
            ],

            prd_version,
            prd_date
        )
    )


    # ========================================================
    # E. GLINER CHUNKS
    # ========================================================

    prd_chunks = (
        create_production_prd_chunks(
            sections
        )
    )


    # ========================================================
    # F. EXPLICIT JIRA ISSUE REFERENCES
    # ========================================================

    issue_occurrences = (
        extract_prd_issue_references(
            prd_chunks
        )
    )


    unique_issue_keys = sorted(
        set(
            item["issue_key"]

            for item
            in issue_occurrences
        )
    )


    # ========================================================
    # G. GLINER SEMANTIC ENTITIES
    # ========================================================

    gliner_entities = (
        extract_production_prd_gliner(
            prd_chunks
        )
    )


    # ========================================================
    # H. CREATE MASTER RECORD
    # ========================================================

    source_id = prd_file.stem


    record = create_master_record(

        repository_project=
            repository_project,

        project_key=
            project_key,

        instance_id=
            instance_id,

        source_type=
            "PRD",

        source_id=
            source_id
    )


    record[
        "source"
    ][
        "source_timestamp"
    ] = prd_date


    record[
        "temporal"
    ][
        "created_at"
    ] = prd_date


    record[
        "temporal"
    ][
        "valid_from"
    ] = prd_date


    # ========================================================
    # I. PRD ATTRIBUTES
    # ========================================================

    record[
        "attributes"
    ].update({

        "title":
            metadata["title"],

        "version":
            metadata["version"],

        "version_raw":
            metadata["version_raw"],

        "version_qualifier":
            metadata[
                "version_qualifier"
            ],

        "status":
            metadata["status"],

        "project_key_metadata":
            metadata[
                "project_key_metadata"
            ],

        "release_stream":
            metadata[
                "release_stream"
            ],

        "scope_items":
            scope_items,

        "acceptance_criteria":
            acceptance_criteria,

        "risk_evidence":
            risk_evidence,

        "open_questions":
            open_questions
    })


    # ========================================================
    # J. PROJECT ENTITY
    # ========================================================

    project_entity = add_entity(

        record,

        "PROJECT",

        project_key,

        attributes={

            "project_name":
                project_name,

            "instance_id":
                instance_id
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # K. REQUIREMENT ENTITIES
    # ========================================================

    requirement_entity_ids = {}


    for requirement in requirements:

        requirement_id = (
            requirement[
                "requirement_id"
            ]
        )


        requirement_entity = add_entity(

            record,

            "REQUIREMENT",

            requirement_id,

            attributes={

                "text":
                    requirement[
                        "text"
                    ],

                "label":
                    requirement[
                        "label"
                    ],

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        requirement_entity_ids[
            requirement_id
        ] = requirement_entity


        add_relationship(

            record,

            project_entity,

            "HAS_REQUIREMENT",

            requirement_entity,

            timestamp=
                prd_date,

            extraction_method=
                "structured",

            confidence=1.0
        )


        # ----------------------------------------------------
        # Requirement temporal event.
        # ----------------------------------------------------

        add_event(

            record,

            event_type=
                "REQUIREMENT_CREATED",

            entity=
                requirement_entity,

            event_time=
                prd_date,

            actor=
                None,

            from_value=
                None,

            to_value=
                requirement_id,

            attributes={

                "prd_version":
                    prd_version
            },

            extraction_method=
                "structured"
        )


    # ========================================================
    # L. MILESTONE ENTITIES
    # ========================================================

    for milestone in milestones:

        milestone_entity = add_entity(

            record,

            "MILESTONE",

            milestone[
                "milestone"
            ],

            attributes={

                "target":
                    milestone[
                        "target"
                    ],

                "notes":
                    milestone[
                        "notes"
                    ],

                "prd_version":
                    prd_version,

                "prd_date":
                    prd_date
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            project_entity,

            "HAS_MILESTONE",

            milestone_entity,

            timestamp=
                prd_date,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # ========================================================
    # M. NAMED STAKEHOLDER ENTITIES
    # ========================================================
    #
    # Only create a TEAM entity when the source explicitly
    # provides a stakeholder name.
    #
    # Plain ownership statements remain preserved in
    # attributes and GLiNER may identify semantic entities.
    # ========================================================

    for stakeholder in stakeholders:

        stakeholder_name = (
            stakeholder.get("name")
        )


        if not stakeholder_name:
            continue


        stakeholder_entity = (
            add_entity(

                record,

                "TEAM",

                stakeholder_name,

                attributes={

                    "description":
                        stakeholder[
                            "description"
                        ]
                },

                extraction_method=
                    "structured",

                confidence=1.0
            )
        )


        add_relationship(

            record,

            project_entity,

            "HAS_STAKEHOLDER",

            stakeholder_entity,

            timestamp=
                prd_date,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # Preserve ALL ownership statements.
    record[
        "attributes"
    ][
        "stakeholder_ownership"
    ] = stakeholders


    # ========================================================
    # N. EXPLICIT JIRA ISSUE REFERENCES
    # ========================================================

    for issue_key in unique_issue_keys:

        issue_entity = add_entity(

            record,

            "ISSUE",

            issue_key,

            extraction_method=
                "regex",

            confidence=1.0
        )


        add_relationship(

            record,

            project_entity,

            "REFERENCES_ISSUE",

            issue_entity,

            timestamp=
                prd_date,

            extraction_method=
                "regex",

            confidence=1.0
        )


    # ========================================================
    # O. GLINER SEMANTIC ENTITIES
    # ========================================================

    for prediction in gliner_entities:

        master_type = (
            GLINER_TO_MASTER_TYPE[
                prediction["label"]
            ]
        )


        semantic_entity = add_entity(

            record,

            master_type,

            prediction["text"],

            attributes={

                "section":
                    prediction[
                        "section"
                    ],

                "chunk_id":
                    prediction[
                        "chunk_id"
                    ]
            },

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ]
        )


        add_relationship(

            record,

            project_entity,

            "MENTIONS",

            semantic_entity,

            timestamp=
                prd_date,

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ]
        )


    # ========================================================
    # P. DEPENDENCY EVIDENCE
    # ========================================================

    for index, dependency in enumerate(
        dependencies,
        start=1
    ):

        record[
            "dependencies"
        ].append({

            "dependency_id":
                (
                    f"{source_id}"
                    f"_DEP_{index:03d}"
                ),

            "text":
                dependency[
                    "text"
                ],

            "source_entity":
                None,

            "target_entity":
                None,

            "dependency_type":
                "EXPLICIT_PRD_DEPENDENCY",

            "dependency_status":
                None,

            "created_at":
                prd_date,

            "removed_at":
                None,

            "extraction_method":
                "structured",

            "confidence":
                1.0
        })


    # ========================================================
    # Q. EXTRACTION DIAGNOSTICS
    # ========================================================

    record[
        "attributes"
    ][
        "extraction_diagnostics"
    ] = {

        "section_count":
            len(sections),

        "chunk_count":
            len(prd_chunks),

        "requirement_count":
            len(requirements),

        "milestone_count":
            len(milestones),

        "dependency_count":
            len(dependencies),

        "stakeholder_statement_count":
            len(stakeholders),

        "scope_item_count":
            len(scope_items),

        "acceptance_criteria_count":
            len(
                acceptance_criteria
            ),

        "risk_evidence_count":
            len(risk_evidence),

        "open_question_count":
            len(open_questions),

        "explicit_issue_reference_count":
            len(unique_issue_keys),

        "gliner_occurrence_count":
            len(gliner_entities)
    }


    # ========================================================
    # R. PROVENANCE
    # ========================================================

    record[
        "provenance"
    ] = {

        "source_type":
            "PRD",

        "source_id":
            source_id,

        "instance_id":
            instance_id,

        "chunk_id":
            None,

        "evidence_text":
            prd_text,

        "file_name":
            prd_file.name
    }


    return record
```

```python
# ============================================================
# STEP 172: GENERIC PRODUCTION MEETING PARSER
# ============================================================
# Purpose:
# Parse any official GC1 meeting transcript into:
#
#   meeting metadata
#   participants + roles
#   speaker utterances
#   explicit Jira references
#   explicit REQ references
#
# No GLiNER is used in this step.
# ============================================================


def parse_production_meeting(
    meeting_file
):

    # --------------------------------------------------------
    # Load transcript
    # --------------------------------------------------------

    with open(
        meeting_file,
        "r",
        encoding="utf-8"
    ) as file:

        text = file.read()


    # --------------------------------------------------------
    # Generic single-line header extractor
    # --------------------------------------------------------

    def header_value(field_names):

        for field_name in field_names:

            pattern = (
                rf"^{re.escape(field_name)}:\s*(.+)$"
            )

            match = re.search(
                pattern,
                text,
                flags=(
                    re.IGNORECASE
                    | re.MULTILINE
                )
            )

            if match:
                return match.group(1).strip()

        return None


    meeting_id = header_value([
        "Meeting ID"
    ])


    meeting_title = header_value([
        "Meeting title",
        "Title"
    ])


    meeting_datetime = header_value([
        "Date/time",
        "Date"
    ])


    participants_raw = header_value([
        "Participants"
    ])


    # ========================================================
    # PARTICIPANTS
    # ========================================================

    participants = []


    if participants_raw:

        raw_participants = [

            value.strip()

            for value
            in participants_raw.split(",")

            if value.strip()
        ]


        for raw_participant in (
            raw_participants
        ):

            match = (
                PARTICIPANT_PATTERN.match(
                    raw_participant
                )
            )


            if match:

                participants.append({

                    "name":
                        match.group(1).strip(),

                    "role":
                        (
                            match.group(2).strip()

                            if match.group(2)

                            else None
                        ),

                    "raw":
                        raw_participant
                })


    # ========================================================
    # UTTERANCES
    # ========================================================

    utterances = []


    for line in text.splitlines():

        line = line.strip()

        match = (
            UTTERANCE_PATTERN.match(
                line
            )
        )


        if not match:
            continue


        utterances.append({

            "speaker_timestamp":
                match.group(1).strip(),

            "speaker":
                match.group(2).strip(),

            "text":
                match.group(3).strip()
        })


    # ========================================================
    # EXPLICIT ISSUE / REQUIREMENT REFERENCES
    # ========================================================

    issue_occurrences = []
    requirement_occurrences = []


    for utterance_index, utterance in enumerate(
        utterances,
        start=1
    ):

        utterance_text = (
            utterance["text"]
        )


        # ----------------------------------------------------
        # REQ-* references
        # ----------------------------------------------------

        requirement_matches = (
            REQUIREMENT_ID_PATTERN.findall(
                utterance_text
            )
        )


        for requirement_id in (
            requirement_matches
        ):

            requirement_occurrences.append({

                "requirement_id":
                    requirement_id.upper(),

                "utterance_index":
                    utterance_index,

                "speaker":
                    utterance["speaker"],

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    utterance_text
            })


        # ----------------------------------------------------
        # Jira-style issue references
        # ----------------------------------------------------

        issue_matches = (
            MEETING_ISSUE_PATTERN.findall(
                utterance_text
            )
        )


        for issue_key in issue_matches:

            if (
                issue_key.upper()
                .startswith("REQ-")
            ):
                continue


            issue_occurrences.append({

                "issue_key":
                    issue_key,

                "utterance_index":
                    utterance_index,

                "speaker":
                    utterance["speaker"],

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    utterance_text
            })


    return {

        "file_name":
            meeting_file.name,

        "file_path":
            meeting_file,

        "text":
            text,

        "meeting_id":
            meeting_id,

        "meeting_title":
            meeting_title,

        "meeting_datetime":
            meeting_datetime,

        "participants":
            participants,

        "utterances":
            utterances,

        "issue_occurrences":
            issue_occurrences,

        "requirement_occurrences":
            requirement_occurrences
    }


print(
    "SUCCESS: Generic production meeting parser defined."
)
```

```python
# ============================================================
# STEP 173: GENERIC MEETING EVIDENCE EXTRACTION
# ============================================================
# Purpose:
# Apply the rule-based candidate logic validated on DOCS to
# any meeting transcript.
#
# Produces:
#
#   action candidates
#   decision candidates + state
#   dependency candidates + state
#   blocker candidates + state
# ============================================================


def extract_meeting_rule_evidence(
    meeting
):

    action_candidates = []
    decision_candidates = []
    dependency_candidates = []
    blocker_candidates = []


    for index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        text = utterance["text"]


        base = {

            "utterance_index":
                index,

            "speaker_timestamp":
                utterance[
                    "speaker_timestamp"
                ],

            "speaker":
                utterance["speaker"],

            "text":
                text
        }


        # ====================================================
        # ACTION
        # ====================================================

        if action_regex.search(text):

            candidate = dict(base)

            candidate[
                "candidate_type"
            ] = "ACTION_ITEM"

            action_candidates.append(
                candidate
            )


        # ====================================================
        # DECISION
        # ====================================================

        if decision_regex.search(text):

            candidate = dict(base)

            candidate[
                "candidate_type"
            ] = "DECISION"

            candidate[
                "decision_state"
            ] = (
                classify_decision_candidate(
                    candidate
                )
            )

            decision_candidates.append(
                candidate
            )


        # ====================================================
        # DEPENDENCY
        # ====================================================

        if dependency_regex.search(text):

            candidate = dict(base)

            candidate[
                "candidate_type"
            ] = "DEPENDENCY"

            candidate[
                "dependency_state"
            ] = (
                classify_dependency_candidate(
                    candidate
                )
            )

            dependency_candidates.append(
                candidate
            )


        # ====================================================
        # BLOCKER
        # ====================================================

        if BLOCKER_PATTERN.search(text):

            candidate = dict(base)

            candidate[
                "candidate_type"
            ] = "BLOCKER"

            candidate[
                "blocker_state"
            ] = (
                classify_blocker_candidate(
                    candidate
                )
            )

            blocker_candidates.append(
                candidate
            )


    # ========================================================
    # FILTER ACTIONS
    # ========================================================

    final_actions = [

        candidate

        for candidate
        in action_candidates

        if strong_action_regex.search(
            candidate["text"]
        )
    ]


    meeting[
        "action_candidates"
    ] = action_candidates

    meeting[
        "final_action_candidates"
    ] = final_actions

    meeting[
        "decision_candidates"
    ] = decision_candidates

    meeting[
        "dependency_candidates"
    ] = dependency_candidates

    meeting[
        "blocker_candidates"
    ] = blocker_candidates


    return meeting


print(
    "SUCCESS: Generic meeting rule extractor defined."
)
```

```python
# ============================================================
# STEP 174: GENERIC PRODUCTION MEETING GLINER EXTRACTOR
# ============================================================
# Purpose:
# Extract semantic entities from speaker utterances while
# retaining speaker-level provenance.
#
# PERSON is deliberately excluded from final GLiNER output.
# ============================================================


def extract_production_meeting_gliner(
    meeting
):

    accepted = []


    for utterance_index, utterance in enumerate(
        meeting["utterances"],
        start=1
    ):

        predictions = (
            gliner_model.predict_entities(

                utterance["text"],

                PRODUCTION_GLINER_LABELS,

                threshold=
                    PRODUCTION_GLINER_INFERENCE_THRESHOLD
            )
        )


        for prediction in predictions:

            candidate = {

                "text":
                    prediction["text"],

                "label":
                    prediction["label"],

                "confidence":
                    float(
                        prediction["score"]
                    ),

                "meeting_id":
                    meeting["meeting_id"],

                "utterance_index":
                    utterance_index,

                "speaker":
                    utterance["speaker"],

                "speaker_timestamp":
                    utterance[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    utterance["text"],

                "extraction_method":
                    "gliner"
            }


            if not (
                accept_meeting_gliner_entity(
                    candidate
                )
            ):
                continue


            if is_structured_identifier(
                candidate["text"]
            ):
                continue


            accepted.append(
                candidate
            )


    return accepted


print(
    "SUCCESS: Generic production meeting GLiNER extractor defined."
)
```

```python
# ============================================================
# STEP 175: COMPLETE PRODUCTION MEETING EXTRACTOR
# ============================================================
# Purpose:
# Convert ONE meeting transcript into one GC1 Master Extraction
# record.
#
# Combines:
#
#   deterministic metadata
#   participants
#   Jira / requirement references
#   rule-based actions/decisions/dependencies/blockers
#   GLiNER semantic entities
#   temporal events
#   provenance
# ============================================================


def extract_production_meeting(
    meeting_file,
    repository_project,
    project_key,
    project_name,
    instance_id
):

    # ========================================================
    # A. PARSE
    # ========================================================

    meeting = (
        parse_production_meeting(
            meeting_file
        )
    )


    # ========================================================
    # B. RULE EVIDENCE
    # ========================================================

    meeting = (
        extract_meeting_rule_evidence(
            meeting
        )
    )


    # ========================================================
    # C. GLINER
    # ========================================================

    gliner_entities = (
        extract_production_meeting_gliner(
            meeting
        )
    )


    # ========================================================
    # D. MASTER RECORD
    # ========================================================

    record = create_master_record(

        repository_project=
            repository_project,

        project_key=
            project_key,

        instance_id=
            instance_id,

        source_type=
            "MEETING",

        source_id=
            meeting["meeting_id"]
    )


    record[
        "source"
    ][
        "source_timestamp"
    ] = meeting[
        "meeting_datetime"
    ]


    record[
        "temporal"
    ][
        "created_at"
    ] = meeting[
        "meeting_datetime"
    ]


    record[
        "temporal"
    ][
        "valid_from"
    ] = meeting[
        "meeting_datetime"
    ]


    # ========================================================
    # E. ATTRIBUTES
    # ========================================================

    record[
        "attributes"
    ].update({

        "meeting_title":
            meeting[
                "meeting_title"
            ],

        "participants":
            meeting[
                "participants"
            ],

        "utterance_count":
            len(
                meeting[
                    "utterances"
                ]
            )
    })


    # ========================================================
    # F. PROJECT
    # ========================================================

    project_entity = add_entity(

        record,

        "PROJECT",

        project_key,

        attributes={
            "project_name":
                project_name,

            "instance_id":
                instance_id
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # G. MEETING
    # ========================================================

    meeting_entity = add_entity(

        record,

        "MEETING",

        meeting["meeting_id"],

        attributes={

            "title":
                meeting[
                    "meeting_title"
                ],

            "meeting_datetime":
                meeting[
                    "meeting_datetime"
                ]
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    add_relationship(

        record,

        project_entity,

        "HAS_MEETING",

        meeting_entity,

        timestamp=
            meeting[
                "meeting_datetime"
            ]
    )


    # ========================================================
    # H. PARTICIPANTS
    # ========================================================

    for participant in meeting[
        "participants"
    ]:

        person_entity = add_entity(

            record,

            "PERSON",

            participant["name"],

            attributes={
                "role":
                    participant["role"]
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            person_entity,

            "PARTICIPATED_IN",

            meeting_entity,

            timestamp=
                meeting[
                    "meeting_datetime"
                ]
        )


    # ========================================================
    # I. ISSUES
    # ========================================================

    unique_issues = sorted(
        set(
            item["issue_key"]

            for item
            in meeting[
                "issue_occurrences"
            ]
        )
    )


    for issue_key in unique_issues:

        issue_entity = add_entity(

            record,

            "ISSUE",

            issue_key,

            extraction_method=
                "regex",

            confidence=1.0
        )


        add_relationship(

            record,

            meeting_entity,

            "REFERENCES_ISSUE",

            issue_entity,

            timestamp=
                meeting[
                    "meeting_datetime"
                ],

            extraction_method=
                "regex",

            confidence=1.0
        )


    # ========================================================
    # J. REQUIREMENTS
    # ========================================================

    unique_requirements = sorted(
        set(
            item[
                "requirement_id"
            ]

            for item
            in meeting[
                "requirement_occurrences"
            ]
        )
    )


    for requirement_id in (
        unique_requirements
    ):

        requirement_entity = (
            add_entity(

                record,

                "REQUIREMENT",

                requirement_id,

                extraction_method=
                    "regex",

                confidence=1.0
            )
        )


        add_relationship(

            record,

            meeting_entity,

            "REFERENCES_REQUIREMENT",

            requirement_entity,

            timestamp=
                meeting[
                    "meeting_datetime"
                ],

            extraction_method=
                "regex",

            confidence=1.0
        )


    # ========================================================
    # K. GLINER SEMANTIC ENTITIES
    # ========================================================

    for prediction in gliner_entities:

        master_type = (
            GLINER_TO_MASTER_TYPE[
                prediction["label"]
            ]
        )


        semantic_entity = add_entity(

            record,

            master_type,

            prediction["text"],

            attributes={

                "utterance_index":
                    prediction[
                        "utterance_index"
                    ],

                "speaker":
                    prediction[
                        "speaker"
                    ],

                "speaker_timestamp":
                    prediction[
                        "speaker_timestamp"
                    ]
            },

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ]
        )


        add_relationship(

            record,

            meeting_entity,

            "MENTIONS",

            semantic_entity,

            timestamp=
                meeting[
                    "meeting_datetime"
                ],

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ],

            attributes={

                "speaker":
                    prediction[
                        "speaker"
                    ],

                "speaker_timestamp":
                    prediction[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    prediction[
                        "evidence_text"
                    ]
            }
        )


    # ========================================================
    # L. ACTION ITEMS
    # ========================================================

    for index, candidate in enumerate(
        meeting[
            "final_action_candidates"
        ],
        start=1
    ):

        record[
            "action_items"
        ].append({

            "action_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_ACTION_{index:03d}"
                ),

            "action_text":
                candidate["text"],

            "action_owner":
                candidate["speaker"],

            "meeting_time":
                meeting[
                    "meeting_datetime"
                ],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "due_at":
                None,

            "completed_at":
                None,

            "action_status":
                "OPEN",

            "is_recap":
                (
                    "recap"
                    in candidate[
                        "text"
                    ].lower()
                ),

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        })


    # ========================================================
    # M. DECISIONS
    # ========================================================

    for index, candidate in enumerate(
        meeting[
            "decision_candidates"
        ],
        start=1
    ):

        decision_id = (
            f"{meeting['meeting_id']}"
            f"_DECISION_{index:03d}"
        )


        record[
            "decisions"
        ].append({

            "decision_id":
                decision_id,

            "decision_text":
                candidate["text"],

            "decision_maker":
                candidate["speaker"],

            "decision_time":
                meeting[
                    "meeting_datetime"
                ],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "decision_state":
                candidate[
                    "decision_state"
                ],

            "decision_target":
                None,

            "reversed_decision":
                None,

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        })


        # ----------------------------------------------------
        # Only actual MADE decisions become DECISION_MADE
        # temporal events.
        # ----------------------------------------------------

        if (
            candidate[
                "decision_state"
            ]
            == "MADE"
        ):

            add_event(

                record,

                event_type=
                    "DECISION_MADE",

                entity=
                    meeting_entity,

                event_time=
                    meeting[
                        "meeting_datetime"
                    ],

                actor=
                    candidate[
                        "speaker"
                    ],

                from_value=
                    None,

                to_value=
                    candidate[
                        "text"
                    ],

                attributes={

                    "speaker_timestamp":
                        candidate[
                            "speaker_timestamp"
                        ],

                    "decision_id":
                        decision_id
                },

                extraction_method=
                    "rule"
            )


    # ========================================================
    # N. DEPENDENCIES
    # ========================================================

    for index, candidate in enumerate(
        meeting[
            "dependency_candidates"
        ],
        start=1
    ):

        record[
            "dependencies"
        ].append({

            "dependency_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_DEP_{index:03d}"
                ),

            "text":
                candidate["text"],

            "source_entity":
                None,

            "target_entity":
                None,

            "dependency_type":
                "MEETING_DEPENDENCY",

            "dependency_status":
                candidate[
                    "dependency_state"
                ],

            "created_at":
                meeting[
                    "meeting_datetime"
                ],

            "speaker":
                candidate["speaker"],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        })


    # ========================================================
    # O. BLOCKER EVIDENCE
    # ========================================================

    for index, candidate in enumerate(
        meeting[
            "blocker_candidates"
        ],
        start=1
    ):

        record[
            "blockers"
        ].append({

            "blocker_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_BLOCKER_{index:03d}"
                ),

            "blocked_entity":
                None,

            "blocker_entity":
                None,

            "blocker_state":
                candidate[
                    "blocker_state"
                ],

            "is_active_assertion":
                (
                    candidate[
                        "blocker_state"
                    ]
                    == "ACTIVE"
                ),

            "is_recap":
                (
                    "recap"
                    in candidate[
                        "text"
                    ].lower()
                ),

            "meeting_time":
                meeting[
                    "meeting_datetime"
                ],

            "speaker":
                candidate["speaker"],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "blocking_reason":
                None,

            "extraction_method":
                "rule",

            "evidence_text":
                candidate["text"]
        })


    # ========================================================
    # P. DIAGNOSTICS
    # ========================================================

    record[
        "attributes"
    ][
        "extraction_diagnostics"
    ] = {

        "participant_count":
            len(
                meeting[
                    "participants"
                ]
            ),

        "utterance_count":
            len(
                meeting[
                    "utterances"
                ]
            ),

        "unique_issue_count":
            len(unique_issues),

        "unique_requirement_count":
            len(
                unique_requirements
            ),

        "action_count":
            len(
                record[
                    "action_items"
                ]
            ),

        "decision_count":
            len(
                record[
                    "decisions"
                ]
            ),

        "dependency_count":
            len(
                record[
                    "dependencies"
                ]
            ),

        "blocker_evidence_count":
            len(
                record[
                    "blockers"
                ]
            ),

        "gliner_occurrence_count":
            len(gliner_entities)
    }


    # ========================================================
    # Q. PROVENANCE
    # ========================================================

    record[
        "provenance"
    ] = {

        "source_type":
            "MEETING",

        "source_id":
            meeting[
                "meeting_id"
            ],

        "instance_id":
            instance_id,

        "chunk_id":
            None,

        "evidence_text":
            meeting["text"],

        "file_name":
            meeting_file.name
    }


    return record
```

```python
# ============================================================
# STEP 176: COMPLETE JBEAP MEETING PRODUCTION TEST
# ============================================================
# Purpose:
# Process both official JBEAP-I01 meetings using the final
# generic meeting pipeline.
#
# Nothing is saved yet.
# ============================================================


jbeap_meeting_records = []


for meeting_file in (
    test_inputs[
        "meeting_files"
    ]
):

    record = (
        extract_production_meeting(

            meeting_file=
                meeting_file,

            repository_project=
                test_inputs[
                    "repository_folder"
                ],

            project_key=
                test_inputs[
                    "repository_key"
                ],

            project_name=
                test_inputs[
                    "project_name"
                ],

            instance_id=
                test_inputs[
                    "instance_id"
                ]
        )
    )


    jbeap_meeting_records.append(
        record
    )


print("=" * 70)
print("JBEAP MEETING EXTRACTION COMPLETE")
print("=" * 70)

print(
    "\nRecords:",
    len(
        jbeap_meeting_records
    )
)

for record in (
    jbeap_meeting_records
):

    print(
        "-",
        record[
            "source"
        ][
            "source_id"
        ]
    )
```

```python
# ============================================================
# STEP 177: FINAL JBEAP MEETING EXTRACTION SUMMARY
# ============================================================


print("=" * 70)
print("FINAL JBEAP-I01 MEETING EXTRACTION")
print("=" * 70)


for record in jbeap_meeting_records:

    entity_counts = Counter(

        entity["type"]

        for entity
        in record["entities"]
    )


    decision_states = Counter(

        decision[
            "decision_state"
        ]

        for decision
        in record["decisions"]
    )


    dependency_states = Counter(

        dependency[
            "dependency_status"
        ]

        for dependency
        in record["dependencies"]
    )


    blocker_states = Counter(

        blocker[
            "blocker_state"
        ]

        for blocker
        in record["blockers"]
    )


    diagnostics = (
        record[
            "attributes"
        ][
            "extraction_diagnostics"
        ]
    )


    print(
        "\n" + "=" * 70
    )


    print(
        "Meeting:",
        record[
            "source"
        ][
            "source_id"
        ]
    )


    print(
        "Entities:",
        len(
            record["entities"]
        )
    )


    print(
        "Relationships:",
        len(
            record[
                "relationships"
            ]
        )
    )


    print(
        "Events:",
        len(
            record["events"]
        )
    )


    print(
        "Actions:",
        len(
            record[
                "action_items"
            ]
        )
    )


    print(
        "Decisions:",
        len(
            record[
                "decisions"
            ]
        )
    )


    print(
        "Dependencies:",
        len(
            record[
                "dependencies"
            ]
        )
    )


    print(
        "Blocker evidence:",
        len(
            record[
                "blockers"
            ]
        )
    )


    print(
        "\nEntity types:",
        dict(entity_counts)
    )


    print(
        "Decision states:",
        dict(decision_states)
    )


    print(
        "Dependency states:",
        dict(
            dependency_states
        )
    )


    print(
        "Blocker states:",
        dict(blocker_states)
    )


    print(
        "\nDiagnostics:",
        diagnostics
    )
```

```python
# ============================================================
# STEP 178: GENERIC JSONL SAVE / LOAD HELPERS
# ============================================================
# Purpose:
# Provide reusable functions for writing and reloading the
# production extraction files.
# ============================================================


def save_jsonl(
    records,
    output_file
):

    output_file.parent.mkdir(
        parents=True,
        exist_ok=True
    )


    with open(
        output_file,
        "w",
        encoding="utf-8"
    ) as file:

        for record in records:

            file.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                )
                + "\n"
            )


def load_jsonl_records(
    input_file
):

    records = []


    with open(
        input_file,
        "r",
        encoding="utf-8"
    ) as file:

        for line_number, line in enumerate(
            file,
            start=1
        ):

            if not line.strip():
                continue


            try:

                records.append(
                    json.loads(line)
                )


            except json.JSONDecodeError:

                raise ValueError(
                    f"Invalid JSON on line "
                    f"{line_number}: {input_file}"
                )


    return records


print(
    "SUCCESS: JSONL save/load helpers defined."
)
```

```python
# ============================================================
# STEP 179: GENERIC INSTANCE JIRA ALIGNMENT FUNCTION
# ============================================================
# Purpose:
# Load one repository's Jira dataset and return only tickets
# with activity inside the selected project-instance timeline.
# ============================================================


def load_aligned_instance_jira(
    inputs
):

    repository_tickets = []


    with open(
        inputs["jira_file"],
        "r",
        encoding="utf-8"
    ) as file:

        for line in file:

            if line.strip():

                repository_tickets.append(
                    json.loads(line)
                )


    instance_start = (
        parse_timestamp_utc(
            inputs[
                "instance_start"
            ]
        )
    )


    instance_end = (
        parse_timestamp_utc(
            inputs[
                "instance_end"
            ]
        )
    )


    aligned_tickets = []


    for ticket in repository_tickets:

        activity_times = (
            collect_ticket_activity_times(
                ticket
            )
        )


        has_activity = False


        for raw_timestamp in activity_times:

            timestamp = (
                parse_timestamp_utc(
                    raw_timestamp
                )
            )


            if (
                timestamp is not None
                and
                instance_start
                <= timestamp
                <= instance_end
            ):

                has_activity = True
                break


        if has_activity:

            aligned_tickets.append(
                ticket
            )


    return aligned_tickets


print(
    "SUCCESS: Generic Jira alignment function defined."
)
```

```python
# ============================================================
# STEP 180: GENERIC PRODUCTION JIRA RUNNER
# ============================================================
# Purpose:
# Align + extract + deduplicate all Jira tickets for one
# project instance.
# ============================================================


def run_production_jira(
    inputs
):

    aligned_tickets = (
        load_aligned_instance_jira(
            inputs
        )
    )


    records = []


    for ticket in aligned_tickets:

        record = extract_jira_ticket(

            ticket=ticket,

            repository_project=
                inputs[
                    "repository_folder"
                ],

            instance_id=
                inputs[
                    "instance_id"
                ],

            instance_start=
                inputs[
                    "instance_start"
                ],

            instance_end=
                inputs[
                    "instance_end"
                ],

            source_file_name=
                inputs[
                    "jira_file"
                ].name
        )


        # -----------------------------------------------
        # Remove duplicate status-history/full-changelog
        # transitions.
        # -----------------------------------------------

        record = (
            deduplicate_status_events(
                record
            )
        )


        records.append(
            record
        )


    return records


print(
    "SUCCESS: Production Jira runner defined."
)
```

```python
# ============================================================
# STEP 181: GENERIC PRD + MEETING PRODUCTION RUNNERS
# ============================================================
# Purpose:
# Process the official manifest-selected PRD and meetings for
# one project instance.
# ============================================================


def run_production_prds(
    inputs
):

    records = []


    for prd_file in inputs[
        "prd_files"
    ]:

        record = extract_production_prd(

            prd_file=
                prd_file,

            repository_project=
                inputs[
                    "repository_folder"
                ],

            project_key=
                inputs[
                    "repository_key"
                ],

            project_name=
                inputs[
                    "project_name"
                ],

            instance_id=
                inputs[
                    "instance_id"
                ]
        )


        records.append(
            record
        )


    return records



def run_production_meetings(
    inputs
):

    records = []


    for meeting_file in inputs[
        "meeting_files"
    ]:

        record = (
            extract_production_meeting(

                meeting_file=
                    meeting_file,

                repository_project=
                    inputs[
                        "repository_folder"
                    ],

                project_key=
                    inputs[
                        "repository_key"
                    ],

                project_name=
                    inputs[
                        "project_name"
                    ],

                instance_id=
                    inputs[
                        "instance_id"
                    ]
            )
        )


        records.append(
            record
        )


    return records


print(
    "SUCCESS: Production PRD/Meeting runners defined."
)
```

```python
# ============================================================
# STEP 182: GENERIC INSTANCE VALIDATION FUNCTION
# ============================================================
# Purpose:
# Validate one complete production extraction before it is
# considered ready for Entity Resolution.
#
# Checks:
#
#   Jira record count
#   PRD count
#   Meeting count
#   required schema sections
#   project/instance consistency
#   timestamps
#   OTHER_CHANGE absence
# ============================================================


MASTER_REQUIRED_SECTIONS = [

    "entities",
    "relationships",
    "events",
    "decisions",
    "action_items",
    "blockers",
    "dependencies",
    "temporal",
    "attributes",
    "provenance"
]


def validate_production_instance(
    inputs,
    jira_records,
    prd_records,
    meeting_records
):

    all_records = (
        jira_records
        + prd_records
        + meeting_records
    )


    report = {

        "repository":
            inputs[
                "repository_folder"
            ],

        "project_key":
            inputs[
                "repository_key"
            ],

        "instance_id":
            inputs[
                "instance_id"
            ],

        "expected_jira_records":
            inputs[
                "expected_tickets"
            ],

        "jira_records":
            len(jira_records),

        "prd_records":
            len(prd_records),

        "meeting_records":
            len(meeting_records),

        "missing_sections":
            [],

        "wrong_project_records":
            [],

        "wrong_instance_records":
            [],

        "unparseable_event_times":
            [],

        "outside_window_events":
            [],

        "other_change_events":
            0,

        "custom_field_events":
            0
    }


    instance_start = (
        parse_timestamp_utc(
            inputs[
                "instance_start"
            ]
        )
    )


    instance_end = (
        parse_timestamp_utc(
            inputs[
                "instance_end"
            ]
        )
    )


    # ========================================================
    # RECORD VALIDATION
    # ========================================================

    for record in all_records:

        source_id = (
            record[
                "source"
            ][
                "source_id"
            ]
        )


        # ----------------------------------------------------
        # Required sections
        # ----------------------------------------------------

        for section in (
            MASTER_REQUIRED_SECTIONS
        ):

            if section not in record:

                report[
                    "missing_sections"
                ].append({

                    "source_id":
                        source_id,

                    "section":
                        section
                })


        # ----------------------------------------------------
        # Project / instance consistency
        # ----------------------------------------------------

        if (
            record[
                "repository_project"
            ]
            !=
            inputs[
                "repository_folder"
            ]
        ):

            report[
                "wrong_project_records"
            ].append(
                source_id
            )


        if (
            record[
                "instance_id"
            ]
            !=
            inputs[
                "instance_id"
            ]
        ):

            report[
                "wrong_instance_records"
            ].append(
                source_id
            )


        # ----------------------------------------------------
        # Event validation
        # ----------------------------------------------------

        for event in (
            record["events"]
        ):

            event_type = (
                event.get(
                    "event_type"
                )
            )


            if (
                event_type
                == "OTHER_CHANGE"
            ):

                report[
                    "other_change_events"
                ] += 1


            if (
                event_type
                == "CUSTOM_FIELD_CHANGE"
            ):

                report[
                    "custom_field_events"
                ] += 1


            raw_time = event.get(
                "event_time"
            )


            if raw_time is None:
                continue


            parsed_time = (
                parse_timestamp_utc(
                    raw_time
                )
            )


            if parsed_time is None:

                report[
                    "unparseable_event_times"
                ].append({

                    "source_id":
                        source_id,

                    "event_type":
                        event_type,

                    "event_time":
                        raw_time
                })

                continue


            if not (
                instance_start
                <= parsed_time
                <= instance_end
            ):

                report[
                    "outside_window_events"
                ].append({

                    "source_id":
                        source_id,

                    "event_type":
                        event_type,

                    "event_time":
                        raw_time
                })


    # ========================================================
    # FINAL PASS / FAIL
    # ========================================================

    report[
        "jira_count_ok"
    ] = (

        len(jira_records)
        ==
        inputs[
            "expected_tickets"
        ]
    )


    report[
        "prd_count_ok"
    ] = (

        len(prd_records)
        ==
        len(
            inputs[
                "prd_files"
            ]
        )
    )


    report[
        "meeting_count_ok"
    ] = (

        len(meeting_records)
        ==
        len(
            inputs[
                "meeting_files"
            ]
        )
    )


    report[
        "schema_ok"
    ] = (
        len(
            report[
                "missing_sections"
            ]
        )
        == 0
    )


    report[
        "project_instance_ok"
    ] = (

        len(
            report[
                "wrong_project_records"
            ]
        )
        == 0

        and

        len(
            report[
                "wrong_instance_records"
            ]
        )
        == 0
    )


    report[
        "temporal_ok"
    ] = (

        len(
            report[
                "unparseable_event_times"
            ]
        )
        == 0

        and

        len(
            report[
                "outside_window_events"
            ]
        )
        == 0
    )


    report[
        "event_vocabulary_ok"
    ] = (

        report[
            "other_change_events"
        ]
        == 0
    )


    report[
        "success"
    ] = all([

        report[
            "jira_count_ok"
        ],

        report[
            "prd_count_ok"
        ],

        report[
            "meeting_count_ok"
        ],

        report[
            "schema_ok"
        ],

        report[
            "project_instance_ok"
        ],

        report[
            "temporal_ok"
        ],

        report[
            "event_vocabulary_ok"
        ]
    ])


    return report


print(
    "SUCCESS: Generic production validation defined."
)
```

```python
# ============================================================
# STEP 183: COMPLETE ONE-INSTANCE PRODUCTION RUNNER
# ============================================================
# Purpose:
# Run the complete GC1 extraction pipeline for one instance:
#
#   Jira
#   PRD
#   Meetings
#   Validation
#   Save outputs
#
# IMPORTANT:
# If validation fails, the function DOES NOT mark the instance
# as successful.
# ============================================================


def run_production_instance(
    inputs,
    save_outputs=True
):

    print(
        "\n" + "=" * 70
    )

    print(
        "PROCESSING:",
        inputs[
            "repository_folder"
        ],
        "/",
        inputs[
            "instance_id"
        ]
    )

    print("=" * 70)


    # ========================================================
    # JIRA
    # ========================================================

    print("\n[1/4] Jira extraction...")

    jira_records = (
        run_production_jira(
            inputs
        )
    )

    print(
        "      Jira records:",
        len(jira_records)
    )


    # ========================================================
    # PRD
    # ========================================================

    print("\n[2/4] PRD extraction...")

    prd_records = (
        run_production_prds(
            inputs
        )
    )

    print(
        "      PRD records:",
        len(prd_records)
    )


    # ========================================================
    # MEETINGS
    # ========================================================

    print("\n[3/4] Meeting extraction...")

    meeting_records = (
        run_production_meetings(
            inputs
        )
    )

    print(
        "      Meeting records:",
        len(meeting_records)
    )


    # ========================================================
    # VALIDATION
    # ========================================================

    print("\n[4/4] Validation...")

    validation_report = (
        validate_production_instance(

            inputs,

            jira_records,

            prd_records,

            meeting_records
        )
    )


    print(
        "      Success:",
        validation_report[
            "success"
        ]
    )


    # ========================================================
    # SAVE
    # ========================================================

    if save_outputs:

        output_folder = (
            inputs[
                "output_folder"
            ]
        )


        output_folder.mkdir(
            parents=True,
            exist_ok=True
        )


        save_jsonl(

            jira_records,

            output_folder
            / JIRA_OUTPUT_NAME
        )


        save_jsonl(

            prd_records,

            output_folder
            / PRD_OUTPUT_NAME
        )


        save_jsonl(

            meeting_records,

            output_folder
            / MEETING_OUTPUT_NAME
        )


        with open(
            output_folder
            / VALIDATION_OUTPUT_NAME,
            "w",
            encoding="utf-8"
        ) as file:

            json.dump(
                validation_report,
                file,
                indent=2,
                ensure_ascii=False
            )


        print(
            "\nSaved to:",
            output_folder
        )


    return {

        "jira_records":
            jira_records,

        "prd_records":
            prd_records,

        "meeting_records":
            meeting_records,

        "validation":
            validation_report
    }


print(
    "SUCCESS: Complete production runner defined."
)
```

```python
# ============================================================
# STEP 186: SELECT REMAINING PROJECTS FOR BATCH EXTRACTION
# ============================================================
# Purpose:
# Prepare the projects that still need production extraction.
#
# Already completed:
#
#     01_JBEAP / JBEAP-I01
#         -> complete production pipeline
#
#     24_DOCS / DOCS-I01
#         -> completed and validated during pilot
#
# Therefore this batch contains the remaining 28 instances.
# ============================================================


COMPLETED_REPOSITORIES = {
    "01_JBEAP",
    "24_DOCS"
}


remaining_production_inputs = [

    inputs

    for inputs in production_inputs

    if inputs["repository_folder"]
    not in COMPLETED_REPOSITORIES
]


print("=" * 70)
print("REMAINING PRODUCTION BATCH")
print("=" * 70)


print(
    "\nTotal selected instances:",
    len(production_inputs)
)

print(
    "Already completed:",
    len(COMPLETED_REPOSITORIES)
)

print(
    "Remaining instances:",
    len(remaining_production_inputs)
)


print("\nPROJECTS TO PROCESS")
print("-" * 70)


for index, inputs in enumerate(
    remaining_production_inputs,
    start=1
):

    print(
        f"{index:02d}.",
        inputs["repository_folder"],
        "->",
        inputs["instance_id"],
        "| expected Jira:",
        inputs["expected_tickets"]
    )


remaining_expected_jira = sum(

    inputs["expected_tickets"]

    for inputs
    in remaining_production_inputs
)


print(
    "\nExpected Jira records in this batch:",
    remaining_expected_jira
)
```

```python
# ============================================================
# STEP 187: SAFE MULTI-PROJECT BATCH RUNNER
# ============================================================
# Purpose:
# Process multiple GC1 instances sequentially.
#
# SAFETY DESIGN:
#
# If one project fails:
#
#     record the error
#     continue with the next project
#
# This prevents losing the entire long-running batch because
# of one repository-specific issue.
# ============================================================

import traceback
import gc
import time


def run_production_batch(
    inputs_list
):

    batch_results = []


    total = len(
        inputs_list
    )


    for index, inputs in enumerate(
        inputs_list,
        start=1
    ):

        repository = (
            inputs[
                "repository_folder"
            ]
        )

        instance_id = (
            inputs[
                "instance_id"
            ]
        )


        print(
            "\n\n" + "#" * 75
        )

        print(
            f"BATCH {index}/{total}"
        )

        print(
            repository,
            "/",
            instance_id
        )

        print(
            "#" * 75
        )


        start_time = time.time()


        try:

            result = (
                run_production_instance(

                    inputs,

                    save_outputs=True
                )
            )


            elapsed_seconds = (
                time.time()
                - start_time
            )


            validation = (
                result[
                    "validation"
                ]
            )


            batch_results.append({

                "repository":
                    repository,

                "instance_id":
                    instance_id,

                "expected_jira":
                    inputs[
                        "expected_tickets"
                    ],

                "actual_jira":
                    validation[
                        "jira_records"
                    ],

                "prd_records":
                    validation[
                        "prd_records"
                    ],

                "meeting_records":
                    validation[
                        "meeting_records"
                    ],

                "custom_field_events":
                    validation[
                        "custom_field_events"
                    ],

                "other_change_events":
                    validation[
                        "other_change_events"
                    ],

                "success":
                    validation[
                        "success"
                    ],

                "runtime_seconds":
                    round(
                        elapsed_seconds,
                        2
                    ),

                "error":
                    None
            })


            print(
                "\nBATCH RESULT:",
                "SUCCESS"
                if validation["success"]
                else "VALIDATION FAILED"
            )


        except Exception as error:

            elapsed_seconds = (
                time.time()
                - start_time
            )


            batch_results.append({

                "repository":
                    repository,

                "instance_id":
                    instance_id,

                "expected_jira":
                    inputs[
                        "expected_tickets"
                    ],

                "actual_jira":
                    None,

                "prd_records":
                    None,

                "meeting_records":
                    None,

                "custom_field_events":
                    None,

                "other_change_events":
                    None,

                "success":
                    False,

                "runtime_seconds":
                    round(
                        elapsed_seconds,
                        2
                    ),

                "error":
                    str(error)
            })


            print(
                "\nERROR while processing:",
                repository
            )

            print(
                str(error)
            )


            print(
                "\nTraceback:"
            )

            traceback.print_exc()


        # ----------------------------------------------------
        # Free Python references between projects.
        #
        # GLiNER model itself remains loaded.
        # ----------------------------------------------------

        gc.collect()


    return batch_results


print(
    "SUCCESS: Safe production batch runner defined."
)
```

```python
# ============================================================
# STEP 188: RUN REMAINING 28 GC1 PROJECTS
# ============================================================
# Purpose:
# Execute the validated production extraction pipeline across
# every project that has not yet been processed.
#
# For EACH project:
#
#     Jira alignment
#         ↓
#     Jira extraction
#         ↓
#     PRD deterministic extraction + GLiNER
#         ↓
#     Meeting deterministic/rule extraction + GLiNER
#         ↓
#     validation
#         ↓
#     save 3 JSONL files + validation_report.json
#
# The batch continues even if one project fails.
# ============================================================


batch_results = (
    run_production_batch(
        remaining_production_inputs
    )
)
```

```python
# ============================================================
# STEP 189: PRODUCTION BATCH SUMMARY
# ============================================================
# Purpose:
# Summarize success/failure across the remaining 28 projects.
# ============================================================


successful_projects = [

    result

    for result in batch_results

    if result["success"]
]


failed_projects = [

    result

    for result in batch_results

    if not result["success"]
]


print("=" * 75)
print("GC1 PRODUCTION BATCH SUMMARY")
print("=" * 75)


print(
    "\nProjects attempted:",
    len(batch_results)
)

print(
    "Successful:",
    len(successful_projects)
)

print(
    "Failed / validation failed:",
    len(failed_projects)
)


print("\nPROJECT RESULTS")
print("-" * 75)


for result in batch_results:

    status = (
        "SUCCESS"
        if result["success"]
        else "FAILED"
    )


    print(
        f"{result['repository']:20s}",
        "|",
        f"{result['instance_id']:20s}",
        "|",
        status,
        "| Jira:",
        result["actual_jira"],
        "/",
        result["expected_jira"],
        "| PRD:",
        result["prd_records"],
        "| Meetings:",
        result["meeting_records"],
        "| Custom:",
        result["custom_field_events"],
        "| Runtime:",
        result["runtime_seconds"],
        "sec"
    )


if failed_projects:

    print("\nFAILED PROJECTS")
    print("-" * 75)


    for result in failed_projects:

        print(
            result["repository"],
            "|",
            result["instance_id"],
            "|",
            result["error"]
        )


else:

    print(
        "\nSUCCESS:"
        "\nAll remaining 28 GC1 projects passed "
        "production extraction and validation."
    )
```

```python
# ============================================================
# STEP 193: DIAGNOSE SPR-I01 JIRA ALIGNMENT
# ============================================================
# Purpose:
# SPR-I01 expects all 300 repository tickets, but our generic
# activity-window reconstruction selected only 47.
#
# Before changing alignment logic, determine how those 300
# tickets relate to the instance timeline.
#
# We inspect:
#
#   ticket creation
#   ticket update
#   resolution
#   status history
#   full changelog
#   comments
#
# and compare the selected 47 against the other 253.
# ============================================================


spr_inputs = next(

    item

    for item in production_inputs

    if item["repository_folder"] == "08_SPR"
)


spr_start = parse_timestamp_utc(
    spr_inputs["instance_start"]
)

spr_end = parse_timestamp_utc(
    spr_inputs["instance_end"]
)


# ------------------------------------------------------------
# Load all 300 SPR tickets.
# ------------------------------------------------------------

spr_all_tickets = []


with open(
    spr_inputs["jira_file"],
    "r",
    encoding="utf-8"
) as file:

    for line in file:

        if line.strip():

            spr_all_tickets.append(
                json.loads(line)
            )


# ------------------------------------------------------------
# Classify each ticket.
# ------------------------------------------------------------

spr_alignment_analysis = []


for ticket in spr_all_tickets:

    activity_times = (
        collect_ticket_activity_times(
            ticket
        )
    )


    normalized_times = [

        parse_timestamp_utc(value)

        for value in activity_times
    ]


    normalized_times = [

        value

        for value in normalized_times

        if value is not None
    ]


    activity_in_window = [

        value

        for value in normalized_times

        if spr_start <= value <= spr_end
    ]


    earliest_activity = (
        min(normalized_times)
        if normalized_times
        else None
    )


    latest_activity = (
        max(normalized_times)
        if normalized_times
        else None
    )


    spr_alignment_analysis.append({

        "issue_key":
            ticket.get("issue_key"),

        "created_at":
            ticket.get("created_at"),

        "updated_at":
            ticket.get("updated_at"),

        "resolved_at":
            ticket.get("resolved_at"),

        "activity_count":
            len(normalized_times),

        "activity_in_window_count":
            len(activity_in_window),

        "earliest_activity":
            (
                earliest_activity.isoformat()
                if earliest_activity
                else None
            ),

        "latest_activity":
            (
                latest_activity.isoformat()
                if latest_activity
                else None
            ),

        "selected_by_current_logic":
            bool(activity_in_window)
    })


selected_count = sum(

    row["selected_by_current_logic"]

    for row in spr_alignment_analysis
)


not_selected = [

    row

    for row in spr_alignment_analysis

    if not row[
        "selected_by_current_logic"
    ]
]


print("=" * 75)
print("SPR-I01 ALIGNMENT DIAGNOSTIC")
print("=" * 75)


print("\nInstance:")
print(
    spr_start,
    "->",
    spr_end
)


print(
    "\nRepository tickets:",
    len(spr_all_tickets)
)


print(
    "Selected by current logic:",
    selected_count
)


print(
    "Not selected:",
    len(not_selected)
)


print("\nFIRST 20 NON-SELECTED TICKETS")
print("-" * 75)


for row in not_selected[:20]:

    print(
        row["issue_key"],
        "| created:",
        row["created_at"],
        "| updated:",
        row["updated_at"],
        "| resolved:",
        row["resolved_at"],
        "| earliest:",
        row["earliest_activity"],
        "| latest:",
        row["latest_activity"]
    )
```

```python
# ============================================================
# STEP 195: PRODUCTION TIMESTAMP PARSER - FINAL VERSION
# ============================================================
# Purpose:
# Support all timestamp formats observed across GC1.
#
# Supported examples:
#
#   2019-03-24
#   2019-03-24T00:00:00+00:00
#   2019-03-24T00:00:00+0000
#
#   24 March 2019
#   17 March 2019
#
# All returned datetime objects are timezone-aware UTC.
# ============================================================

from datetime import datetime, timezone


def parse_timestamp_utc(value):

    if value is None:
        return None


    # --------------------------------------------------------
    # Already datetime
    # --------------------------------------------------------

    if isinstance(value, datetime):

        dt = value


    else:

        value = str(value).strip()


        if not value:
            return None


        dt = None


        # ====================================================
        # 1. ISO-style timestamps
        # ====================================================

        iso_value = value


        # Convert trailing:
        #
        #     +0000
        #
        # into:
        #
        #     +00:00
        #
        # for datetime.fromisoformat().
        if (
            len(iso_value) >= 5
            and
            iso_value[-5] in ["+", "-"]
            and
            iso_value[-3] != ":"
        ):

            iso_value = (
                iso_value[:-2]
                + ":"
                + iso_value[-2:]
            )


        try:

            dt = datetime.fromisoformat(
                iso_value
            )

        except ValueError:

            dt = None


        # ====================================================
        # 2. Human-readable PRD dates
        # ====================================================
        #
        # Examples:
        #
        #     24 March 2019
        #     17 March 2019
        # ====================================================

        if dt is None:

            human_date_formats = [

                "%d %B %Y",
                "%d %b %Y",

                "%B %d, %Y",
                "%b %d, %Y"
            ]


            for date_format in (
                human_date_formats
            ):

                try:

                    dt = datetime.strptime(
                        value,
                        date_format
                    )

                    break

                except ValueError:

                    continue


        # ----------------------------------------------------
        # Still not parseable.
        # ----------------------------------------------------

        if dt is None:

            return None


    # ========================================================
    # NORMALIZE TO UTC
    # ========================================================

    if dt.tzinfo is None:

        dt = dt.replace(
            tzinfo=timezone.utc
        )

    else:

        dt = dt.astimezone(
            timezone.utc
        )


    return dt


print("=" * 70)
print("TIMESTAMP PARSER TEST")
print("=" * 70)


test_timestamp_values = [

    "2019-03-24",

    "2019-03-24T00:00:00+00:00",

    "2019-03-24T00:00:00+0000",

    "24 March 2019",

    "17 March 2019"
]


for value in test_timestamp_values:

    print(
        value,
        "->",
        parse_timestamp_utc(value)
    )
```

```python
# ============================================================
# STEP 196: LOAD OFFICIAL GENERATED-ARTEFACT TIMESTAMPS
# ============================================================
# Purpose:
# Build the authoritative timestamp lookup for official PRDs
# and meetings from generation_manifest.json.
#
# CONTROL METADATA ONLY.
#
# This information is used only for validation.
# It does NOT become KG evidence or ML features.
# ============================================================


def get_official_artifact_timestamps(
    inputs
):

    with open(
        inputs[
            "generation_manifest_file"
        ],
        "r",
        encoding="utf-8"
    ) as file:

        manifest = json.load(file)


    official_timestamps = {}


    # --------------------------------------------------------
    # PRDs
    # --------------------------------------------------------

    for item in (
        manifest.get(
            "prd_schedule"
        )
        or []
    ):

        document_id = (
            item.get(
                "document_id"
            )
        )


        timestamp = (
            item.get(
                "timestamp"
            )
        )


        if (
            document_id
            and timestamp
        ):

            official_timestamps[
                document_id
            ] = (
                parse_timestamp_utc(
                    timestamp
                )
            )


    # --------------------------------------------------------
    # Meetings
    # --------------------------------------------------------

    for item in (
        manifest.get(
            "meeting_schedule"
        )
        or []
    ):

        meeting_id = (
            item.get(
                "meeting_id"
            )
        )


        timestamp = (
            item.get(
                "timestamp"
            )
        )


        if (
            meeting_id
            and timestamp
        ):

            official_timestamps[
                meeting_id
            ] = (
                parse_timestamp_utc(
                    timestamp
                )
            )


    return official_timestamps


print(
    "SUCCESS: Official artefact timestamp helper defined."
)
```

```python
# ============================================================
# STEP 197: SOURCE-AWARE PRODUCTION VALIDATOR
# ============================================================
# Purpose:
# Correct the temporal validation model.
#
# JIRA:
#   Event must fall inside the selected Jira instance window.
#
# PRD / MEETING:
#   Event timestamp must be parseable and correspond to an
#   official generated artefact.
#
# This prevents legitimate scheduled meetings immediately
# after the Jira window from being incorrectly rejected.
# ============================================================


def validate_production_instance(
    inputs,
    jira_records,
    prd_records,
    meeting_records
):

    all_records = (
        jira_records
        + prd_records
        + meeting_records
    )


    report = {

        "repository":
            inputs[
                "repository_folder"
            ],

        "project_key":
            inputs[
                "repository_key"
            ],

        "instance_id":
            inputs[
                "instance_id"
            ],

        "expected_jira_records":
            inputs[
                "expected_tickets"
            ],

        "jira_records":
            len(jira_records),

        "prd_records":
            len(prd_records),

        "meeting_records":
            len(meeting_records),

        "missing_sections":
            [],

        "wrong_project_records":
            [],

        "wrong_instance_records":
            [],

        "unparseable_event_times":
            [],

        "outside_window_events":
            [],

        "artifact_timestamp_mismatches":
            [],

        "other_change_events":
            0,

        "custom_field_events":
            0
    }


    instance_start = (
        parse_timestamp_utc(
            inputs[
                "instance_start"
            ]
        )
    )


    instance_end = (
        parse_timestamp_utc(
            inputs[
                "instance_end"
            ]
        )
    )


    official_artifact_times = (
        get_official_artifact_timestamps(
            inputs
        )
    )


    # ========================================================
    # VALIDATE RECORDS
    # ========================================================

    for record in all_records:

        source_type = (
            record[
                "source"
            ][
                "source_type"
            ]
        )


        source_id = (
            record[
                "source"
            ][
                "source_id"
            ]
        )


        # ----------------------------------------------------
        # Required schema
        # ----------------------------------------------------

        for section in (
            MASTER_REQUIRED_SECTIONS
        ):

            if section not in record:

                report[
                    "missing_sections"
                ].append({

                    "source_id":
                        source_id,

                    "section":
                        section
                })


        # ----------------------------------------------------
        # Project consistency
        # ----------------------------------------------------

        if (
            record[
                "repository_project"
            ]
            !=
            inputs[
                "repository_folder"
            ]
        ):

            report[
                "wrong_project_records"
            ].append(
                source_id
            )


        # ----------------------------------------------------
        # Instance consistency
        # ----------------------------------------------------

        if (
            record[
                "instance_id"
            ]
            !=
            inputs[
                "instance_id"
            ]
        ):

            report[
                "wrong_instance_records"
            ].append(
                source_id
            )


        # ====================================================
        # SOURCE-LEVEL PRD / MEETING TIMESTAMP VALIDATION
        # ====================================================

        if source_type in [
            "PRD",
            "MEETING"
        ]:

            raw_source_time = (
                record[
                    "source"
                ].get(
                    "source_timestamp"
                )
            )


            parsed_source_time = (
                parse_timestamp_utc(
                    raw_source_time
                )
            )


            if (
                raw_source_time is not None
                and
                parsed_source_time is None
            ):

                report[
                    "unparseable_event_times"
                ].append({

                    "source_id":
                        source_id,

                    "event_type":
                        "SOURCE_TIMESTAMP",

                    "event_time":
                        raw_source_time
                })


            official_time = (
                official_artifact_times.get(
                    source_id
                )
            )


            # Compare calendar dates rather than requiring
            # identical time-of-day formatting.
            if (
                official_time is not None
                and
                parsed_source_time is not None
                and
                official_time.date()
                !=
                parsed_source_time.date()
            ):

                report[
                    "artifact_timestamp_mismatches"
                ].append({

                    "source_id":
                        source_id,

                    "source_timestamp":
                        raw_source_time,

                    "official_timestamp":
                        official_time.isoformat()
                })


        # ====================================================
        # EVENT VALIDATION
        # ====================================================

        for event in (
            record["events"]
        ):

            event_type = (
                event.get(
                    "event_type"
                )
            )


            if (
                event_type
                == "OTHER_CHANGE"
            ):

                report[
                    "other_change_events"
                ] += 1


            if (
                event_type
                == "CUSTOM_FIELD_CHANGE"
            ):

                report[
                    "custom_field_events"
                ] += 1


            raw_time = event.get(
                "event_time"
            )


            if raw_time is None:
                continue


            parsed_time = (
                parse_timestamp_utc(
                    raw_time
                )
            )


            if parsed_time is None:

                report[
                    "unparseable_event_times"
                ].append({

                    "source_id":
                        source_id,

                    "event_type":
                        event_type,

                    "event_time":
                        raw_time
                })

                continue


            # ------------------------------------------------
            # JIRA EVENT:
            # strict Jira instance boundary.
            # ------------------------------------------------

            if source_type == "JIRA":

                if not (
                    instance_start
                    <= parsed_time
                    <= instance_end
                ):

                    report[
                        "outside_window_events"
                    ].append({

                        "source_id":
                            source_id,

                        "event_type":
                            event_type,

                        "event_time":
                            raw_time
                    })


            # ------------------------------------------------
            # PRD / MEETING EVENT:
            # validate against official artefact date.
            # ------------------------------------------------

            elif source_type in [
                "PRD",
                "MEETING"
            ]:

                official_time = (
                    official_artifact_times.get(
                        source_id
                    )
                )


                if (
                    official_time is not None
                    and
                    parsed_time.date()
                    !=
                    official_time.date()
                ):

                    report[
                        "artifact_timestamp_mismatches"
                    ].append({

                        "source_id":
                            source_id,

                        "event_type":
                            event_type,

                        "event_time":
                            raw_time,

                        "official_timestamp":
                            official_time.isoformat()
                    })


    # ========================================================
    # FINAL FLAGS
    # ========================================================

    report[
        "jira_count_ok"
    ] = (

        len(jira_records)
        ==
        inputs[
            "expected_tickets"
        ]
    )


    report[
        "prd_count_ok"
    ] = (

        len(prd_records)
        ==
        len(
            inputs[
                "prd_files"
            ]
        )
    )


    report[
        "meeting_count_ok"
    ] = (

        len(meeting_records)
        ==
        len(
            inputs[
                "meeting_files"
            ]
        )
    )


    report[
        "schema_ok"
    ] = (

        len(
            report[
                "missing_sections"
            ]
        )
        == 0
    )


    report[
        "project_instance_ok"
    ] = (

        len(
            report[
                "wrong_project_records"
            ]
        )
        == 0

        and

        len(
            report[
                "wrong_instance_records"
            ]
        )
        == 0
    )


    report[
        "temporal_ok"
    ] = (

        len(
            report[
                "unparseable_event_times"
            ]
        )
        == 0

        and

        len(
            report[
                "outside_window_events"
            ]
        )
        == 0

        and

        len(
            report[
                "artifact_timestamp_mismatches"
            ]
        )
        == 0
    )


    report[
        "event_vocabulary_ok"
    ] = (

        report[
            "other_change_events"
        ]
        == 0
    )


    report[
        "success"
    ] = all([

        report[
            "jira_count_ok"
        ],

        report[
            "prd_count_ok"
        ],

        report[
            "meeting_count_ok"
        ],

        report[
            "schema_ok"
        ],

        report[
            "project_instance_ok"
        ],

        report[
            "temporal_ok"
        ],

        report[
            "event_vocabulary_ok"
        ]
    ])


    return report


print(
    "SUCCESS: Source-aware production validator defined."
)
```

```python
# ============================================================
# STEP 198: REVALIDATE FOUR TEMPORAL-FAILURE PROJECTS
# ============================================================
# Purpose:
# Reuse existing saved extraction files.
#
# NO GLiNER rerun.
# NO Jira extraction rerun.
# ============================================================


TEMPORAL_RECHECK_PROJECTS = [

    "11_OSSRH",
    "15_MONGOID",
    "21_JRACLOUD",
    "26_JSDCLOUD"
]


temporal_recheck_results = {}


for repository in (
    TEMPORAL_RECHECK_PROJECTS
):

    inputs = next(

        item

        for item in production_inputs

        if item[
            "repository_folder"
        ] == repository
    )


    output_folder = (
        inputs[
            "output_folder"
        ]
    )


    jira_records = (
        load_jsonl_records(

            output_folder
            / JIRA_OUTPUT_NAME
        )
    )


    prd_records = (
        load_jsonl_records(

            output_folder
            / PRD_OUTPUT_NAME
        )
    )


    meeting_records = (
        load_jsonl_records(

            output_folder
            / MEETING_OUTPUT_NAME
        )
    )


    report = (
        validate_production_instance(

            inputs,

            jira_records,

            prd_records,

            meeting_records
        )
    )


    temporal_recheck_results[
        repository
    ] = report


    # --------------------------------------------------------
    # Replace old validation report with corrected validation.
    # --------------------------------------------------------

    with open(
        output_folder
        / VALIDATION_OUTPUT_NAME,
        "w",
        encoding="utf-8"
    ) as file:

        json.dump(
            report,
            file,
            indent=2,
            ensure_ascii=False
        )


    print(
        repository,
        "| success:",
        report["success"],
        "| unparseable:",
        len(
            report[
                "unparseable_event_times"
            ]
        ),
        "| outside:",
        len(
            report[
                "outside_window_events"
            ]
        ),
        "| artifact mismatch:",
        len(
            report[
                "artifact_timestamp_mismatches"
            ]
        )
    )
```

```python
# ============================================================
# STEP 199: NORMALIZE INSTANCE WINDOW BOUNDARIES
# ============================================================
# Purpose:
# Treat a manifest end timestamp at exactly 00:00:00 as an
# inclusive calendar-day boundary.
#
# Example:
#
#   SPR manifest:
#       2019-01-14T00:00:00+00:00
#
# should represent:
#
#       through the end of 2019-01-14
#
# rather than excluding activity later that same day.
#
# Other manifests already ending at:
#
#   23:59:59.999999
#
# remain unchanged.
# ============================================================

from datetime import time


def normalize_instance_boundaries(
    start_value,
    end_value
):

    start_dt = parse_timestamp_utc(
        start_value
    )

    end_dt = parse_timestamp_utc(
        end_value
    )


    if (
        end_dt is not None
        and
        end_dt.hour == 0
        and
        end_dt.minute == 0
        and
        end_dt.second == 0
        and
        end_dt.microsecond == 0
    ):

        end_dt = end_dt.replace(

            hour=23,
            minute=59,
            second=59,
            microsecond=999999
        )


    return (
        start_dt,
        end_dt
    )


# ------------------------------------------------------------
# Test SPR
# ------------------------------------------------------------

spr_start_normalized, spr_end_normalized = (
    normalize_instance_boundaries(

        spr_inputs["instance_start"],
        spr_inputs["instance_end"]
    )
)


print("=" * 70)
print("SPR NORMALIZED INSTANCE WINDOW")
print("=" * 70)

print(
    "\nOriginal start:",
    spr_inputs["instance_start"]
)

print(
    "Original end  :",
    spr_inputs["instance_end"]
)

print(
    "\nNormalized start:",
    spr_start_normalized
)

print(
    "Normalized end  :",
    spr_end_normalized
)
```

```python
# ============================================================
# STEP 200: RECHECK SPR ALIGNMENT WITH INCLUSIVE END DATE
# ============================================================


spr_realigned_tickets = []


for ticket in spr_all_tickets:

    activity_times = (
        collect_ticket_activity_times(
            ticket
        )
    )


    has_activity = False


    for raw_timestamp in activity_times:

        timestamp = parse_timestamp_utc(
            raw_timestamp
        )


        if (
            timestamp is not None
            and
            spr_start_normalized
            <= timestamp
            <= spr_end_normalized
        ):

            has_activity = True
            break


    if has_activity:

        spr_realigned_tickets.append(
            ticket
        )


print("=" * 70)
print("SPR ALIGNMENT RECHECK")
print("=" * 70)

print(
    "\nRepository tickets:",
    len(spr_all_tickets)
)

print(
    "Previously selected:",
    47
)

print(
    "Selected after boundary normalization:",
    len(spr_realigned_tickets)
)

print(
    "Expected:",
    spr_inputs["expected_tickets"]
)

print(
    "Difference:",
    len(spr_realigned_tickets)
    - spr_inputs["expected_tickets"]
)


if (
    len(spr_realigned_tickets)
    ==
    spr_inputs["expected_tickets"]
):

    print(
        "\nSUCCESS:"
        "\nSPR's 300-ticket selection is explained "
        "by inclusive end-of-day handling."
    )

else:

    print(
        "\nCHECK REQUIRED:"
        "\nThere is still another SPR selection difference."
    )
```

```python
# ============================================================
# STEP 201: FINAL GENERIC JIRA ALIGNMENT FUNCTION
# ============================================================
# Purpose:
# Select Jira tickets that have activity inside the official
# project-instance window.
#
# IMPORTANT FIX:
# Some instance manifests, such as SPR-I01, represent the final
# calendar day as:
#
#     2019-01-14T00:00:00+00:00
#
# but Jira activity continues later on that same day.
#
# normalize_instance_boundaries() converts an exact-midnight
# end timestamp to:
#
#     2019-01-14T23:59:59.999999+00:00
#
# Other manifests already ending at 23:59:59.999999 remain
# unchanged.
#
# This gives:
#
#     DOCS-I01  -> 167 / 167
#     JBEAP-I01 -> 131 / 131
#     SPR-I01   -> 300 / 300
# ============================================================


def load_aligned_instance_jira(
    inputs
):

    # --------------------------------------------------------
    # Load all sampled Jira tickets for this repository.
    # --------------------------------------------------------

    repository_tickets = []


    with open(
        inputs["jira_file"],
        "r",
        encoding="utf-8"
    ) as file:

        for line in file:

            if line.strip():

                repository_tickets.append(
                    json.loads(line)
                )


    # --------------------------------------------------------
    # Normalize instance boundaries.
    #
    # This is the critical SPR fix.
    # --------------------------------------------------------

    instance_start, instance_end = (
        normalize_instance_boundaries(

            inputs["instance_start"],
            inputs["instance_end"]
        )
    )


    aligned_tickets = []


    # --------------------------------------------------------
    # Keep tickets having at least one activity timestamp
    # inside the selected instance window.
    # --------------------------------------------------------

    for ticket in repository_tickets:

        activity_times = (
            collect_ticket_activity_times(
                ticket
            )
        )


        has_activity = False


        for raw_timestamp in activity_times:

            timestamp = (
                parse_timestamp_utc(
                    raw_timestamp
                )
            )


            if (
                timestamp is not None
                and
                instance_start
                <= timestamp
                <= instance_end
            ):

                has_activity = True
                break


        if has_activity:

            aligned_tickets.append(
                ticket
            )


    return aligned_tickets


print("=" * 70)
print("FINAL GENERIC JIRA ALIGNMENT DEFINED")
print("=" * 70)

print(
    "\nSUCCESS:"
    "\nProduction Jira alignment now uses "
    "inclusive end-of-day boundaries."
)
```

```python
# ============================================================
# STEP 202: FINAL PRODUCTION JIRA EXTRACTION FUNCTION
# ============================================================
# Purpose:
# Final production Jira extractor used across all GC1 projects.
#
# Includes:
#
#   structured entities
#   relationships
#   dependencies
#   blockers
#   status history
#   complete Jira changelog
#   custom-field fallback
#   temporal filtering
#   provenance
#
# IMPORTANT FIX:
#
# Uses normalize_instance_boundaries() so Jira events occurring
# later on a final day represented as midnight are preserved.
# ============================================================


def extract_jira_ticket(
    ticket,
    repository_project,
    instance_id,
    instance_start,
    instance_end,
    source_file_name
):

    # ========================================================
    # A. NORMALIZE INSTANCE BOUNDARIES
    # ========================================================

    instance_start_utc, instance_end_utc = (
        normalize_instance_boundaries(
            instance_start,
            instance_end
        )
    )


    # ========================================================
    # B. CREATE MASTER RECORD
    # ========================================================

    record = create_master_record(

        repository_project=
            repository_project,

        project_key=
            ticket.get("project_key"),

        instance_id=
            instance_id,

        source_type=
            "JIRA",

        source_id=
            ticket.get("issue_key")
    )


    # ========================================================
    # C. PROJECT ENTITY
    # ========================================================

    project_entity = add_entity(

        record,

        "PROJECT",

        ticket.get("project_key"),

        attributes={

            "project_id":
                ticket.get("project_id"),

            "project_name":
                ticket.get("project_name"),

            "source_repository":
                ticket.get("source_repository")
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # D. ISSUE ENTITY
    # ========================================================

    issue_entity = add_entity(

        record,

        "ISSUE",

        ticket.get("issue_key"),

        attributes={

            "issue_id":
                ticket.get("issue_id"),

            "issue_type":
                ticket.get("issue_type"),

            "summary":
                ticket.get("summary"),

            "description":
                ticket.get("description")
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    add_relationship(

        record,

        project_entity,

        "HAS_ISSUE",

        issue_entity,

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # E. PEOPLE
    # ========================================================

    person_fields = [

        (
            "assignee",
            "ASSIGNED_TO"
        ),

        (
            "reporter",
            "REPORTED_BY"
        ),

        (
            "creator",
            "CREATED_BY"
        )
    ]


    for (
        field_name,
        relationship_type
    ) in person_fields:

        person_name = ticket.get(
            field_name
        )


        if not person_name:
            continue


        person_entity = add_entity(

            record,

            "PERSON",

            person_name,

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            issue_entity,

            relationship_type,

            person_entity,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # ========================================================
    # F. COMPONENTS
    # ========================================================

    for component in (
        ticket.get("components")
        or []
    ):

        if isinstance(
            component,
            dict
        ):

            component_name = (
                component.get("name")
            )

            component_id = (
                component.get("id")
            )

        else:

            component_name = str(
                component
            )

            component_id = None


        if not component_name:
            continue


        component_entity = add_entity(

            record,

            "COMPONENT",

            component_name,

            attributes={

                "component_id":
                    component_id
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            issue_entity,

            "HAS_COMPONENT",

            component_entity,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # ========================================================
    # G. ISSUE RELATIONSHIPS / DEPENDENCIES
    # ========================================================

    relation_sources = [

        (
            "blocks",
            "BLOCKS"
        ),

        (
            "blocked_by",
            "BLOCKED_BY"
        ),

        (
            "depends_on",
            "DEPENDS_ON"
        ),

        (
            "other_issue_links",
            "RELATED_TO"
        )
    ]


    for (
        field_name,
        relationship_type
    ) in relation_sources:

        linked_items = (
            ticket.get(field_name)
            or []
        )


        for linked in linked_items:

            if isinstance(
                linked,
                dict
            ):

                linked_key = (

                    linked.get(
                        "issue_key"
                    )

                    or

                    linked.get(
                        "key"
                    )
                )

                linked_attributes = (
                    linked
                )

            else:

                linked_key = linked

                linked_attributes = {}


            if not linked_key:
                continue


            linked_entity = add_entity(

                record,

                "ISSUE",

                linked_key,

                extraction_method=
                    "structured",

                confidence=1.0
            )


            add_relationship(

                record,

                issue_entity,

                relationship_type,

                linked_entity,

                attributes=
                    linked_attributes,

                extraction_method=
                    "structured",

                confidence=1.0
            )


            if relationship_type in [

                "BLOCKS",
                "BLOCKED_BY",
                "DEPENDS_ON"

            ]:

                record[
                    "dependencies"
                ].append({

                    "source_entity":
                        issue_entity,

                    "target_entity":
                        linked_entity,

                    "dependency_type":
                        relationship_type,

                    "dependency_status":
                        None,

                    "created_at":
                        None,

                    "removed_at":
                        None,

                    "extraction_method":
                        "structured",

                    "confidence":
                        1.0
                })


    # ========================================================
    # H. TEMPORAL FIELDS
    # ========================================================

    record["temporal"] = {

        "created_at":
            ticket.get(
                "created_at"
            ),

        "updated_at":
            ticket.get(
                "updated_at"
            ),

        "due_at":
            ticket.get(
                "due_at"
            ),

        "resolved_at":
            ticket.get(
                "resolved_at"
            ),

        "blocked_at":
            ticket.get(
                "blocked_at"
            ),

        "unblocked_at":
            ticket.get(
                "unblocked_at"
            ),

        "valid_from":
            (
                instance_start_utc.isoformat()

                if instance_start_utc

                else None
            ),

        "valid_to":
            (
                instance_end_utc.isoformat()

                if instance_end_utc

                else None
            )
    }


    # ========================================================
    # I. EXPLICIT BLOCKER INFORMATION
    # ========================================================

    if ticket.get(
        "blocked_at"
    ):

        record[
            "blockers"
        ].append({

            "blocked_entity":
                issue_entity,

            "blocker_entity":
                None,

            "blocked_at":
                ticket.get(
                    "blocked_at"
                ),

            "unblocked_at":
                ticket.get(
                    "unblocked_at"
                ),

            "blocking_reason":
                None,

            "blocking_status":
                (
                    "RESOLVED"

                    if ticket.get(
                        "unblocked_at"
                    )

                    else "ACTIVE"
                ),

            "extraction_method":
                "structured",

            "confidence":
                1.0
        })


    # ========================================================
    # J. STATUS HISTORY
    # ========================================================

    for status_event in (
        ticket.get(
            "status_history"
        )
        or []
    ):

        event_time = (
            parse_timestamp_utc(

                status_event.get(
                    "changed_at"
                )
            )
        )


        if not (
            event_time
            and
            instance_start_utc
            <= event_time
            <= instance_end_utc
        ):

            continue


        add_event(

            record,

            event_type=
                "STATUS_CHANGE",

            entity=
                issue_entity,

            event_time=
                status_event.get(
                    "changed_at"
                ),

            actor=
                status_event.get(
                    "changed_by"
                ),

            from_value=
                status_event.get(
                    "from_status"
                ),

            to_value=
                status_event.get(
                    "to_status"
                ),

            attributes={

                "source_structure":
                    "status_history"
            },

            extraction_method=
                "structured"
        )


    # ========================================================
    # K. FULL CHANGELOG
    # ========================================================

    for history in (
        ticket.get(
            "full_changelog"
        )
        or []
    ):

        history_time = (
            parse_timestamp_utc(
                history.get(
                    "created"
                )
            )
        )


        if not (
            history_time
            and
            instance_start_utc
            <= history_time
            <= instance_end_utc
        ):

            continue


        history_author = (
            history.get("author")
            or {}
        )


        if isinstance(
            history_author,
            dict
        ):

            author = (

                history_author.get(
                    "displayName"
                )

                or

                history_author.get(
                    "name"
                )

                or

                history_author.get(
                    "key"
                )
            )

        else:

            author = str(
                history_author
            )


        for change_item in (
            history.get("items")
            or []
        ):

            field_name = str(

                change_item.get(
                    "field"
                )

                or ""

            ).strip()


            if not field_name:
                continue


            mapped_event_type = (
                resolve_jira_event_type(
                    field_name
                )
            )


            is_custom_fallback = (

                mapped_event_type
                ==
                "CUSTOM_FIELD_CHANGE"
            )


            add_event(

                record,

                event_type=
                    mapped_event_type,

                entity=
                    issue_entity,

                event_time=
                    history.get(
                        "created"
                    ),

                actor=
                    author,

                from_value=
                    change_item.get(
                        "fromString"
                    ),

                to_value=
                    change_item.get(
                        "toString"
                    ),

                attributes={

                    "jira_field":
                        field_name,

                    "field_type":
                        change_item.get(
                            "fieldtype"
                        ),

                    "field_id":
                        change_item.get(
                            "fieldId"
                        ),

                    "source_structure":
                        "full_changelog",

                    "custom_field_fallback":
                        is_custom_fallback
                },

                extraction_method=
                    "structured"
            )


    # ========================================================
    # L. COMMENTS INSIDE INSTANCE WINDOW
    # ========================================================

    comments_in_window = []


    for comment in (
        ticket.get("comments")
        or []
    ):

        raw_comment_time = (

            comment.get(
                "created_at"
            )

            or

            comment.get(
                "created"
            )
        )


        comment_time = (
            parse_timestamp_utc(
                raw_comment_time
            )
        )


        if (
            comment_time
            and
            instance_start_utc
            <= comment_time
            <= instance_end_utc
        ):

            comments_in_window.append(
                comment
            )


    # ========================================================
    # M. STRUCTURED ISSUE ATTRIBUTES
    # ========================================================

    record[
        "attributes"
    ] = {

        "issue_type":
            ticket.get(
                "issue_type"
            ),

        "summary":
            ticket.get(
                "summary"
            ),

        "description":
            ticket.get(
                "description"
            ),

        "related_requirement":
            ticket.get(
                "related_requirement"
            ),

        "epic_feature":
            ticket.get(
                "epic_feature"
            ),

        "current_status":
            ticket.get(
                "status"
            ),

        "priority":
            ticket.get(
                "priority"
            ),

        "resolution":
            ticket.get(
                "resolution"
            ),

        "team":
            ticket.get(
                "team"
            ),

        "labels":
            (
                ticket.get(
                    "labels"
                )
                or []
            ),

        "fix_versions":
            (
                ticket.get(
                    "fix_versions"
                )
                or []
            ),

        "affected_versions":
            (
                ticket.get(
                    "affected_versions"
                )
                or []
            ),

        "comments":
            comments_in_window
    }


    # ========================================================
    # N. PROVENANCE
    # ========================================================

    evidence_parts = []


    if ticket.get(
        "summary"
    ):

        evidence_parts.append(
            str(
                ticket[
                    "summary"
                ]
            )
        )


    if ticket.get(
        "description"
    ):

        evidence_parts.append(
            str(
                ticket[
                    "description"
                ]
            )
        )


    record[
        "provenance"
    ] = {

        "source_type":
            "JIRA",

        "source_id":
            ticket.get(
                "issue_key"
            ),

        "instance_id":
            instance_id,

        "chunk_id":
            None,

        "evidence_text":
            "\n".join(
                evidence_parts
            ),

        "file_name":
            source_file_name
    }


    return record


print("=" * 70)
print("FINAL PRODUCTION JIRA EXTRACTOR DEFINED")
print("=" * 70)

print(
    "\nSUCCESS:"
    "\nJira extraction now uses the corrected "
    "inclusive instance boundary."
)
```

```python
# ============================================================
# STEP 203: FINAL SOURCE-AWARE PRODUCTION VALIDATOR
# ============================================================
# Purpose:
# Final validation used across all GC1 project instances.
#
# Temporal rules:
#
# JIRA
# ----
# Jira events must remain inside the selected instance window.
# Midnight end dates are interpreted as inclusive calendar
# days using normalize_instance_boundaries().
#
# PRD / MEETING
# -------------
# Generated artefacts are validated against their official
# generation_manifest schedule rather than being forced into
# the Jira activity window.
#
# This handles:
#
#   OSSRH meeting after Jira window
#   MONGOID meeting after Jira window
#   JRACLOUD human-readable PRD date
#   JSDCLOUD human-readable PRD date
#   SPR midnight end-date boundary
# ============================================================


def validate_production_instance(
    inputs,
    jira_records,
    prd_records,
    meeting_records
):

    all_records = (
        jira_records
        + prd_records
        + meeting_records
    )


    report = {

        "repository":
            inputs[
                "repository_folder"
            ],

        "project_key":
            inputs[
                "repository_key"
            ],

        "instance_id":
            inputs[
                "instance_id"
            ],

        "expected_jira_records":
            inputs[
                "expected_tickets"
            ],

        "jira_records":
            len(
                jira_records
            ),

        "prd_records":
            len(
                prd_records
            ),

        "meeting_records":
            len(
                meeting_records
            ),

        "missing_sections":
            [],

        "wrong_project_records":
            [],

        "wrong_instance_records":
            [],

        "unparseable_event_times":
            [],

        "outside_window_events":
            [],

        "artifact_timestamp_mismatches":
            [],

        "other_change_events":
            0,

        "custom_field_events":
            0
    }


    # ========================================================
    # CORRECTED INSTANCE WINDOW
    # ========================================================

    instance_start, instance_end = (
        normalize_instance_boundaries(

            inputs[
                "instance_start"
            ],

            inputs[
                "instance_end"
            ]
        )
    )


    # ========================================================
    # OFFICIAL PRD / MEETING SCHEDULE
    # ========================================================

    official_artifact_times = (
        get_official_artifact_timestamps(
            inputs
        )
    )


    # ========================================================
    # VALIDATE EACH SOURCE RECORD
    # ========================================================

    for record in all_records:

        source_type = (
            record[
                "source"
            ][
                "source_type"
            ]
        )


        source_id = (
            record[
                "source"
            ][
                "source_id"
            ]
        )


        # ====================================================
        # REQUIRED MASTER SECTIONS
        # ====================================================

        for section in (
            MASTER_REQUIRED_SECTIONS
        ):

            if section not in record:

                report[
                    "missing_sections"
                ].append({

                    "source_id":
                        source_id,

                    "section":
                        section
                })


        # ====================================================
        # PROJECT CONSISTENCY
        # ====================================================

        if (
            record[
                "repository_project"
            ]
            !=
            inputs[
                "repository_folder"
            ]
        ):

            report[
                "wrong_project_records"
            ].append(
                source_id
            )


        # ====================================================
        # INSTANCE CONSISTENCY
        # ====================================================

        if (
            record[
                "instance_id"
            ]
            !=
            inputs[
                "instance_id"
            ]
        ):

            report[
                "wrong_instance_records"
            ].append(
                source_id
            )


        # ====================================================
        # PRD / MEETING SOURCE TIMESTAMP
        # ====================================================

        if source_type in [
            "PRD",
            "MEETING"
        ]:

            raw_source_time = (
                record[
                    "source"
                ].get(
                    "source_timestamp"
                )
            )


            parsed_source_time = (
                parse_timestamp_utc(
                    raw_source_time
                )
            )


            if (
                raw_source_time
                is not None
                and
                parsed_source_time
                is None
            ):

                report[
                    "unparseable_event_times"
                ].append({

                    "source_id":
                        source_id,

                    "event_type":
                        "SOURCE_TIMESTAMP",

                    "event_time":
                        raw_source_time
                })


            official_time = (
                official_artifact_times.get(
                    source_id
                )
            )


            # ------------------------------------------------
            # Compare calendar date only.
            #
            # Different artefacts may represent the same
            # official date with different time-of-day detail.
            # ------------------------------------------------

            if (
                official_time
                is not None

                and
                parsed_source_time
                is not None

                and
                official_time.date()
                !=
                parsed_source_time.date()
            ):

                report[
                    "artifact_timestamp_mismatches"
                ].append({

                    "source_id":
                        source_id,

                    "source_timestamp":
                        raw_source_time,

                    "official_timestamp":
                        official_time.isoformat()
                })


        # ====================================================
        # EVENT VALIDATION
        # ====================================================

        for event in (
            record[
                "events"
            ]
        ):

            event_type = (
                event.get(
                    "event_type"
                )
            )


            # ------------------------------------------------
            # Event vocabulary checks
            # ------------------------------------------------

            if (
                event_type
                == "OTHER_CHANGE"
            ):

                report[
                    "other_change_events"
                ] += 1


            if (
                event_type
                == "CUSTOM_FIELD_CHANGE"
            ):

                report[
                    "custom_field_events"
                ] += 1


            # ------------------------------------------------
            # Timestamp parsing
            # ------------------------------------------------

            raw_time = (
                event.get(
                    "event_time"
                )
            )


            if raw_time is None:
                continue


            parsed_time = (
                parse_timestamp_utc(
                    raw_time
                )
            )


            if parsed_time is None:

                report[
                    "unparseable_event_times"
                ].append({

                    "source_id":
                        source_id,

                    "event_type":
                        event_type,

                    "event_time":
                        raw_time
                })

                continue


            # =================================================
            # JIRA EVENT VALIDATION
            # =================================================

            if source_type == "JIRA":

                if not (
                    instance_start
                    <= parsed_time
                    <= instance_end
                ):

                    report[
                        "outside_window_events"
                    ].append({

                        "source_id":
                            source_id,

                        "event_type":
                            event_type,

                        "event_time":
                            raw_time
                    })


            # =================================================
            # PRD / MEETING EVENT VALIDATION
            # =================================================

            elif source_type in [
                "PRD",
                "MEETING"
            ]:

                official_time = (
                    official_artifact_times.get(
                        source_id
                    )
                )


                if (
                    official_time
                    is not None

                    and
                    parsed_time.date()
                    !=
                    official_time.date()
                ):

                    report[
                        "artifact_timestamp_mismatches"
                    ].append({

                        "source_id":
                            source_id,

                        "event_type":
                            event_type,

                        "event_time":
                            raw_time,

                        "official_timestamp":
                            official_time.isoformat()
                    })


    # ========================================================
    # VALIDATION FLAGS
    # ========================================================

    report[
        "jira_count_ok"
    ] = (

        len(jira_records)
        ==
        inputs[
            "expected_tickets"
        ]
    )


    report[
        "prd_count_ok"
    ] = (

        len(prd_records)
        ==
        len(
            inputs[
                "prd_files"
            ]
        )
    )


    report[
        "meeting_count_ok"
    ] = (

        len(meeting_records)
        ==
        len(
            inputs[
                "meeting_files"
            ]
        )
    )


    report[
        "schema_ok"
    ] = (

        len(
            report[
                "missing_sections"
            ]
        )
        == 0
    )


    report[
        "project_instance_ok"
    ] = (

        len(
            report[
                "wrong_project_records"
            ]
        )
        == 0

        and

        len(
            report[
                "wrong_instance_records"
            ]
        )
        == 0
    )


    report[
        "temporal_ok"
    ] = (

        len(
            report[
                "unparseable_event_times"
            ]
        )
        == 0

        and

        len(
            report[
                "outside_window_events"
            ]
        )
        == 0

        and

        len(
            report[
                "artifact_timestamp_mismatches"
            ]
        )
        == 0
    )


    report[
        "event_vocabulary_ok"
    ] = (

        report[
            "other_change_events"
        ]
        == 0
    )


    # ========================================================
    # FINAL RESULT
    # ========================================================

    report[
        "success"
    ] = all([

        report[
            "jira_count_ok"
        ],

        report[
            "prd_count_ok"
        ],

        report[
            "meeting_count_ok"
        ],

        report[
            "schema_ok"
        ],

        report[
            "project_instance_ok"
        ],

        report[
            "temporal_ok"
        ],

        report[
            "event_vocabulary_ok"
        ]
    ])


    return report


print("=" * 70)
print("FINAL SOURCE-AWARE VALIDATOR DEFINED")
print("=" * 70)

print(
    "\nSUCCESS:"
    "\nProduction validation now supports "
    "inclusive Jira boundaries and official "
    "PRD/meeting schedules."
)
```

```python
# ============================================================
# STEP 207: FINAL ALL-30 SAVED OUTPUT AUDIT
# ============================================================
# Purpose:
# Reload the FINAL files actually stored on Google Drive for
# every one of the 30 project instances.
#
# This verifies the persisted dataset rather than relying on
# old in-memory batch_results.
#
# NO extraction is rerun.
# NO GLiNER is rerun.
# ============================================================


global_audit_rows = []


for inputs in production_inputs:

    repository = inputs[
        "repository_folder"
    ]

    instance_id = inputs[
        "instance_id"
    ]

    output_folder = inputs[
        "output_folder"
    ]


    jira_file = (
        output_folder
        / JIRA_OUTPUT_NAME
    )

    prd_file = (
        output_folder
        / PRD_OUTPUT_NAME
    )

    meeting_file = (
        output_folder
        / MEETING_OUTPUT_NAME
    )


    # --------------------------------------------------------
    # Check files exist.
    # --------------------------------------------------------

    files_exist = all([

        jira_file.exists(),
        prd_file.exists(),
        meeting_file.exists()
    ])


    if not files_exist:

        global_audit_rows.append({

            "repository":
                repository,

            "instance":
                instance_id,

            "expected_jira":
                inputs[
                    "expected_tickets"
                ],

            "jira":
                None,

            "prd":
                None,

            "meetings":
                None,

            "entities":
                None,

            "relationships":
                None,

            "events":
                None,

            "dependencies":
                None,

            "blockers":
                None,

            "actions":
                None,

            "decisions":
                None,

            "custom_events":
                None,

            "other_events":
                None,

            "success":
                False,

            "problem":
                "MISSING_OUTPUT_FILE"
        })

        continue


    # --------------------------------------------------------
    # Reload persisted extraction.
    # --------------------------------------------------------

    jira_records = (
        load_jsonl_records(
            jira_file
        )
    )

    prd_records = (
        load_jsonl_records(
            prd_file
        )
    )

    meeting_records = (
        load_jsonl_records(
            meeting_file
        )
    )


    all_records = (
        jira_records
        + prd_records
        + meeting_records
    )


    # --------------------------------------------------------
    # Re-run FINAL validator.
    # --------------------------------------------------------

    report = (
        validate_production_instance(

            inputs,

            jira_records,

            prd_records,

            meeting_records
        )
    )


    # --------------------------------------------------------
    # Dataset statistics.
    # --------------------------------------------------------

    total_entities = sum(

        len(record["entities"])

        for record in all_records
    )


    total_relationships = sum(

        len(record["relationships"])

        for record in all_records
    )


    total_events = sum(

        len(record["events"])

        for record in all_records
    )


    total_dependencies = sum(

        len(record["dependencies"])

        for record in all_records
    )


    total_blockers = sum(

        len(record["blockers"])

        for record in all_records
    )


    total_actions = sum(

        len(record["action_items"])

        for record in all_records
    )


    total_decisions = sum(

        len(record["decisions"])

        for record in all_records
    )


    custom_events = sum(

        1

        for record in all_records

        for event in record["events"]

        if event.get("event_type")
        == "CUSTOM_FIELD_CHANGE"
    )


    other_events = sum(

        1

        for record in all_records

        for event in record["events"]

        if event.get("event_type")
        == "OTHER_CHANGE"
    )


    global_audit_rows.append({

        "repository":
            repository,

        "instance":
            instance_id,

        "expected_jira":
            inputs[
                "expected_tickets"
            ],

        "jira":
            len(jira_records),

        "prd":
            len(prd_records),

        "meetings":
            len(meeting_records),

        "entities":
            total_entities,

        "relationships":
            total_relationships,

        "events":
            total_events,

        "dependencies":
            total_dependencies,

        "blockers":
            total_blockers,

        "actions":
            total_actions,

        "decisions":
            total_decisions,

        "custom_events":
            custom_events,

        "other_events":
            other_events,

        "success":
            report["success"],

        "problem":
            None
            if report["success"]
            else "VALIDATION_FAILED"
    })


print("=" * 80)
print("GC1 FINAL ALL-30 SAVED OUTPUT AUDIT")
print("=" * 80)


for row in global_audit_rows:

    print(

        f"{row['repository']:20s}",

        "| Jira:",
        row["jira"],
        "/",
        row["expected_jira"],

        "| PRD:",
        row["prd"],

        "| Meetings:",
        row["meetings"],

        "| Success:",
        row["success"]
    )
```

```python
# ============================================================
# STEP 212: FINAL PRODUCTION MEETING EXTRACTOR
# ============================================================
# Purpose:
# Final meeting extractor for GC1.
#
# IMPORTANT TEMPORAL RULE:
#
#   raw transcript date
#       -> preserved as source metadata
#
#   official generation-manifest timestamp
#       -> canonical temporal timestamp used for:
#            source_timestamp
#            temporal.created_at
#            temporal.valid_from
#            relationships
#            DECISION_MADE events
#            dependency/blocker evidence
#
# This prevents stale dates written inside generated text from
# corrupting Temporal KG placement.
# ============================================================


def extract_production_meeting(
    meeting_file,
    repository_project,
    project_key,
    project_name,
    instance_id,
    official_timestamp=None
):

    # ========================================================
    # A. PARSE TRANSCRIPT
    # ========================================================

    meeting = parse_production_meeting(
        meeting_file
    )


    # ========================================================
    # B. RULE-BASED EVIDENCE
    # ========================================================

    meeting = extract_meeting_rule_evidence(
        meeting
    )


    # ========================================================
    # C. GLINER
    # ========================================================

    gliner_entities = (
        extract_production_meeting_gliner(
            meeting
        )
    )


    # ========================================================
    # D. TEMPORAL RESOLUTION
    # ========================================================

    raw_transcript_datetime = (
        meeting[
            "meeting_datetime"
        ]
    )


    parsed_official_time = (
        parse_timestamp_utc(
            official_timestamp
        )
        if official_timestamp
        else None
    )


    # --------------------------------------------------------
    # Official generation timestamp takes precedence.
    #
    # If unavailable, fall back to transcript timestamp.
    # --------------------------------------------------------

    if parsed_official_time:

        canonical_meeting_time = (
            parsed_official_time.isoformat()
        )

    else:

        canonical_meeting_time = (
            raw_transcript_datetime
        )


    # ========================================================
    # E. MASTER RECORD
    # ========================================================

    record = create_master_record(

        repository_project=
            repository_project,

        project_key=
            project_key,

        instance_id=
            instance_id,

        source_type=
            "MEETING",

        source_id=
            meeting["meeting_id"]
    )


    record[
        "source"
    ][
        "source_timestamp"
    ] = canonical_meeting_time


    record[
        "temporal"
    ][
        "created_at"
    ] = canonical_meeting_time


    record[
        "temporal"
    ][
        "valid_from"
    ] = canonical_meeting_time


    # ========================================================
    # F. ATTRIBUTES
    # ========================================================

    record[
        "attributes"
    ].update({

        "meeting_title":
            meeting[
                "meeting_title"
            ],

        "participants":
            meeting[
                "participants"
            ],

        "utterance_count":
            len(
                meeting[
                    "utterances"
                ]
            ),

        # -----------------------------------------------
        # Preserve BOTH temporal representations.
        # -----------------------------------------------

        "transcript_datetime_raw":
            raw_transcript_datetime,

        "official_generation_timestamp":
            (
                parsed_official_time.isoformat()

                if parsed_official_time

                else None
            ),

        "canonical_meeting_timestamp":
            canonical_meeting_time,

        "timestamp_source":
            (
                "generation_manifest"

                if parsed_official_time

                else "transcript"
            )
    })


    # ========================================================
    # G. PROJECT ENTITY
    # ========================================================

    project_entity = add_entity(

        record,

        "PROJECT",

        project_key,

        attributes={

            "project_name":
                project_name,

            "instance_id":
                instance_id
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # H. MEETING ENTITY
    # ========================================================

    meeting_entity = add_entity(

        record,

        "MEETING",

        meeting["meeting_id"],

        attributes={

            "title":
                meeting[
                    "meeting_title"
                ],

            "meeting_datetime":
                canonical_meeting_time,

            "transcript_datetime_raw":
                raw_transcript_datetime
        },

        extraction_method=
            "structured",

        confidence=1.0
    )


    add_relationship(

        record,

        project_entity,

        "HAS_MEETING",

        meeting_entity,

        timestamp=
            canonical_meeting_time,

        extraction_method=
            "structured",

        confidence=1.0
    )


    # ========================================================
    # I. PARTICIPANTS
    # ========================================================

    for participant in meeting[
        "participants"
    ]:

        person_entity = add_entity(

            record,

            "PERSON",

            participant[
                "name"
            ],

            attributes={

                "role":
                    participant[
                        "role"
                    ]
            },

            extraction_method=
                "structured",

            confidence=1.0
        )


        add_relationship(

            record,

            person_entity,

            "PARTICIPATED_IN",

            meeting_entity,

            timestamp=
                canonical_meeting_time,

            extraction_method=
                "structured",

            confidence=1.0
        )


    # ========================================================
    # J. EXPLICIT JIRA ISSUE REFERENCES
    # ========================================================

    unique_issues = sorted(
        set(

            item[
                "issue_key"
            ]

            for item
            in meeting[
                "issue_occurrences"
            ]
        )
    )


    for issue_key in unique_issues:

        issue_entity = add_entity(

            record,

            "ISSUE",

            issue_key,

            extraction_method=
                "regex",

            confidence=1.0
        )


        add_relationship(

            record,

            meeting_entity,

            "REFERENCES_ISSUE",

            issue_entity,

            timestamp=
                canonical_meeting_time,

            extraction_method=
                "regex",

            confidence=1.0
        )


    # ========================================================
    # K. REQUIREMENT REFERENCES
    # ========================================================

    unique_requirements = sorted(
        set(

            item[
                "requirement_id"
            ]

            for item
            in meeting[
                "requirement_occurrences"
            ]
        )
    )


    for requirement_id in (
        unique_requirements
    ):

        requirement_entity = add_entity(

            record,

            "REQUIREMENT",

            requirement_id,

            extraction_method=
                "regex",

            confidence=1.0
        )


        add_relationship(

            record,

            meeting_entity,

            "REFERENCES_REQUIREMENT",

            requirement_entity,

            timestamp=
                canonical_meeting_time,

            extraction_method=
                "regex",

            confidence=1.0
        )


    # ========================================================
    # L. GLINER SEMANTIC ENTITIES
    # ========================================================

    for prediction in gliner_entities:

        master_type = (
            GLINER_TO_MASTER_TYPE[
                prediction[
                    "label"
                ]
            ]
        )


        semantic_entity = add_entity(

            record,

            master_type,

            prediction[
                "text"
            ],

            attributes={

                "utterance_index":
                    prediction[
                        "utterance_index"
                    ],

                "speaker":
                    prediction[
                        "speaker"
                    ],

                "speaker_timestamp":
                    prediction[
                        "speaker_timestamp"
                    ]
            },

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ]
        )


        add_relationship(

            record,

            meeting_entity,

            "MENTIONS",

            semantic_entity,

            timestamp=
                canonical_meeting_time,

            extraction_method=
                "gliner",

            confidence=
                prediction[
                    "confidence"
                ],

            attributes={

                "speaker":
                    prediction[
                        "speaker"
                    ],

                "speaker_timestamp":
                    prediction[
                        "speaker_timestamp"
                    ],

                "evidence_text":
                    prediction[
                        "evidence_text"
                    ]
            }
        )


    # ========================================================
    # M. ACTION ITEMS
    # ========================================================

    for index, candidate in enumerate(

        meeting[
            "final_action_candidates"
        ],

        start=1
    ):

        record[
            "action_items"
        ].append({

            "action_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_ACTION_{index:03d}"
                ),

            "action_text":
                candidate[
                    "text"
                ],

            "action_owner":
                candidate[
                    "speaker"
                ],

            "meeting_time":
                canonical_meeting_time,

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "due_at":
                None,

            "completed_at":
                None,

            "action_status":
                "OPEN",

            "is_recap":
                (
                    "recap"
                    in candidate[
                        "text"
                    ].lower()
                ),

            "extraction_method":
                "rule",

            "evidence_text":
                candidate[
                    "text"
                ]
        })


    # ========================================================
    # N. DECISIONS
    # ========================================================

    for index, candidate in enumerate(

        meeting[
            "decision_candidates"
        ],

        start=1
    ):

        decision_id = (

            f"{meeting['meeting_id']}"
            f"_DECISION_{index:03d}"
        )


        record[
            "decisions"
        ].append({

            "decision_id":
                decision_id,

            "decision_text":
                candidate[
                    "text"
                ],

            "decision_maker":
                candidate[
                    "speaker"
                ],

            "decision_time":
                canonical_meeting_time,

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "decision_state":
                candidate[
                    "decision_state"
                ],

            "decision_target":
                None,

            "reversed_decision":
                None,

            "extraction_method":
                "rule",

            "evidence_text":
                candidate[
                    "text"
                ]
        })


        # ----------------------------------------------------
        # Only MADE decisions become temporal decision events.
        # ----------------------------------------------------

        if (
            candidate[
                "decision_state"
            ]
            == "MADE"
        ):

            add_event(

                record,

                event_type=
                    "DECISION_MADE",

                entity=
                    meeting_entity,

                event_time=
                    canonical_meeting_time,

                actor=
                    candidate[
                        "speaker"
                    ],

                from_value=
                    None,

                to_value=
                    candidate[
                        "text"
                    ],

                attributes={

                    "speaker_timestamp":
                        candidate[
                            "speaker_timestamp"
                        ],

                    "decision_id":
                        decision_id
                },

                extraction_method=
                    "rule"
            )


    # ========================================================
    # O. DEPENDENCIES
    # ========================================================

    for index, candidate in enumerate(

        meeting[
            "dependency_candidates"
        ],

        start=1
    ):

        record[
            "dependencies"
        ].append({

            "dependency_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_DEP_{index:03d}"
                ),

            "text":
                candidate[
                    "text"
                ],

            "source_entity":
                None,

            "target_entity":
                None,

            "dependency_type":
                "MEETING_DEPENDENCY",

            "dependency_status":
                candidate[
                    "dependency_state"
                ],

            "created_at":
                canonical_meeting_time,

            "speaker":
                candidate[
                    "speaker"
                ],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "extraction_method":
                "rule",

            "evidence_text":
                candidate[
                    "text"
                ]
        })


    # ========================================================
    # P. BLOCKER EVIDENCE
    # ========================================================

    for index, candidate in enumerate(

        meeting[
            "blocker_candidates"
        ],

        start=1
    ):

        record[
            "blockers"
        ].append({

            "blocker_id":
                (
                    f"{meeting['meeting_id']}"
                    f"_BLOCKER_{index:03d}"
                ),

            "blocked_entity":
                None,

            "blocker_entity":
                None,

            "blocker_state":
                candidate[
                    "blocker_state"
                ],

            "is_active_assertion":
                (
                    candidate[
                        "blocker_state"
                    ]
                    == "ACTIVE"
                ),

            "is_recap":
                (
                    "recap"
                    in candidate[
                        "text"
                    ].lower()
                ),

            "meeting_time":
                canonical_meeting_time,

            "speaker":
                candidate[
                    "speaker"
                ],

            "speaker_timestamp":
                candidate[
                    "speaker_timestamp"
                ],

            "blocking_reason":
                None,

            "extraction_method":
                "rule",

            "evidence_text":
                candidate[
                    "text"
                ]
        })


    # ========================================================
    # Q. DIAGNOSTICS
    # ========================================================

    record[
        "attributes"
    ][
        "extraction_diagnostics"
    ] = {

        "participant_count":
            len(
                meeting[
                    "participants"
                ]
            ),

        "utterance_count":
            len(
                meeting[
                    "utterances"
                ]
            ),

        "unique_issue_count":
            len(
                unique_issues
            ),

        "unique_requirement_count":
            len(
                unique_requirements
            ),

        "action_count":
            len(
                record[
                    "action_items"
                ]
            ),

        "decision_count":
            len(
                record[
                    "decisions"
                ]
            ),

        "dependency_count":
            len(
                record[
                    "dependencies"
                ]
            ),

        "blocker_evidence_count":
            len(
                record[
                    "blockers"
                ]
            ),

        "gliner_occurrence_count":
            len(
                gliner_entities
            )
    }


    # ========================================================
    # R. PROVENANCE
    # ========================================================

    record[
        "provenance"
    ] = {

        "source_type":
            "MEETING",

        "source_id":
            meeting[
                "meeting_id"
            ],

        "instance_id":
            instance_id,

        "chunk_id":
            None,

        "evidence_text":
            meeting[
                "text"
            ],

        "file_name":
            meeting_file.name,

        # Preserve original displayed date as provenance.
        "transcript_datetime_raw":
            raw_transcript_datetime,

        "official_generation_timestamp":
            (
                parsed_official_time.isoformat()

                if parsed_official_time

                else None
            )
    }


    return record


print(
    "SUCCESS: Final production meeting extractor defined."
)
```

```python
# ============================================================
# STEP 213: FINAL PRODUCTION MEETING RUNNER
# ============================================================
# Purpose:
# Process official meetings using their official generation
# timestamps.
# ============================================================


def run_production_meetings(
    inputs
):

    records = []


    official_times = (
        get_official_artifact_timestamps(
            inputs
        )
    )


    for meeting_file in (
        inputs[
            "meeting_files"
        ]
    ):

        meeting_id = (
            meeting_file.stem
        )


        official_time = (
            official_times.get(
                meeting_id
            )
        )


        record = (
            extract_production_meeting(

                meeting_file=
                    meeting_file,

                repository_project=
                    inputs[
                        "repository_folder"
                    ],

                project_key=
                    inputs[
                        "repository_key"
                    ],

                project_name=
                    inputs[
                        "project_name"
                    ],

                instance_id=
                    inputs[
                        "instance_id"
                    ],

                official_timestamp=
                    official_time
            )
        )


        records.append(
            record
        )


    return records


print(
    "SUCCESS: Final production meeting runner defined."
)
```

```python
# ============================================================
# STEP 214: REGENERATE JBEAP MEETINGS ONLY
# ============================================================
# Purpose:
# Correct JBEAP meeting temporal placement without rerunning
# Jira or PRD extraction.
# ============================================================


jbeap_inputs = next(

    item

    for item in production_inputs

    if item[
        "repository_folder"
    ] == "01_JBEAP"
)


corrected_jbeap_meetings = (
    run_production_meetings(
        jbeap_inputs
    )
)


jbeap_meeting_output = (

    jbeap_inputs[
        "output_folder"
    ]

    / MEETING_OUTPUT_NAME
)


save_jsonl(

    corrected_jbeap_meetings,

    jbeap_meeting_output
)


print("=" * 70)
print("JBEAP MEETINGS CORRECTED")
print("=" * 70)


print(
    "\nRecords:",
    len(
        corrected_jbeap_meetings
    )
)


for record in corrected_jbeap_meetings:

    print(
        record[
            "source"
        ][
            "source_id"
        ],
        "| canonical:",
        record[
            "source"
        ][
            "source_timestamp"
        ],
        "| transcript:",
        record[
            "attributes"
        ][
            "transcript_datetime_raw"
        ],
        "| source:",
        record[
            "attributes"
        ][
            "timestamp_source"
        ]
    )
```

```python
# ============================================================
# STEP 215: FINAL JBEAP REVALIDATION
# ============================================================


jbeap_output_folder = (
    jbeap_inputs[
        "output_folder"
    ]
)


jbeap_jira_final = (
    load_jsonl_records(

        jbeap_output_folder
        / JIRA_OUTPUT_NAME
    )
)


jbeap_prd_final = (
    load_jsonl_records(

        jbeap_output_folder
        / PRD_OUTPUT_NAME
    )
)


jbeap_meetings_final = (
    load_jsonl_records(

        jbeap_output_folder
        / MEETING_OUTPUT_NAME
    )
)


jbeap_final_report = (
    validate_production_instance(

        jbeap_inputs,

        jbeap_jira_final,

        jbeap_prd_final,

        jbeap_meetings_final
    )
)


print("=" * 75)
print("JBEAP FINAL REVALIDATION")
print("=" * 75)


for key in [

    "jira_count_ok",
    "prd_count_ok",
    "meeting_count_ok",
    "schema_ok",
    "project_instance_ok",
    "temporal_ok",
    "event_vocabulary_ok"

]:

    print(
        f"{key:30s}",
        jbeap_final_report[
            key
        ]
    )


print(
    "\nArtifact timestamp mismatches:",
    len(
        jbeap_final_report[
            "artifact_timestamp_mismatches"
        ]
    )
)


print(
    "Unparseable timestamps:",
    len(
        jbeap_final_report[
            "unparseable_event_times"
        ]
    )
)


print(
    "Outside-window events:",
    len(
        jbeap_final_report[
            "outside_window_events"
        ]
    )
)


print(
    "\nFINAL SUCCESS:",
    jbeap_final_report[
        "success"
    ]
)
```
